```python
You are the ML/RAG engineer for the SAATHI — BIS Navigator project.

IMPORTANT:
The following BIS datasets have already been uploaded to this Kaggle notebook:

1. bis_standards_master(1).csv
2. bis_all_standards_authored_details.csv
3. bis_cross_references(1).csv
4. bis_indian_references(1).csv
5. bis_international_references(1).csv
6. bis_referred_by(1).csv
7. bis_failed_records(1).csv

We are NOT training a neural network yet.

Our goal is to build a production-oriented RAG system where an LLM can answer BIS questions using these datasets as authoritative knowledge and perform multi-step reasoning over standards and their relationships.

FIRST TASK — DATASET INGESTION AND STRUCTURAL AUDIT ONLY.

Write and execute Python code that:

1. Locate these CSV files inside /kaggle/input recursively.

2. Create a dictionary:

DATASET_PATHS = {
    "standards_master": "...",
    "authored_details": "...",
    "cross_references": "...",
    "indian_references": "...",
    "international_references": "...",
    "referred_by": "...",
    "failed_records": "..."
}

using the actual discovered paths.

3. Load all seven datasets into pandas DataFrames using sensible memory-efficient settings.

Use these names:

standards_master
authored_details
cross_references
indian_references
international_references
referred_by
failed_records

4. Print for every dataset:

- shape
- column names
- data types
- memory usage
- null count per column
- duplicate row count

5. Specifically analyze the primary identifier:

internal_id

For every dataset containing internal_id:

- number of unique internal_id values
- number of null internal_id values
- number of duplicated internal_id values
- percentage of records with a valid internal_id

6. Check whether internal_id is a reliable joining key between:

standards_master
authored_details
cross_references
indian_references
international_references
referred_by

Calculate:

- internal IDs present in master
- internal IDs missing from master
- internal IDs shared between datasets
- percentage of each child dataset whose internal_id exists in standards_master

7. For standards_master specifically, inspect:

IS_number
IS_title
superseding_IS
degree_of_equivalence
aspect
language
technical_department
technical_committee
member_secretary
group
sub_group
sub_sub_group
certification
catalogue_withdrawn_status
detail_url
download_url
composition_url

Report null percentages and unique counts.

8. For authored_details:

- calculate how many records have at least one non-empty line_1 through line_10
- calculate how many have all ten lines empty
- calculate the average number of populated detail lines
- show 5 representative examples

9. For every reference dataset:

cross_references
indian_references
international_references
referred_by

show:

- number of rows
- number of unique internal_ids
- number of unique reference_number values
- null counts
- 5 representative rows

10. For failed_records:

DO NOT use these records as authoritative BIS knowledge.

Instead report:

- total failed records
- unique internal_ids
- error frequency by error type/message
- 10 most common errors

11. Check for duplicate BIS standards.

Specifically inspect duplicates based on:

IS_number
internal_id

Do NOT delete anything.

12. Check whether the master dataset contains multiple records representing the same IS_number.

Report examples if found.

13. Save the resulting audit information into a Python dictionary called DATA_AUDIT.

14. Do not create embeddings.
15. Do not train a model.
16. Do not create a vector database.
17. Do not modify the source CSV files.
18. Do not fabricate missing information.

At the end, print a concise section:

=============================
SAATHI DATA AUDIT COMPLETE
=============================

Include:

Total master standards:
Total authored-detail records:
Total cross references:
Total Indian references:
Total international references:
Total referred-by records:
Total failed records:

Master internal_id uniqueness:
Master IS_number uniqueness:

Child-record linkage percentages:

Then stop.
```

```python
import os
import glob
import pandas as pd
import numpy as np
import json

print("=" * 70)
print("SAATHI - BIS DATASET DISCOVERY")
print("=" * 70)

# Find all CSV files inside Kaggle input
csv_files = glob.glob("/kaggle/input/**/*.csv", recursive=True)

print(f"\nCSV files discovered: {len(csv_files)}\n")

for path in csv_files:
    print(f"- {path}")

# Expected datasets
expected_files = {
    "standards_master": "bis_standards_master",
    "authored_details": "bis_all_standards_authored_details",
    "cross_references": "bis_cross_references",
    "indian_references": "bis_indian_references",
    "international_references": "bis_international_references",
    "referred_by": "bis_referred_by",
    "failed_records": "bis_failed_records"
}

DATASET_PATHS = {}

print("\n" + "=" * 70)
print("MATCHING EXPECTED DATASETS")
print("=" * 70)

for logical_name, filename_fragment in expected_files.items():

    matches = [
        path for path in csv_files
        if filename_fragment.lower() in os.path.basename(path).lower()
    ]

    if matches:
        DATASET_PATHS[logical_name] = matches[0]
        print(f"\n[FOUND] {logical_name}")
        print(f"Path: {matches[0]}")
    else:
        print(f"\n[MISSING] {logical_name}")

print("\n" + "=" * 70)
print("DATASET SUMMARY")
print("=" * 70)

for logical_name, path in DATASET_PATHS.items():

    try:
        df_sample = pd.read_csv(path, nrows=5)

        # Get actual row count without loading entire dataset
        with open(path, "rb") as f:
            row_count = sum(1 for _ in f) - 1

        file_size_mb = os.path.getsize(path) / (1024 * 1024)

        print(f"\n{logical_name}")
        print("-" * 50)
        print(f"File       : {os.path.basename(path)}")
        print(f"Rows       : {row_count:,}")
        print(f"Columns    : {len(df_sample.columns)}")
        print(f"Size       : {file_size_mb:.2f} MB")
        print(f"Columns    : {list(df_sample.columns)}")

        print("\nSample:")
        display(df_sample.head(2))

    except Exception as e:
        print(f"ERROR reading {path}")
        print(str(e))

print("\n" + "=" * 70)
print("DATASET PATHS")
print("=" * 70)

for key, value in DATASET_PATHS.items():
    print(f"{key}: {value}")

print("\n" + "=" * 70)
print("CELL 1 COMPLETE")
print("=" * 70)
```

```python
import pandas as pd
import numpy as np
import json
import os

print("=" * 80)
print("SAATHI - CELL 2: FULL BIS DATA INTEGRITY AUDIT")
print("=" * 80)

# ============================================================
# 1. LOAD DATASETS
# ============================================================

print("\n[1/8] Loading datasets...")

standards_master = pd.read_csv(
    DATASET_PATHS["standards_master"],
    low_memory=False
)

authored_details = pd.read_csv(
    DATASET_PATHS["authored_details"],
    low_memory=False
)

cross_references = pd.read_csv(
    DATASET_PATHS["cross_references"],
    low_memory=False
)

indian_references = pd.read_csv(
    DATASET_PATHS["indian_references"],
    low_memory=False
)

international_references = pd.read_csv(
    DATASET_PATHS["international_references"],
    low_memory=False
)

referred_by = pd.read_csv(
    DATASET_PATHS["referred_by"],
    low_memory=False
)

failed_records = pd.read_csv(
    DATASET_PATHS["failed_records"],
    low_memory=False
)

datasets = {
    "standards_master": standards_master,
    "authored_details": authored_details,
    "cross_references": cross_references,
    "indian_references": indian_references,
    "international_references": international_references,
    "referred_by": referred_by,
    "failed_records": failed_records
}

print("All datasets loaded successfully.")


# ============================================================
# 2. BASIC DATASET STATISTICS
# ============================================================

print("\n[2/8] Basic statistics")
print("-" * 80)

for name, df in datasets.items():

    memory_mb = df.memory_usage(deep=True).sum() / (1024 ** 2)

    print(
        f"{name:30s} "
        f"rows={len(df):>9,}  "
        f"columns={len(df.columns):>3}  "
        f"memory={memory_mb:>8.2f} MB"
    )


# ============================================================
# 3. NULL / EMPTY VALUE ANALYSIS
# ============================================================

print("\n[3/8] Null-value analysis")
print("-" * 80)

DATA_AUDIT = {}

for name, df in datasets.items():

    null_counts = df.isna().sum()

    # Treat empty strings and whitespace as empty values too
    empty_counts = {}

    for col in df.columns:
        if df[col].dtype == "object":
            empty_counts[col] = (
                df[col]
                .fillna("")
                .astype(str)
                .str.strip()
                .eq("")
                .sum()
            )
        else:
            empty_counts[col] = 0

    DATA_AUDIT[name] = {
        "rows": int(len(df)),
        "columns": int(len(df.columns)),
        "null_counts": null_counts.to_dict(),
        "empty_counts": empty_counts
    }

    print(f"\n{name}")

    important_nulls = (
        null_counts[null_counts > 0]
        .sort_values(ascending=False)
        .head(15)
    )

    if len(important_nulls) == 0:
        print("  No NULL values.")
    else:
        for col, count in important_nulls.items():
            pct = count / len(df) * 100
            print(f"  {col:35s}: {count:>8,} ({pct:6.2f}%)")


# ============================================================
# 4. DUPLICATE ANALYSIS
# ============================================================

print("\n[4/8] Duplicate analysis")
print("-" * 80)

for name, df in datasets.items():

    duplicate_rows = int(df.duplicated().sum())

    DATA_AUDIT[name]["duplicate_rows"] = duplicate_rows

    print(
        f"{name:30s}: "
        f"{duplicate_rows:,} completely duplicated rows"
    )


# ============================================================
# 5. INTERNAL_ID ANALYSIS
# ============================================================

print("\n[5/8] internal_id integrity analysis")
print("-" * 80)

internal_id_stats = {}

for name, df in datasets.items():

    if "internal_id" not in df.columns:
        continue

    total = len(df)

    null_count = int(df["internal_id"].isna().sum())

    unique_count = int(df["internal_id"].nunique(dropna=True))

    duplicated_values = int(
        df["internal_id"]
        .dropna()
        .duplicated()
        .sum()
    )

    valid_percentage = (
        ((total - null_count) / total) * 100
        if total > 0 else 0
    )

    internal_id_stats[name] = {
        "total": total,
        "null": null_count,
        "unique": unique_count,
        "duplicated_values": duplicated_values,
        "valid_percentage": valid_percentage
    }

    print(f"\n{name}")
    print(f"  Total records       : {total:,}")
    print(f"  NULL internal_id   : {null_count:,}")
    print(f"  Unique internal_id : {unique_count:,}")
    print(f"  Duplicate values   : {duplicated_values:,}")
    print(f"  Valid percentage   : {valid_percentage:.2f}%")

DATA_AUDIT["internal_id"] = internal_id_stats


# ============================================================
# 6. MASTER LINKAGE ANALYSIS
# ============================================================

print("\n[6/8] Master dataset linkage analysis")
print("-" * 80)

master_ids = set(
    standards_master["internal_id"]
    .dropna()
    .astype(str)
)

linkage_results = {}

child_datasets = [
    "authored_details",
    "cross_references",
    "indian_references",
    "international_references",
    "referred_by"
]

for name in child_datasets:

    df = datasets[name]

    if "internal_id" not in df.columns:
        continue

    child_ids = set(
        df["internal_id"]
        .dropna()
        .astype(str)
    )

    matched_ids = child_ids.intersection(master_ids)

    unmatched_ids = child_ids - master_ids

    row_matches = (
        df["internal_id"]
        .dropna()
        .astype(str)
        .isin(master_ids)
        .sum()
    )

    valid_rows = df["internal_id"].notna().sum()

    match_percentage = (
        row_matches / valid_rows * 100
        if valid_rows > 0 else 0
    )

    linkage_results[name] = {
        "unique_child_ids": len(child_ids),
        "matched_unique_ids": len(matched_ids),
        "unmatched_unique_ids": len(unmatched_ids),
        "matched_rows": int(row_matches),
        "valid_rows": int(valid_rows),
        "match_percentage": match_percentage
    }

    print(f"\n{name}")
    print(f"  Unique child IDs      : {len(child_ids):,}")
    print(f"  Matched master IDs    : {len(matched_ids):,}")
    print(f"  Unmatched master IDs  : {len(unmatched_ids):,}")
    print(f"  Matched rows          : {row_matches:,}")
    print(f"  Valid rows            : {valid_rows:,}")
    print(f"  Linkage percentage    : {match_percentage:.2f}%")

DATA_AUDIT["linkage"] = linkage_results


# ============================================================
# 7. BIS NUMBER DUPLICATE ANALYSIS
# ============================================================

print("\n[7/8] BIS number integrity analysis")
print("-" * 80)

if "IS_number" in standards_master.columns:

    is_numbers = (
        standards_master["IS_number"]
        .dropna()
        .astype(str)
        .str.strip()
    )

    duplicate_is_numbers = (
        is_numbers[is_numbers.duplicated(keep=False)]
        .value_counts()
    )

    unique_is_numbers = is_numbers.nunique()

    print(f"Non-null IS numbers : {len(is_numbers):,}")
    print(f"Unique IS numbers   : {unique_is_numbers:,}")
    print(
        f"IS numbers appearing multiple times: "
        f"{len(duplicate_is_numbers):,}"
    )

    if len(duplicate_is_numbers) > 0:

        print("\nExamples of duplicated IS numbers:")

        duplicate_values = duplicate_is_numbers.head(20)

        for value, count in duplicate_values.items():
            print(f"  {value} -> {count} records")

    DATA_AUDIT["IS_number"] = {
        "non_null": int(len(is_numbers)),
        "unique": int(unique_is_numbers),
        "duplicated_values": int(len(duplicate_is_numbers))
    }


# ============================================================
# 8. AUTHORED DETAILS ANALYSIS
# ============================================================

print("\n[8/8] Authored-details analysis")
print("-" * 80)

detail_columns = [
    f"line_{i}"
    for i in range(1, 11)
    if f"line_{i}" in authored_details.columns
]

if detail_columns:

    detail_df = authored_details[detail_columns].fillna("")

    populated_counts = (
        detail_df.astype(str)
        .apply(lambda col: col.str.strip().ne(""))
        .sum(axis=1)
    )

    completely_empty = int((populated_counts == 0).sum())

    average_lines = float(populated_counts.mean())

    fully_populated = int((populated_counts == len(detail_columns)).sum())

    print(f"Detail columns              : {len(detail_columns)}")
    print(f"Records with no details     : {completely_empty:,}")
    print(f"Records with all details    : {fully_populated:,}")
    print(f"Average populated lines     : {average_lines:.2f}")

    DATA_AUDIT["authored_details_analysis"] = {
        "detail_columns": detail_columns,
        "empty_records": completely_empty,
        "fully_populated_records": fully_populated,
        "average_populated_lines": average_lines
    }

    print("\nExample authored records:")

    display(
        authored_details[
            ["internal_id", "is_number", "is_title"] + detail_columns
        ].head(3)
    )


# ============================================================
# FINAL SUMMARY
# ============================================================

print("\n")
print("=" * 80)
print("SAATHI DATA INTEGRITY AUDIT COMPLETE")
print("=" * 80)

print(f"\nTotal master standards       : {len(standards_master):,}")
print(f"Total authored details       : {len(authored_details):,}")
print(f"Total cross references       : {len(cross_references):,}")
print(f"Total Indian references     : {len(indian_references):,}")
print(f"Total international refs    : {len(international_references):,}")
print(f"Total referred-by records   : {len(referred_by):,}")
print(f"Total failed records        : {len(failed_records):,}")

print("\nMaster integrity:")

if "internal_id" in internal_id_stats:
    master_internal = internal_id_stats["standards_master"]

    print(
        f"  internal_id unique       : "
        f"{master_internal['unique']:,}"
    )

if "IS_number" in DATA_AUDIT:
    print(
        f"  IS_number unique         : "
        f"{DATA_AUDIT['IS_number']['unique']:,}"
    )

print("\nChild dataset linkage:")

for name, result in linkage_results.items():
    print(
        f"  {name:25s}: "
        f"{result['match_percentage']:.2f}% linked"
    )

print("\nDATA_AUDIT object created successfully.")

print("\nNext step: use this audit before any cleaning or embedding.")
```

```python
# ============================================================
# SAATHI - CELL 3
# CANONICAL BIS KNOWLEDGE RECORD CONSTRUCTION
# ============================================================

import pandas as pd
import numpy as np
import json
import os
import re
from collections import defaultdict

print("=" * 80)
print("SAATHI - CELL 3: BUILDING CANONICAL BIS KNOWLEDGE RECORDS")
print("=" * 80)


# ============================================================
# 1. HELPER FUNCTIONS
# ============================================================

def clean_value(value):
    """
    Convert pandas values into clean JSON-safe values.
    """
    if pd.isna(value):
        return None

    if isinstance(value, str):
        value = value.strip()

        if value == "":
            return None

    return value


def clean_text(value):
    """
    Normalize textual fields without changing their meaning.
    """
    if value is None:
        return ""

    value = str(value)

    # Normalize whitespace
    value = re.sub(r"\s+", " ", value)

    return value.strip()


def unique_clean_list(values):
    """
    Preserve order while removing duplicates and empty values.
    """
    result = []
    seen = set()

    for value in values:

        if value is None:
            continue

        value = clean_text(value)

        if not value:
            continue

        key = value.lower()

        if key not in seen:
            seen.add(key)
            result.append(value)

    return result


# ============================================================
# 2. PREPARE REFERENCE DATA
# ============================================================

print("\n[1/6] Preparing reference indexes...")

# Cross references
cross_ref_index = defaultdict(list)

for row in cross_references.itertuples(index=False):

    cross_ref_index[row.internal_id].append({
        "reference_number": clean_value(row.reference_number),
        "reference_is_number": clean_value(row.reference_is_number),
        "reference_title": clean_value(row.reference_title),
        "reference_committee": clean_value(row.reference_committee)
    })


# Indian references
indian_ref_index = defaultdict(list)

for row in indian_references.itertuples(index=False):

    indian_ref_index[row.internal_id].append({
        "reference_number": clean_value(row.reference_number),
        "reference_is_number": clean_value(row.reference_is_number),
        "reference_title": clean_value(row.reference_title),
        "reference_committee": clean_value(row.reference_committee)
    })


# International references
international_ref_index = defaultdict(list)

for row in international_references.itertuples(index=False):

    international_ref_index[row.internal_id].append({
        "reference_number": clean_value(row.reference_number),
        "international_standard": clean_value(row.international_standard)
    })


# Referred-by
referred_by_index = defaultdict(list)

for row in referred_by.itertuples(index=False):

    referred_by_index[row.internal_id].append({
        "reference_number": clean_value(row.reference_number),
        "reference_is_number": clean_value(row.reference_is_number),
        "reference_title": clean_value(row.reference_title),
        "reference_committee": clean_value(row.reference_committee)
    })


print("Reference indexes created.")


# ============================================================
# 3. PREPARE AUTHORED DETAILS INDEX
# ============================================================

print("\n[2/6] Preparing authored-details index...")

detail_columns = [
    f"line_{i}"
    for i in range(1, 11)
]

authored_index = {}

for row in authored_details.itertuples(index=False):

    details = []

    for column in detail_columns:

        value = getattr(row, column)

        value = clean_text(value)

        if value:
            details.append(value)

    authored_index[row.internal_id] = details

print(f"Authored detail records indexed: {len(authored_index):,}")


# ============================================================
# 4. BUILD CANONICAL RECORDS
# ============================================================

print("\n[3/6] Building canonical records...")

canonical_records = []

master_columns = [
    "internal_id",
    "IS_number",
    "IS_title",
    "superseding_IS",
    "degree_of_equivalence",
    "number_of_revisions",
    "number_of_amendments",
    "aspect",
    "language",
    "reaffirmation_year",
    "technical_department",
    "technical_committee",
    "member_secretary",
    "group",
    "sub_group",
    "sub_sub_group",
    "certification",
    "amendment_count",
    "gazette_document_count",
    "license_count",
    "product_manual_sit_count",
    "laboratory_count",
    "corrigendum_count",
    "catalogue_id",
    "catalogue_is_no",
    "catalogue_title",
    "catalogue_amendments",
    "catalogue_technical_committee",
    "catalogue_aspect",
    "catalogue_reaffirmation",
    "catalogue_withdrawn_status",
    "detail_url",
    "download_url",
    "composition_url",
    "extracted_at"
]

for row in standards_master.itertuples(index=False):

    internal_id = row.internal_id

    record = {}

    # --------------------------------------------------------
    # Identity
    # --------------------------------------------------------

    record["identity"] = {
        "internal_id": clean_value(row.internal_id),
        "IS_number": clean_value(row.IS_number),
        "title": clean_value(row.IS_title)
    }

    # --------------------------------------------------------
    # Classification
    # --------------------------------------------------------

    record["classification"] = {
        "aspect": clean_value(row.aspect),
        "language": clean_value(row.language),
        "technical_department": clean_value(row.technical_department),
        "technical_committee": clean_value(row.technical_committee),
        "member_secretary": clean_value(row.member_secretary),
        "group": clean_value(row.group),
        "sub_group": clean_value(row.sub_group),
        "sub_sub_group": clean_value(row.sub_sub_group)
    }

    # --------------------------------------------------------
    # Status / history
    # --------------------------------------------------------

    record["status"] = {
        "superseding_IS": clean_value(row.superseding_IS),
        "degree_of_equivalence": clean_value(
            row.degree_of_equivalence
        ),
        "number_of_revisions": clean_value(
            row.number_of_revisions
        ),
        "number_of_amendments": clean_value(
            row.number_of_amendments
        ),
        "reaffirmation_year": clean_value(
            row.reaffirmation_year
        ),
        "amendment_count": clean_value(
            row.amendment_count
        ),
        "catalogue_amendments": clean_value(
            row.catalogue_amendments
        ),
        "catalogue_reaffirmation": clean_value(
            row.catalogue_reaffirmation
        ),
        "catalogue_withdrawn_status": clean_value(
            row.catalogue_withdrawn_status
        )
    }

    # --------------------------------------------------------
    # Certification / operational information
    # --------------------------------------------------------

    record["certification"] = {
        "certification": clean_value(row.certification),
        "gazette_document_count": clean_value(
            row.gazette_document_count
        ),
        "license_count": clean_value(
            row.license_count
        ),
        "product_manual_sit_count": clean_value(
            row.product_manual_sit_count
        ),
        "laboratory_count": clean_value(
            row.laboratory_count
        ),
        "corrigendum_count": clean_value(
            row.corrigendum_count
        )
    }

    # --------------------------------------------------------
    # Catalogue information
    # --------------------------------------------------------

    record["catalogue"] = {
        "catalogue_id": clean_value(row.catalogue_id),
        "catalogue_is_no": clean_value(row.catalogue_is_no),
        "catalogue_title": clean_value(row.catalogue_title),
        "catalogue_technical_committee": clean_value(
            row.catalogue_technical_committee
        ),
        "catalogue_aspect": clean_value(
            row.catalogue_aspect
        )
    }

    # --------------------------------------------------------
    # Official links
    # --------------------------------------------------------

    record["links"] = {
        "detail_url": clean_value(row.detail_url),
        "download_url": clean_value(row.download_url),
        "composition_url": clean_value(row.composition_url)
    }

    # --------------------------------------------------------
    # Extraction metadata
    # --------------------------------------------------------

    record["metadata"] = {
        "extracted_at": clean_value(row.extracted_at)
    }

    # --------------------------------------------------------
    # Authored explanation
    # --------------------------------------------------------

    details = authored_index.get(internal_id, [])

    record["authored_details"] = details

    # --------------------------------------------------------
    # References
    # --------------------------------------------------------

    record["references"] = {
        "cross_references": cross_ref_index.get(
            internal_id, []
        ),
        "indian_references": indian_ref_index.get(
            internal_id, []
        ),
        "international_references": international_ref_index.get(
            internal_id, []
        ),
        "referred_by": referred_by_index.get(
            internal_id, []
        )
    }

    canonical_records.append(record)


print(
    f"Canonical records created: "
    f"{len(canonical_records):,}"
)


# ============================================================
# 5. VALIDATE CANONICAL RECORDS
# ============================================================

print("\n[4/6] Validating canonical records...")

assert len(canonical_records) == len(standards_master)

# Internal IDs must remain unique
canonical_ids = [
    record["identity"]["internal_id"]
    for record in canonical_records
]

assert len(canonical_ids) == len(set(canonical_ids))

print("✓ Record count matches master dataset.")
print("✓ internal_id remains unique.")
print("✓ No master records were lost.")


# ============================================================
# 6. CREATE SEARCHABLE TEXT REPRESENTATION
# ============================================================

print("\n[5/6] Creating semantic search documents...")


def build_search_document(record):

    identity = record["identity"]
    classification = record["classification"]
    status = record["status"]
    certification = record["certification"]
    catalogue = record["catalogue"]
    links = record["links"]

    parts = []

    # Identity
    parts.append(
        f"Indian Standard Number: "
        f"{identity.get('IS_number') or 'Not available'}"
    )

    parts.append(
        f"Title: "
        f"{identity.get('title') or 'Not available'}"
    )

    parts.append(
        f"Internal ID: "
        f"{identity.get('internal_id')}"
    )

    # Classification
    for key, label in [
        ("aspect", "Aspect"),
        ("language", "Language"),
        ("technical_department", "Technical Department"),
        ("technical_committee", "Technical Committee"),
        ("member_secretary", "Member Secretary"),
        ("group", "Group"),
        ("sub_group", "Sub Group"),
        ("sub_sub_group", "Sub Sub Group")
    ]:

        value = classification.get(key)

        if value is not None:
            parts.append(f"{label}: {value}")

    # Status
    for key, label in [
        ("superseding_IS", "Superseding Standard"),
        ("degree_of_equivalence", "Degree of Equivalence"),
        ("number_of_revisions", "Number of Revisions"),
        ("number_of_amendments", "Number of Amendments"),
        ("reaffirmation_year", "Reaffirmation Year"),
        ("catalogue_withdrawn_status", "Withdrawn Status")
    ]:

        value = status.get(key)

        if value is not None:
            parts.append(f"{label}: {value}")

    # Certification
    for key, label in [
        ("certification", "Certification"),
        ("gazette_document_count", "Gazette Documents"),
        ("license_count", "Licenses"),
        ("laboratory_count", "Laboratories"),
        ("corrigendum_count", "Corrigenda")
    ]:

        value = certification.get(key)

        if value is not None:
            parts.append(f"{label}: {value}")

    # Authored explanation
    details = record["authored_details"]

    if details:
        parts.append("Official BIS-derived description:")

        for detail in details:
            parts.append(detail)

    # References
    cross_refs = record["references"]["cross_references"]

    if cross_refs:

        parts.append("Cross References:")

        for ref in cross_refs:

            ref_text = []

            if ref.get("reference_is_number"):
                ref_text.append(
                    str(ref["reference_is_number"])
                )

            if ref.get("reference_title"):
                ref_text.append(
                    str(ref["reference_title"])
                )

            if ref_text:
                parts.append(
                    " - " + " | ".join(ref_text)
                )

    indian_refs = record["references"]["indian_references"]

    if indian_refs:

        parts.append("Indian References:")

        for ref in indian_refs:

            ref_text = []

            if ref.get("reference_is_number"):
                ref_text.append(
                    str(ref["reference_is_number"])
                )

            if ref.get("reference_title"):
                ref_text.append(
                    str(ref["reference_title"])
                )

            if ref_text:
                parts.append(
                    " - " + " | ".join(ref_text)
                )

    international_refs = record["references"][
        "international_references"
    ]

    if international_refs:

        parts.append("International References:")

        for ref in international_refs:

            if ref.get("international_standard"):
                parts.append(
                    " - " + str(
                        ref["international_standard"]
                    )
                )

    referred = record["references"]["referred_by"]

    if referred:

        parts.append("Referenced By:")

        for ref in referred:

            ref_text = []

            if ref.get("reference_is_number"):
                ref_text.append(
                    str(ref["reference_is_number"])
                )

            if ref.get("reference_title"):
                ref_text.append(
                    str(ref["reference_title"])
                )

            if ref_text:
                parts.append(
                    " - " + " | ".join(ref_text)
                )

    return "\n".join(parts)


for record in canonical_records:

    record["search_document"] = build_search_document(
        record
    )


print("Semantic search documents created.")


# ============================================================
# FINAL SAMPLE
# ============================================================

print("\n[6/6] Inspecting sample canonical records...")

for i in range(3):

    record = canonical_records[i]

    print("\n" + "=" * 80)

    print(
        f"Record {i + 1}: "
        f"{record['identity']['IS_number']}"
    )

    print(
        f"Title: "
        f"{record['identity']['title']}"
    )

    print(
        f"Authored detail lines: "
        f"{len(record['authored_details'])}"
    )

    print(
        f"Cross references: "
        f"{len(record['references']['cross_references'])}"
    )

    print(
        f"Indian references: "
        f"{len(record['references']['indian_references'])}"
    )

    print(
        f"International references: "
        f"{len(record['references']['international_references'])}"
    )

    print(
        f"Referred by: "
        f"{len(record['references']['referred_by'])}"
    )

    print("\nSEARCH DOCUMENT PREVIEW:\n")

    print(
        record["search_document"][:3000]
    )


# ============================================================
# SAVE CANONICAL DATA
# ============================================================

print("\nSaving canonical dataset...")

canonical_output_path = "/kaggle/working/bis_canonical_records.json"

with open(
    canonical_output_path,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        canonical_records,
        f,
        ensure_ascii=False,
        indent=2
    )

print(
    f"Saved: {canonical_output_path}"
)

print("\n" + "=" * 80)
print("CELL 3 COMPLETE")
print("=" * 80)
```

```python
# ============================================================
# SAATHI - CELL 4
# BUILD RETRIEVAL-OPTIMIZED RAG CORPUS
# ============================================================

import json
import os
import re
import pandas as pd
import numpy as np

print("=" * 80)
print("SAATHI - CELL 4: BUILDING RETRIEVAL CORPUS")
print("=" * 80)


# ============================================================
# 1. LOAD CANONICAL RECORDS
# ============================================================

print("\n[1/7] Loading canonical records...")

canonical_path = "/kaggle/working/bis_canonical_records.json"

with open(
    canonical_path,
    "r",
    encoding="utf-8"
) as f:
    canonical_records = json.load(f)

print(
    f"Loaded {len(canonical_records):,} canonical records."
)


# ============================================================
# 2. TEXT CLEANING
# ============================================================

print("\n[2/7] Preparing text normalization...")


def normalize_text(text):

    if text is None:
        return ""

    text = str(text)

    # Normalize whitespace
    text = re.sub(r"\s+", " ", text)

    return text.strip()


def safe_value(value):

    if value is None:
        return ""

    return normalize_text(value)


# ============================================================
# 3. BUILD CORE STANDARD DOCUMENT
# ============================================================

print("\n[3/7] Building core standard documents...")

rag_documents = []

for record in canonical_records:

    identity = record["identity"]
    classification = record["classification"]
    status = record["status"]
    certification = record["certification"]

    internal_id = identity["internal_id"]
    is_number = safe_value(identity["IS_number"])
    title = safe_value(identity["title"])

    # --------------------------------------------------------
    # Core document
    # --------------------------------------------------------

    parts = []

    parts.append(
        f"Indian Standard: {is_number}"
    )

    parts.append(
        f"Title: {title}"
    )

    parts.append(
        f"Internal ID: {internal_id}"
    )

    # Classification
    classification_labels = [
        ("aspect", "Aspect"),
        ("language", "Language"),
        ("technical_department", "Technical Department"),
        ("technical_committee", "Technical Committee"),
        ("member_secretary", "Member Secretary"),
        ("group", "Group"),
        ("sub_group", "Sub Group"),
        ("sub_sub_group", "Sub Sub Group")
    ]

    for key, label in classification_labels:

        value = safe_value(
            classification.get(key)
        )

        if value:
            parts.append(
                f"{label}: {value}"
            )

    # Status
    status_labels = [
        ("superseding_IS", "Superseding Standard"),
        ("degree_of_equivalence", "Degree of Equivalence"),
        ("number_of_revisions", "Number of Revisions"),
        ("number_of_amendments", "Number of Amendments"),
        ("reaffirmation_year", "Reaffirmation Year"),
        ("catalogue_withdrawn_status", "Withdrawn Status")
    ]

    for key, label in status_labels:

        value = safe_value(
            status.get(key)
        )

        if value:
            parts.append(
                f"{label}: {value}"
            )

    # Certification
    certification_labels = [
        ("certification", "Certification"),
        ("gazette_document_count", "Gazette Documents"),
        ("license_count", "Licenses"),
        ("product_manual_sit_count", "Product Manual SIT Count"),
        ("laboratory_count", "Laboratories"),
        ("corrigendum_count", "Corrigenda")
    ]

    for key, label in certification_labels:

        value = safe_value(
            certification.get(key)
        )

        if value:
            parts.append(
                f"{label}: {value}"
            )

    # Authored details
    details = record.get(
        "authored_details",
        []
    )

    if details:

        parts.append(
            "\nBIS-derived description:"
        )

        for detail in details:

            detail = normalize_text(detail)

            if detail:
                parts.append(detail)

    core_text = "\n".join(parts)

    rag_documents.append({
        "document_id": f"{internal_id}_core",
        "internal_id": internal_id,
        "IS_number": is_number,
        "title": title,
        "document_type": "standard_core",
        "text": core_text,
        "metadata": {
            "internal_id": internal_id,
            "IS_number": is_number,
            "title": title,
            "technical_department": classification.get(
                "technical_department"
            ),
            "technical_committee": classification.get(
                "technical_committee"
            ),
            "aspect": classification.get("aspect"),
            "language": classification.get("language"),
            "withdrawn_status": status.get(
                "catalogue_withdrawn_status"
            ),
            "degree_of_equivalence": status.get(
                "degree_of_equivalence"
            )
        }
    })


print(
    f"Core documents created: "
    f"{len(rag_documents):,}"
)


# ============================================================
# 4. BUILD REFERENCE DOCUMENTS
# ============================================================

print("\n[4/7] Building reference documents...")

reference_document_count = 0

for record in canonical_records:

    identity = record["identity"]

    internal_id = identity["internal_id"]
    is_number = safe_value(identity["IS_number"])
    title = safe_value(identity["title"])

    references = record["references"]

    # --------------------------------------------------------
    # Cross references
    # --------------------------------------------------------

    cross_refs = references.get(
        "cross_references",
        []
    )

    if cross_refs:

        lines = [
            f"Indian Standard: {is_number}",
            f"Title: {title}",
            "",
            "Cross references:"
        ]

        for ref in cross_refs:

            ref_number = safe_value(
                ref.get("reference_is_number")
            )

            ref_title = safe_value(
                ref.get("reference_title")
            )

            committee = safe_value(
                ref.get("reference_committee")
            )

            line = "- "

            if ref_number:
                line += ref_number

            if ref_title:
                line += f" | {ref_title}"

            if committee:
                line += f" | Committee: {committee}"

            lines.append(line)

        rag_documents.append({
            "document_id": f"{internal_id}_cross_references",
            "internal_id": internal_id,
            "IS_number": is_number,
            "title": title,
            "document_type": "cross_references",
            "text": "\n".join(lines),
            "metadata": {
                "internal_id": internal_id,
                "IS_number": is_number,
                "title": title,
                "document_type": "cross_references"
            }
        })

        reference_document_count += 1

    # --------------------------------------------------------
    # Indian references
    # --------------------------------------------------------

    indian_refs = references.get(
        "indian_references",
        []
    )

    if indian_refs:

        lines = [
            f"Indian Standard: {is_number}",
            f"Title: {title}",
            "",
            "Indian standard references:"
        ]

        for ref in indian_refs:

            ref_number = safe_value(
                ref.get("reference_is_number")
            )

            ref_title = safe_value(
                ref.get("reference_title")
            )

            committee = safe_value(
                ref.get("reference_committee")
            )

            line = "- "

            if ref_number:
                line += ref_number

            if ref_title:
                line += f" | {ref_title}"

            if committee:
                line += f" | Committee: {committee}"

            lines.append(line)

        rag_documents.append({
            "document_id": f"{internal_id}_indian_references",
            "internal_id": internal_id,
            "IS_number": is_number,
            "title": title,
            "document_type": "indian_references",
            "text": "\n".join(lines),
            "metadata": {
                "internal_id": internal_id,
                "IS_number": is_number,
                "title": title,
                "document_type": "indian_references"
            }
        })

        reference_document_count += 1

    # --------------------------------------------------------
    # International references
    # --------------------------------------------------------

    international_refs = references.get(
        "international_references",
        []
    )

    if international_refs:

        lines = [
            f"Indian Standard: {is_number}",
            f"Title: {title}",
            "",
            "International standards referenced:"
        ]

        for ref in international_refs:

            international_standard = safe_value(
                ref.get("international_standard")
            )

            if international_standard:

                lines.append(
                    f"- {international_standard}"
                )

        rag_documents.append({
            "document_id": f"{internal_id}_international_references",
            "internal_id": internal_id,
            "IS_number": is_number,
            "title": title,
            "document_type": "international_references",
            "text": "\n".join(lines),
            "metadata": {
                "internal_id": internal_id,
                "IS_number": is_number,
                "title": title,
                "document_type": "international_references"
            }
        })

        reference_document_count += 1


# ============================================================
# 5. BUILD REVERSE RELATIONSHIP DOCUMENTS
# ============================================================

print("\n[5/7] Building reverse relationship documents...")

referred_documents = 0

for record in canonical_records:

    identity = record["identity"]

    internal_id = identity["internal_id"]
    is_number = safe_value(identity["IS_number"])
    title = safe_value(identity["title"])

    referred = record["references"].get(
        "referred_by",
        []
    )

    if not referred:
        continue

    lines = [
        f"Indian Standard: {is_number}",
        f"Title: {title}",
        "",
        "Standards that refer to this standard:"
    ]

    for ref in referred:

        ref_number = safe_value(
            ref.get("reference_is_number")
        )

        ref_title = safe_value(
            ref.get("reference_title")
        )

        committee = safe_value(
            ref.get("reference_committee")
        )

        line = "- "

        if ref_number:
            line += ref_number

        if ref_title:
            line += f" | {ref_title}"

        if committee:
            line += f" | Committee: {committee}"

        lines.append(line)

    rag_documents.append({
        "document_id": f"{internal_id}_referred_by",
        "internal_id": internal_id,
        "IS_number": is_number,
        "title": title,
        "document_type": "referred_by",
        "text": "\n".join(lines),
        "metadata": {
            "internal_id": internal_id,
            "IS_number": is_number,
            "title": title,
            "document_type": "referred_by"
        }
    })

    referred_documents += 1


# ============================================================
# 6. VALIDATE CORPUS
# ============================================================

print("\n[6/7] Validating RAG corpus...")

total_documents = len(rag_documents)

empty_documents = sum(
    1
    for doc in rag_documents
    if not doc["text"].strip()
)

unique_document_ids = len(
    set(doc["document_id"] for doc in rag_documents)
)

document_types = {}

for doc in rag_documents:

    doc_type = doc["document_type"]

    document_types[doc_type] = (
        document_types.get(doc_type, 0) + 1
    )

print(
    f"Total RAG documents : "
    f"{total_documents:,}"
)

print(
    f"Unique document IDs : "
    f"{unique_document_ids:,}"
)

print(
    f"Empty documents     : "
    f"{empty_documents:,}"
)

print("\nDocument types:")

for doc_type, count in document_types.items():

    print(
        f"  {doc_type:25s}: "
        f"{count:,}"
    )

assert empty_documents == 0
assert total_documents == unique_document_ids


# ============================================================
# 7. SAVE RAG CORPUS
# ============================================================

print("\n[7/7] Saving RAG corpus...")

rag_output_path = (
    "/kaggle/working/bis_rag_corpus.json"
)

with open(
    rag_output_path,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        rag_documents,
        f,
        ensure_ascii=False,
        indent=2
    )

print(
    f"Saved: {rag_output_path}"
)


# Save a lightweight JSONL version too
jsonl_path = (
    "/kaggle/working/bis_rag_corpus.jsonl"
)

with open(
    jsonl_path,
    "w",
    encoding="utf-8"
) as f:

    for doc in rag_documents:

        f.write(
            json.dumps(
                doc,
                ensure_ascii=False
            )
            + "\n"
        )

print(
    f"Saved: {jsonl_path}"
)


# ============================================================
# SAMPLE DOCUMENTS
# ============================================================

print("\nSample retrieval documents:")

for doc in rag_documents[:3]:

    print("\n" + "-" * 80)

    print(
        "Document ID:",
        doc["document_id"]
    )

    print(
        "Type:",
        doc["document_type"]
    )

    print(
        "Standard:",
        doc["IS_number"]
    )

    print(
        "\nText preview:\n"
    )

    print(
        doc["text"][:2000]
    )


print("\n" + "=" * 80)
print("CELL 4 COMPLETE")
print("=" * 80)
```

```python
# ============================================================
# SAATHI - CELL 5
# EMBEDDING ENVIRONMENT + MODEL BENCHMARK
# ============================================================

import os
import sys
import subprocess
import importlib.util

print("=" * 80)
print("SAATHI - CELL 5: EMBEDDING MODEL SETUP")
print("=" * 80)


# ============================================================
# 1. INSTALL REQUIRED PACKAGES
# ============================================================

print("\n[1/5] Checking required packages...")

required_packages = {
    "sentence_transformers": "sentence-transformers",
    "faiss": "faiss-cpu"
}

for import_name, package_name in required_packages.items():

    if importlib.util.find_spec(import_name) is None:

        print(f"Installing {package_name}...")

        subprocess.check_call([
            sys.executable,
            "-m",
            "pip",
            "install",
            "-q",
            package_name
        ])

    else:
        print(f"{package_name}: already installed")


# ============================================================
# 2. IMPORT
# ============================================================

print("\n[2/5] Importing embedding libraries...")

from sentence_transformers import SentenceTransformer
import faiss

print("SentenceTransformers:", "OK")
print("FAISS:", "OK")


# ============================================================
# 3. LOAD RAG CORPUS
# ============================================================

print("\n[3/5] Loading RAG corpus...")

rag_path = "/kaggle/working/bis_rag_corpus.json"

with open(
    rag_path,
    "r",
    encoding="utf-8"
) as f:

    rag_documents = json.load(f)

print(
    f"RAG documents loaded: "
    f"{len(rag_documents):,}"
)


# ============================================================
# 4. LOAD EMBEDDING MODEL
# ============================================================

print("\n[4/5] Loading embedding model...")

MODEL_NAME = "BAAI/bge-m3"

print(f"Model: {MODEL_NAME}")

embedding_model = SentenceTransformer(
    MODEL_NAME
)

print("Embedding model loaded successfully.")

embedding_dimension = embedding_model.get_sentence_embedding_dimension()

print(
    f"Embedding dimension: "
    f"{embedding_dimension}"
)


# ============================================================
# 5. BENCHMARK REAL BIS QUERIES
# ============================================================

print("\n[5/5] Running retrieval benchmark...")

# ------------------------------------------------------------
# Create a small temporary benchmark corpus
# ------------------------------------------------------------

benchmark_documents = [
    doc
    for doc in rag_documents
    if doc["document_type"] == "standard_core"
]

print(
    f"Benchmark corpus: "
    f"{len(benchmark_documents):,} core standards"
)


# Use a manageable subset for the first benchmark
benchmark_size = min(
    10000,
    len(benchmark_documents)
)

benchmark_subset = benchmark_documents[
    :benchmark_size
]

benchmark_texts = [
    doc["text"]
    for doc in benchmark_subset
]

print(
    f"Embedding benchmark subset: "
    f"{len(benchmark_texts):,} documents"
)


# ------------------------------------------------------------
# Generate temporary embeddings
# ------------------------------------------------------------

benchmark_embeddings = embedding_model.encode(
    benchmark_texts,
    batch_size=32,
    show_progress_bar=True,
    normalize_embeddings=True,
    convert_to_numpy=True
)

print(
    "Benchmark embeddings shape:",
    benchmark_embeddings.shape
)


# ------------------------------------------------------------
# Create FAISS index
# ------------------------------------------------------------

dimension = benchmark_embeddings.shape[1]

index = faiss.IndexFlatIP(dimension)

index.add(
    benchmark_embeddings.astype(
        "float32"
    )
)

print(
    f"FAISS benchmark index contains "
    f"{index.ntotal:,} vectors"
)


# ============================================================
# TEST QUERIES
# ============================================================

test_queries = [

    "What is IS/IEC 60691:2023?",

    "Which Indian standards are related to thermal links?",

    "What is the degree of equivalence of IS/IEC 60691:2023?",

    "Which BIS standard deals with coaxial communication cables?",

    "Find standards related to electrical fuses",

    "Which standards are referenced by IS/IEC 60691:2023?",

    "What technical committee is responsible for thermal links?",

    "Which standards are withdrawn?"
]


print("\n")
print("=" * 80)
print("RETRIEVAL RESULTS")
print("=" * 80)


for query in test_queries:

    print("\n" + "=" * 80)
    print("QUERY:")
    print(query)

    query_embedding = embedding_model.encode(
        [query],
        normalize_embeddings=True,
        convert_to_numpy=True
    )

    scores, indices = index.search(
        query_embedding.astype("float32"),
        5
    )

    print("\nTOP 5 RESULTS:\n")

    for rank, (score, idx) in enumerate(
        zip(scores[0], indices[0]),
        start=1
    ):

        doc = benchmark_subset[idx]

        print(
            f"{rank}. "
            f"[score={score:.4f}] "
            f"{doc['IS_number']}"
        )

        print(
            f"   {doc['title']}"
        )

        print(
            f"   Type: "
            f"{doc['document_type']}"
        )

        print(
            f"   Internal ID: "
            f"{doc['internal_id']}"
        )


print("\n" + "=" * 80)
print("CELL 5 COMPLETE")
print("=" * 80)

print(
    "\nIMPORTANT:"
    "\nThis cell intentionally embeds only a 10,000-document "
    "benchmark subset."
    "\nWe will NOT generate the complete 116,597-document "
    "embedding index until we inspect these results."
)
```

```python
# ============================================================
# SAATHI — CELL 5
# FULL BIS RAG EMBEDDING INDEX
# 116,597 DOCUMENTS — GPU BATCHED
# ============================================================

import os
import json
import numpy as np
import torch

from sentence_transformers import SentenceTransformer
import faiss

# ------------------------------------------------------------
# CONFIG
# ------------------------------------------------------------

CORPUS_PATH = "/kaggle/working/bis_rag_corpus.json"

OUTPUT_EMBEDDINGS = "/kaggle/working/bis_embeddings.npy"
OUTPUT_INDEX = "/kaggle/working/bis_faiss.index"
OUTPUT_METADATA = "/kaggle/working/bis_rag_metadata.json"

MODEL_NAME = "BAAI/bge-small-en-v1.5"

# Increase if GPU has enough VRAM
BATCH_SIZE = 256

# ------------------------------------------------------------
# DEVICE
# ------------------------------------------------------------

device = "cuda" if torch.cuda.is_available() else "cpu"

print("=" * 70)
print("SAATHI — FULL RAG EMBEDDING PIPELINE")
print("=" * 70)

print(f"Device : {device}")

if device == "cuda":
    print(f"GPU    : {torch.cuda.get_device_name(0)}")
    print(
        f"VRAM   : "
        f"{torch.cuda.get_device_properties(0).total_memory / 1024**3:.2f} GB"
    )
else:
    print("WARNING: GPU NOT DETECTED")
    print("Enable GPU in Kaggle Settings → Accelerator → GPU")

# ------------------------------------------------------------
# LOAD CORPUS
# ------------------------------------------------------------

print("\nLoading RAG corpus...")

with open(CORPUS_PATH, "r", encoding="utf-8") as f:
    corpus = json.load(f)

print(f"Total documents: {len(corpus):,}")

# ------------------------------------------------------------
# EXTRACT TEXT
# ------------------------------------------------------------

documents = []
metadata = []

for doc in corpus:

    text = doc.get("text", "").strip()

    if not text:
        continue

    documents.append(text)

    metadata.append({
        "doc_id": doc.get("doc_id"),
        "internal_id": doc.get("internal_id"),
        "document_type": doc.get("document_type"),
        "is_number": doc.get("is_number"),
        "is_title": doc.get("is_title")
    })

print(f"Documents to embed: {len(documents):,}")

# ------------------------------------------------------------
# LOAD EMBEDDING MODEL
# ------------------------------------------------------------

print("\nLoading embedding model...")
print(MODEL_NAME)

model = SentenceTransformer(
    MODEL_NAME,
    device=device
)

# Use half precision on GPU
if device == "cuda":
    model.half()

print("Embedding model loaded.")

# ------------------------------------------------------------
# FULL BATCH EMBEDDING
# ------------------------------------------------------------

print("\nStarting FULL corpus embedding...")
print(f"Batch size: {BATCH_SIZE:,}")
print(f"Documents:  {len(documents):,}")

embeddings = model.encode(
    documents,
    batch_size=BATCH_SIZE,
    show_progress_bar=True,
    convert_to_numpy=True,
    normalize_embeddings=True,
    device=device
)

# ------------------------------------------------------------
# SAVE EMBEDDINGS
# ------------------------------------------------------------

print("\nSaving embeddings...")

embeddings = embeddings.astype("float32")

np.save(
    OUTPUT_EMBEDDINGS,
    embeddings
)

print(f"Embedding shape: {embeddings.shape}")
print(f"Saved: {OUTPUT_EMBEDDINGS}")

# ------------------------------------------------------------
# BUILD FAISS INDEX
# ------------------------------------------------------------

print("\nBuilding FAISS index...")

dimension = embeddings.shape[1]

index = faiss.IndexFlatIP(dimension)

index.add(embeddings)

print(f"FAISS dimension : {dimension}")
print(f"FAISS vectors   : {index.ntotal:,}")

# ------------------------------------------------------------
# SAVE FAISS INDEX
# ------------------------------------------------------------

faiss.write_index(
    index,
    OUTPUT_INDEX
)

print(f"Saved: {OUTPUT_INDEX}")

# ------------------------------------------------------------
# SAVE METADATA
# ------------------------------------------------------------

with open(
    OUTPUT_METADATA,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        metadata,
        f,
        ensure_ascii=False
    )

print(f"Saved: {OUTPUT_METADATA}")

# ------------------------------------------------------------
# FINAL REPORT
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("FULL RAG INDEX COMPLETE")
print("=" * 70)

print(f"Documents indexed : {index.ntotal:,}")
print(f"Embedding size    : {dimension}")
print(f"Embeddings file   : {OUTPUT_EMBEDDINGS}")
print(f"FAISS index       : {OUTPUT_INDEX}")
print(f"Metadata file     : {OUTPUT_METADATA}")

print("=" * 70)
```

```python
# ============================================================
# SAATHI — REBUILD FULL RAG CORPUS
# 35,119 STANDARDS → ~116K RETRIEVAL DOCUMENTS
# ============================================================

import os
import json
import pandas as pd
from collections import defaultdict

# ------------------------------------------------------------
# PATHS
# ------------------------------------------------------------

BASE = "/kaggle/input/datasets/bhaskaraaryan/saaaaathi/BIS_CSV"

MASTER_PATH = os.path.join(BASE, "bis_standards_master.csv")
CROSS_PATH = os.path.join(BASE, "bis_cross_references.csv")
INDIAN_PATH = os.path.join(BASE, "bis_indian_references.csv")
INTERNATIONAL_PATH = os.path.join(BASE, "bis_international_references.csv")
REFERRED_PATH = os.path.join(BASE, "bis_referred_by.csv")

OUT_JSON = "/kaggle/working/bis_rag_corpus.json"
OUT_JSONL = "/kaggle/working/bis_rag_corpus.jsonl"

print("=" * 70)
print("SAATHI — FULL RAG CORPUS REBUILD")
print("=" * 70)

# ------------------------------------------------------------
# LOAD DATA
# ------------------------------------------------------------

print("\nLoading datasets...")

master = pd.read_csv(MASTER_PATH, low_memory=False)
cross = pd.read_csv(CROSS_PATH, low_memory=False)
indian = pd.read_csv(INDIAN_PATH, low_memory=False)
international = pd.read_csv(INTERNATIONAL_PATH, low_memory=False)
referred = pd.read_csv(REFERRED_PATH, low_memory=False)

print(f"Master standards      : {len(master):,}")
print(f"Cross references      : {len(cross):,}")
print(f"Indian references     : {len(indian):,}")
print(f"International refs    : {len(international):,}")
print(f"Referred-by records   : {len(referred):,}")

# ------------------------------------------------------------
# HELPERS
# ------------------------------------------------------------

def clean(value):
    if pd.isna(value):
        return ""
    return str(value).strip()

def add_line(lines, label, value):
    value = clean(value)
    if value:
        lines.append(f"{label}: {value}")

# ------------------------------------------------------------
# BUILD REFERENCE INDEXES
# ------------------------------------------------------------

print("\nBuilding reference indexes...")

cross_idx = defaultdict(list)
indian_idx = defaultdict(list)
international_idx = defaultdict(list)
referred_idx = defaultdict(list)

for row in cross.itertuples(index=False):
    cross_idx[clean(row.internal_id)].append({
        "reference_number": clean(row.reference_number),
        "reference_is_number": clean(row.reference_is_number),
        "reference_title": clean(row.reference_title),
        "reference_committee": clean(row.reference_committee)
    })

for row in indian.itertuples(index=False):
    indian_idx[clean(row.internal_id)].append({
        "reference_number": clean(row.reference_number),
        "reference_is_number": clean(row.reference_is_number),
        "reference_title": clean(row.reference_title),
        "reference_committee": clean(row.reference_committee)
    })

for row in international.itertuples(index=False):
    international_idx[clean(row.internal_id)].append({
        "reference_number": clean(row.reference_number),
        "international_standard": clean(row.international_standard)
    })

for row in referred.itertuples(index=False):
    referred_idx[clean(row.internal_id)].append({
        "reference_number": clean(row.reference_number),
        "reference_is_number": clean(row.reference_is_number),
        "reference_title": clean(row.reference_title),
        "reference_committee": clean(row.reference_committee)
    })

print(f"Cross-reference IDs    : {len(cross_idx):,}")
print(f"Indian-reference IDs   : {len(indian_idx):,}")
print(f"International IDs      : {len(international_idx):,}")
print(f"Referred-by IDs        : {len(referred_idx):,}")

# ------------------------------------------------------------
# CREATE RETRIEVAL DOCUMENTS
# ------------------------------------------------------------

print("\nCreating retrieval documents...")

documents = []

# ============================================================
# 1. STANDARD CORE DOCUMENT
# ============================================================

for row in master.itertuples(index=False):

    internal_id = clean(row.internal_id)
    is_number = clean(row.IS_number)
    title = clean(row.IS_title)

    lines = []

    # Identity
    add_line(lines, "BIS Standard", is_number)
    add_line(lines, "Title", title)

    # Classification
    add_line(lines, "Technical Department", row.technical_department)
    add_line(lines, "Technical Committee", row.technical_committee)
    add_line(lines, "Aspect", row.aspect)
    add_line(lines, "Group", row.group)
    add_line(lines, "Sub Group", row.sub_group)
    add_line(lines, "Sub Sub Group", row.sub_sub_group)
    add_line(lines, "Language", row.language)

    # Status / history
    add_line(lines, "Superseding IS", row.superseding_IS)
    add_line(lines, "Degree of Equivalence", row.degree_of_equivalence)
    add_line(lines, "Number of Revisions", row.number_of_revisions)
    add_line(lines, "Number of Amendments", row.number_of_amendments)
    add_line(lines, "Reaffirmation Year", row.reaffirmation_year)
    add_line(lines, "Withdrawn Status", row.catalogue_withdrawn_status)

    # Certification / operational information
    add_line(lines, "Certification", row.certification)
    add_line(lines, "Amendment Count", row.amendment_count)
    add_line(lines, "Gazette Document Count", row.gazette_document_count)
    add_line(lines, "License Count", row.license_count)
    add_line(lines, "Product Manual SIT Count", row.product_manual_sit_count)
    add_line(lines, "Laboratory Count", row.laboratory_count)
    add_line(lines, "Corrigendum Count", row.corrigendum_count)

    # Catalogue information
    add_line(lines, "Catalogue ID", row.catalogue_id)
    add_line(lines, "Catalogue IS Number", row.catalogue_is_no)
    add_line(lines, "Catalogue Title", row.catalogue_title)
    add_line(lines, "Catalogue Amendments", row.catalogue_amendments)
    add_line(lines, "Catalogue Technical Committee",
             row.catalogue_technical_committee)
    add_line(lines, "Catalogue Aspect", row.catalogue_aspect)
    add_line(lines, "Catalogue Reaffirmation", row.catalogue_reaffirmation)

    # Official links
    add_line(lines, "Detail URL", row.detail_url)
    add_line(lines, "Download URL", row.download_url)
    add_line(lines, "Composition URL", row.composition_url)

    documents.append({
        "doc_id": f"{internal_id}_core",
        "internal_id": internal_id,
        "document_type": "standard_core",
        "is_number": is_number,
        "is_title": title,
        "text": "\n".join(lines)
    })

# ============================================================
# 2. CROSS REFERENCES
# ============================================================

for internal_id, refs in cross_idx.items():

    if not internal_id:
        continue

    # One retrieval document for all references of this standard
    lines = [
        f"BIS Standard Internal ID: {internal_id}",
        "Document Type: Cross References"
    ]

    for ref in refs:
        parts = []

        if ref["reference_number"]:
            parts.append(f"Reference Number: {ref['reference_number']}")

        if ref["reference_is_number"]:
            parts.append(f"Referenced IS: {ref['reference_is_number']}")

        if ref["reference_title"]:
            parts.append(f"Referenced Title: {ref['reference_title']}")

        if ref["reference_committee"]:
            parts.append(f"Reference Committee: {ref['reference_committee']}")

        if parts:
            lines.append(" | ".join(parts))

    documents.append({
        "doc_id": f"{internal_id}_cross_references",
        "internal_id": internal_id,
        "document_type": "cross_references",
        "text": "\n".join(lines)
    })

# ============================================================
# 3. INDIAN REFERENCES
# ============================================================

for internal_id, refs in indian_idx.items():

    if not internal_id:
        continue

    lines = [
        f"BIS Standard Internal ID: {internal_id}",
        "Document Type: Indian References"
    ]

    for ref in refs:
        parts = []

        if ref["reference_number"]:
            parts.append(f"Reference Number: {ref['reference_number']}")

        if ref["reference_is_number"]:
            parts.append(f"Referenced IS: {ref['reference_is_number']}")

        if ref["reference_title"]:
            parts.append(f"Referenced Title: {ref['reference_title']}")

        if ref["reference_committee"]:
            parts.append(f"Reference Committee: {ref['reference_committee']}")

        if parts:
            lines.append(" | ".join(parts))

    documents.append({
        "doc_id": f"{internal_id}_indian_references",
        "internal_id": internal_id,
        "document_type": "indian_references",
        "text": "\n".join(lines)
    })

# ============================================================
# 4. INTERNATIONAL REFERENCES
# ============================================================

for internal_id, refs in international_idx.items():

    if not internal_id:
        continue

    lines = [
        f"BIS Standard Internal ID: {internal_id}",
        "Document Type: International References"
    ]

    for ref in refs:
        parts = []

        if ref["reference_number"]:
            parts.append(f"Reference Number: {ref['reference_number']}")

        if ref["international_standard"]:
            parts.append(
                f"International Standard: {ref['international_standard']}"
            )

        if parts:
            lines.append(" | ".join(parts))

    documents.append({
        "doc_id": f"{internal_id}_international_references",
        "internal_id": internal_id,
        "document_type": "international_references",
        "text": "\n".join(lines)
    })

# ============================================================
# 5. REFERRED-BY
# ============================================================

for internal_id, refs in referred_idx.items():

    if not internal_id:
        continue

    lines = [
        f"BIS Standard Internal ID: {internal_id}",
        "Document Type: Referred By"
    ]

    for ref in refs:
        parts = []

        if ref["reference_number"]:
            parts.append(f"Reference Number: {ref['reference_number']}")

        if ref["reference_is_number"]:
            parts.append(f"Referencing IS: {ref['reference_is_number']}")

        if ref["reference_title"]:
            parts.append(f"Referencing Title: {ref['reference_title']}")

        if ref["reference_committee"]:
            parts.append(f"Reference Committee: {ref['reference_committee']}")

        if parts:
            lines.append(" | ".join(parts))

    documents.append({
        "doc_id": f"{internal_id}_referred_by",
        "internal_id": internal_id,
        "document_type": "referred_by",
        "text": "\n".join(lines)
    })

# ------------------------------------------------------------
# VALIDATION
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("CORPUS VALIDATION")
print("=" * 70)

print(f"Total retrieval documents: {len(documents):,}")

types = defaultdict(int)

for doc in documents:
    types[doc["document_type"]] += 1

for doc_type, count in sorted(types.items()):
    print(f"{doc_type:30s}: {count:,}")

empty = sum(
    1 for doc in documents
    if not doc.get("text", "").strip()
)

unique_ids = len({
    doc["doc_id"]
    for doc in documents
})

print(f"\nUnique document IDs       : {unique_ids:,}")
print(f"Empty documents           : {empty:,}")

# ------------------------------------------------------------
# SAVE JSON
# ------------------------------------------------------------

print("\nSaving JSON...")

with open(OUT_JSON, "w", encoding="utf-8") as f:
    json.dump(
        documents,
        f,
        ensure_ascii=False,
        separators=(",", ":")
    )

print(f"Saved: {OUT_JSON}")

# ------------------------------------------------------------
# SAVE JSONL
# ------------------------------------------------------------

print("Saving JSONL...")

with open(OUT_JSONL, "w", encoding="utf-8") as f:

    for doc in documents:
        f.write(
            json.dumps(
                doc,
                ensure_ascii=False,
                separators=(",", ":")
            ) + "\n"
        )

print(f"Saved: {OUT_JSONL}")

# ------------------------------------------------------------
# FINAL
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("FULL RAG CORPUS READY")
print("=" * 70)

print(f"Standards           : {len(master):,}")
print(f"Retrieval documents : {len(documents):,}")
print(f"Empty documents     : {empty:,}")
print(f"JSON                : {OUT_JSON}")
print(f"JSONL               : {OUT_JSONL}")
print("=" * 70)
```

```python
# ============================================================
# SAATHI — CELL 6
# FULL 116,597 DOCUMENT GPU EMBEDDING + FAISS
# ============================================================

import json
import os
import time
import numpy as np
import torch
import faiss

from sentence_transformers import SentenceTransformer

# ------------------------------------------------------------
# CONFIG
# ------------------------------------------------------------

CORPUS_PATH = "/kaggle/working/bis_rag_corpus.json"

EMBEDDINGS_PATH = "/kaggle/working/bis_embeddings.npy"
FAISS_PATH = "/kaggle/working/bis_faiss.index"
METADATA_PATH = "/kaggle/working/bis_rag_metadata.json"

MODEL_NAME = "BAAI/bge-small-en-v1.5"

# Tesla T4 16GB / ~14.5GB usable
BATCH_SIZE = 256

# ------------------------------------------------------------
# GPU
# ------------------------------------------------------------

assert torch.cuda.is_available(), "CUDA/GPU is not available!"

device = "cuda"

print("=" * 70)
print("SAATHI — FULL GPU EMBEDDING")
print("=" * 70)

print("GPU:", torch.cuda.get_device_name(0))
print(
    "VRAM:",
    round(
        torch.cuda.get_device_properties(0).total_memory / 1024**3,
        2
    ),
    "GB"
)

# ------------------------------------------------------------
# LOAD CORPUS
# ------------------------------------------------------------

print("\nLoading corpus...")

with open(CORPUS_PATH, "r", encoding="utf-8") as f:
    corpus = json.load(f)

print(f"Loaded {len(corpus):,} documents")

assert len(corpus) == 116597, \
    f"Expected 116,597 documents, found {len(corpus):,}"

# ------------------------------------------------------------
# EXTRACT TEXT + METADATA
# ------------------------------------------------------------

texts = []
metadata = []

for doc in corpus:

    text = doc["text"].strip()

    if not text:
        continue

    texts.append(text)

    metadata.append({
        "doc_id": doc["doc_id"],
        "internal_id": doc["internal_id"],
        "document_type": doc["document_type"],
        "is_number": doc.get("is_number", ""),
        "is_title": doc.get("is_title", "")
    })

print(f"Texts ready: {len(texts):,}")

# ------------------------------------------------------------
# LOAD MODEL
# ------------------------------------------------------------

print("\nLoading embedding model...")

model = SentenceTransformer(
    MODEL_NAME,
    device=device
)

# T4 benefits from FP16
model.half()

print("Model loaded:", MODEL_NAME)

# ------------------------------------------------------------
# EMBEDDING
# ------------------------------------------------------------

print("\nStarting FULL EMBEDDING JOB")
print("-" * 70)

start_time = time.time()

embeddings = model.encode(
    texts,
    batch_size=BATCH_SIZE,
    show_progress_bar=True,
    convert_to_numpy=True,
    normalize_embeddings=True,
    device=device
)

elapsed = time.time() - start_time

print("\nEmbedding complete!")

print(f"Documents : {len(texts):,}")
print(f"Dimensions: {embeddings.shape[1]}")
print(f"Time      : {elapsed / 60:.2f} minutes")

# ------------------------------------------------------------
# FLOAT32
# ------------------------------------------------------------

embeddings = embeddings.astype(np.float32)

# ------------------------------------------------------------
# SAVE EMBEDDINGS
# ------------------------------------------------------------

print("\nSaving embeddings...")

np.save(
    EMBEDDINGS_PATH,
    embeddings
)

size_gb = os.path.getsize(EMBEDDINGS_PATH) / (1024**3)

print(
    f"Saved: {EMBEDDINGS_PATH}"
)

print(
    f"File size: {size_gb:.2f} GB"
)

# ------------------------------------------------------------
# BUILD FAISS
# ------------------------------------------------------------

print("\nBuilding FAISS index...")

dimension = embeddings.shape[1]

index = faiss.IndexFlatIP(dimension)

index.add(embeddings)

print("FAISS vectors:", f"{index.ntotal:,}")
print("Dimensions:", dimension)

# ------------------------------------------------------------
# SAVE FAISS
# ------------------------------------------------------------

faiss.write_index(
    index,
    FAISS_PATH
)

print(
    f"Saved: {FAISS_PATH}"
)

# ------------------------------------------------------------
# SAVE METADATA
# ------------------------------------------------------------

print("\nSaving metadata...")

with open(
    METADATA_PATH,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        metadata,
        f,
        ensure_ascii=False,
        separators=(",", ":")
    )

print(
    f"Saved: {METADATA_PATH}"
)

# ------------------------------------------------------------
# FINAL
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("FULL RAG VECTOR INDEX COMPLETE")
print("=" * 70)

print(f"Documents indexed : {index.ntotal:,}")
print(f"Embedding dimension: {dimension}")
print(f"Embedding time     : {elapsed / 60:.2f} minutes")
print(f"Embeddings         : {EMBEDDINGS_PATH}")
print(f"FAISS              : {FAISS_PATH}")
print(f"Metadata           : {METADATA_PATH}")

print("=" * 70)
```

```python
# ============================================================
# SAATHI — CELL 7
# RETRIEVAL ENGINE
# ============================================================

import json
import numpy as np
import faiss
import torch

from sentence_transformers import SentenceTransformer

# ------------------------------------------------------------
# PATHS
# ------------------------------------------------------------

CORPUS_PATH = "/kaggle/working/bis_rag_corpus.json"
FAISS_PATH = "/kaggle/working/bis_faiss.index"
METADATA_PATH = "/kaggle/working/bis_rag_metadata.json"

MODEL_NAME = "BAAI/bge-small-en-v1.5"

# ------------------------------------------------------------
# LOAD EVERYTHING
# ------------------------------------------------------------

print("=" * 70)
print("SAATHI — RETRIEVAL ENGINE")
print("=" * 70)

print("\nLoading corpus...")

with open(CORPUS_PATH, "r", encoding="utf-8") as f:
    corpus = json.load(f)

print(f"Corpus: {len(corpus):,}")

print("\nLoading metadata...")

with open(METADATA_PATH, "r", encoding="utf-8") as f:
    metadata = json.load(f)

print(f"Metadata: {len(metadata):,}")

print("\nLoading FAISS index...")

index = faiss.read_index(FAISS_PATH)

print(f"FAISS vectors: {index.ntotal:,}")

# ------------------------------------------------------------
# LOAD EMBEDDING MODEL
# ------------------------------------------------------------

print("\nLoading query embedding model...")

device = "cuda" if torch.cuda.is_available() else "cpu"

model = SentenceTransformer(
    MODEL_NAME,
    device=device
)

if device == "cuda":
    model.half()

print("Model ready.")
print("Device:", device)

# ------------------------------------------------------------
# RETRIEVAL FUNCTION
# ------------------------------------------------------------

def retrieve(query, top_k=10):

    query_embedding = model.encode(
        [query],
        convert_to_numpy=True,
        normalize_embeddings=True
    ).astype("float32")

    scores, indices = index.search(
        query_embedding,
        top_k
    )

    results = []

    for score, idx in zip(scores[0], indices[0]):

        if idx < 0:
            continue

        doc = corpus[idx]
        meta = metadata[idx]

        results.append({
            "rank": len(results) + 1,
            "score": float(score),
            "doc_id": meta["doc_id"],
            "internal_id": meta["internal_id"],
            "document_type": meta["document_type"],
            "is_number": meta.get("is_number", ""),
            "is_title": meta.get("is_title", ""),
            "text": doc["text"]
        })

    return results


# ------------------------------------------------------------
# RETRIEVAL TEST FUNCTION
# ------------------------------------------------------------

def show_results(query, top_k=5):

    print("\n" + "=" * 70)
    print("QUERY")
    print("=" * 70)

    print(query)

    results = retrieve(query, top_k)

    print("\nRESULTS")
    print("-" * 70)

    for r in results:

        print(
            f"\n#{r['rank']} "
            f"[score={r['score']:.4f}] "
            f"[{r['document_type']}]"
        )

        print(
            f"IS: {r['is_number']}"
        )

        print(
            f"Title: {r['is_title']}"
        )

        print(
            f"Internal ID: {r['internal_id']}"
        )

        print(
            "Text:"
        )

        print(
            r["text"][:800]
        )


# ------------------------------------------------------------
# SAVE ENGINE OBJECTS
# ------------------------------------------------------------

print("\nRetrieval engine ready.")

# ------------------------------------------------------------
# TEST REAL BIS QUESTIONS
# ------------------------------------------------------------

show_results(
    "What is IS/IEC 60691:2023?",
    top_k=5
)

show_results(
    "Which BIS standard deals with thermal links?",
    top_k=5
)

show_results(
    "Which standard deals with coaxial communication cables?",
    top_k=5
)

print("\n" + "=" * 70)
print("RETRIEVAL ENGINE COMPLETE")
print("=" * 70)
```

```python
# ============================================================
# SAATHI — REBUILD RAG CORPUS WITH AUTHORED BIS DETAILS
# ============================================================

import os
import json
import pandas as pd
from collections import defaultdict

# ------------------------------------------------------------
# PATHS
# ------------------------------------------------------------

BASE = "/kaggle/input/datasets/bhaskaraaryan/saaaaathi/BIS_CSV"

MASTER_PATH = os.path.join(BASE, "bis_standards_master.csv")
CROSS_PATH = os.path.join(BASE, "bis_cross_references.csv")
INDIAN_PATH = os.path.join(BASE, "bis_indian_references.csv")
INTERNATIONAL_PATH = os.path.join(BASE, "bis_international_references.csv")
REFERRED_PATH = os.path.join(BASE, "bis_referred_by.csv")

AUTHORED_PATH = (
    "/kaggle/input/datasets/bhaskaraaryan/hehe1526/"
    "bis_all_standards_authored_details.csv"
)

OUT_JSON = "/kaggle/working/bis_rag_corpus.json"
OUT_JSONL = "/kaggle/working/bis_rag_corpus.jsonl"

# ------------------------------------------------------------
# HELPERS
# ------------------------------------------------------------

def clean(value):
    if pd.isna(value):
        return ""
    return str(value).strip()

def add_line(lines, label, value):
    value = clean(value)
    if value:
        lines.append(f"{label}: {value}")

# ------------------------------------------------------------
# LOAD DATA
# ------------------------------------------------------------

print("=" * 70)
print("SAATHI — RAG CORPUS REBUILD")
print("WITH FULL BIS AUTHORED DETAILS")
print("=" * 70)

print("\nLoading datasets...")

master = pd.read_csv(MASTER_PATH, low_memory=False)
authored = pd.read_csv(AUTHORED_PATH, low_memory=False)
cross = pd.read_csv(CROSS_PATH, low_memory=False)
indian = pd.read_csv(INDIAN_PATH, low_memory=False)
international = pd.read_csv(INTERNATIONAL_PATH, low_memory=False)
referred = pd.read_csv(REFERRED_PATH, low_memory=False)

print(f"Master standards       : {len(master):,}")
print(f"Authored details       : {len(authored):,}")
print(f"Cross references       : {len(cross):,}")
print(f"Indian references      : {len(indian):,}")
print(f"International refs     : {len(international):,}")
print(f"Referred-by records    : {len(referred):,}")

# ------------------------------------------------------------
# VALIDATE AUTHORED DATA
# ------------------------------------------------------------

print("\nValidating authored details...")

assert len(authored) == len(master), (
    f"Master/authored count mismatch: "
    f"{len(master):,} vs {len(authored):,}"
)

assert authored["internal_id"].is_unique, \
    "Authored internal_id contains duplicates"

assert master["internal_id"].is_unique, \
    "Master internal_id contains duplicates"

master_ids = set(master["internal_id"].astype(str))
authored_ids = set(authored["internal_id"].astype(str))

missing_authored = master_ids - authored_ids
extra_authored = authored_ids - master_ids

assert not missing_authored, \
    f"Missing authored records: {len(missing_authored)}"

assert not extra_authored, \
    f"Extra authored records: {len(extra_authored)}"

print("✓ 35,119 master records matched")
print("✓ 35,119 authored-detail records matched")

# ------------------------------------------------------------
# INDEX AUTHORED DETAILS
# ------------------------------------------------------------

print("\nIndexing authored details...")

authored_idx = {}

detail_columns = [
    "line_1",
    "line_2",
    "line_3",
    "line_4",
    "line_5",
    "line_6",
    "line_7",
    "line_8",
    "line_9",
    "line_10"
]

for row in authored.itertuples(index=False):

    internal_id = clean(row.internal_id)

    details = []

    for col in detail_columns:
        value = clean(getattr(row, col))
        if value:
            details.append(value)

    authored_idx[internal_id] = details

print(f"Indexed authored records: {len(authored_idx):,}")

# ------------------------------------------------------------
# BUILD REFERENCE INDEXES
# ------------------------------------------------------------

print("\nBuilding reference indexes...")

cross_idx = defaultdict(list)
indian_idx = defaultdict(list)
international_idx = defaultdict(list)
referred_idx = defaultdict(list)

for row in cross.itertuples(index=False):
    cross_idx[clean(row.internal_id)].append({
        "reference_number": clean(row.reference_number),
        "reference_is_number": clean(row.reference_is_number),
        "reference_title": clean(row.reference_title),
        "reference_committee": clean(row.reference_committee)
    })

for row in indian.itertuples(index=False):
    indian_idx[clean(row.internal_id)].append({
        "reference_number": clean(row.reference_number),
        "reference_is_number": clean(row.reference_is_number),
        "reference_title": clean(row.reference_title),
        "reference_committee": clean(row.reference_committee)
    })

for row in international.itertuples(index=False):
    international_idx[clean(row.internal_id)].append({
        "reference_number": clean(row.reference_number),
        "international_standard": clean(row.international_standard)
    })

for row in referred.itertuples(index=False):
    referred_idx[clean(row.internal_id)].append({
        "reference_number": clean(row.reference_number),
        "reference_is_number": clean(row.reference_is_number),
        "reference_title": clean(row.reference_title),
        "reference_committee": clean(row.reference_committee)
    })

print(f"Cross-reference IDs    : {len(cross_idx):,}")
print(f"Indian-reference IDs   : {len(indian_idx):,}")
print(f"International IDs      : {len(international_idx):,}")
print(f"Referred-by IDs        : {len(referred_idx):,}")

# ------------------------------------------------------------
# CREATE DOCUMENTS
# ------------------------------------------------------------

print("\nCreating retrieval documents...")

documents = []

# ============================================================
# STANDARD CORE
# ============================================================

for row in master.itertuples(index=False):

    internal_id = clean(row.internal_id)
    is_number = clean(row.IS_number)
    title = clean(row.IS_title)

    lines = []

    # Identity
    add_line(lines, "BIS Standard", is_number)
    add_line(lines, "Title", title)

    # Classification
    add_line(lines, "Technical Department", row.technical_department)
    add_line(lines, "Technical Committee", row.technical_committee)
    add_line(lines, "Aspect", row.aspect)
    add_line(lines, "Group", row.group)
    add_line(lines, "Sub Group", row.sub_group)
    add_line(lines, "Sub Sub Group", row.sub_sub_group)
    add_line(lines, "Language", row.language)

    # Status/history
    add_line(lines, "Superseding IS", row.superseding_IS)
    add_line(lines, "Degree of Equivalence", row.degree_of_equivalence)
    add_line(lines, "Number of Revisions", row.number_of_revisions)
    add_line(lines, "Number of Amendments", row.number_of_amendments)
    add_line(lines, "Reaffirmation Year", row.reaffirmation_year)
    add_line(lines, "Withdrawn Status", row.catalogue_withdrawn_status)

    # Certification / operations
    add_line(lines, "Certification", row.certification)
    add_line(lines, "Amendment Count", row.amendment_count)
    add_line(lines, "Gazette Document Count", row.gazette_document_count)
    add_line(lines, "License Count", row.license_count)
    add_line(lines, "Product Manual SIT Count", row.product_manual_sit_count)
    add_line(lines, "Laboratory Count", row.laboratory_count)
    add_line(lines, "Corrigendum Count", row.corrigendum_count)

    # Catalogue
    add_line(lines, "Catalogue ID", row.catalogue_id)
    add_line(lines, "Catalogue IS Number", row.catalogue_is_no)
    add_line(lines, "Catalogue Title", row.catalogue_title)
    add_line(lines, "Catalogue Amendments", row.catalogue_amendments)
    add_line(lines, "Catalogue Technical Committee",
             row.catalogue_technical_committee)
    add_line(lines, "Catalogue Aspect", row.catalogue_aspect)
    add_line(lines, "Catalogue Reaffirmation", row.catalogue_reaffirmation)

    # Official links
    add_line(lines, "Detail URL", row.detail_url)
    add_line(lines, "Download URL", row.download_url)
    add_line(lines, "Composition URL", row.composition_url)

    # --------------------------------------------------------
    # AUTHORED BIS DETAILS — IMPORTANT
    # --------------------------------------------------------

    details = authored_idx.get(internal_id, [])

    if details:

        lines.append("BIS Authored Details:")

        for i, detail in enumerate(details, 1):
            lines.append(
                f"Detail {i}: {detail}"
            )

    documents.append({
        "doc_id": f"{internal_id}_core",
        "internal_id": internal_id,
        "document_type": "standard_core",
        "is_number": is_number,
        "is_title": title,
        "text": "\n".join(lines)
    })

# ============================================================
# CROSS REFERENCES
# ============================================================

for internal_id, refs in cross_idx.items():

    if not internal_id:
        continue

    lines = [
        f"BIS Standard Internal ID: {internal_id}",
        "Document Type: Cross References"
    ]

    for ref in refs:

        parts = []

        if ref["reference_number"]:
            parts.append(
                f"Reference Number: {ref['reference_number']}"
            )

        if ref["reference_is_number"]:
            parts.append(
                f"Referenced IS: {ref['reference_is_number']}"
            )

        if ref["reference_title"]:
            parts.append(
                f"Referenced Title: {ref['reference_title']}"
            )

        if ref["reference_committee"]:
            parts.append(
                f"Reference Committee: {ref['reference_committee']}"
            )

        if parts:
            lines.append(" | ".join(parts))

    documents.append({
        "doc_id": f"{internal_id}_cross_references",
        "internal_id": internal_id,
        "document_type": "cross_references",
        "text": "\n".join(lines)
    })

# ============================================================
# INDIAN REFERENCES
# ============================================================

for internal_id, refs in indian_idx.items():

    if not internal_id:
        continue

    lines = [
        f"BIS Standard Internal ID: {internal_id}",
        "Document Type: Indian References"
    ]

    for ref in refs:

        parts = []

        if ref["reference_number"]:
            parts.append(
                f"Reference Number: {ref['reference_number']}"
            )

        if ref["reference_is_number"]:
            parts.append(
                f"Referenced IS: {ref['reference_is_number']}"
            )

        if ref["reference_title"]:
            parts.append(
                f"Referenced Title: {ref['reference_title']}"
            )

        if ref["reference_committee"]:
            parts.append(
                f"Reference Committee: {ref['reference_committee']}"
            )

        if parts:
            lines.append(" | ".join(parts))

    documents.append({
        "doc_id": f"{internal_id}_indian_references",
        "internal_id": internal_id,
        "document_type": "indian_references",
        "text": "\n".join(lines)
    })

# ============================================================
# INTERNATIONAL REFERENCES
# ============================================================

for internal_id, refs in international_idx.items():

    if not internal_id:
        continue

    lines = [
        f"BIS Standard Internal ID: {internal_id}",
        "Document Type: International References"
    ]

    for ref in refs:

        parts = []

        if ref["reference_number"]:
            parts.append(
                f"Reference Number: {ref['reference_number']}"
            )

        if ref["international_standard"]:
            parts.append(
                f"International Standard: {ref['international_standard']}"
            )

        if parts:
            lines.append(" | ".join(parts))

    documents.append({
        "doc_id": f"{internal_id}_international_references",
        "internal_id": internal_id,
        "document_type": "international_references",
        "text": "\n".join(lines)
    })

# ============================================================
# REFERRED BY
# ============================================================

for internal_id, refs in referred_idx.items():

    if not internal_id:
        continue

    lines = [
        f"BIS Standard Internal ID: {internal_id}",
        "Document Type: Referred By"
    ]

    for ref in refs:

        parts = []

        if ref["reference_number"]:
            parts.append(
                f"Reference Number: {ref['reference_number']}"
            )

        if ref["reference_is_number"]:
            parts.append(
                f"Referencing IS: {ref['reference_is_number']}"
            )

        if ref["reference_title"]:
            parts.append(
                f"Referencing Title: {ref['reference_title']}"
            )

        if ref["reference_committee"]:
            parts.append(
                f"Reference Committee: {ref['reference_committee']}"
            )

        if parts:
            lines.append(" | ".join(parts))

    documents.append({
        "doc_id": f"{internal_id}_referred_by",
        "internal_id": internal_id,
        "document_type": "referred_by",
        "text": "\n".join(lines)
    })

# ------------------------------------------------------------
# VALIDATION
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("FINAL CORPUS VALIDATION")
print("=" * 70)

print(f"Total documents: {len(documents):,}")

counts = defaultdict(int)

for doc in documents:
    counts[doc["document_type"]] += 1

for doc_type in sorted(counts):
    print(f"{doc_type:30s}: {counts[doc_type]:,}")

empty = sum(
    1 for doc in documents
    if not doc["text"].strip()
)

core_docs = [
    d for d in documents
    if d["document_type"] == "standard_core"
]

core_with_details = sum(
    1 for d in core_docs
    if "BIS Authored Details:" in d["text"]
)

print(f"\nEmpty documents       : {empty:,}")
print(f"Core documents        : {len(core_docs):,}")
print(f"Core with BIS details : {core_with_details:,}")

assert len(documents) == 116597
assert empty == 0
assert len(core_docs) == 35119
assert core_with_details == 35119

# ------------------------------------------------------------
# SAVE
# ------------------------------------------------------------

print("\nSaving JSON...")

with open(OUT_JSON, "w", encoding="utf-8") as f:
    json.dump(
        documents,
        f,
        ensure_ascii=False,
        separators=(",", ":")
    )

print("Saved:", OUT_JSON)

print("\nSaving JSONL...")

with open(OUT_JSONL, "w", encoding="utf-8") as f:
    for doc in documents:
        f.write(
            json.dumps(
                doc,
                ensure_ascii=False,
                separators=(",", ":")
            ) + "\n"
        )

print("Saved:", OUT_JSONL)

print("\n" + "=" * 70)
print("CORPUS READY FOR RE-EMBEDDING")
print("=" * 70)
```

```python
# ============================================================
# SAATHI — CELL 7
# HYBRID RETRIEVAL ENGINE
#
# Exact BIS lookup + lexical retrieval + FAISS semantic search
# ============================================================

import json
import re
import numpy as np
import faiss
import torch

from collections import defaultdict
from sentence_transformers import SentenceTransformer


# ============================================================
# CONFIG
# ============================================================

CORPUS_PATH = "/kaggle/working/bis_rag_corpus.json"
FAISS_PATH = "/kaggle/working/bis_faiss.index"
METADATA_PATH = "/kaggle/working/bis_rag_metadata.json"

MODEL_NAME = "BAAI/bge-small-en-v1.5"

SEMANTIC_TOP_K = 30
FINAL_TOP_K = 10


# ============================================================
# LOAD DATA
# ============================================================

print("=" * 70)
print("SAATHI — HYBRID RETRIEVAL ENGINE")
print("=" * 70)

print("\nLoading corpus...")

with open(CORPUS_PATH, "r", encoding="utf-8") as f:
    corpus = json.load(f)

print(f"Corpus documents: {len(corpus):,}")

assert len(corpus) == 116597


print("\nLoading metadata...")

with open(METADATA_PATH, "r", encoding="utf-8") as f:
    metadata = json.load(f)

print(f"Metadata records: {len(metadata):,}")

assert len(metadata) == 116597


print("\nLoading FAISS index...")

index = faiss.read_index(FAISS_PATH)

print(f"FAISS vectors: {index.ntotal:,}")

assert index.ntotal == 116597


# ============================================================
# BUILD EXACT LOOKUP INDEX
# ============================================================

print("\nBuilding exact BIS-number lookup...")

exact_lookup = defaultdict(list)

for i, doc in enumerate(corpus):

    # Standard core
    if doc.get("document_type") == "standard_core":

        is_number = str(
            doc.get("is_number", "")
        ).strip()

        if is_number:
            exact_lookup[
                is_number.upper()
            ].append(i)

        # Also index catalogue number
        text = doc.get("text", "")

        catalogue_match = re.search(
            r"Catalogue IS Number:\s*(.+)",
            text,
            re.IGNORECASE
        )

        if catalogue_match:

            catalogue_number = (
                catalogue_match.group(1)
                .strip()
                .upper()
            )

            if catalogue_number:
                exact_lookup[
                    catalogue_number
                ].append(i)


print(
    f"Exact lookup keys: {len(exact_lookup):,}"
)


# ============================================================
# NORMALIZATION
# ============================================================

def normalize_is_number(value):

    value = str(value).upper().strip()

    # Normalize whitespace
    value = re.sub(r"\s+", " ", value)

    # Normalize common spacing around punctuation
    value = re.sub(r"\s*:\s*", ":", value)
    value = re.sub(r"\s*-\s*", "-", value)

    return value


# ============================================================
# EXTRACT POSSIBLE BIS NUMBER FROM QUERY
# ============================================================

def extract_is_numbers(query):

    query_upper = query.upper()

    candidates = []

    # IS / IEC forms
    patterns = [

        r"\bIS/IEC\s*[:\-]?\s*\d+(?:[-:]\d+)*",
        r"\bIS\s*[:\-]?\s*\d+(?:[-:]\d+)*",
        r"\bIEC\s*\d+(?:[-:]\d+)*",

    ]

    for pattern in patterns:

        matches = re.findall(
            pattern,
            query_upper
        )

        candidates.extend(matches)

    # Clean
    cleaned = []

    for value in candidates:

        value = normalize_is_number(value)

        if value not in cleaned:
            cleaned.append(value)

    return cleaned


# ============================================================
# SEMANTIC SEARCH
# ============================================================

print("\nLoading embedding model...")

device = "cuda" if torch.cuda.is_available() else "cpu"

model = SentenceTransformer(
    MODEL_NAME,
    device=device
)

if device == "cuda":
    model.half()

print("Model:", MODEL_NAME)
print("Device:", device)


def semantic_search(query, top_k=SEMANTIC_TOP_K):

    query_embedding = model.encode(
        [query],
        convert_to_numpy=True,
        normalize_embeddings=True
    ).astype("float32")

    scores, indices = index.search(
        query_embedding,
        top_k
    )

    results = []

    for score, idx in zip(
        scores[0],
        indices[0]
    ):

        if idx < 0:
            continue

        results.append({
            "index": int(idx),
            "score": float(score),
            "source": "semantic"
        })

    return results


# ============================================================
# EXACT SEARCH
# ============================================================

def exact_search(query):

    numbers = extract_is_numbers(query)

    results = []

    seen = set()

    for number in numbers:

        # Direct lookup
        matches = exact_lookup.get(
            number,
            []
        )

        # Try normalized keys
        if not matches:

            normalized_query = normalize_is_number(
                number
            )

            for key, indices in exact_lookup.items():

                if normalize_is_number(key) == normalized_query:
                    matches.extend(indices)

        for idx in matches:

            if idx in seen:
                continue

            seen.add(idx)

            results.append({
                "index": idx,
                "score": 1.0,
                "source": "exact",
                "matched_number": number
            })

    return results


# ============================================================
# RESULT FUSION
# ============================================================

def hybrid_search(
    query,
    semantic_top_k=SEMANTIC_TOP_K,
    final_top_k=FINAL_TOP_K
):

    exact_results = exact_search(query)

    semantic_results = semantic_search(
        query,
        semantic_top_k
    )

    combined = {}

    # --------------------------------------------------------
    # EXACT RESULTS
    # --------------------------------------------------------

    for result in exact_results:

        idx = result["index"]

        combined[idx] = {
            "index": idx,
            "exact_score": 1.0,
            "semantic_score": 0.0,
            "sources": ["exact"]
        }

    # --------------------------------------------------------
    # SEMANTIC RESULTS
    # --------------------------------------------------------

    for rank, result in enumerate(
        semantic_results,
        start=1
    ):

        idx = result["index"]

        semantic_score = result["score"]

        if idx not in combined:

            combined[idx] = {
                "index": idx,
                "exact_score": 0.0,
                "semantic_score": semantic_score,
                "sources": ["semantic"]
            }

        else:

            combined[idx]["semantic_score"] = (
                semantic_score
            )

            combined[idx]["sources"].append(
                "semantic"
            )

    # --------------------------------------------------------
    # FINAL SCORING
    # --------------------------------------------------------

    ranked = []

    for item in combined.values():

        # Exact matches receive a very strong boost.
        exact_component = (
            item["exact_score"] * 10.0
        )

        semantic_component = (
            item["semantic_score"]
        )

        final_score = (
            exact_component +
            semantic_component
        )

        item["final_score"] = final_score

        ranked.append(item)

    ranked.sort(
        key=lambda x: x["final_score"],
        reverse=True
    )

    # --------------------------------------------------------
    # BUILD FINAL DOCUMENT RESULTS
    # --------------------------------------------------------

    results = []

    for rank, item in enumerate(
        ranked[:final_top_k],
        start=1
    ):

        idx = item["index"]

        doc = corpus[idx]
        meta = metadata[idx]

        results.append({

            "rank": rank,

            "score": item["final_score"],

            "exact_score": item["exact_score"],

            "semantic_score": item["semantic_score"],

            "sources": item["sources"],

            "doc_id": meta["doc_id"],

            "internal_id": meta["internal_id"],

            "document_type": meta["document_type"],

            "is_number": meta.get(
                "is_number",
                ""
            ),

            "is_title": meta.get(
                "is_title",
                ""
            ),

            "text": doc["text"]

        })

    return results


# ============================================================
# DISPLAY FUNCTION
# ============================================================

def show_hybrid_results(
    query,
    top_k=10
):

    print("\n" + "=" * 70)
    print("QUERY")
    print("=" * 70)

    print(query)

    detected = extract_is_numbers(query)

    if detected:

        print("\nDetected BIS number(s):")

        for number in detected:
            print("  →", number)

    results = hybrid_search(
        query,
        final_top_k=top_k
    )

    print("\nRESULTS")
    print("-" * 70)

    for result in results:

        print(
            f"\n#{result['rank']} "
            f"[score={result['score']:.4f}] "
            f"[{result['document_type']}]"
        )

        print(
            f"IS: {result['is_number']}"
        )

        print(
            f"Title: {result['is_title']}"
        )

        print(
            f"Internal ID: {result['internal_id']}"
        )

        print(
            f"Source: {', '.join(result['sources'])}"
        )

        print(
            "Text:"
        )

        print(
            result["text"][:1000]
        )


# ============================================================
# TEST 1 — EXACT STANDARD NUMBER
# ============================================================

show_hybrid_results(
    "What is IS/IEC 60691:2023?",
    top_k=5
)


# ============================================================
# TEST 2 — NATURAL LANGUAGE
# ============================================================

show_hybrid_results(
    "Which BIS standard deals with thermal links?",
    top_k=5
)


# ============================================================
# TEST 3 — TECHNICAL TOPIC
# ============================================================

show_hybrid_results(
    "Which standard deals with coaxial communication cables?",
    top_k=5
)


# ============================================================
# TEST 4 — REFERENCE QUESTION
# ============================================================

show_hybrid_results(
    "Which BIS standards refer to IS/IEC 61196-1:2005?",
    top_k=5
)


# ============================================================
# COMPLETE
# ============================================================

print("\n" + "=" * 70)
print("HYBRID RETRIEVAL ENGINE READY")
print("=" * 70)
```

```python
# ============================================================
# SAATHI — CELL 8
# RELATIONSHIP-AWARE RETRIEVAL
#
# Exact standard lookup
# "Referenced by" lookup
# "References" lookup
# Indian references
# International references
# ============================================================

import os
import re
import json
import pandas as pd
from collections import defaultdict


# ============================================================
# PATHS
# ============================================================

BASE = "/kaggle/input/datasets/bhaskaraaryan/saaaaathi/BIS_CSV"

MASTER_PATH = os.path.join(
    BASE,
    "bis_standards_master.csv"
)

CROSS_PATH = os.path.join(
    BASE,
    "bis_cross_references.csv"
)

INDIAN_PATH = os.path.join(
    BASE,
    "bis_indian_references.csv"
)

INTERNATIONAL_PATH = os.path.join(
    BASE,
    "bis_international_references.csv"
)

REFERRED_PATH = os.path.join(
    BASE,
    "bis_referred_by.csv"
)


# ============================================================
# HELPERS
# ============================================================

def clean(value):

    if pd.isna(value):
        return ""

    return str(value).strip()


def normalize_is(value):

    value = clean(value).upper()

    value = re.sub(
        r"\s+",
        " ",
        value
    )

    value = re.sub(
        r"\s*:\s*",
        ":",
        value
    )

    value = re.sub(
        r"\s*-\s*",
        "-",
        value
    )

    return value.strip()


# ============================================================
# LOAD DATA
# ============================================================

print("=" * 70)
print("SAATHI — RELATIONSHIP-AWARE RETRIEVAL")
print("=" * 70)

print("\nLoading BIS datasets...")

master = pd.read_csv(
    MASTER_PATH,
    low_memory=False
)

cross = pd.read_csv(
    CROSS_PATH,
    low_memory=False
)

indian = pd.read_csv(
    INDIAN_PATH,
    low_memory=False
)

international = pd.read_csv(
    INTERNATIONAL_PATH,
    low_memory=False
)

referred = pd.read_csv(
    REFERRED_PATH,
    low_memory=False
)

print(
    f"Master standards       : {len(master):,}"
)

print(
    f"Cross references       : {len(cross):,}"
)

print(
    f"Indian references      : {len(indian):,}"
)

print(
    f"International refs     : {len(international):,}"
)

print(
    f"Referred-by records    : {len(referred):,}"
)


# ============================================================
# STANDARD LOOKUP
# ============================================================

print("\nBuilding standard lookup...")

standard_by_number = {}

standard_by_normalized_number = {}

standard_by_internal_id = {}

for row in master.itertuples(index=False):

    internal_id = clean(
        row.internal_id
    )

    is_number = clean(
        row.IS_number
    )

    title = clean(
        row.IS_title
    )

    record = {
        "internal_id": internal_id,
        "is_number": is_number,
        "title": title,
        "technical_department": clean(
            row.technical_department
        ),
        "technical_committee": clean(
            row.technical_committee
        ),
        "aspect": clean(
            row.aspect
        ),
        "superseding_is": clean(
            row.superseding_IS
        ),
        "withdrawn_status": clean(
            row.catalogue_withdrawn_status
        ),
        "detail_url": clean(
            row.detail_url
        ),
        "download_url": clean(
            row.download_url
        )
    }

    if is_number:

        standard_by_number[
            is_number.upper()
        ] = record

        standard_by_normalized_number[
            normalize_is(is_number)
        ] = record

    standard_by_internal_id[
        internal_id
    ] = record


print(
    f"Standards indexed: "
    f"{len(standard_by_internal_id):,}"
)


# ============================================================
# RELATIONSHIP INDEX 1
#
# STANDARD → CROSS REFERENCES
# ============================================================

print("\nBuilding cross-reference graph...")

references_by_standard = defaultdict(list)

for row in cross.itertuples(index=False):

    source_id = clean(
        row.internal_id
    )

    if not source_id:
        continue

    target_number = clean(
        row.reference_is_number
    )

    references_by_standard[
        source_id
    ].append({

        "reference_number": clean(
            row.reference_number
        ),

        "target_is": target_number,

        "target_normalized": normalize_is(
            target_number
        ),

        "target_title": clean(
            row.reference_title
        ),

        "committee": clean(
            row.reference_committee
        ),

        "relationship": "cross_reference"
    })


print(
    f"Standards with cross references: "
    f"{len(references_by_standard):,}"
)


# ============================================================
# RELATIONSHIP INDEX 2
#
# STANDARD → INDIAN REFERENCES
# ============================================================

print("\nBuilding Indian-reference graph...")

indian_refs_by_standard = defaultdict(list)

for row in indian.itertuples(index=False):

    source_id = clean(
        row.internal_id
    )

    if not source_id:
        continue

    target_number = clean(
        row.reference_is_number
    )

    indian_refs_by_standard[
        source_id
    ].append({

        "reference_number": clean(
            row.reference_number
        ),

        "target_is": target_number,

        "target_normalized": normalize_is(
            target_number
        ),

        "target_title": clean(
            row.reference_title
        ),

        "committee": clean(
            row.reference_committee
        ),

        "relationship": "indian_reference"
    })


print(
    f"Standards with Indian references: "
    f"{len(indian_refs_by_standard):,}"
)


# ============================================================
# RELATIONSHIP INDEX 3
#
# STANDARD → INTERNATIONAL REFERENCES
# ============================================================

print("\nBuilding international-reference graph...")

international_refs_by_standard = defaultdict(list)

for row in international.itertuples(index=False):

    source_id = clean(
        row.internal_id
    )

    if not source_id:
        continue

    international_refs_by_standard[
        source_id
    ].append({

        "reference_number": clean(
            row.reference_number
        ),

        "international_standard": clean(
            row.international_standard
        ),

        "relationship": "international_reference"
    })


print(
    f"Standards with international references: "
    f"{len(international_refs_by_standard):,}"
)


# ============================================================
# RELATIONSHIP INDEX 4
#
# TARGET STANDARD → STANDARDS REFERENCING IT
# ============================================================

print("\nBuilding reverse reference graph...")

referred_by_standard = defaultdict(list)

for row in referred.itertuples(index=False):

    target_id = clean(
        row.internal_id
    )

    if not target_id:
        continue

    referencing_number = clean(
        row.reference_is_number
    )

    referred_by_standard[
        target_id
    ].append({

        "reference_number": clean(
            row.reference_number
        ),

        "referencing_is": referencing_number,

        "referencing_normalized": normalize_is(
            referencing_number
        ),

        "referencing_title": clean(
            row.reference_title
        ),

        "committee": clean(
            row.reference_committee
        ),

        "relationship": "referred_by"
    })


print(
    f"Standards with referred-by records: "
    f"{len(referred_by_standard):,}"
)


# ============================================================
# FIND STANDARD BY NUMBER
# ============================================================

def find_standard(is_number):

    normalized = normalize_is(
        is_number
    )

    # Direct normalized lookup
    result = standard_by_normalized_number.get(
        normalized
    )

    if result:
        return result

    # Try substring matching for catalogue formatting
    for key, record in standard_by_normalized_number.items():

        if key == normalized:
            return record

    return None


# ============================================================
# FIND STANDARD INTERNAL ID
# ============================================================

def get_internal_id(is_number):

    standard = find_standard(
        is_number
    )

    if standard:
        return standard["internal_id"]

    return None


# ============================================================
# GET RELATIONSHIPS
# ============================================================

def get_relationships(is_number):

    standard = find_standard(
        is_number
    )

    if not standard:

        return {
            "standard": None,
            "cross_references": [],
            "indian_references": [],
            "international_references": [],
            "referred_by": []
        }

    internal_id = standard[
        "internal_id"
    ]

    return {

        "standard": standard,

        "cross_references":
            references_by_standard.get(
                internal_id,
                []
            ),

        "indian_references":
            indian_refs_by_standard.get(
                internal_id,
                []
            ),

        "international_references":
            international_refs_by_standard.get(
                internal_id,
                []
            ),

        "referred_by":
            referred_by_standard.get(
                internal_id,
                []
            )
    }


# ============================================================
# DISPLAY RELATIONSHIPS
# ============================================================

def show_relationships(is_number):

    print("\n" + "=" * 70)
    print(
        "RELATIONSHIP QUERY:",
        is_number
    )
    print("=" * 70)

    data = get_relationships(
        is_number
    )

    standard = data["standard"]

    if not standard:

        print(
            "\nStandard not found:"
        )

        print(
            is_number
        )

        return

    print(
        f"\nStandard: {standard['is_number']}"
    )

    print(
        f"Title: {standard['title']}"
    )

    print(
        f"Internal ID: {standard['internal_id']}"
    )


    # --------------------------------------------------------
    # CROSS REFERENCES
    # --------------------------------------------------------

    refs = data[
        "cross_references"
    ]

    print(
        f"\nCross references: {len(refs):,}"
    )

    for ref in refs[:20]:

        print(
            f"  → {ref['target_is']} | "
            f"{ref['target_title']}"
        )


    # --------------------------------------------------------
    # INDIAN REFERENCES
    # --------------------------------------------------------

    refs = data[
        "indian_references"
    ]

    print(
        f"\nIndian references: {len(refs):,}"
    )

    for ref in refs[:20]:

        print(
            f"  → {ref['target_is']} | "
            f"{ref['target_title']}"
        )


    # --------------------------------------------------------
    # INTERNATIONAL
    # --------------------------------------------------------

    refs = data[
        "international_references"
    ]

    print(
        f"\nInternational references: {len(refs):,}"
    )

    for ref in refs[:20]:

        print(
            f"  → {ref['international_standard']}"
        )


    # --------------------------------------------------------
    # REFERRED BY
    # --------------------------------------------------------

    refs = data[
        "referred_by"
    ]

    print(
        f"\nReferred by: {len(refs):,}"
    )

    for ref in refs[:20]:

        print(
            f"  ← {ref['referencing_is']} | "
            f"{ref['referencing_title']}"
        )


# ============================================================
# TEST 1
# ============================================================

show_relationships(
    "IS/IEC 61196-1:2005"
)


# ============================================================
# TEST 2
# ============================================================

show_relationships(
    "IS/IEC 60691:2023"
)


# ============================================================
# COMPLETE
# ============================================================

print("\n" + "=" * 70)
print("RELATIONSHIP ENGINE READY")
print("=" * 70)
```

```python
# ============================================================
# SAATHI — CELL 9
# INTENT-AWARE HYBRID RETRIEVAL
#
# EXACT STANDARD
# RELATIONSHIP / REVERSE RELATIONSHIP
# SEMANTIC FAISS
# MIXED QUERIES
# ============================================================

import re
import json
import numpy as np
import pandas as pd
import faiss

from collections import defaultdict


# ============================================================
# LOAD EXISTING RAG ARTIFACTS
# ============================================================

CORPUS_PATH = "/kaggle/working/bis_rag_corpus.json"
METADATA_PATH = "/kaggle/working/bis_rag_metadata.json"
FAISS_PATH = "/kaggle/working/bis_faiss.index"

print("=" * 70)
print("SAATHI — CELL 9")
print("INTENT-AWARE HYBRID RETRIEVAL")
print("=" * 70)

with open(
    CORPUS_PATH,
    "r",
    encoding="utf-8"
) as f:

    corpus = json.load(f)


with open(
    METADATA_PATH,
    "r",
    encoding="utf-8"
) as f:

    metadata = json.load(f)


index = faiss.read_index(
    FAISS_PATH
)

print(
    f"\nCorpus documents : {len(corpus):,}"
)

print(
    f"Metadata records : {len(metadata):,}"
)

print(
    f"FAISS vectors    : {index.ntotal:,}"
)


# ============================================================
# ROBUST BIS IDENTIFIER NORMALIZATION
# ============================================================

def normalize_identifier(value):

    if value is None:
        return None

    text = str(value).upper().strip()

    # Normalize unicode punctuation
    text = text.replace("–", "-")
    text = text.replace("—", "-")

    # Remove brackets around formatting
    text = re.sub(
        r"[\[\]\(\)]",
        " ",
        text
    )

    # Collapse whitespace
    text = re.sub(
        r"\s+",
        " ",
        text
    ).strip()


    # --------------------------------------------------------
    # Handle malformed forms such as:
    #
    # IS/IEC :2023IEC 60691:2023
    # IS/IEC :2015IEC 60691 : 2015
    #
    # We look for the final IEC number/year pair.
    # --------------------------------------------------------

    m = re.search(
        r"IEC\s+(\d+(?:-\d+)?)\s*:\s*(\d{4})",
        text
    )

    if m:

        number = m.group(1)
        year = m.group(2)

        prefix = "IS/IEC"

        if "IS/IEC" in text:
            prefix = "IS/IEC"

        elif re.search(
            r"\bIEC\b",
            text
        ):
            prefix = "IEC"

        return f"{prefix} {number}:{year}"


    # --------------------------------------------------------
    # Normal IS forms
    # --------------------------------------------------------

    m = re.search(
        r"\bIS\s*/?\s*IEC\s*"
        r"(\d+(?:-\d+)?)\s*:\s*(\d{4})",
        text
    )

    if m:

        return (
            f"IS/IEC "
            f"{m.group(1)}:"
            f"{m.group(2)}"
        )


    # --------------------------------------------------------
    # Plain IS number
    # --------------------------------------------------------

    m = re.search(
        r"\bIS\s+"
        r"(\d+(?:-\d+)?)\s*:\s*(\d{4})",
        text
    )

    if m:

        return (
            f"IS "
            f"{m.group(1)}:"
            f"{m.group(2)}"
        )


    return re.sub(
        r"\s+",
        " ",
        text
    )


# ============================================================
# EXTRACT IDENTIFIERS FROM QUERY
# ============================================================

def extract_identifiers(query):

    text = query.upper()

    found = []

    # --------------------------------------------------------
    # IS/IEC
    # --------------------------------------------------------

    patterns = [

        r"\bIS\s*/\s*IEC\s*:?\s*"
        r"(?:\d{4}\s*)?"
        r"(\d+(?:-\d+)?)\s*:\s*(\d{4})",

        r"\bIEC\s+"
        r"(\d+(?:-\d+)?)\s*:\s*(\d{4})",

        r"\bIS\s+"
        r"(\d+(?:-\d+)?)\s*:\s*(\d{4})"
    ]


    for pattern in patterns:

        for match in re.finditer(
            pattern,
            text
        ):

            number = match.group(1)
            year = match.group(2)

            prefix = (
                "IS/IEC"
                if "IS/IEC" in match.group(0)
                else (
                    "IEC"
                    if "IEC" in match.group(0)
                    else "IS"
                )
            )

            identifier = (
                f"{prefix} "
                f"{number}:"
                f"{year}"
            )

            if identifier not in found:

                found.append(
                    identifier
                )


    return found


# ============================================================
# BUILD CANONICAL STANDARD INDEX
# ============================================================

standard_records = {}

for meta in metadata:

    if meta.get("doc_type") != "standard_core":
        continue

    internal_id = str(
        meta.get(
            "internal_id",
            ""
        )
    )

    if not internal_id:
        continue

    is_number = meta.get(
        "is_number",
        ""
    )

    normalized = normalize_identifier(
        is_number
    )

    if normalized:

        standard_records[
            normalized
        ] = {
            "internal_id": internal_id,
            "is_number": is_number,
            "title": meta.get(
                "title",
                ""
            )
        }


print(
    f"\nCanonical standard identifiers: "
    f"{len(standard_records):,}"
)


# ============================================================
# SEMANTIC RETRIEVAL
# ============================================================

def semantic_search(
    query,
    top_k=20
):

    # This expects the embedding model created
    # in the previous retrieval cell to still exist.

    query_embedding = model.encode(
        [query],
        normalize_embeddings=True,
        show_progress_bar=False
    )

    query_embedding = np.asarray(
        query_embedding,
        dtype="float32"
    )

    scores, indices = index.search(
        query_embedding,
        top_k
    )

    results = []

    for score, idx in zip(
        scores[0],
        indices[0]
    ):

        if idx < 0:
            continue

        results.append({

            "rank": len(results) + 1,

            "score": float(score),

            "source": "semantic",

            "document_index": int(idx),

            "document": corpus[idx]
        })

    return results


# ============================================================
# EXACT STANDARD RETRIEVAL
# ============================================================

def exact_standard_search(
    identifier
):

    canonical = normalize_identifier(
        identifier
    )

    record = standard_records.get(
        canonical
    )

    if not record:
        return []


    # Locate corresponding standard_core document
    results = []

    for i, meta in enumerate(metadata):

        if meta.get(
            "doc_type"
        ) != "standard_core":

            continue

        if str(
            meta.get(
                "internal_id",
                ""
            )
        ) != record["internal_id"]:

            continue

        results.append({

            "rank": 1,

            "score": 100.0,

            "source": "exact",

            "document_index": i,

            "document": corpus[i],

            "identifier": canonical,

            "internal_id":
                record["internal_id"],

            "title":
                record["title"]
        })

        break


    return results


# ============================================================
# QUERY INTENT DETECTION
# ============================================================

def detect_intent(query):

    q = query.lower().strip()

    identifiers = extract_identifiers(
        query
    )


    # --------------------------------------------------------
    # REVERSE REFERENCE
    # --------------------------------------------------------

    reverse_patterns = [

        "which standards refer to",
        "which standards reference",
        "which bis standards refer to",
        "which bis standards reference",
        "who refers to",
        "who references",
        "referred by",
        "referenced by",
        "standards referring to",
        "standards referencing"
    ]

    if any(
        p in q
        for p in reverse_patterns
    ):

        return {
            "intent": "reverse_reference",
            "identifiers": identifiers
        }


    # --------------------------------------------------------
    # OUTGOING REFERENCES
    # --------------------------------------------------------

    outgoing_patterns = [

        "what does it reference",
        "what does it refer to",
        "what standards does",
        "references of",
        "references made by",
        "standards referenced by",
        "standards referred to by",
        "references from"
    ]

    if any(
        p in q
        for p in outgoing_patterns
    ):

        return {
            "intent": "outgoing_reference",
            "identifiers": identifiers
        }


    # --------------------------------------------------------
    # INTERNATIONAL
    # --------------------------------------------------------

    international_patterns = [

        "international standard",
        "international standards",
        "iec reference",
        "iec references",
        "based on iec"
    ]

    if any(
        p in q
        for p in international_patterns
    ) and identifiers:

        return {
            "intent": "international_reference",
            "identifiers": identifiers
        }


    # --------------------------------------------------------
    # EXACT STANDARD
    # --------------------------------------------------------

    exact_patterns = [

        "what is",
        "tell me about",
        "details of",
        "details for",
        "information about",
        "give me details",
        "describe"
    ]

    if identifiers and any(
        p in q
        for p in exact_patterns
    ):

        return {
            "intent": "exact_standard",
            "identifiers": identifiers
        }


    # --------------------------------------------------------
    # IDENTIFIER WITHOUT EXPLICIT INTENT
    # --------------------------------------------------------

    if identifiers:

        return {
            "intent": "exact_standard",
            "identifiers": identifiers
        }


    # --------------------------------------------------------
    # PURE SEMANTIC
    # --------------------------------------------------------

    return {
        "intent": "semantic",
        "identifiers": []
    }


# ============================================================
# RELATIONSHIP RETRIEVAL
# ============================================================

def relationship_search(
    identifier,
    relationship_type
):

    canonical = normalize_identifier(
        identifier
    )

    standard = standard_records.get(
        canonical
    )

    if not standard:
        return []


    internal_id = standard[
        "internal_id"
    ]

    results = []


    # --------------------------------------------------------
    # Scan relationship documents in corpus.
    #
    # We intentionally use metadata/document text here,
    # because the relationship documents already contain
    # source and target information.
    # --------------------------------------------------------

    for i, meta in enumerate(metadata):

        doc_type = meta.get(
            "doc_type"
        )

        if relationship_type == "reverse":

            allowed = {
                "referred_by",
                "cross_references",
                "indian_references"
            }

        elif relationship_type == "outgoing":

            allowed = {
                "cross_references",
                "indian_references",
                "international_references"
            }

        else:

            allowed = {
                "international_references"
            }


        if doc_type not in allowed:
            continue


        doc = corpus[i]

        # ----------------------------------------------------
        # For reverse lookup, relationship document must point
        # to the target standard.
        # ----------------------------------------------------

        if relationship_type == "reverse":

            target_found = (
                canonical.lower()
                in doc.lower()
            )

            if not target_found:
                continue


            # Do not return the queried standard itself
            doc_is_number = normalize_identifier(
                meta.get(
                    "is_number",
                    ""
                )
            )

            if doc_is_number == canonical:
                continue


        # ----------------------------------------------------
        # Outgoing relationship:
        # document must belong to the queried source standard.
        # ----------------------------------------------------

        elif relationship_type == "outgoing":

            source_id = str(
                meta.get(
                    "internal_id",
                    ""
                )
            )

            if source_id != internal_id:
                continue


        # ----------------------------------------------------
        # International relationship
        # ----------------------------------------------------

        elif relationship_type == "international":

            source_id = str(
                meta.get(
                    "internal_id",
                    ""
                )
            )

            if source_id != internal_id:
                continue


        results.append({

            "rank": len(results) + 1,

            "score": 50.0,

            "source": "relationship",

            "document_index": i,

            "document": doc,

            "relationship_type":
                relationship_type
        })


    return results


# ============================================================
# MAIN RETRIEVER
# ============================================================

def retrieve(
    query,
    top_k=10
):

    intent_data = detect_intent(
        query
    )

    intent = intent_data[
        "intent"
    ]

    identifiers = intent_data[
        "identifiers"
    ]


    results = []


    # ========================================================
    # EXACT STANDARD
    # ========================================================

    if intent == "exact_standard":

        for identifier in identifiers:

            results.extend(
                exact_standard_search(
                    identifier
                )
            )

        # Add semantic support
        if len(results) < top_k:

            semantic = semantic_search(
                query,
                top_k=top_k
            )

            exact_indexes = {
                r["document_index"]
                for r in results
            }

            for result in semantic:

                if result[
                    "document_index"
                ] in exact_indexes:

                    continue

                results.append(
                    result
                )

                if len(results) >= top_k:
                    break


    # ========================================================
    # REVERSE REFERENCE
    # ========================================================

    elif intent == "reverse_reference":

        for identifier in identifiers:

            results.extend(
                relationship_search(
                    identifier,
                    "reverse"
                )
            )


    # ========================================================
    # OUTGOING REFERENCE
    # ========================================================

    elif intent == "outgoing_reference":

        for identifier in identifiers:

            results.extend(
                relationship_search(
                    identifier,
                    "outgoing"
                )
            )


    # ========================================================
    # INTERNATIONAL
    # ========================================================

    elif intent == "international_reference":

        for identifier in identifiers:

            results.extend(
                relationship_search(
                    identifier,
                    "international"
                )
            )


    # ========================================================
    # PURE SEMANTIC
    # ========================================================

    elif intent == "semantic":

        results = semantic_search(
            query,
            top_k=top_k
        )


    # ========================================================
    # DEDUPLICATE
    # ========================================================

    seen = set()

    deduped = []

    for result in results:

        idx = result.get(
            "document_index"
        )

        if idx in seen:
            continue

        seen.add(idx)

        deduped.append(
            result
        )

    results = deduped[
        :top_k
    ]


    # Re-rank
    for rank, result in enumerate(
        results,
        start=1
    ):

        result["rank"] = rank


    return {
        "query": query,
        "intent": intent,
        "identifiers": identifiers,
        "results": results
    }


# ============================================================
# RESULT DISPLAY
# ============================================================

def display_results(
    response
):

    print("\n" + "=" * 70)

    print(
        "QUERY:",
        response["query"]
    )

    print(
        "INTENT:",
        response["intent"]
    )

    print(
        "IDENTIFIERS:",
        response["identifiers"]
    )

    print(
        "RESULTS:",
        len(response["results"])
    )

    print("=" * 70)


    for result in response["results"]:

        print(
            f"\n#{result['rank']} "
            f"[{result['source']}] "
            f"score={result['score']:.4f}"
        )

        print(
            result["document"][:900]
        )

        print("-" * 70)


# ============================================================
# TEST 1 — EXACT STANDARD
# ============================================================

response = retrieve(
    "What is IS/IEC 60691:2023?",
    top_k=5
)

display_results(
    response
)


# ============================================================
# TEST 2 — SEMANTIC
# ============================================================

response = retrieve(
    "Which BIS standard deals with thermal links?",
    top_k=5
)

display_results(
    response
)


# ============================================================
# TEST 3 — REVERSE RELATIONSHIP
# ============================================================

response = retrieve(
    "Which BIS standards refer to IS/IEC 61196-1:2005?",
    top_k=10
)

display_results(
    response
)


# ============================================================
# TEST 4 — OUTGOING RELATIONSHIP
# ============================================================

response = retrieve(
    "What standards does IS/IEC 61196-1:2005 refer to?",
    top_k=10
)

display_results(
    response
)


# ============================================================
# TEST 5 — MIXED QUERY
# ============================================================

response = retrieve(
    "Tell me about IS/IEC 60691:2023 and its references.",
    top_k=10
)

display_results(
    response
)


print("\n" + "=" * 70)
print("CELL 9 COMPLETE")
print("=" * 70)
```

```python
# ============================================================
# SAATHI — CELL 9 FIXED
# INTENT-AWARE HYBRID RETRIEVAL
#
# Uses:
#   - MASTER CSV for authoritative standard lookup
#   - RELATIONSHIP GRAPHS from Cell 8
#   - EXISTING FAISS INDEX from Cell 7
#   - EXISTING RAG CORPUS
#
# DOES NOT REBUILD EMBEDDINGS
# ============================================================

import re
import json
import numpy as np
import pandas as pd
import faiss

from collections import defaultdict


# ============================================================
# PATHS
# ============================================================

BASE = "/kaggle/input/datasets/bhaskaraaryan/saaaaathi/BIS_CSV"

MASTER_PATH = os.path.join(
    BASE,
    "bis_standards_master.csv"
)

CORPUS_PATH = "/kaggle/working/bis_rag_corpus.json"
METADATA_PATH = "/kaggle/working/bis_rag_metadata.json"
FAISS_PATH = "/kaggle/working/bis_faiss.index"


# ============================================================
# HEADER
# ============================================================

print("=" * 70)
print("SAATHI — CELL 9 FIXED")
print("INTENT-AWARE HYBRID RETRIEVAL")
print("=" * 70)


# ============================================================
# LOAD EXISTING RAG ARTIFACTS
# ============================================================

with open(
    CORPUS_PATH,
    "r",
    encoding="utf-8"
) as f:

    corpus = json.load(f)


with open(
    METADATA_PATH,
    "r",
    encoding="utf-8"
) as f:

    metadata = json.load(f)


index = faiss.read_index(
    FAISS_PATH
)


print(
    f"\nCorpus documents : {len(corpus):,}"
)

print(
    f"Metadata records : {len(metadata):,}"
)

print(
    f"FAISS vectors    : {index.ntotal:,}"
)


# ============================================================
# LOAD MASTER
#
# IMPORTANT:
# We use the master CSV rather than assuming metadata schema.
# ============================================================

print("\nLoading master standards...")

master = pd.read_csv(
    MASTER_PATH,
    low_memory=False
)

print(
    f"Master standards : {len(master):,}"
)


# ============================================================
# CORPUS DOCUMENT TEXT HELPER
# ============================================================

def document_to_text(document):

    if document is None:
        return ""

    if isinstance(
        document,
        str
    ):
        return document

    if isinstance(
        document,
        dict
    ):

        parts = []

        for key, value in document.items():

            if value is None:
                continue

            if isinstance(
                value,
                (dict, list)
            ):

                value = json.dumps(
                    value,
                    ensure_ascii=False
                )

            parts.append(
                f"{key}: {value}"
            )

        return "\n".join(
            parts
        )

    if isinstance(
        document,
        list
    ):

        return "\n".join(
            str(x)
            for x in document
        )

    return str(
        document
    )


# ============================================================
# ROBUST IDENTIFIER NORMALIZATION
# ============================================================

def normalize_identifier(value):

    if value is None:
        return None

    text = str(
        value
    ).upper().strip()

    # Normalize separators
    text = text.replace(
        "–",
        "-"
    )

    text = text.replace(
        "—",
        "-"
    )

    # Collapse whitespace
    text = re.sub(
        r"\s+",
        " ",
        text
    ).strip()


    # --------------------------------------------------------
    # IS/IEC
    #
    # Handles:
    #
    # IS/IEC 60691:2023
    # IS/IEC :2023IEC 60691:2023
    # IS/IEC :2015IEC 60691 : 2015
    # IS/IEC 61196-1:2005
    # --------------------------------------------------------

    matches = list(
        re.finditer(
            r"IEC\s+"
            r"(\d+(?:-\d+)?)"
            r"\s*:\s*"
            r"(\d{4})",
            text
        )
    )

    if matches:

        # Use the LAST IEC number in malformed strings.
        match = matches[-1]

        number = match.group(1)
        year = match.group(2)

        return (
            f"IS/IEC "
            f"{number}:"
            f"{year}"
        )


    # --------------------------------------------------------
    # IS
    # --------------------------------------------------------

    matches = list(
        re.finditer(
            r"\bIS\s+"
            r"(\d+(?:-\d+)?)"
            r"\s*:\s*"
            r"(\d{4})",
            text
        )
    )

    if matches:

        match = matches[-1]

        number = match.group(1)
        year = match.group(2)

        return (
            f"IS "
            f"{number}:"
            f"{year}"
        )


    return text


# ============================================================
# EXTRACT BIS IDENTIFIERS FROM USER QUERY
# ============================================================

def extract_identifiers(query):

    text = str(
        query
    ).upper()

    found = []


    # --------------------------------------------------------
    # IMPORTANT:
    # Capture the COMPLETE number before the year.
    #
    # Previous bug:
    #
    # IS/IEC 60691:2023
    # was interpreted as:
    #
    # IS/IEC 1:2023
    #
    # because the regex was too permissive.
    # --------------------------------------------------------

    patterns = [

        (
            "IS/IEC",
            r"\bIS\s*/\s*IEC\s*"
            r"(\d+(?:-\d+)?)"
            r"\s*:\s*"
            r"(\d{4})"
        ),

        (
            "IEC",
            r"\bIEC\s+"
            r"(\d+(?:-\d+)?)"
            r"\s*:\s*"
            r"(\d{4})"
        ),

        (
            "IS",
            r"\bIS\s+"
            r"(\d+(?:-\d+)?)"
            r"\s*:\s*"
            r"(\d{4})"
        )
    ]


    for prefix, pattern in patterns:

        for match in re.finditer(
            pattern,
            text
        ):

            number = match.group(1)

            year = match.group(2)

            identifier = (
                f"{prefix} "
                f"{number}:"
                f"{year}"
            )

            if identifier not in found:

                found.append(
                    identifier
                )


    return found


# ============================================================
# BUILD AUTHORITATIVE STANDARD LOOKUP
# ============================================================

print(
    "\nBuilding canonical standard lookup..."
)

standard_records = {}

standard_by_internal_id = {}


for row in master.itertuples(
    index=False
):

    internal_id = str(
        getattr(
            row,
            "internal_id",
            ""
        )
    ).strip()

    is_number = str(
        getattr(
            row,
            "IS_number",
            ""
        )
    ).strip()

    title = str(
        getattr(
            row,
            "IS_title",
            ""
        )
    ).strip()


    if not internal_id:
        continue

    record = {

        "internal_id":
            internal_id,

        "is_number":
            is_number,

        "title":
            title,

        "technical_department":
            str(
                getattr(
                    row,
                    "technical_department",
                    ""
                )
            ),

        "technical_committee":
            str(
                getattr(
                    row,
                    "technical_committee",
                    ""
                )
            ),

        "aspect":
            str(
                getattr(
                    row,
                    "aspect",
                    ""
                )
            ),

        "superseding_is":
            str(
                getattr(
                    row,
                    "superseding_IS",
                    ""
                )
            ),

        "withdrawn_status":
            str(
                getattr(
                    row,
                    "catalogue_withdrawn_status",
                    ""
                )
            ),

        "detail_url":
            str(
                getattr(
                    row,
                    "detail_url",
                    ""
                )
            ),

        "download_url":
            str(
                getattr(
                    row,
                    "download_url",
                    ""
                )
            )
    }


    standard_by_internal_id[
        internal_id
    ] = record


    normalized = normalize_identifier(
        is_number
    )


    if normalized:

        standard_records[
            normalized
        ] = record


print(
    f"Canonical standard identifiers: "
    f"{len(standard_records):,}"
)


# ============================================================
# DEBUG 60691
# ============================================================

print("\nChecking 60691 normalization...")

for candidate in [
    "IS/IEC 60691:2023",
    "IS/IEC :2023IEC 60691:2023",
    "IS/IEC :2015IEC 60691 : 2015"
]:

    print(
        f"{candidate}"
        f"  ->  "
        f"{normalize_identifier(candidate)}"
    )


# ============================================================
# EXACT STANDARD LOOKUP
# ============================================================

def exact_standard_lookup(
    identifier
):

    canonical = normalize_identifier(
        identifier
    )

    record = standard_records.get(
        canonical
    )

    if not record:

        return []


    results = []


    # Find corresponding standard_core
    # using the master internal_id.
    for i, meta in enumerate(
        metadata
    ):

        meta_id = str(
            meta.get(
                "internal_id",
                ""
            )
        ).strip()

        if meta_id != record[
            "internal_id"
        ]:

            continue


        results.append({

            "rank": 1,

            "score": 100.0,

            "source": "exact",

            "document_index": i,

            "document":
                corpus[i],

            "identifier":
                canonical,

            "internal_id":
                record[
                    "internal_id"
                ],

            "title":
                record[
                    "title"
                ]
        })

        break


    # If metadata does not expose internal_id,
    # create an evidence result directly from master.
    if not results:

        results.append({

            "rank": 1,

            "score": 100.0,

            "source": "exact_master",

            "document_index": None,

            "document": {

                "IS_number":
                    record["is_number"],

                "IS_title":
                    record["title"],

                "internal_id":
                    record["internal_id"],

                "technical_department":
                    record[
                        "technical_department"
                    ],

                "technical_committee":
                    record[
                        "technical_committee"
                    ],

                "aspect":
                    record["aspect"],

                "superseding_IS":
                    record[
                        "superseding_is"
                    ],

                "withdrawn_status":
                    record[
                        "withdrawn_status"
                    ],

                "detail_url":
                    record["detail_url"],

                "download_url":
                    record["download_url"]
            },

            "identifier":
                canonical,

            "internal_id":
                record[
                    "internal_id"
                ],

            "title":
                record[
                    "title"
                ]
        })


    return results


# ============================================================
# INTENT DETECTION
# ============================================================

def detect_intent(
    query
):

    q = str(
        query
    ).lower().strip()

    identifiers = extract_identifiers(
        query
    )


    # --------------------------------------------------------
    # REVERSE RELATIONSHIP
    # --------------------------------------------------------

    reverse_patterns = [

        "which standards refer to",
        "which standards reference",
        "which bis standards refer to",
        "which bis standards reference",
        "who refers to",
        "who references",
        "referred by",
        "referenced by",
        "standards referring to",
        "standards referencing"
    ]

    if any(
        pattern in q
        for pattern in reverse_patterns
    ):

        return {
            "intent":
                "reverse_reference",

            "identifiers":
                identifiers
        }


    # --------------------------------------------------------
    # OUTGOING RELATIONSHIP
    # --------------------------------------------------------

    outgoing_patterns = [

        "what does it reference",
        "what does it refer to",
        "what standards does",
        "references of",
        "references made by",
        "standards referenced by",
        "standards referred to by",
        "references from"
    ]

    if any(
        pattern in q
        for pattern in outgoing_patterns
    ):

        return {
            "intent":
                "outgoing_reference",

            "identifiers":
                identifiers
        }


    # --------------------------------------------------------
    # INTERNATIONAL
    # --------------------------------------------------------

    international_patterns = [

        "international standard",
        "international standards",
        "iec reference",
        "iec references",
        "based on iec"
    ]

    if (
        identifiers
        and any(
            pattern in q
            for pattern in international_patterns
        )
    ):

        return {
            "intent":
                "international_reference",

            "identifiers":
                identifiers
        }


    # --------------------------------------------------------
    # EXACT STANDARD
    # --------------------------------------------------------

    exact_patterns = [

        "what is",
        "tell me about",
        "details of",
        "details for",
        "information about",
        "give me details",
        "describe"
    ]

    if (
        identifiers
        and any(
            pattern in q
            for pattern in exact_patterns
        )
    ):

        return {
            "intent":
                "exact_standard",

            "identifiers":
                identifiers
        }


    # --------------------------------------------------------
    # IDENTIFIER PRESENT
    # --------------------------------------------------------

    if identifiers:

        return {
            "intent":
                "exact_standard",

            "identifiers":
                identifiers
        }


    # --------------------------------------------------------
    # SEMANTIC
    # --------------------------------------------------------

    return {
        "intent":
            "semantic",

        "identifiers":
            []
    }


# ============================================================
# SEMANTIC SEARCH
# ============================================================

def semantic_search(
    query,
    top_k=20
):

    query_embedding = model.encode(
        [query],
        normalize_embeddings=True,
        show_progress_bar=False
    )

    query_embedding = np.asarray(
        query_embedding,
        dtype="float32"
    )


    scores, indices = index.search(
        query_embedding,
        top_k
    )


    results = []


    for score, idx in zip(
        scores[0],
        indices[0]
    ):

        if idx < 0:
            continue

        results.append({

            "rank":
                len(results) + 1,

            "score":
                float(score),

            "source":
                "semantic",

            "document_index":
                int(idx),

            "document":
                corpus[int(idx)]
        })


    return results


# ============================================================
# RELATIONSHIP SEARCH
#
# Uses the relationship structures built in CELL 8.
# ============================================================

def relationship_search(
    identifier,
    relationship_type
):

    canonical = normalize_identifier(
        identifier
    )

    standard = standard_records.get(
        canonical
    )

    if not standard:

        return []


    internal_id = standard[
        "internal_id"
    ]


    results = []


    # ========================================================
    # REVERSE
    # ========================================================

    if relationship_type == "reverse":

        refs = referred_by_standard.get(
            internal_id,
            []
        )


        for ref in refs:

            referencing_is = ref.get(
                "referencing_is",
                ""
            )

            referencing_title = ref.get(
                "referencing_title",
                ""
            )


            # Build explicit evidence rather than depending
            # on FAISS document formatting.

            results.append({

                "rank":
                    len(results) + 1,

                "score":
                    100.0,

                "source":
                    "relationship",

                "relationship_type":
                    "referred_by",

                "document_index":
                    None,

                "document": {

                    "relationship":
                        "referred_by",

                    "target_standard":
                        standard[
                            "is_number"
                        ],

                    "target_title":
                        standard[
                            "title"
                        ],

                    "referencing_standard":
                        referencing_is,

                    "referencing_title":
                        referencing_title,

                    "reference_number":
                        ref.get(
                            "reference_number",
                            ""
                        ),

                    "committee":
                        ref.get(
                            "committee",
                            ""
                        )
                }
            })


    # ========================================================
    # OUTGOING
    # ========================================================

    elif relationship_type == "outgoing":

        refs = []

        refs.extend(
            references_by_standard.get(
                internal_id,
                []
            )
        )

        refs.extend(
            indian_refs_by_standard.get(
                internal_id,
                []
            )
        )


        for ref in refs:

            results.append({

                "rank":
                    len(results) + 1,

                "score":
                    100.0,

                "source":
                    "relationship",

                "relationship_type":
                    ref.get(
                        "relationship",
                        "reference"
                    ),

                "document_index":
                    None,

                "document": {

                    "source_standard":
                        standard[
                            "is_number"
                        ],

                    "source_title":
                        standard[
                            "title"
                        ],

                    "referenced_standard":
                        ref.get(
                            "target_is",
                            ""
                        ),

                    "referenced_title":
                        ref.get(
                            "target_title",
                            ""
                        ),

                    "reference_number":
                        ref.get(
                            "reference_number",
                            ""
                        ),

                    "committee":
                        ref.get(
                            "committee",
                            ""
                        )
                }
            })


        # International references
        international_refs = (
            international_refs_by_standard.get(
                internal_id,
                []
            )
        )


        for ref in international_refs:

            results.append({

                "rank":
                    len(results) + 1,

                "score":
                    100.0,

                "source":
                    "relationship",

                "relationship_type":
                    "international_reference",

                "document_index":
                    None,

                "document": {

                    "source_standard":
                        standard[
                            "is_number"
                        ],

                    "source_title":
                        standard[
                            "title"
                        ],

                    "international_standard":
                        ref.get(
                            "international_standard",
                            ""
                        ),

                    "reference_number":
                        ref.get(
                            "reference_number",
                            ""
                        )
                }
            })


    # ========================================================
    # INTERNATIONAL ONLY
    # ========================================================

    elif relationship_type == "international":

        refs = (
            international_refs_by_standard.get(
                internal_id,
                []
            )
        )


        for ref in refs:

            results.append({

                "rank":
                    len(results) + 1,

                "score":
                    100.0,

                "source":
                    "relationship",

                "relationship_type":
                    "international_reference",

                "document_index":
                    None,

                "document": {

                    "source_standard":
                        standard[
                            "is_number"
                        ],

                    "source_title":
                        standard[
                            "title"
                        ],

                    "international_standard":
                        ref.get(
                            "international_standard",
                            ""
                        ),

                    "reference_number":
                        ref.get(
                            "reference_number",
                            ""
                        )
                }
            })


    return results


# ============================================================
# MAIN RETRIEVER
# ============================================================

def retrieve(
    query,
    top_k=10
):

    intent_data = detect_intent(
        query
    )

    intent = intent_data[
        "intent"
    ]

    identifiers = intent_data[
        "identifiers"
    ]


    results = []


    # ========================================================
    # EXACT
    # ========================================================

    if intent == "exact_standard":

        for identifier in identifiers:

            results.extend(
                exact_standard_lookup(
                    identifier
                )
            )


        # Semantic support
        if len(results) < top_k:

            semantic = semantic_search(
                query,
                top_k=top_k
            )


            existing_ids = {
                r.get(
                    "document_index"
                )
                for r in results
                if r.get(
                    "document_index"
                ) is not None
            }


            for result in semantic:

                if result.get(
                    "document_index"
                ) in existing_ids:

                    continue

                results.append(
                    result
                )

                if len(results) >= top_k:
                    break


    # ========================================================
    # REVERSE
    # ========================================================

    elif intent == "reverse_reference":

        for identifier in identifiers:

            results.extend(
                relationship_search(
                    identifier,
                    "reverse"
                )
            )


    # ========================================================
    # OUTGOING
    # ========================================================

    elif intent == "outgoing_reference":

        for identifier in identifiers:

            results.extend(
                relationship_search(
                    identifier,
                    "outgoing"
                )
            )


    # ========================================================
    # INTERNATIONAL
    # ========================================================

    elif intent == "international_reference":

        for identifier in identifiers:

            results.extend(
                relationship_search(
                    identifier,
                    "international"
                )
            )


    # ========================================================
    # SEMANTIC
    # ========================================================

    elif intent == "semantic":

        results = semantic_search(
            query,
            top_k=top_k
        )


    # ========================================================
    # DEDUPLICATE
    # ========================================================

    seen = set()

    deduped = []


    for result in results:

        # Relationship results have no document index,
        # therefore use their serialized evidence as key.

        if result.get(
            "document_index"
        ) is not None:

            key = (
                "idx",
                result[
                    "document_index"
                ]
            )

        else:

            key = (
                "relationship",
                json.dumps(
                    result.get(
                        "document",
                        {}
                    ),
                    sort_keys=True,
                    ensure_ascii=False
                )
            )


        if key in seen:
            continue

        seen.add(key)

        deduped.append(
            result
        )


    results = deduped[
        :top_k
    ]


    for rank, result in enumerate(
        results,
        start=1
    ):

        result[
            "rank"
        ] = rank


    return {

        "query":
            query,

        "intent":
            intent,

        "identifiers":
            identifiers,

        "results":
            results
    }


# ============================================================
# RESULT FORMATTER
# ============================================================

def display_results(
    response
):

    print("\n" + "=" * 70)

    print(
        "QUERY:",
        response[
            "query"
        ]
    )

    print(
        "INTENT:",
        response[
            "intent"
        ]
    )

    print(
        "IDENTIFIERS:",
        response[
            "identifiers"
        ]
    )

    print(
        "RESULTS:",
        len(
            response[
                "results"
            ]
        )
    )

    print("=" * 70)


    for result in response[
        "results"
    ]:

        print(
            f"\n#{result['rank']} "
            f"[{result['source']}] "
            f"score={result['score']:.4f}"
        )


        document = result.get(
            "document",
            ""
        )


        # Safely render dictionaries/lists/strings
        if isinstance(
            document,
            str
        ):

            text = document

        else:

            text = json.dumps(
                document,
                indent=2,
                ensure_ascii=False
            )


        print(
            text[:1500]
        )

        print(
            "-" * 70
        )


# ============================================================
# TEST 1
# EXACT STANDARD
# ============================================================

response = retrieve(
    "What is IS/IEC 60691:2023?",
    top_k=5
)

display_results(
    response
)


# ============================================================
# TEST 2
# SEMANTIC
# ============================================================

response = retrieve(
    "Which BIS standard deals with thermal links?",
    top_k=5
)

display_results(
    response
)


# ============================================================
# TEST 3
# REVERSE RELATIONSHIP
# ============================================================

response = retrieve(
    "Which BIS standards refer to IS/IEC 61196-1:2005?",
    top_k=10
)

display_results(
    response
)


# ============================================================
# TEST 4
# OUTGOING RELATIONSHIP
# ============================================================

response = retrieve(
    "What standards does IS/IEC 61196-1:2005 refer to?",
    top_k=10
)

display_results(
    response
)


# ============================================================
# TEST 5
# MIXED QUERY
# ============================================================

response = retrieve(
    "Tell me about IS/IEC 60691:2023 and its references.",
    top_k=10
)

display_results(
    response
)


print("\n" + "=" * 70)
print("CELL 9 FIXED — COMPLETE")
print("=" * 70)
```

```python
# ============================================================
# SAATHI — CELL 10
# STRUCTURED EVIDENCE RETRIEVAL
#
# Fixes:
#   1. Duplicate IS/IEC + IEC identifier detection
#   2. Mixed / compound queries
#   3. Clean evidence grouping
#   4. Relationship-aware evidence
#
# Uses everything already built in Cells 7–9.
# DOES NOT rebuild embeddings.
# ============================================================

import re
import json
from collections import defaultdict


# ============================================================
# 1. CLEAN IDENTIFIER EXTRACTION
# ============================================================

def extract_identifiers_clean(query):

    text = str(query).upper()

    found = []
    canonical_seen = set()

    patterns = [
        (
            "IS/IEC",
            r"\bIS\s*/\s*IEC\s*"
            r"(\d+(?:-\d+)?)"
            r"\s*:\s*(\d{4})"
        ),
        (
            "IEC",
            r"\bIEC\s+"
            r"(\d+(?:-\d+)?)"
            r"\s*:\s*(\d{4})"
        ),
        (
            "IS",
            r"\bIS\s+"
            r"(\d+(?:-\d+)?)"
            r"\s*:\s*(\d{4})"
        )
    ]

    for prefix, pattern in patterns:

        for match in re.finditer(
            pattern,
            text
        ):

            number = match.group(1)
            year = match.group(2)

            # Always canonicalize IEC-only references
            # to IS/IEC when they appear in a BIS-style query.
            if prefix == "IEC":
                canonical = (
                    f"IS/IEC {number}:{year}"
                )
            else:
                canonical = (
                    f"{prefix} "
                    f"{number}:{year}"
                )

            normalized = normalize_identifier(
                canonical
            )

            if normalized in canonical_seen:
                continue

            canonical_seen.add(
                normalized
            )

            found.append(
                normalized
            )

    return found


# ============================================================
# 2. DETECT WHETHER QUERY IS COMPOUND
# ============================================================

def detect_compound_query(query):

    q = str(
        query
    ).lower()

    relationship_words = [

        "reference",
        "references",
        "refer",
        "referred",
        "referencing",
        "referenced",
        "based on",
        "supersed",
        "amendment",
        "revision"
    ]

    information_words = [

        "what is",
        "tell me about",
        "details",
        "information",
        "describe",
        "explain"
    ]

    has_relationship = any(
        word in q
        for word in relationship_words
    )

    has_information = any(
        word in q
        for word in information_words
    )

    return (
        has_relationship
        and has_information
    )


# ============================================================
# 3. DETERMINE RELATIONSHIP DIRECTION
# ============================================================

def detect_relationship_direction(query):

    q = str(
        query
    ).lower()

    reverse_patterns = [

        "which standards refer to",
        "which standards reference",
        "which bis standards refer to",
        "which bis standards reference",
        "who refers to",
        "who references",
        "referred by",
        "referenced by",
        "standards referring to",
        "standards referencing"
    ]

    outgoing_patterns = [

        "what does it reference",
        "what does it refer to",
        "what standards does",
        "references of",
        "references made by",
        "standards referenced by",
        "standards referred to by",
        "references from",
        "its references"
    ]

    if any(
        p in q
        for p in reverse_patterns
    ):

        return "reverse"

    if any(
        p in q
        for p in outgoing_patterns
    ):

        return "outgoing"

    return None


# ============================================================
# 4. STANDARD EVIDENCE
# ============================================================

def build_standard_evidence(
    identifier
):

    results = exact_standard_lookup(
        identifier
    )

    if not results:
        return []

    result = results[0]

    document = result.get(
        "document",
        {}
    )

    if isinstance(
        document,
        dict
    ):

        return [{
            "evidence_type":
                "standard",

            "source":
                "exact",

            "identifier":
                identifier,

            "internal_id":
                result.get(
                    "internal_id"
                ),

            "title":
                result.get(
                    "title",
                    document.get(
                        "is_title",
                        ""
                    )
                ),

            "content":
                document
        }]

    return [{
        "evidence_type":
            "standard",

        "source":
            "exact",

        "identifier":
            identifier,

        "internal_id":
            result.get(
                "internal_id"
            ),

        "title":
            result.get(
                "title",
                ""
            ),

        "content":
            document
    }]


# ============================================================
# 5. RELATIONSHIP EVIDENCE
# ============================================================

def build_relationship_evidence(
    identifier,
    direction,
    limit=50
):

    relationship_results = (
        relationship_search(
            identifier,
            direction
        )
    )

    evidence = []

    for result in relationship_results[
        :limit
    ]:

        evidence.append({

            "evidence_type":
                "relationship",

            "source":
                "relationship",

            "relationship":
                result.get(
                    "relationship_type",
                    ""
                ),

            "identifier":
                identifier,

            "content":
                result.get(
                    "document",
                    {}
                )
        })

    return evidence


# ============================================================
# 6. SEMANTIC SUPPORT
# ============================================================

def build_semantic_evidence(
    query,
    limit=5
):

    semantic_results = semantic_search(
        query,
        top_k=limit
    )

    evidence = []

    for result in semantic_results:

        evidence.append({

            "evidence_type":
                "semantic",

            "source":
                "faiss",

            "score":
                result.get(
                    "score"
                ),

            "document_index":
                result.get(
                    "document_index"
                ),

            "content":
                result.get(
                    "document",
                    {}
                )
        })

    return evidence


# ============================================================
# 7. STRUCTURED RETRIEVER
# ============================================================

def retrieve_evidence(
    query,
    semantic_fallback=True,
    max_relationships=50
):

    identifiers = extract_identifiers_clean(
        query
    )

    compound = detect_compound_query(
        query
    )

    direction = detect_relationship_direction(
        query
    )


    evidence = []

    retrieval_plan = []


    # ========================================================
    # NO IDENTIFIER
    # ========================================================

    if not identifiers:

        retrieval_plan.append(
            "semantic"
        )

        evidence.extend(
            build_semantic_evidence(
                query
            )
        )


    # ========================================================
    # IDENTIFIER PRESENT
    # ========================================================

    else:

        for identifier in identifiers:

            # ------------------------------------------------
            # ALWAYS retrieve exact standard
            # ------------------------------------------------

            retrieval_plan.append(
                f"standard:{identifier}"
            )

            evidence.extend(
                build_standard_evidence(
                    identifier
                )
            )


            # ------------------------------------------------
            # COMPOUND QUERY
            #
            # "Tell me about X and its references"
            # ------------------------------------------------

            if compound:

                if direction is not None:

                    retrieval_plan.append(
                        f"{direction}_references:{identifier}"
                    )

                    evidence.extend(
                        build_relationship_evidence(
                            identifier,
                            direction,
                            max_relationships
                        )
                    )

                else:

                    # "its references" normally means
                    # outgoing references.

                    retrieval_plan.append(
                        f"outgoing_references:{identifier}"
                    )

                    evidence.extend(
                        build_relationship_evidence(
                            identifier,
                            "outgoing",
                            max_relationships
                        )
                    )


            # ------------------------------------------------
            # PURE RELATIONSHIP QUERY
            # ------------------------------------------------

            elif direction is not None:

                retrieval_plan.append(
                    f"{direction}_references:{identifier}"
                )

                evidence.extend(
                    build_relationship_evidence(
                        identifier,
                        direction,
                        max_relationships
                    )
                )


    # ========================================================
    # SEMANTIC FALLBACK
    # ========================================================

    if (
        semantic_fallback
        and not compound
        and len(evidence) == 0
    ):

        retrieval_plan.append(
            "semantic_fallback"
        )

        evidence.extend(
            build_semantic_evidence(
                query
            )
        )


    # ========================================================
    # DEDUPLICATE
    # ========================================================

    unique = []

    seen = set()


    for item in evidence:

        content = item.get(
            "content",
            {}
        )

        try:

            content_key = json.dumps(
                content,
                sort_keys=True,
                ensure_ascii=False
            )

        except Exception:

            content_key = str(
                content
            )


        key = (
            item.get(
                "evidence_type"
            ),
            item.get(
                "identifier"
            ),
            content_key
        )


        if key in seen:
            continue

        seen.add(
            key
        )

        unique.append(
            item
        )


    # ========================================================
    # RETURN STRUCTURED RESULT
    # ========================================================

    return {

        "query":
            query,

        "identifiers":
            identifiers,

        "compound":
            compound,

        "relationship_direction":
            direction,

        "retrieval_plan":
            retrieval_plan,

        "evidence_count":
            len(unique),

        "evidence":
            unique
    }


# ============================================================
# 8. DISPLAY STRUCTURED EVIDENCE
# ============================================================

def display_evidence(
    result,
    max_items=15
):

    print("\n" + "=" * 70)

    print(
        "QUERY:",
        result["query"]
    )

    print(
        "IDENTIFIERS:",
        result["identifiers"]
    )

    print(
        "COMPOUND:",
        result["compound"]
    )

    print(
        "RELATIONSHIP:",
        result[
            "relationship_direction"
        ]
    )

    print(
        "EVIDENCE COUNT:",
        result[
            "evidence_count"
        ]
    )

    print(
        "\nRETRIEVAL PLAN:"
    )

    for step in result[
        "retrieval_plan"
    ]:

        print(
            f"  → {step}"
        )

    print(
        "\nEVIDENCE:"
    )

    print("=" * 70)


    for i, item in enumerate(
        result["evidence"][
            :max_items
        ],
        start=1
    ):

        print(
            f"\n[{i}] "
            f"{item['evidence_type']}"
        )

        if item.get(
            "relationship"
        ):

            print(
                "Relationship:",
                item[
                    "relationship"
                ]
            )


        if item.get(
            "identifier"
        ):

            print(
                "Identifier:",
                item[
                    "identifier"
                ]
            )


        content = item.get(
            "content",
            {}
        )


        if isinstance(
            content,
            dict
        ):

            print(
                json.dumps(
                    content,
                    indent=2,
                    ensure_ascii=False
                )[:1500]
            )

        else:

            print(
                str(
                    content
                )[:1500]
            )

        print(
            "-" * 70
        )


# ============================================================
# TEST 1
# EXACT STANDARD
# ============================================================

result = retrieve_evidence(
    "What is IS/IEC 60691:2023?"
)

display_evidence(
    result,
    max_items=5
)


# ============================================================
# TEST 2
# SEMANTIC
# ============================================================

result = retrieve_evidence(
    "Which BIS standard deals with thermal links?"
)

display_evidence(
    result,
    max_items=5
)


# ============================================================
# TEST 3
# REVERSE RELATIONSHIP
# ============================================================

result = retrieve_evidence(
    "Which BIS standards refer to IS/IEC 61196-1:2005?",
    max_relationships=50
)

display_evidence(
    result,
    max_items=15
)


# ============================================================
# TEST 4
# OUTGOING RELATIONSHIP
# ============================================================

result = retrieve_evidence(
    "What standards does IS/IEC 61196-1:2005 refer to?",
    max_relationships=50
)

display_evidence(
    result,
    max_items=15
)


# ============================================================
# TEST 5
# COMPOUND QUERY
# ============================================================

result = retrieve_evidence(
    "Tell me about IS/IEC 60691:2023 and its references.",
    max_relationships=50
)

display_evidence(
    result,
    max_items=15
)


# ============================================================
# SAVE A SMALL STRUCTURED OUTPUT
#
# This is NOT the final production evidence store.
# It is just the result of the final test query.
# ============================================================

with open(
    "/kaggle/working/cell10_last_evidence.json",
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        result,
        f,
        indent=2,
        ensure_ascii=False
    )


print("\n" + "=" * 70)
print("CELL 10 COMPLETE")
print("=" * 70)
```

```python
# ============================================================
# SAATHI — CELL 11
# GROUNDED LLM ANSWER GENERATION
#
# Cell 10 Evidence
#        ↓
# Gemini
#        ↓
# Grounded BIS answer + evidence citations
#
# DOES NOT MODIFY:
#   - BIS corpus
#   - embeddings
#   - FAISS
#   - relationship indexes
# ============================================================

import os
import json
import re
import textwrap

# ============================================================
# 1. GEMINI SDK
# ============================================================

try:
    from google import genai
except ImportError:

    print("Installing Google GenAI SDK...")

    !pip -q install -U google-genai

    from google import genai


# ============================================================
# 2. API KEY
# ============================================================

# IMPORTANT:
# Put your Gemini API key in Kaggle Secrets as:
#
# GEMINI_API_KEY
#
# Then enable the secret for this notebook.

try:

    from kaggle_secrets import UserSecretsClient

    secrets = UserSecretsClient()

    GEMINI_API_KEY = secrets.get_secret(
        "GEMINI_API_KEY"
    )

except Exception:

    GEMINI_API_KEY = os.environ.get(
        "AQ.Ab8RN6Ju1zJZXO64UIxc3tu7l73vj2fWf_jmt7vexL8LPrxYUw"
    )


if not GEMINI_API_KEY:

    raise RuntimeError(
        "GEMINI_API_KEY was not found. "
        "Add GEMINI_API_KEY to Kaggle Secrets."
    )


# ============================================================
# 3. GEMINI CLIENT
# ============================================================

client = genai.Client(
    api_key=GEMINI_API_KEY
)


# ============================================================
# 4. MODEL
# ============================================================

MODEL_NAME = "gemini-2.5-flash"


print("=" * 70)
print("SAATHI — CELL 11")
print("GROUNDED LLM ANSWER GENERATION")
print("=" * 70)

print(
    f"\nGemini model: {MODEL_NAME}"
)


# ============================================================
# 5. PREPARE EVIDENCE FOR LLM
# ============================================================

def prepare_evidence_for_llm(
    retrieval_result,
    max_chars_per_item=5000,
    max_total_chars=30000
):

    evidence = retrieval_result.get(
        "evidence",
        []
    )

    prepared = []

    total_chars = 0


    for idx, item in enumerate(
        evidence,
        start=1
    ):

        evidence_type = item.get(
            "evidence_type",
            ""
        )

        identifier = item.get(
            "identifier",
            ""
        )

        relationship = item.get(
            "relationship",
            ""
        )

        content = item.get(
            "content",
            {}
        )


        if isinstance(
            content,
            dict
        ):

            content_text = json.dumps(
                content,
                indent=2,
                ensure_ascii=False
            )

        else:

            content_text = str(
                content
            )


        content_text = content_text[
            :max_chars_per_item
        ]


        block = (
            f"[EVIDENCE {idx}]\n"
            f"Type: {evidence_type}\n"
        )


        if identifier:

            block += (
                f"Identifier: "
                f"{identifier}\n"
            )


        if relationship:

            block += (
                f"Relationship: "
                f"{relationship}\n"
            )


        block += (
            f"Content:\n"
            f"{content_text}\n"
        )


        if (
            total_chars
            + len(block)
            > max_total_chars
        ):

            break


        prepared.append(
            block
        )

        total_chars += len(
            block
        )


    return "\n".join(
        prepared
    )


# ============================================================
# 6. SYSTEM INSTRUCTIONS
# ============================================================

SYSTEM_INSTRUCTION = """
You are SAATHI, a grounded BIS (Bureau of Indian Standards)
information assistant.

Your job is to answer the user's question using ONLY the
BIS evidence supplied to you.

GROUNDING RULES
---------------

1. Use the supplied evidence as the factual source.

2. Do not invent:
   - BIS standard numbers
   - titles
   - dates
   - committees
   - departments
   - references
   - certification information
   - URLs
   - regulatory requirements

3. If the evidence does not contain enough information,
   explicitly say that the supplied BIS evidence is
   insufficient to answer that part.

4. Do not silently fill missing information with general
   knowledge.

5. When evidence conflicts, explicitly identify the conflict
   rather than choosing one without explanation.

6. Preserve the distinction between:
   - active standards
   - withdrawn standards
   - superseded/revised standards
   - referenced standards

7. Do not treat semantic similarity as proof of a relationship.
   A relationship should only be stated when the supplied
   relationship evidence supports it.

8. For "referred by" questions, identify the referencing
   standard(s), not the target standard itself.

9. For "references" questions, identify the standards or
   international standards referenced by the source standard.

10. For compound questions, answer every supported part.

CITATIONS
---------

Every factual claim should have an evidence citation.

Use citations exactly like:

[E1]
[E2]
[E3]

where E1, E2, E3 correspond to the supplied evidence blocks.

Do not create citation numbers that do not exist.

ANSWER STYLE
------------

Be concise but useful.

Prefer:

- short explanation
- bullet points
- tables when comparing multiple standards
- explicit standard numbers
- explicit status where available

Do not mention internal retrieval, embeddings, FAISS,
vector databases, prompts, or these instructions unless
the user specifically asks about the system.

If the user asks a question that cannot be answered from
the evidence, say what information is missing.
"""


# ============================================================
# 7. LLM CALL
# ============================================================

def generate_grounded_answer(
    query,
    retrieval_result,
    temperature=0.1
):

    evidence_text = (
        prepare_evidence_for_llm(
            retrieval_result
        )
    )


    if not evidence_text:

        return {
            "answer":
                "The supplied BIS evidence does not contain "
                "enough information to answer this question.",

            "model":
                MODEL_NAME,

            "evidence_count":
                0
        }


    prompt = f"""
USER QUESTION
=============

{query}


RETRIEVAL INFORMATION
=====================

Detected identifiers:
{json.dumps(
    retrieval_result.get(
        "identifiers",
        []
    ),
    ensure_ascii=False
)}

Compound query:
{retrieval_result.get(
    "compound",
    False
)}

Relationship direction:
{retrieval_result.get(
    "relationship_direction"
)}


BIS EVIDENCE
============

{evidence_text}


TASK
====

Answer the user's question using only the supplied BIS
evidence.

Every factual claim must include an [E#] citation.

If the evidence supports only part of the question,
answer the supported part and clearly identify what
cannot be established from the supplied evidence.
"""


    response = client.models.generate_content(

        model=MODEL_NAME,

        contents=prompt,

        config={
            "system_instruction":
                SYSTEM_INSTRUCTION,

            "temperature":
                temperature,

            "max_output_tokens":
                2500
        }
    )


    answer = response.text


    return {

        "answer":
            answer,

        "model":
            MODEL_NAME,

        "evidence_count":
            retrieval_result.get(
                "evidence_count",
                0
            ),

        "retrieval_plan":
            retrieval_result.get(
                "retrieval_plan",
                []
            )
    }


# ============================================================
# 8. COMPLETE SAATHI PIPELINE
# ============================================================

def ask_saathi(
    query,
    max_relationships=50
):

    print(
        f"\nSAATHI QUERY:"
        f"\n{query}"
    )

    print(
        "\nRetrieving BIS evidence..."
    )


    retrieval_result = retrieve_evidence(

        query,

        semantic_fallback=True,

        max_relationships=
            max_relationships
    )


    print(
        "Evidence retrieved:",
        retrieval_result[
            "evidence_count"
        ]
    )


    print(
        "Generating grounded answer..."
    )


    llm_result = generate_grounded_answer(

        query,

        retrieval_result
    )


    return {

        "query":
            query,

        "retrieval":
            retrieval_result,

        "answer":
            llm_result
    }


# ============================================================
# 9. DISPLAY ANSWER
# ============================================================

def display_saathi_result(
    result
):

    print("\n")
    print("=" * 70)

    print(
        "SAATHI ANSWER"
    )

    print("=" * 70)

    print(
        result[
            "answer"
        ][
            "answer"
        ]
    )


    print("\n")
    print("=" * 70)

    print(
        "RETRIEVAL PLAN"
    )

    print("=" * 70)

    for step in result[
        "retrieval"
    ][
        "retrieval_plan"
    ]:

        print(
            f"→ {step}"
        )


    print("\n")
    print(
        "Evidence used:",
        result[
            "retrieval"
        ][
            "evidence_count"
        ]
    )


# ============================================================
# 10. TEST 1 — EXACT STANDARD
# ============================================================

result = ask_saathi(
    "What is IS/IEC 60691:2023?"
)

display_saathi_result(
    result
)


# ============================================================
# 11. TEST 2 — SEMANTIC
# ============================================================

result = ask_saathi(
    "Which BIS standard deals with thermal links?"
)

display_saathi_result(
    result
)


# ============================================================
# 12. TEST 3 — REVERSE RELATIONSHIP
# ============================================================

result = ask_saathi(
    "Which BIS standards refer to IS/IEC 61196-1:2005?",
    max_relationships=50
)

display_saathi_result(
    result
)


# ============================================================
# 13. TEST 4 — COMPOUND QUERY
# ============================================================

result = ask_saathi(
    "Tell me about IS/IEC 60691:2023 and its references.",
    max_relationships=50
)

display_saathi_result(
    result
)


# ============================================================
# 14. SAVE LAST RESULT
# ============================================================

with open(
    "/kaggle/working/saathi_cell11_result.json",
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        result,
        f,
        indent=2,
        ensure_ascii=False
    )


print("\n" + "=" * 70)
print("CELL 11 COMPLETE")
print("=" * 70)
```

```python
# ============================================================
# SAATHI — CELL 11
# GROUNDED LLM ANSWER GENERATION
#
# Cell 10 Evidence
#        ↓
# Gemini
#        ↓
# Grounded BIS answer + evidence citations
#
# DOES NOT MODIFY:
#   - BIS corpus
#   - embeddings
#   - FAISS
#   - relationship indexes
# ============================================================

import os
import json
import re
import textwrap

# ============================================================
# 1. GEMINI SDK
# ============================================================

try:
    from google import genai
except ImportError:

    print("Installing Google GenAI SDK...")

    !pip -q install -U google-genai

    from google import genai


# ============================================================
# 2. API KEY
# ============================================================

# IMPORTANT:
# Put your Gemini API key in Kaggle Secrets as:
#
# GEMINI_API_KEY
#
# Then enable the secret for this notebook.

try:

    from kaggle_secrets import UserSecretsClient

    secrets = UserSecretsClient()

    GEMINI_API_KEY = secrets.get_secret(
        "GEMINI_API_KEY"
    )

except Exception:

    GEMINI_API_KEY = os.environ.get(
        "GEMINI_API_KEY"
    )


if not GEMINI_API_KEY:

    raise RuntimeError(
        "GEMINI_API_KEY was not found. "
        "Add GEMINI_API_KEY to Kaggle Secrets."
    )


# ============================================================
# 3. GEMINI CLIENT
# ============================================================

client = genai.Client(
    api_key=GEMINI_API_KEY
)


# ============================================================
# 4. MODEL
# ============================================================

MODEL_NAME = "gemini-2.5-flash"


print("=" * 70)
print("SAATHI — CELL 11")
print("GROUNDED LLM ANSWER GENERATION")
print("=" * 70)

print(
    f"\nGemini model: {MODEL_NAME}"
)


# ============================================================
# 5. PREPARE EVIDENCE FOR LLM
# ============================================================

def prepare_evidence_for_llm(
    retrieval_result,
    max_chars_per_item=5000,
    max_total_chars=30000
):

    evidence = retrieval_result.get(
        "evidence",
        []
    )

    prepared = []

    total_chars = 0


    for idx, item in enumerate(
        evidence,
        start=1
    ):

        evidence_type = item.get(
            "evidence_type",
            ""
        )

        identifier = item.get(
            "identifier",
            ""
        )

        relationship = item.get(
            "relationship",
            ""
        )

        content = item.get(
            "content",
            {}
        )


        if isinstance(
            content,
            dict
        ):

            content_text = json.dumps(
                content,
                indent=2,
                ensure_ascii=False
            )

        else:

            content_text = str(
                content
            )


        content_text = content_text[
            :max_chars_per_item
        ]


        block = (
            f"[EVIDENCE {idx}]\n"
            f"Type: {evidence_type}\n"
        )


        if identifier:

            block += (
                f"Identifier: "
                f"{identifier}\n"
            )


        if relationship:

            block += (
                f"Relationship: "
                f"{relationship}\n"
            )


        block += (
            f"Content:\n"
            f"{content_text}\n"
        )


        if (
            total_chars
            + len(block)
            > max_total_chars
        ):

            break


        prepared.append(
            block
        )

        total_chars += len(
            block
        )


    return "\n".join(
        prepared
    )


# ============================================================
# 6. SYSTEM INSTRUCTIONS
# ============================================================

SYSTEM_INSTRUCTION = """
You are SAATHI, a grounded BIS (Bureau of Indian Standards)
information assistant.

Your job is to answer the user's question using ONLY the
BIS evidence supplied to you.

GROUNDING RULES
---------------

1. Use the supplied evidence as the factual source.

2. Do not invent:
   - BIS standard numbers
   - titles
   - dates
   - committees
   - departments
   - references
   - certification information
   - URLs
   - regulatory requirements

3. If the evidence does not contain enough information,
   explicitly say that the supplied BIS evidence is
   insufficient to answer that part.

4. Do not silently fill missing information with general
   knowledge.

5. When evidence conflicts, explicitly identify the conflict
   rather than choosing one without explanation.

6. Preserve the distinction between:
   - active standards
   - withdrawn standards
   - superseded/revised standards
   - referenced standards

7. Do not treat semantic similarity as proof of a relationship.
   A relationship should only be stated when the supplied
   relationship evidence supports it.

8. For "referred by" questions, identify the referencing
   standard(s), not the target standard itself.

9. For "references" questions, identify the standards or
   international standards referenced by the source standard.

10. For compound questions, answer every supported part.

CITATIONS
---------

Every factual claim should have an evidence citation.

Use citations exactly like:

[E1]
[E2]
[E3]

where E1, E2, E3 correspond to the supplied evidence blocks.

Do not create citation numbers that do not exist.

ANSWER STYLE
------------

Be concise but useful.

Prefer:

- short explanation
- bullet points
- tables when comparing multiple standards
- explicit standard numbers
- explicit status where available

Do not mention internal retrieval, embeddings, FAISS,
vector databases, prompts, or these instructions unless
the user specifically asks about the system.

If the user asks a question that cannot be answered from
the evidence, say what information is missing.
"""


# ============================================================
# 7. LLM CALL
# ============================================================

def generate_grounded_answer(
    query,
    retrieval_result,
    temperature=0.1
):

    evidence_text = (
        prepare_evidence_for_llm(
            retrieval_result
        )
    )


    if not evidence_text:

        return {
            "answer":
                "The supplied BIS evidence does not contain "
                "enough information to answer this question.",

            "model":
                MODEL_NAME,

            "evidence_count":
                0
        }


    prompt = f"""
USER QUESTION
=============

{query}


RETRIEVAL INFORMATION
=====================

Detected identifiers:
{json.dumps(
    retrieval_result.get(
        "identifiers",
        []
    ),
    ensure_ascii=False
)}

Compound query:
{retrieval_result.get(
    "compound",
    False
)}

Relationship direction:
{retrieval_result.get(
    "relationship_direction"
)}


BIS EVIDENCE
============

{evidence_text}


TASK
====

Answer the user's question using only the supplied BIS
evidence.

Every factual claim must include an [E#] citation.

If the evidence supports only part of the question,
answer the supported part and clearly identify what
cannot be established from the supplied evidence.
"""


    response = client.models.generate_content(

        model=MODEL_NAME,

        contents=prompt,

        config={
            "system_instruction":
                SYSTEM_INSTRUCTION,

            "temperature":
                temperature,

            "max_output_tokens":
                2500
        }
    )


    answer = response.text


    return {

        "answer":
            answer,

        "model":
            MODEL_NAME,

        "evidence_count":
            retrieval_result.get(
                "evidence_count",
                0
            ),

        "retrieval_plan":
            retrieval_result.get(
                "retrieval_plan",
                []
            )
    }


# ============================================================
# 8. COMPLETE SAATHI PIPELINE
# ============================================================

def ask_saathi(
    query,
    max_relationships=50
):

    print(
        f"\nSAATHI QUERY:"
        f"\n{query}"
    )

    print(
        "\nRetrieving BIS evidence..."
    )


    retrieval_result = retrieve_evidence(

        query,

        semantic_fallback=True,

        max_relationships=
            max_relationships
    )


    print(
        "Evidence retrieved:",
        retrieval_result[
            "evidence_count"
        ]
    )


    print(
        "Generating grounded answer..."
    )


    llm_result = generate_grounded_answer(

        query,

        retrieval_result
    )


    return {

        "query":
            query,

        "retrieval":
            retrieval_result,

        "answer":
            llm_result
    }


# ============================================================
# 9. DISPLAY ANSWER
# ============================================================

def display_saathi_result(
    result
):

    print("\n")
    print("=" * 70)

    print(
        "SAATHI ANSWER"
    )

    print("=" * 70)

    print(
        result[
            "answer"
        ][
            "answer"
        ]
    )


    print("\n")
    print("=" * 70)

    print(
        "RETRIEVAL PLAN"
    )

    print("=" * 70)

    for step in result[
        "retrieval"
    ][
        "retrieval_plan"
    ]:

        print(
            f"→ {step}"
        )


    print("\n")
    print(
        "Evidence used:",
        result[
            "retrieval"
        ][
            "evidence_count"
        ]
    )


# ============================================================
# 10. TEST 1 — EXACT STANDARD
# ============================================================

result = ask_saathi(
    "What is IS/IEC 60691:2023?"
)

display_saathi_result(
    result
)


# ============================================================
# 11. TEST 2 — SEMANTIC
# ============================================================

result = ask_saathi(
    "Which BIS standard deals with thermal links?"
)

display_saathi_result(
    result
)


# ============================================================
# 12. TEST 3 — REVERSE RELATIONSHIP
# ============================================================

result = ask_saathi(
    "Which BIS standards refer to IS/IEC 61196-1:2005?",
    max_relationships=50
)

display_saathi_result(
    result
)


# ============================================================
# 13. TEST 4 — COMPOUND QUERY
# ============================================================

result = ask_saathi(
    "Tell me about IS/IEC 60691:2023 and its references.",
    max_relationships=50
)

display_saathi_result(
    result
)


# ============================================================
# 14. SAVE LAST RESULT
# ============================================================

with open(
    "/kaggle/working/saathi_cell11_result.json",
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        result,
        f,
        indent=2,
        ensure_ascii=False
    )


print("\n" + "=" * 70)
print("CELL 11 COMPLETE")
print("=" * 70)
```

```python
# ============================================================
# SAATHI — CELL 12
# ANSWER QUALITY + CITATION VALIDATION
# ============================================================

import re
import json
from collections import Counter


print("=" * 70)
print("SAATHI — CELL 12")
print("ANSWER QUALITY + CITATION VALIDATION")
print("=" * 70)


# ============================================================
# 1. CITATION VALIDATION
# ============================================================

def extract_citations(answer):

    if not answer:
        return []

    matches = re.findall(
        r"\[E(\d+)\]",
        answer
    )

    return [
        int(x)
        for x in matches
    ]


def validate_citations(
    answer,
    evidence_count
):

    citations = extract_citations(
        answer
    )

    invalid = [
        c
        for c in citations
        if c < 1 or c > evidence_count
    ]

    return {
        "citations_found":
            citations,

        "unique_citations":
            sorted(set(citations)),

        "citation_count":
            len(citations),

        "invalid_citations":
            invalid,

        "all_citations_valid":
            len(invalid) == 0
    }


# ============================================================
# 2. IDENTIFIER EXTRACTION
# ============================================================

def extract_bis_identifiers(text):

    if not text:
        return []

    patterns = [

        r"\bIS/IEC\s*[0-9]+(?:\s*:\s*[0-9]{4})?",

        r"\bIS\s*[0-9]+(?:\s*\([^)]+\))?"
        r"(?:\s*:\s*[0-9]{4})?",

        r"\bIEC\s*[0-9]+"
        r"(?:\s*:\s*[0-9]{4})?"
    ]

    found = []

    for pattern in patterns:

        found.extend(
            re.findall(
                pattern,
                text,
                flags=re.IGNORECASE
            )
        )

    return list(
        dict.fromkeys(
            x.strip()
            for x in found
        )
    )


# ============================================================
# 3. MALFORMED IDENTIFIER DETECTION
# ============================================================

def detect_malformed_identifiers(
    text
):

    if not text:
        return []

    suspicious_patterns = [

        r"IS/IEC\s*:\s*\d{4}IEC",

        r"IS/IEC\s*:\s*\d{4}\s*IEC",

        r"IS/IEC\s+\d+\s*:\s*\d{4}\s*IEC",

        r"IEC\s+\d+\s*:\s*\d{4}\s*IEC",

        r"\d{4}IEC\s+\d+"
    ]

    found = []

    for pattern in suspicious_patterns:

        found.extend(
            re.findall(
                pattern,
                text,
                flags=re.IGNORECASE
            )
        )

    return list(
        dict.fromkeys(found)
    )


# ============================================================
# 4. EVIDENCE COVERAGE
# ============================================================

def check_evidence_coverage(
    answer,
    evidence_count
):

    citations = extract_citations(
        answer
    )

    if evidence_count == 0:

        return {
            "has_evidence":
                False,

            "coverage":
                0.0
        }

    unique_citations = set(
        c
        for c in citations
        if 1 <= c <= evidence_count
    )

    coverage = (
        len(unique_citations)
        / evidence_count
    )

    return {
        "has_evidence":
            evidence_count > 0,

        "coverage":
            round(
                coverage,
                4
            )
    }


# ============================================================
# 5. RELATIONSHIP-SPECIFIC VALIDATION
# ============================================================

def validate_relationship_answer(
    query,
    retrieval_result,
    answer
):

    direction = retrieval_result.get(
        "relationship_direction"
    )

    result = {
        "relationship_type":
            direction,

        "passed":
            True,

        "issues":
            []
    }


    if direction == "reverse":

        # Reverse question:
        # "Which standards refer to X?"

        if not re.search(
            r"\b(refer|references|referenced|cite|cites)\b",
            answer,
            re.IGNORECASE
        ):

            result[
                "issues"
            ].append(
                "Answer does not clearly describe the reference relationship."
            )


    elif direction == "outgoing":

        if not re.search(
            r"\b(refer|references|reference|cites|cited)\b",
            answer,
            re.IGNORECASE
        ):

            result[
                "issues"
            ].append(
                "Answer does not clearly describe outgoing references."
            )


    result["passed"] = (
        len(
            result["issues"]
        ) == 0
    )

    return result


# ============================================================
# 6. STATUS CONSISTENCY CHECK
# ============================================================

def validate_status_terms(
    answer,
    retrieval_result
):

    answer_lower = answer.lower()

    evidence_text = json.dumps(
        retrieval_result.get(
            "evidence",
            []
        ),
        ensure_ascii=False
    ).lower()


    issues = []


    # If answer claims "active", ensure
    # active evidence exists somewhere.

    if (
        "active" in answer_lower
        and "active" not in evidence_text
    ):

        issues.append(
            "Answer mentions ACTIVE but evidence does not contain an active status."
        )


    # If answer claims "withdrawn", ensure
    # withdrawn evidence exists.

    if (
        "withdrawn" in answer_lower
        and "withdrawn" not in evidence_text
    ):

        issues.append(
            "Answer mentions WITHDRAWN but evidence does not contain a withdrawn status."
        )


    return {
        "passed":
            len(issues) == 0,

        "issues":
            issues
    }


# ============================================================
# 7. ANSWER QUALITY VALIDATOR
# ============================================================

def validate_saathi_answer(
    result
):

    retrieval = result[
        "retrieval"
    ]

    answer = result[
        "answer"
    ][
        "answer"
    ]


    evidence_count = retrieval.get(
        "evidence_count",
        0
    )


    citation_result = (
        validate_citations(
            answer,
            evidence_count
        )
    )


    malformed = (
        detect_malformed_identifiers(
            answer
        )
    )


    coverage = (
        check_evidence_coverage(
            answer,
            evidence_count
        )
    )


    relationship = (
        validate_relationship_answer(
            result["query"],
            retrieval,
            answer
        )
    )


    status = (
        validate_status_terms(
            answer,
            retrieval
        )
    )


    return {

        "query":
            result["query"],

        "answer":
            answer,

        "evidence_count":
            evidence_count,

        "citations":
            citation_result,

        "evidence_coverage":
            coverage,

        "malformed_identifiers":
            malformed,

        "relationship_validation":
            relationship,

        "status_validation":
            status
    }


# ============================================================
# 8. RUN VALIDATION ON CELL 11 TESTS
# ============================================================

test_queries = [

    "What is IS/IEC 60691:2023?",

    "Which BIS standard deals with thermal links?",

    "Which BIS standards refer to IS/IEC 61196-1:2005?",

    "Tell me about IS/IEC 60691:2023 and its references."
]


validation_results = []


for query in test_queries:

    print("\n")
    print("=" * 70)

    print(
        "VALIDATING:"
    )

    print(
        query
    )

    print("=" * 70)


    test_result = ask_saathi(
        query,
        max_relationships=50
    )


    validation = (
        validate_saathi_answer(
            test_result
        )
    )


    validation_results.append(
        validation
    )


    print(
        "\nAnswer:"
    )

    print(
        validation["answer"]
    )


    print(
        "\nEvidence:",
        validation[
            "evidence_count"
        ]
    )


    print(
        "Citations:",
        validation[
            "citations"
        ][
            "unique_citations"
        ]
    )


    print(
        "Invalid citations:",
        validation[
            "citations"
        ][
            "invalid_citations"
        ]
    )


    print(
        "Malformed identifiers:",
        validation[
            "malformed_identifiers"
        ]
    )


    print(
        "Relationship check:",
        validation[
            "relationship_validation"
        ][
            "passed"
        ]
    )


    print(
        "Status check:",
        validation[
            "status_validation"
        ][
            "passed"
        ]
    )


# ============================================================
# 9. OVERALL SUMMARY
# ============================================================

print("\n")
print("=" * 70)
print("CELL 12 VALIDATION SUMMARY")
print("=" * 70)


total_tests = len(
    validation_results
)


citation_passes = sum(
    r[
        "citations"
    ][
        "all_citations_valid"
    ]
    for r in validation_results
)


relationship_passes = sum(
    r[
        "relationship_validation"
    ][
        "passed"
    ]
    for r in validation_results
)


status_passes = sum(
    r[
        "status_validation"
    ][
        "passed"
    ]
    for r in validation_results
)


malformed_count = sum(
    len(
        r[
            "malformed_identifiers"
        ]
    )
    for r in validation_results
)


print(
    f"\nTests executed: {total_tests}"
)

print(
    f"Citation validation passed: "
    f"{citation_passes}/{total_tests}"
)

print(
    f"Relationship validation passed: "
    f"{relationship_passes}/{total_tests}"
)

print(
    f"Status validation passed: "
    f"{status_passes}/{total_tests}"
)

print(
    f"Malformed identifier occurrences: "
    f"{malformed_count}"
)


# ============================================================
# 10. SAVE VALIDATION REPORT
# ============================================================

validation_path = (
    "/kaggle/working/"
    "saathi_cell12_validation.json"
)


with open(
    validation_path,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        validation_results,
        f,
        indent=2,
        ensure_ascii=False
    )


print(
    f"\nValidation report saved:"
    f"\n{validation_path}"
)


print("\n" + "=" * 70)
print("CELL 12 COMPLETE")
print("=" * 70)
```

```python
# ============================================================
# SAATHI — CELL 11.1
# EVIDENCE NORMALIZATION + GEMINI RELIABILITY PATCH
# ============================================================

import re
import time
import json
from google.genai import errors


print("=" * 70)
print("SAATHI — CELL 11.1")
print("EVIDENCE NORMALIZATION + GEMINI RELIABILITY PATCH")
print("=" * 70)


# ============================================================
# 1. BIS IDENTIFIER NORMALIZER
# ============================================================

def normalize_bis_identifier(value):

    if value is None:
        return ""

    s = str(value).strip()

    # Normalize whitespace
    s = re.sub(r"\s+", " ", s)

    # --------------------------------------------------------
    # Fix malformed portal identifiers such as:
    #
    # IS/IEC :2023IEC 60691:2023
    # IS/IEC :2015IEC 60691 : 2015
    # --------------------------------------------------------

    malformed = re.search(
        r"IS/IEC\s*:\s*\d{4}\s*IEC\s*"
        r"(\d+)\s*(?::\s*(\d{4}))?",
        s,
        re.IGNORECASE
    )

    if malformed:

        number = malformed.group(1)
        year = malformed.group(2)

        if year:
            return f"IS/IEC {number}:{year}"

        return f"IS/IEC {number}"


    # --------------------------------------------------------
    # Normal standard form
    # --------------------------------------------------------

    match = re.search(
        r"\bIS/IEC\s+(\d+)"
        r"\s*:\s*(\d{4})\b",
        s,
        re.IGNORECASE
    )

    if match:

        return (
            f"IS/IEC "
            f"{match.group(1)}:"
            f"{match.group(2)}"
        )


    # --------------------------------------------------------
    # IS/IEC with spaces around colon
    # --------------------------------------------------------

    match = re.search(
        r"\bIS/IEC\s+(\d+)"
        r"\s*:\s*(\d{4})\b",
        s,
        re.IGNORECASE
    )

    if match:

        return (
            f"IS/IEC "
            f"{match.group(1)}:"
            f"{match.group(2)}"
        )


    # --------------------------------------------------------
    # Preserve legitimate identifiers such as:
    #
    # IS/IEC 691:1993
    # --------------------------------------------------------

    match = re.search(
        r"\bIS/IEC\s+(\d+)"
        r"\s*:\s*(\d{4})",
        s,
        re.IGNORECASE
    )

    if match:

        return (
            f"IS/IEC "
            f"{match.group(1)}:"
            f"{match.group(2)}"
        )


    return s


# ============================================================
# 2. NORMALIZE REFERENCE TEXT
# ============================================================

def normalize_reference_text(value):

    if value is None:
        return ""

    s = str(value).strip()

    s = re.sub(
        r"\s+",
        " ",
        s
    )

    # Fix common malformed concatenation:
    #
    # IS 616:2017IEC 60065 : 2014
    #
    # → IS 616:2017 — IEC 60065:2014
    #
    # We deliberately do not rewrite the entire reference,
    # because the original BIS relationship data must remain
    # authoritative.

    s = re.sub(
        r"(\d{4})IEC\s*",
        r"\1 — IEC ",
        s,
        flags=re.IGNORECASE
    )

    # Normalize spaces around colon for IEC references

    s = re.sub(
        r"\bIEC\s+(\d+)\s*:\s*(\d{4})\b",
        r"IEC \1:\2",
        s,
        flags=re.IGNORECASE
    )

    # Normalize obvious IS/IEC identifiers embedded in text

    s = re.sub(
        r"\bIS/IEC\s+(\d+)\s*:\s*(\d{4})\b",
        r"IS/IEC \1:\2",
        s,
        flags=re.IGNORECASE
    )

    return s


# ============================================================
# 3. CLEAN EVIDENCE OBJECT
# ============================================================

def normalize_evidence_item(
    item
):

    cleaned = dict(item)

    content = cleaned.get(
        "content"
    )


    # --------------------------------------------------------
    # Dictionary evidence
    # --------------------------------------------------------

    if isinstance(
        content,
        dict
    ):

        new_content = {}

        for key, value in content.items():

            key_lower = str(
                key
            ).lower()


            if (
                "is_number"
                in key_lower
                or "reference_number"
                in key_lower
                or key_lower == "identifier"
                or key_lower == "is_no"
            ):

                new_content[key] = (
                    normalize_bis_identifier(
                        value
                    )
                )


            elif (
                "reference"
                in key_lower
                or "international_standard"
                in key_lower
            ):

                new_content[key] = (
                    normalize_reference_text(
                        value
                    )
                )


            else:

                new_content[key] = value


        cleaned["content"] = new_content


    # --------------------------------------------------------
    # Identifier field
    # --------------------------------------------------------

    if cleaned.get(
        "identifier"
    ):

        cleaned[
            "identifier"
        ] = normalize_bis_identifier(
            cleaned[
                "identifier"
            ]
        )


    return cleaned


# ============================================================
# 4. CLEAN COMPLETE RETRIEVAL RESULT
# ============================================================

def normalize_retrieval_evidence(
    retrieval_result
):

    cleaned_result = dict(
        retrieval_result
    )

    evidence = retrieval_result.get(
        "evidence",
        []
    )

    cleaned_result[
        "evidence"
    ] = [

        normalize_evidence_item(
            item
        )

        for item in evidence
    ]


    return cleaned_result


# ============================================================
# 5. REPLACE CELL 11 EVIDENCE PREPARATION
# ============================================================

def prepare_evidence_for_llm(
    retrieval_result,
    max_chars_per_item=5000,
    max_total_chars=30000
):

    # Normalize only the copy sent to Gemini.
    cleaned_result = (
        normalize_retrieval_evidence(
            retrieval_result
        )
    )

    evidence = cleaned_result.get(
        "evidence",
        []
    )

    prepared = []

    total_chars = 0


    for idx, item in enumerate(
        evidence,
        start=1
    ):

        evidence_type = item.get(
            "evidence_type",
            ""
        )

        identifier = item.get(
            "identifier",
            ""
        )

        relationship = item.get(
            "relationship",
            ""
        )

        content = item.get(
            "content",
            {}
        )


        if isinstance(
            content,
            dict
        ):

            content_text = json.dumps(
                content,
                indent=2,
                ensure_ascii=False
            )

        else:

            content_text = str(
                content
            )


        content_text = (
            content_text[
                :max_chars_per_item
            ]
        )


        block = (
            f"[EVIDENCE {idx}]\n"
            f"Type: {evidence_type}\n"
        )


        if identifier:

            block += (
                f"Identifier: "
                f"{identifier}\n"
            )


        if relationship:

            block += (
                f"Relationship: "
                f"{relationship}\n"
            )


        block += (
            f"Content:\n"
            f"{content_text}\n"
        )


        if (
            total_chars
            + len(block)
            > max_total_chars
        ):

            break


        prepared.append(
            block
        )

        total_chars += len(
            block
        )


    return "\n".join(
        prepared
    )


# ============================================================
# 6. ROBUST GEMINI CALL
# ============================================================

def generate_grounded_answer(
    query,
    retrieval_result,
    temperature=0.1,
    max_retries=4
):

    evidence_text = (
        prepare_evidence_for_llm(
            retrieval_result
        )
    )


    if not evidence_text:

        return {

            "answer":
                "The supplied BIS evidence does not contain "
                "enough information to answer this question.",

            "model":
                MODEL_NAME,

            "evidence_count":
                0
        }


    prompt = f"""
USER QUESTION
=============

{query}


RETRIEVAL INFORMATION
=====================

Detected identifiers:
{json.dumps(
    retrieval_result.get(
        "identifiers",
        []
    ),
    ensure_ascii=False
)}

Compound query:
{retrieval_result.get(
    "compound",
    False
)}

Relationship direction:
{retrieval_result.get(
    "relationship_direction"
)}


BIS EVIDENCE
============

{evidence_text}


TASK
====

Answer the user's question using only the supplied BIS
evidence.

Every factual claim must include an [E#] citation.

If the evidence supports only part of the question,
answer the supported part and clearly identify what
cannot be established from the supplied evidence.

Use the CLEANED identifiers from the evidence rather than
reproducing malformed source strings.
"""


    last_error = None


    for attempt in range(
        max_retries
    ):

        try:

            response = (
                client.models.generate_content(

                    model=MODEL_NAME,

                    contents=prompt,

                    config={
                        "system_instruction":
                            SYSTEM_INSTRUCTION,

                        "temperature":
                            temperature,

                        "max_output_tokens":
                            2500
                    }
                )
            )


            return {

                "answer":
                    response.text,

                "model":
                    MODEL_NAME,

                "evidence_count":
                    retrieval_result.get(
                        "evidence_count",
                        0
                    ),

                "retrieval_plan":
                    retrieval_result.get(
                        "retrieval_plan",
                        []
                    ),

                "attempts":
                    attempt + 1
            }


        except Exception as e:

            last_error = e

            error_text = str(
                e
            )


            # Retry only transient/server-side
            # Gemini failures.

            is_retryable = (

                "503" in error_text

                or "UNAVAILABLE"
                in error_text

                or "500"
                in error_text

                or "INTERNAL"
                in error_text

                or "429"
                in error_text

                or "RESOURCE_EXHAUSTED"
                in error_text
            )


            if not is_retryable:

                raise


            if attempt < (
                max_retries - 1
            ):

                wait_seconds = (
                    2 ** attempt
                )

                print(
                    f"Gemini temporarily unavailable "
                    f"(attempt {attempt + 1}/"
                    f"{max_retries}). "
                    f"Retrying in "
                    f"{wait_seconds}s..."
                )

                time.sleep(
                    wait_seconds
                )


    # If all retries fail, return a structured
    # failure instead of crashing the entire pipeline.

    return {

        "answer":
            "Gemini was temporarily unavailable while "
            "generating the grounded answer. The BIS "
            "evidence retrieval itself completed successfully.",

        "model":
            MODEL_NAME,

        "evidence_count":
            retrieval_result.get(
                "evidence_count",
                0
            ),

        "retrieval_plan":
            retrieval_result.get(
                "retrieval_plan",
                []
            ),

        "generation_error":
            str(last_error),

        "attempts":
            max_retries
    }


# ============================================================
# 7. TEST NORMALIZATION DIRECTLY
# ============================================================

print("\n")
print("=" * 70)
print("IDENTIFIER NORMALIZATION TEST")
print("=" * 70)


normalization_tests = [

    "IS/IEC :2023IEC 60691:2023",

    "IS/IEC :2015IEC 60691 : 2015",

    "IS/IEC 691:1993",

    "IS/IEC 60691:2023",

    "IS/IEC 61196 : PART 1 : SEC 101 : 2005"
]


for value in normalization_tests:

    print(
        f"\nRAW:     {value}"
    )

    print(
        f"CLEAN:   "
        f"{normalize_bis_identifier(value)}"
    )


# ============================================================
# 8. TEST CELL 11 AGAIN
# ============================================================

print("\n")
print("=" * 70)
print("CELL 11.1 INTEGRATION TEST")
print("=" * 70)


patch_test_queries = [

    "What is IS/IEC 60691:2023?",

    "Which BIS standard deals with thermal links?",

    "Tell me about IS/IEC 60691:2023 and its references."
]


patch_results = []


for query in patch_test_queries:

    print("\n")
    print("-" * 70)

    print(
        f"QUERY: {query}"
    )

    print("-" * 70)


    retrieval_result = retrieve_evidence(
        query,
        semantic_fallback=True,
        max_relationships=50
    )


    llm_result = generate_grounded_answer(

        query,

        retrieval_result
    )


    result = {

        "query":
            query,

        "retrieval":
            retrieval_result,

        "answer":
            llm_result
    }


    patch_results.append(
        result
    )


    print(
        "\nANSWER:\n"
    )

    print(
        llm_result[
            "answer"
        ]
    )


    print(
        "\nGeneration attempts:",
        llm_result.get(
            "attempts",
            "N/A"
        )
    )


# ============================================================
# 9. SAVE PATCH TEST
# ============================================================

patch_path = (
    "/kaggle/working/"
    "saathi_cell11_1_patch_test.json"
)


with open(
    patch_path,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        patch_results,
        f,
        indent=2,
        ensure_ascii=False
    )


print("\n")
print("=" * 70)
print("CELL 11.1 COMPLETE")
print("=" * 70)

print(
    f"\nSaved:"
    f"\n{patch_path}"
)
```

```python
# ============================================================
# SAATHI — CELL 12
# ANSWER QUALITY + CITATION VALIDATION
# ============================================================

import re
import json


print("=" * 70)
print("SAATHI — CELL 12")
print("ANSWER QUALITY + CITATION VALIDATION")
print("=" * 70)


# ============================================================
# 1. CITATION VALIDATION
# ============================================================

def extract_citations(answer):

    if not answer:
        return []

    matches = re.findall(
        r"\[E(\d+)\]",
        answer
    )

    return [
        int(x)
        for x in matches
    ]


def validate_citations(
    answer,
    evidence_count
):

    citations = extract_citations(
        answer
    )

    invalid = [
        c
        for c in citations
        if c < 1 or c > evidence_count
    ]

    return {
        "citations_found":
            citations,

        "unique_citations":
            sorted(set(citations)),

        "citation_count":
            len(citations),

        "invalid_citations":
            invalid,

        "all_citations_valid":
            len(invalid) == 0
    }


# ============================================================
# 2. BIS IDENTIFIER EXTRACTION
# ============================================================

def extract_bis_identifiers(text):

    if not text:
        return []

    patterns = [

        # IS/IEC 60691:2023
        r"\bIS/IEC\s*[0-9]+"
        r"(?:\s*:\s*[0-9]{4})?",

        # IS 8504 (Part 6):2012
        r"\bIS\s*[0-9]+"
        r"(?:\s*\([^)]+\))?"
        r"(?:\s*:\s*[0-9]{4})?",

        # IEC 61210:2010
        r"\bIEC\s*[0-9]+"
        r"(?:\s*:\s*[0-9]{4})?"
    ]

    found = []

    for pattern in patterns:

        found.extend(
            re.findall(
                pattern,
                text,
                flags=re.IGNORECASE
            )
        )

    return list(
        dict.fromkeys(
            x.strip()
            for x in found
        )
    )


# ============================================================
# 3. MALFORMED IDENTIFIER DETECTION
# ============================================================

def detect_malformed_identifiers(
    text
):

    if not text:
        return []

    suspicious_patterns = [

        # Examples:
        # IS/IEC :2023IEC
        # IS/IEC :2015IEC

        r"IS/IEC\s*:\s*\d{4}\s*IEC",

        # Examples:
        # 2023IEC 60691
        # 2015IEC 60691

        r"\d{4}\s*IEC\s*\d+",

        # Example:
        # IS/IEC :2023

        r"IS/IEC\s*:\s*\d{4}"
    ]

    found = []

    for pattern in suspicious_patterns:

        found.extend(
            re.findall(
                pattern,
                text,
                flags=re.IGNORECASE
            )
        )

    return list(
        dict.fromkeys(
            found
        )
    )


# ============================================================
# 4. EVIDENCE COVERAGE
# ============================================================

def check_evidence_coverage(
    answer,
    evidence_count
):

    citations = extract_citations(
        answer
    )

    if evidence_count == 0:

        return {
            "has_evidence":
                False,

            "coverage":
                0.0
        }

    unique_citations = set(
        c
        for c in citations
        if 1 <= c <= evidence_count
    )

    coverage = (
        len(unique_citations)
        / evidence_count
    )

    return {
        "has_evidence":
            evidence_count > 0,

        "coverage":
            round(
                coverage,
                4
            )
    }


# ============================================================
# 5. RELATIONSHIP VALIDATION
# ============================================================

def validate_relationship_answer(
    query,
    retrieval_result,
    answer
):

    direction = retrieval_result.get(
        "relationship_direction"
    )

    result = {
        "relationship_type":
            direction,

        "passed":
            True,

        "issues":
            []
    }


    # No relationship query
    if direction is None:

        return result


    # --------------------------------------------------------
    # Reverse relationship
    # "Which standards refer to X?"
    # --------------------------------------------------------

    if direction == "reverse":

        if not re.search(
            r"\b(refer|references|referenced|"
            r"cite|cites|referencing)\b",
            answer,
            re.IGNORECASE
        ):

            result[
                "issues"
            ].append(
                "Answer does not clearly describe "
                "the reverse reference relationship."
            )


    # --------------------------------------------------------
    # Outgoing relationship
    # "What standards does X refer to?"
    # --------------------------------------------------------

    elif direction == "outgoing":

        if not re.search(
            r"\b(refer|references|reference|"
            r"cites|cited)\b",
            answer,
            re.IGNORECASE
        ):

            result[
                "issues"
            ].append(
                "Answer does not clearly describe "
                "the outgoing reference relationship."
            )


    result["passed"] = (
        len(
            result["issues"]
        ) == 0
    )

    return result


# ============================================================
# 6. STATUS CONSISTENCY CHECK
# ============================================================

def validate_status_terms(
    answer,
    retrieval_result
):

    answer_lower = (
        answer.lower()
        if answer
        else ""
    )

    evidence_text = json.dumps(
        retrieval_result.get(
            "evidence",
            []
        ),
        ensure_ascii=False
    ).lower()


    issues = []


    # --------------------------------------------------------
    # Active claim
    # --------------------------------------------------------

    if (
        re.search(
            r"\bactive\b",
            answer_lower
        )
        and
        "active" not in evidence_text
    ):

        issues.append(
            "Answer mentions ACTIVE but the "
            "evidence does not contain an active status."
        )


    # --------------------------------------------------------
    # Withdrawn claim
    # --------------------------------------------------------

    if (
        re.search(
            r"\bwithdrawn\b",
            answer_lower
        )
        and
        "withdrawn" not in evidence_text
    ):

        issues.append(
            "Answer mentions WITHDRAWN but the "
            "evidence does not contain a withdrawn status."
        )


    return {

        "passed":
            len(issues) == 0,

        "issues":
            issues
    }


# ============================================================
# 7. GENERATION FAILURE CHECK
# ============================================================

def validate_generation(
    result
):

    answer_data = result.get(
        "answer",
        {}
    )

    if "generation_error" in answer_data:

        return {

            "passed":
                False,

            "error":
                answer_data[
                    "generation_error"
                ]
        }


    answer = answer_data.get(
        "answer",
        ""
    )


    if not answer.strip():

        return {

            "passed":
                False,

            "error":
                "Empty answer returned."
        }


    return {

        "passed":
            True,

        "error":
            None
    }


# ============================================================
# 8. COMPLETE ANSWER VALIDATOR
# ============================================================

def validate_saathi_answer(
    result
):

    retrieval = result[
        "retrieval"
    ]

    answer_data = result[
        "answer"
    ]

    answer = answer_data.get(
        "answer",
        ""
    )


    evidence_count = retrieval.get(
        "evidence_count",
        0
    )


    citation_result = (
        validate_citations(
            answer,
            evidence_count
        )
    )


    malformed = (
        detect_malformed_identifiers(
            answer
        )
    )


    coverage = (
        check_evidence_coverage(
            answer,
            evidence_count
        )
    )


    relationship = (
        validate_relationship_answer(
            result["query"],
            retrieval,
            answer
        )
    )


    status = (
        validate_status_terms(
            answer,
            retrieval
        )
    )


    generation = (
        validate_generation(
            result
        )
    )


    # --------------------------------------------------------
    # Overall validation
    #
    # Coverage is intentionally NOT required to be 100%.
    # A summarized answer does not need to cite every retrieved
    # evidence record.
    # --------------------------------------------------------

    overall_pass = (

        generation["passed"]

        and
        citation_result[
            "all_citations_valid"
        ]

        and
        len(malformed) == 0

        and
        relationship["passed"]

        and
        status["passed"]
    )


    return {

        "query":
            result["query"],

        "answer":
            answer,

        "evidence_count":
            evidence_count,

        "citations":
            citation_result,

        "evidence_coverage":
            coverage,

        "malformed_identifiers":
            malformed,

        "relationship_validation":
            relationship,

        "status_validation":
            status,

        "generation_validation":
            generation,

        "overall_pass":
            overall_pass
    }


# ============================================================
# 9. TEST QUERIES
# ============================================================

test_queries = [

    # Exact lookup
    "What is IS/IEC 60691:2023?",

    # Semantic search
    "Which BIS standard deals with thermal links?",

    # Reverse relationship
    "Which BIS standards refer to IS/IEC 61196-1:2005?",

    # Compound query
    "Tell me about IS/IEC 60691:2023 and its references."
]


# ============================================================
# 10. RUN VALIDATION
# ============================================================

validation_results = []


for query in test_queries:

    print("\n")
    print("=" * 70)

    print(
        "VALIDATING:"
    )

    print(
        query
    )

    print("=" * 70)


    try:

        test_result = ask_saathi(
            query,
            max_relationships=50
        )


        validation = (
            validate_saathi_answer(
                test_result
            )
        )


        validation_results.append(
            validation
        )


        # ----------------------------------------------------
        # Answer
        # ----------------------------------------------------

        print(
            "\nANSWER:\n"
        )

        print(
            validation[
                "answer"
            ]
        )


        # ----------------------------------------------------
        # Validation information
        # ----------------------------------------------------

        print(
            "\nEvidence:",
            validation[
                "evidence_count"
            ]
        )


        print(
            "Citations:",
            validation[
                "citations"
            ][
                "unique_citations"
            ]
        )


        print(
            "Invalid citations:",
            validation[
                "citations"
            ][
                "invalid_citations"
            ]
        )


        print(
            "Malformed identifiers:",
            validation[
                "malformed_identifiers"
            ]
        )


        print(
            "Relationship check:",
            validation[
                "relationship_validation"
            ][
                "passed"
            ]
        )


        print(
            "Status check:",
            validation[
                "status_validation"
            ][
                "passed"
            ]
        )


        print(
            "Generation check:",
            validation[
                "generation_validation"
            ][
                "passed"
            ]
        )


        print(
            "OVERALL:",
            "PASS"
            if validation[
                "overall_pass"
            ]
            else "FAIL"
        )


    except Exception as e:

        print(
            "\nVALIDATION ERROR:"
        )

        print(
            str(e)
        )


        validation_results.append({

            "query":
                query,

            "overall_pass":
                False,

            "error":
                str(e)
        })


# ============================================================
# 11. OVERALL SUMMARY
# ============================================================

print("\n")
print("=" * 70)
print("CELL 12 VALIDATION SUMMARY")
print("=" * 70)


total_tests = len(
    validation_results
)


passed_tests = sum(
    1
    for r in validation_results
    if r.get(
        "overall_pass",
        False
    )
)


citation_passes = sum(

    1

    for r in validation_results

    if r.get(
        "citations",
        {}
    ).get(
        "all_citations_valid",
        False
    )
)


relationship_passes = sum(

    1

    for r in validation_results

    if r.get(
        "relationship_validation",
        {}
    ).get(
        "passed",
        False
    )
)


status_passes = sum(

    1

    for r in validation_results

    if r.get(
        "status_validation",
        {}
    ).get(
        "passed",
        False
    )
)


generation_passes = sum(

    1

    for r in validation_results

    if r.get(
        "generation_validation",
        {}
    ).get(
        "passed",
        False
    )
)


malformed_count = sum(

    len(
        r.get(
            "malformed_identifiers",
            []
        )
    )

    for r in validation_results
)


print(
    f"\nTests executed: "
    f"{total_tests}"
)

print(
    f"Overall tests passed: "
    f"{passed_tests}/{total_tests}"
)

print(
    f"Citation validation passed: "
    f"{citation_passes}/{total_tests}"
)

print(
    f"Relationship validation passed: "
    f"{relationship_passes}/{total_tests}"
)

print(
    f"Status validation passed: "
    f"{status_passes}/{total_tests}"
)

print(
    f"Generation validation passed: "
    f"{generation_passes}/{total_tests}"
)

print(
    f"Malformed identifier occurrences: "
    f"{malformed_count}"
)


# ============================================================
# 12. PRINT FAILED TESTS
# ============================================================

failed = [

    r

    for r in validation_results

    if not r.get(
        "overall_pass",
        False
    )
]


if failed:

    print("\n")
    print("=" * 70)
    print("FAILED TEST DETAILS")
    print("=" * 70)


    for r in failed:

        print(
            f"\nQuery: "
            f"{r.get('query')}"
        )


        if "error" in r:

            print(
                f"Error: "
                f"{r['error']}"
            )

            continue


        if r.get(
            "malformed_identifiers"
        ):

            print(
                "Malformed identifiers:",
                r[
                    "malformed_identifiers"
                ]
            )


        if not r.get(
            "citations",
            {}
        ).get(
            "all_citations_valid",
            True
        ):

            print(
                "Invalid citations:",
                r[
                    "citations"
                ][
                    "invalid_citations"
                ]
            )


        relationship_issues = (
            r.get(
                "relationship_validation",
                {}
            ).get(
                "issues",
                []
            )
        )


        if relationship_issues:

            print(
                "Relationship issues:",
                relationship_issues
            )


        status_issues = (
            r.get(
                "status_validation",
                {}
            ).get(
                "issues",
                []
            )
        )


        if status_issues:

            print(
                "Status issues:",
                status_issues
            )


        generation_error = (
            r.get(
                "generation_validation",
                {}
            ).get(
                "error"
            )
        )


        if generation_error:

            print(
                "Generation error:",
                generation_error
            )


else:

    print("\n")
    print(
        "ALL VALIDATION TESTS PASSED."
    )


# ============================================================
# 13. SAVE VALIDATION REPORT
# ============================================================

validation_path = (
    "/kaggle/working/"
    "saathi_cell12_validation.json"
)


with open(
    validation_path,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        validation_results,
        f,
        indent=2,
        ensure_ascii=False
    )


print(
    f"\nValidation report saved:"
    f"\n{validation_path}"
)


print("\n" + "=" * 70)
print("CELL 12 COMPLETE")
print("=" * 70)
```

```python
# ============================================================
# SAATHI — CELL 13
# API-READY INFERENCE LAYER
# ============================================================

import json
import re
from datetime import datetime


print("=" * 70)
print("SAATHI — CELL 13")
print("API-READY INFERENCE LAYER")
print("=" * 70)


# ============================================================
# 1. BUILD SOURCE RECORD
# ============================================================

def build_source_record(
    evidence_item,
    evidence_number
):

    content = evidence_item.get(
        "content",
        {}
    )

    if not isinstance(
        content,
        dict
    ):
        content = {
            "text": str(content)
        }


    identifier = (
        evidence_item.get(
            "identifier"
        )
        or content.get(
            "is_number"
        )
        or content.get(
            "reference_number"
        )
        or ""
    )


    title = (
        content.get(
            "is_title"
        )
        or content.get(
            "reference_title"
        )
        or ""
    )


    return {

        "evidence_id":
            f"E{evidence_number}",

        "type":
            evidence_item.get(
                "evidence_type",
                ""
            ),

        "relationship":
            evidence_item.get(
                "relationship",
                None
            ),

        "identifier":
            str(identifier),

        "title":
            str(title),

        "source":
            content
    }


# ============================================================
# 2. EXTRACT SOURCE RECORDS
# ============================================================

def build_sources(
    retrieval_result
):

    evidence = retrieval_result.get(
        "evidence",
        []
    )

    sources = []


    for index, item in enumerate(
        evidence,
        start=1
    ):

        sources.append(
            build_source_record(
                item,
                index
            )
        )


    return sources


# ============================================================
# 3. EXTRACT CITATIONS FROM ANSWER
# ============================================================

def extract_answer_citations(
    answer
):

    if not answer:
        return []


    citations = re.findall(
        r"\[E(\d+)\]",
        answer
    )


    return list(
        dict.fromkeys(
            f"E{c}"
            for c in citations
        )
    )


# ============================================================
# 4. CLASSIFY QUERY
# ============================================================

def classify_query(
    retrieval_result
):

    relationship = retrieval_result.get(
        "relationship_direction"
    )

    compound = retrieval_result.get(
        "compound",
        False
    )

    identifiers = retrieval_result.get(
        "identifiers",
        []
    )


    if compound:

        query_type = "compound"

    elif relationship == "reverse":

        query_type = "reverse_reference"

    elif relationship == "outgoing":

        query_type = "outgoing_reference"

    elif identifiers:

        query_type = "exact_standard"

    else:

        query_type = "semantic"


    return query_type


# ============================================================
# 5. DETERMINE RESPONSE STATUS
# ============================================================

def determine_response_status(
    llm_result,
    retrieval_result
):

    if llm_result.get(
        "generation_error"
    ):

        return "generation_error"


    if not retrieval_result.get(
        "evidence"
    ):

        return "no_evidence"


    return "success"


# ============================================================
# 6. BUILD API RESPONSE
# ============================================================

def build_api_response(
    query,
    retrieval_result,
    llm_result
):

    answer = llm_result.get(
        "answer",
        ""
    )


    sources = build_sources(
        retrieval_result
    )


    citations = (
        extract_answer_citations(
            answer
        )
    )


    query_type = classify_query(
        retrieval_result
    )


    status = determine_response_status(
        llm_result,
        retrieval_result
    )


    response = {

        # ----------------------------------------------------
        # Request
        # ----------------------------------------------------

        "query":
            query,

        "timestamp":
            datetime.utcnow().isoformat()
            + "Z",


        # ----------------------------------------------------
        # Answer
        # ----------------------------------------------------

        "answer":
            answer,


        # ----------------------------------------------------
        # Query understanding
        # ----------------------------------------------------

        "query_type":
            query_type,

        "identifiers":
            retrieval_result.get(
                "identifiers",
                []
            ),

        "relationship_direction":
            retrieval_result.get(
                "relationship_direction"
            ),

        "compound":
            retrieval_result.get(
                "compound",
                False
            ),


        # ----------------------------------------------------
        # Retrieval
        # ----------------------------------------------------

        "retrieval_plan":
            retrieval_result.get(
                "retrieval_plan",
                []
            ),

        "evidence_count":
            retrieval_result.get(
                "evidence_count",
                0
            ),


        # ----------------------------------------------------
        # Citations
        # ----------------------------------------------------

        "citations":
            citations,

        "sources":
            sources,


        # ----------------------------------------------------
        # Generation
        # ----------------------------------------------------

        "model":
            llm_result.get(
                "model",
                MODEL_NAME
            ),

        "generation_attempts":
            llm_result.get(
                "attempts",
                1
            ),


        # ----------------------------------------------------
        # Overall status
        # ----------------------------------------------------

        "status":
            status
    }


    if llm_result.get(
        "generation_error"
    ):

        response[
            "generation_error"
        ] = llm_result[
            "generation_error"
        ]


    return response


# ============================================================
# 7. MAIN SAATHI INFERENCE FUNCTION
# ============================================================

def saathi_inference(
    query,
    max_relationships=50
):

    if not isinstance(
        query,
        str
    ):

        raise TypeError(
            "query must be a string"
        )


    query = query.strip()


    if not query:

        raise ValueError(
            "query cannot be empty"
        )


    # --------------------------------------------------------
    # STEP 1 — Retrieval
    # --------------------------------------------------------

    retrieval_result = retrieve_evidence(

        query,

        semantic_fallback=True,

        max_relationships=
            max_relationships
    )


    # --------------------------------------------------------
    # STEP 2 — Grounded LLM
    # --------------------------------------------------------

    llm_result = generate_grounded_answer(

        query,

        retrieval_result
    )


    # --------------------------------------------------------
    # STEP 3 — Structured response
    # --------------------------------------------------------

    response = build_api_response(

        query,

        retrieval_result,

        llm_result
    )


    return response


# ============================================================
# 8. PRETTY PRINTER
# ============================================================

def display_api_response(
    response
):

    print("\n")
    print("=" * 70)
    print("SAATHI API RESPONSE")
    print("=" * 70)


    print(
        "\nStatus:",
        response["status"]
    )


    print(
        "Query type:",
        response["query_type"]
    )


    print(
        "Evidence:",
        response["evidence_count"]
    )


    print(
        "\nANSWER"
    )

    print("-" * 70)

    print(
        response["answer"]
    )


    print(
        "\nCITATIONS"
    )

    print("-" * 70)

    print(
        response["citations"]
    )


    print(
        "\nRETRIEVAL PLAN"
    )

    print("-" * 70)

    for step in response[
        "retrieval_plan"
    ]:

        print(
            f"→ {step}"
        )


    print(
        "\nSOURCE COUNT:",
        len(
            response[
                "sources"
            ]
        )
    )


# ============================================================
# 9. TEST 1 — EXACT LOOKUP
# ============================================================

print("\n")
print("=" * 70)
print("TEST 1 — EXACT STANDARD")
print("=" * 70)


response_1 = saathi_inference(
    "What is IS/IEC 60691:2023?"
)


display_api_response(
    response_1
)


# ============================================================
# 10. TEST 2 — SEMANTIC QUERY
# ============================================================

print("\n")
print("=" * 70)
print("TEST 2 — SEMANTIC SEARCH")
print("=" * 70)


response_2 = saathi_inference(
    "Which BIS standard deals with thermal links?"
)


display_api_response(
    response_2
)


# ============================================================
# 11. TEST 3 — REVERSE REFERENCE
# ============================================================

print("\n")
print("=" * 70)
print("TEST 3 — REVERSE REFERENCES")
print("=" * 70)


response_3 = saathi_inference(
    "Which BIS standards refer to IS/IEC 61196-1:2005?",
    max_relationships=50
)


display_api_response(
    response_3
)


# ============================================================
# 12. TEST 4 — COMPOUND QUERY
# ============================================================

print("\n")
print("=" * 70)
print("TEST 4 — COMPOUND QUERY")
print("=" * 70)


response_4 = saathi_inference(
    "Tell me about IS/IEC 60691:2023 and its references.",
    max_relationships=50
)


display_api_response(
    response_4
)


# ============================================================
# 13. SAVE API RESPONSES
# ============================================================

api_test_results = {

    "exact":
        response_1,

    "semantic":
        response_2,

    "reverse_reference":
        response_3,

    "compound":
        response_4
}


api_test_path = (
    "/kaggle/working/"
    "saathi_cell13_api_test_results.json"
)


with open(
    api_test_path,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        api_test_results,
        f,
        indent=2,
        ensure_ascii=False
    )


# ============================================================
# 14. JSON PREVIEW
# ============================================================

print("\n")
print("=" * 70)
print("API JSON STRUCTURE")
print("=" * 70)


preview = {

    "query":
        response_1["query"],

    "answer":
        response_1["answer"],

    "query_type":
        response_1["query_type"],

    "identifiers":
        response_1["identifiers"],

    "relationship_direction":
        response_1[
            "relationship_direction"
        ],

    "retrieval_plan":
        response_1[
            "retrieval_plan"
        ],

    "evidence_count":
        response_1[
            "evidence_count"
        ],

    "citations":
        response_1[
            "citations"
        ],

    "source_count":
        len(
            response_1[
                "sources"
            ]
        ),

    "status":
        response_1["status"]
}


print(
    json.dumps(
        preview,
        indent=2,
        ensure_ascii=False
    )
)


print("\n")
print("=" * 70)
print("CELL 13 COMPLETE")
print("=" * 70)

print(
    f"\nSaved:"
    f"\n{api_test_path}"
)
```

```python
# ============================================================
# SAATHI — CELL 14
# PRODUCTION SOURCE / CITATION BUILDER
# ============================================================

import json
import re
from datetime import datetime, timezone


print("=" * 70)
print("SAATHI — CELL 14")
print("PRODUCTION SOURCE / CITATION BUILDER")
print("=" * 70)


# ============================================================
# 1. SAFE VALUE HELPER
# ============================================================

def first_non_empty(*values):

    for value in values:

        if value is None:
            continue

        if isinstance(value, str):

            if value.strip():
                return value.strip()

        else:
            return value

    return None


# ============================================================
# 2. EXTRACT URL FROM EVIDENCE
# ============================================================

def extract_source_url(content):

    if not isinstance(
        content,
        dict
    ):
        return None


    # Prefer the official BIS detail page.
    url = first_non_empty(

        content.get(
            "detail_url"
        ),

        content.get(
            "url"
        ),

        content.get(
            "source_url"
        )
    )


    if url:
        return str(url)


    return None


# ============================================================
# 3. CLEAN SOURCE CONTENT
# ============================================================

def build_clean_source(
    evidence_item,
    evidence_number
):

    content = evidence_item.get(
        "content",
        {}
    )


    if not isinstance(
        content,
        dict
    ):

        content = {
            "text":
                str(content)
        }


    evidence_id = (
        f"E{evidence_number}"
    )


    # --------------------------------------------------------
    # Identifier
    # --------------------------------------------------------

    identifier = first_non_empty(

        evidence_item.get(
            "identifier"
        ),

        content.get(
            "is_number"
        ),

        content.get(
            "reference_number"
        ),

        content.get(
            "international_standard"
        )
    )


    # --------------------------------------------------------
    # Title
    # --------------------------------------------------------

    title = first_non_empty(

        content.get(
            "is_title"
        ),

        content.get(
            "reference_title"
        ),

        content.get(
            "title"
        )
    )


    # --------------------------------------------------------
    # Status
    # --------------------------------------------------------

    status = first_non_empty(

        content.get(
            "status"
        ),

        content.get(
            "catalogue_withdrawn_status"
        ),

        content.get(
            "withdrawn_status"
        )
    )


    # --------------------------------------------------------
    # Committee
    # --------------------------------------------------------

    committee = first_non_empty(

        content.get(
            "technical_committee"
        ),

        content.get(
            "reference_committee"
        ),

        content.get(
            "catalogue_technical_committee"
        )
    )


    # --------------------------------------------------------
    # Department
    # --------------------------------------------------------

    department = first_non_empty(

        content.get(
            "technical_department"
        )
    )


    # --------------------------------------------------------
    # Relationship
    # --------------------------------------------------------

    relationship = evidence_item.get(
        "relationship"
    )


    # --------------------------------------------------------
    # Source URL
    # --------------------------------------------------------

    source_url = extract_source_url(
        content
    )


    # --------------------------------------------------------
    # Source type
    # --------------------------------------------------------

    source_type = evidence_item.get(
        "evidence_type",
        "unknown"
    )


    return {

        "evidence_id":
            evidence_id,

        "type":
            source_type,

        "relationship":
            relationship,

        "identifier":
            identifier,

        "title":
            title,

        "status":
            status,

        "technical_committee":
            committee,

        "technical_department":
            department,

        "source_url":
            source_url
    }


# ============================================================
# 4. BUILD CLEAN SOURCES
# ============================================================

def build_production_sources(
    retrieval_result
):

    evidence = retrieval_result.get(
        "evidence",
        []
    )


    sources = []


    for index, item in enumerate(
        evidence,
        start=1
    ):

        sources.append(

            build_clean_source(
                item,
                index
            )

        )


    return sources


# ============================================================
# 5. EXTRACT CITATIONS FROM ANSWER
# ============================================================

def extract_answer_citations(
    answer
):

    if not answer:
        return []


    matches = re.findall(
        r"\[E(\d+)\]",
        answer
    )


    return list(
        dict.fromkeys(
            f"E{number}"
            for number in matches
        )
    )


# ============================================================
# 6. ONLY RETURN SOURCES ACTUALLY CITED
# ============================================================

def build_cited_sources(
    answer,
    sources
):

    citation_ids = set(
        extract_answer_citations(
            answer
        )
    )


    return [

        source

        for source in sources

        if source[
            "evidence_id"
        ] in citation_ids

    ]


# ============================================================
# 7. BUILD FRONTEND-FRIENDLY RESPONSE
# ============================================================

def build_production_response(
    query,
    retrieval_result,
    llm_result
):

    answer = llm_result.get(
        "answer",
        ""
    )


    all_sources = (
        build_production_sources(
            retrieval_result
        )
    )


    citations = (
        extract_answer_citations(
            answer
        )
    )


    cited_sources = (
        build_cited_sources(
            answer,
            all_sources
        )
    )


    relationship = retrieval_result.get(
        "relationship_direction"
    )


    compound = retrieval_result.get(
        "compound",
        False
    )


    identifiers = retrieval_result.get(
        "identifiers",
        []
    )


    if compound:

        query_type = "compound"

    elif relationship == "reverse":

        query_type = "reverse_reference"

    elif relationship == "outgoing":

        query_type = "outgoing_reference"

    elif identifiers:

        query_type = "exact_standard"

    else:

        query_type = "semantic"


    if llm_result.get(
        "generation_error"
    ):

        status = "generation_error"

    elif not retrieval_result.get(
        "evidence"
    ):

        status = "no_evidence"

    else:

        status = "success"


    return {

        "query":
            query,

        "answer":
            answer,

        "query_type":
            query_type,

        "identifiers":
            identifiers,

        "relationship_direction":
            relationship,

        "compound":
            compound,

        "retrieval_plan":
            retrieval_result.get(
                "retrieval_plan",
                []
            ),

        "evidence_count":
            retrieval_result.get(
                "evidence_count",
                0
            ),

        "citations":
            citations,

        "sources":
            cited_sources,

        "total_sources_available":
            len(all_sources),

        "model":
            llm_result.get(
                "model",
                MODEL_NAME
            ),

        "generation_attempts":
            llm_result.get(
                "attempts",
                1
            ),

        "status":
            status,

        "timestamp":
            datetime.now(
                timezone.utc
            ).isoformat()
    }


# ============================================================
# 8. MAIN PRODUCTION INFERENCE FUNCTION
# ============================================================

def saathi_production_inference(
    query,
    max_relationships=50
):

    if not isinstance(
        query,
        str
    ):

        raise TypeError(
            "query must be a string"
        )


    query = query.strip()


    if not query:

        raise ValueError(
            "query cannot be empty"
        )


    # --------------------------------------------------------
    # Retrieval
    # --------------------------------------------------------

    retrieval_result = retrieve_evidence(

        query,

        semantic_fallback=True,

        max_relationships=
            max_relationships
    )


    # --------------------------------------------------------
    # Grounded generation
    # --------------------------------------------------------

    llm_result = generate_grounded_answer(

        query,

        retrieval_result
    )


    # --------------------------------------------------------
    # Production response
    # --------------------------------------------------------

    return build_production_response(

        query,

        retrieval_result,

        llm_result
    )


# ============================================================
# 9. DISPLAY PRODUCTION RESPONSE
# ============================================================

def display_production_response(
    response
):

    print("\n")
    print("=" * 70)

    print(
        "PRODUCTION RESPONSE"
    )

    print("=" * 70)


    print(
        "\nStatus:",
        response["status"]
    )


    print(
        "Query type:",
        response["query_type"]
    )


    print(
        "Evidence count:",
        response["evidence_count"]
    )


    print(
        "Citations:",
        response["citations"]
    )


    print(
        "\nANSWER"
    )

    print("-" * 70)

    print(
        response["answer"]
    )


    print(
        "\nCITED SOURCES"
    )

    print("-" * 70)


    for source in response[
        "sources"
    ]:

        print(
            f"\n[{source['evidence_id']}]"
        )

        print(
            "Type:",
            source["type"]
        )

        print(
            "Identifier:",
            source["identifier"]
        )

        print(
            "Title:",
            source["title"]
        )

        print(
            "Status:",
            source["status"]
        )

        print(
            "Committee:",
            source["technical_committee"]
        )

        print(
            "URL:",
            source["source_url"]
        )


# ============================================================
# 10. TEST 1 — EXACT STANDARD
# ============================================================

print("\n")
print("=" * 70)
print("TEST 1 — EXACT STANDARD")
print("=" * 70)


production_1 = (
    saathi_production_inference(
        "What is IS/IEC 60691:2023?"
    )
)


display_production_response(
    production_1
)


# ============================================================
# 11. TEST 2 — SEMANTIC
# ============================================================

print("\n")
print("=" * 70)
print("TEST 2 — SEMANTIC")
print("=" * 70)


production_2 = (
    saathi_production_inference(
        "Which BIS standard deals with thermal links?"
    )
)


display_production_response(
    production_2
)


# ============================================================
# 12. TEST 3 — COMPOUND
# ============================================================

print("\n")
print("=" * 70)
print("TEST 3 — COMPOUND")
print("=" * 70)


production_3 = (
    saathi_production_inference(
        "Tell me about IS/IEC 60691:2023 and its references."
    )
)


display_production_response(
    production_3
)


# ============================================================
# 13. SAVE PRODUCTION TEST RESULTS
# ============================================================

production_results = {

    "exact":
        production_1,

    "semantic":
        production_2,

    "compound":
        production_3
}


production_path = (
    "/kaggle/working/"
    "saathi_cell14_production_results.json"
)


with open(
    production_path,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        production_results,
        f,
        indent=2,
        ensure_ascii=False
    )


# ============================================================
# 14. JSON PREVIEW
# ============================================================

print("\n")
print("=" * 70)
print("FRONTEND RESPONSE PREVIEW")
print("=" * 70)


print(
    json.dumps(
        production_1,
        indent=2,
        ensure_ascii=False
    )
)


print("\n")
print("=" * 70)
print("CELL 14 COMPLETE")
print("=" * 70)

print(
    f"\nSaved:"
    f"\n{production_path}"
)
```

```python
# ============================================================
# SAATHI — CELL 14.1
# PATCH PRODUCTION SOURCE METADATA
# ============================================================

import json
import re
from datetime import datetime, timezone


print("=" * 70)
print("SAATHI — CELL 14.1")
print("PATCHING PRODUCTION SOURCE METADATA")
print("=" * 70)


# ============================================================
# 1. RECURSIVE VALUE SEARCH
# ============================================================

def recursive_find_value(obj, keys):

    """
    Search recursively through dictionaries/lists for the first
    non-empty value matching one of the requested keys.
    """

    if isinstance(obj, dict):

        # Prefer direct matches first.
        for key in keys:

            if key in obj:

                value = obj[key]

                if value is not None:

                    if isinstance(value, str):

                        if value.strip():

                            return value.strip()

                    else:

                        return value


        # Then recursively inspect nested values.
        for value in obj.values():

            result = recursive_find_value(
                value,
                keys
            )

            if result is not None:

                return result


    elif isinstance(obj, list):

        for item in obj:

            result = recursive_find_value(
                item,
                keys
            )

            if result is not None:

                return result


    return None


# ============================================================
# 2. GET COMPLETE EVIDENCE FIELD
# ============================================================

def get_evidence_value(
    evidence_item,
    content,
    keys
):

    """
    Search both the evidence object and its content object.
    """

    value = recursive_find_value(
        evidence_item,
        keys
    )

    if value is not None:
        return value


    return recursive_find_value(
        content,
        keys
    )


# ============================================================
# 3. URL EXTRACTION
# ============================================================

def extract_production_url(
    evidence_item,
    content
):

    return get_evidence_value(

        evidence_item,

        content,

        [
            "detail_url",
            "source_url",
            "url"
        ]
    )


# ============================================================
# 4. BUILD ONE PRODUCTION SOURCE
# ============================================================

def build_clean_source_v2(
    evidence_item,
    evidence_number
):

    content = evidence_item.get(
        "content",
        {}
    )


    if not isinstance(
        content,
        dict
    ):

        content = {
            "text": str(content)
        }


    evidence_id = (
        f"E{evidence_number}"
    )


    # --------------------------------------------------------
    # IDENTIFIER
    # --------------------------------------------------------

    identifier = get_evidence_value(

        evidence_item,

        content,

        [
            "is_number",
            "IS_number",
            "identifier",
            "reference_number",
            "international_standard"
        ]
    )


    # --------------------------------------------------------
    # TITLE
    # --------------------------------------------------------

    title = get_evidence_value(

        evidence_item,

        content,

        [
            "is_title",
            "IS_title",
            "title",
            "reference_title",
            "catalogue_title"
        ]
    )


    # --------------------------------------------------------
    # STATUS
    # --------------------------------------------------------

    status = get_evidence_value(

        evidence_item,

        content,

        [
            "status",
            "catalogue_withdrawn_status",
            "withdrawn_status"
        ]
    )


    # --------------------------------------------------------
    # TECHNICAL DEPARTMENT
    # --------------------------------------------------------

    technical_department = get_evidence_value(

        evidence_item,

        content,

        [
            "technical_department"
        ]
    )


    # --------------------------------------------------------
    # TECHNICAL COMMITTEE
    # --------------------------------------------------------

    technical_committee = get_evidence_value(

        evidence_item,

        content,

        [
            "technical_committee",
            "catalogue_technical_committee",
            "reference_committee"
        ]
    )


    # --------------------------------------------------------
    # ASPECT
    # --------------------------------------------------------

    aspect = get_evidence_value(

        evidence_item,

        content,

        [
            "aspect",
            "catalogue_aspect"
        ]
    )


    # --------------------------------------------------------
    # DEGREE OF EQUIVALENCE
    # --------------------------------------------------------

    degree_of_equivalence = get_evidence_value(

        evidence_item,

        content,

        [
            "degree_of_equivalence"
        ]
    )


    # --------------------------------------------------------
    # REVISIONS
    # --------------------------------------------------------

    number_of_revisions = get_evidence_value(

        evidence_item,

        content,

        [
            "number_of_revisions"
        ]
    )


    # --------------------------------------------------------
    # AMENDMENTS
    # --------------------------------------------------------

    number_of_amendments = get_evidence_value(

        evidence_item,

        content,

        [
            "number_of_amendments",
            "amendment_count"
        ]
    )


    # --------------------------------------------------------
    # REAFFIRMATION
    # --------------------------------------------------------

    reaffirmation_year = get_evidence_value(

        evidence_item,

        content,

        [
            "reaffirmation_year",
            "catalogue_reaffirmation"
        ]
    )


    # --------------------------------------------------------
    # SOURCE URL
    # --------------------------------------------------------

    source_url = extract_production_url(

        evidence_item,

        content
    )


    # --------------------------------------------------------
    # DOWNLOAD URL
    # --------------------------------------------------------

    download_url = get_evidence_value(

        evidence_item,

        content,

        [
            "download_url"
        ]
    )


    # --------------------------------------------------------
    # COMPOSITION URL
    # --------------------------------------------------------

    composition_url = get_evidence_value(

        evidence_item,

        content,

        [
            "composition_url"
        ]
    )


    # --------------------------------------------------------
    # RELATIONSHIP
    # --------------------------------------------------------

    relationship = (
        evidence_item.get(
            "relationship"
        )
    )


    # --------------------------------------------------------
    # SOURCE TYPE
    # --------------------------------------------------------

    source_type = (

        evidence_item.get(
            "evidence_type"
        )

        or

        evidence_item.get(
            "type"
        )

        or

        "unknown"
    )


    # --------------------------------------------------------
    # RETURN CLEAN OBJECT
    # --------------------------------------------------------

    return {

        "evidence_id":
            evidence_id,

        "type":
            source_type,

        "relationship":
            relationship,

        "identifier":
            identifier,

        "title":
            title,

        "status":
            status,

        "technical_department":
            technical_department,

        "technical_committee":
            technical_committee,

        "aspect":
            aspect,

        "degree_of_equivalence":
            degree_of_equivalence,

        "number_of_revisions":
            number_of_revisions,

        "number_of_amendments":
            number_of_amendments,

        "reaffirmation_year":
            reaffirmation_year,

        "source_url":
            source_url,

        "download_url":
            download_url,

        "composition_url":
            composition_url
    }


# ============================================================
# 5. BUILD ALL PRODUCTION SOURCES
# ============================================================

def build_production_sources_v2(
    retrieval_result
):

    evidence = retrieval_result.get(
        "evidence",
        []
    )


    sources = []


    for index, item in enumerate(
        evidence,
        start=1
    ):

        sources.append(

            build_clean_source_v2(

                item,

                index
            )

        )


    return sources


# ============================================================
# 6. EXTRACT CITATIONS
# ============================================================

def extract_answer_citations_v2(
    answer
):

    if not answer:
        return []


    matches = re.findall(
        r"\[E(\d+)\]",
        answer
    )


    return list(
        dict.fromkeys(

            f"E{number}"

            for number in matches

        )
    )


# ============================================================
# 7. BUILD ONLY CITED SOURCES
# ============================================================

def build_cited_sources_v2(
    answer,
    sources
):

    citation_ids = set(

        extract_answer_citations_v2(
            answer
        )

    )


    return [

        source

        for source in sources

        if source[
            "evidence_id"
        ] in citation_ids

    ]


# ============================================================
# 8. PATCHED PRODUCTION RESPONSE
# ============================================================

def build_production_response_v2(

    query,

    retrieval_result,

    llm_result

):

    answer = llm_result.get(
        "answer",
        ""
    )


    all_sources = (
        build_production_sources_v2(
            retrieval_result
        )
    )


    citations = (
        extract_answer_citations_v2(
            answer
        )
    )


    cited_sources = (
        build_cited_sources_v2(

            answer,

            all_sources

        )
    )


    relationship = (
        retrieval_result.get(
            "relationship_direction"
        )
    )


    compound = retrieval_result.get(
        "compound",
        False
    )


    identifiers = retrieval_result.get(
        "identifiers",
        []
    )


    # --------------------------------------------------------
    # Query classification
    # --------------------------------------------------------

    if compound:

        query_type = "compound"

    elif relationship == "reverse":

        query_type = "reverse_reference"

    elif relationship == "outgoing":

        query_type = "outgoing_reference"

    elif identifiers:

        query_type = "exact_standard"

    else:

        query_type = "semantic"


    # --------------------------------------------------------
    # Status
    # --------------------------------------------------------

    if llm_result.get(
        "generation_error"
    ):

        status = "generation_error"

    elif not retrieval_result.get(
        "evidence"
    ):

        status = "no_evidence"

    else:

        status = "success"


    return {

        "query":
            query,

        "answer":
            answer,

        "query_type":
            query_type,

        "identifiers":
            identifiers,

        "relationship_direction":
            relationship,

        "compound":
            compound,

        "retrieval_plan":
            retrieval_result.get(
                "retrieval_plan",
                []
            ),

        "evidence_count":
            retrieval_result.get(
                "evidence_count",
                0
            ),

        "citations":
            citations,

        "sources":
            cited_sources,

        "total_sources_available":
            len(all_sources),

        "model":
            llm_result.get(
                "model",
                MODEL_NAME
            ),

        "generation_attempts":
            llm_result.get(
                "attempts",
                1
            ),

        "status":
            status,

        "timestamp":
            datetime.now(
                timezone.utc
            ).isoformat()

    }


# ============================================================
# 9. PATCHED INFERENCE FUNCTION
# ============================================================

def saathi_production_inference_v2(
    query,
    max_relationships=50
):

    if not isinstance(
        query,
        str
    ):

        raise TypeError(
            "query must be a string"
        )


    query = query.strip()


    if not query:

        raise ValueError(
            "query cannot be empty"
        )


    # --------------------------------------------------------
    # Existing retrieval layer
    # --------------------------------------------------------

    retrieval_result = retrieve_evidence(

        query,

        semantic_fallback=True,

        max_relationships=
            max_relationships

    )


    # --------------------------------------------------------
    # Existing grounded Gemini layer
    # --------------------------------------------------------

    llm_result = generate_grounded_answer(

        query,

        retrieval_result

    )


    # --------------------------------------------------------
    # New production source layer
    # --------------------------------------------------------

    return build_production_response_v2(

        query,

        retrieval_result,

        llm_result

    )


# ============================================================
# 10. TEST EXACT STANDARD
# ============================================================

print("\n")
print("=" * 70)
print("TEST — EXACT STANDARD")
print("=" * 70)


test_response = (
    saathi_production_inference_v2(
        "What is IS/IEC 60691:2023?"
    )
)


print("\nANSWER")
print("-" * 70)

print(
    test_response["answer"]
)


print("\nPRODUCTION SOURCES")
print("-" * 70)


for source in test_response[
    "sources"
]:

    print(
        json.dumps(
            source,
            indent=2,
            ensure_ascii=False
        )
    )


# ============================================================
# 11. VERIFY REQUIRED METADATA
# ============================================================

print("\n")
print("=" * 70)
print("METADATA CHECK")
print("=" * 70)


if test_response[
    "sources"
]:

    source = test_response[
        "sources"
    ][0]


    required_fields = [

        "evidence_id",
        "identifier",
        "title",
        "status",
        "technical_department",
        "technical_committee",
        "aspect",
        "degree_of_equivalence",
        "source_url"

    ]


    print(
        "\nField availability:"
    )


    for field in required_fields:

        value = source.get(
            field
        )


        print(

            f"  {field:<28} "
            f"{'OK' if value is not None else 'MISSING'}"

        )


    metadata_present = all(

        source.get(field)
        is not None

        for field in required_fields

    )


else:

    metadata_present = False


print("\n")


if metadata_present:

    print(
        "✅ PRODUCTION SOURCE METADATA CHECK PASSED"
    )

else:

    print(
        "⚠️ SOME SOURCE METADATA IS STILL MISSING"
    )


# ============================================================
# 12. SAVE PATCH TEST
# ============================================================

patch_test_path = (
    "/kaggle/working/"
    "saathi_cell14_1_patch_test.json"
)


with open(
    patch_test_path,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        test_response,
        f,
        indent=2,
        ensure_ascii=False
    )


print(
    f"\nSaved:\n{patch_test_path}"
)


print("\n")
print("=" * 70)
print("CELL 14.1 COMPLETE")
print("=" * 70)
```

```python
# ============================================================
# SAATHI — CELL 14.2
# RAW EVIDENCE STRUCTURE INSPECTOR
# ============================================================

import json


print("=" * 70)
print("SAATHI — CELL 14.2")
print("RAW EVIDENCE STRUCTURE INSPECTOR")
print("=" * 70)


# ============================================================
# 1. RETRIEVE WITHOUT GEMINI
# ============================================================

test_query = "What is IS/IEC 60691:2023?"


print("\nRunning retrieval only...")
print("(Gemini will NOT be called.)")


retrieval_debug = retrieve_evidence(

    test_query,

    semantic_fallback=True

)


print(
    "\nEvidence count:",
    retrieval_debug.get(
        "evidence_count",
        len(
            retrieval_debug.get(
                "evidence",
                []
            )
        )
    )
)


# ============================================================
# 2. GET FIRST EVIDENCE ITEM
# ============================================================

debug_evidence = retrieval_debug.get(
    "evidence",
    []
)


if not debug_evidence:

    print(
        "\n❌ No evidence returned."
    )

else:

    first_evidence = debug_evidence[0]


    print("\n")
    print("=" * 70)
    print("FIRST EVIDENCE ITEM — COMPLETE RAW OBJECT")
    print("=" * 70)


    print(
        json.dumps(
            first_evidence,
            indent=2,
            ensure_ascii=False,
            default=str
        )
    )


# ============================================================
# 3. PRINT TOP-LEVEL KEYS
# ============================================================

if debug_evidence:

    print("\n")
    print("=" * 70)
    print("TOP-LEVEL EVIDENCE KEYS")
    print("=" * 70)


    print(
        list(
            first_evidence.keys()
        )
    )


# ============================================================
# 4. INSPECT CONTENT
# ============================================================

if debug_evidence:

    content = first_evidence.get(
        "content"
    )


    print("\n")
    print("=" * 70)
    print("CONTENT OBJECT")
    print("=" * 70)


    print(
        json.dumps(
            content,
            indent=2,
            ensure_ascii=False,
            default=str
        )
    )


# ============================================================
# 5. RECURSIVE KEY LOCATOR
# ============================================================

TARGET_KEYS = {

    "is_number",
    "IS_number",
    "is_title",
    "IS_title",
    "status",
    "catalogue_withdrawn_status",
    "withdrawn_status",
    "technical_department",
    "technical_committee",
    "catalogue_technical_committee",
    "aspect",
    "catalogue_aspect",
    "degree_of_equivalence",
    "number_of_revisions",
    "number_of_amendments",
    "amendment_count",
    "detail_url",
    "source_url",
    "download_url",
    "composition_url"

}


def locate_keys(
    obj,
    path="root"
):

    found = []


    if isinstance(
        obj,
        dict
    ):

        for key, value in obj.items():

            current_path = (
                f"{path}.{key}"
            )


            if key in TARGET_KEYS:

                found.append({

                    "path":
                        current_path,

                    "key":
                        key,

                    "value":
                        value

                })


            found.extend(

                locate_keys(

                    value,

                    current_path

                )

            )


    elif isinstance(
        obj,
        list
    ):

        for index, item in enumerate(
            obj
        ):

            current_path = (
                f"{path}[{index}]"
            )


            found.extend(

                locate_keys(

                    item,

                    current_path

                )

            )


    return found


# ============================================================
# 6. SHOW WHERE THE METADATA ACTUALLY LIVES
# ============================================================

if debug_evidence:

    print("\n")
    print("=" * 70)
    print("METADATA LOCATIONS")
    print("=" * 70)


    locations = locate_keys(
        first_evidence
    )


    if locations:

        for item in locations:

            print(
                f"\n{item['path']}"
            )

            print(
                f"  {item['key']} = "
                f"{repr(item['value'])}"
            )

    else:

        print(
            "No target metadata keys were found."
        )


# ============================================================
# 7. RETRIEVAL RESULT STRUCTURE
# ============================================================

print("\n")
print("=" * 70)
print("RETRIEVAL RESULT TOP-LEVEL KEYS")
print("=" * 70)


print(
    list(
        retrieval_debug.keys()
    )
)


# ============================================================
# 8. SAVE DEBUG OBJECT
# ============================================================

debug_path = (
    "/kaggle/working/"
    "saathi_cell14_2_raw_evidence.json"
)


with open(
    debug_path,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        retrieval_debug,
        f,
        indent=2,
        ensure_ascii=False,
        default=str
    )


print("\n")
print("=" * 70)
print("DEBUG FILE SAVED")
print("=" * 70)

print(
    debug_path
)


print("\n")
print("=" * 70)
print("CELL 14.2 COMPLETE")
print("=" * 70)
```

```python
# ============================================================
# SAATHI — CELL 14.3
# PRODUCTION SOURCE BUILDER — TEXT METADATA PARSER
# ============================================================

import json
import re
from datetime import datetime, timezone


print("=" * 70)
print("SAATHI — CELL 14.3")
print("PRODUCTION SOURCE BUILDER — TEXT METADATA PARSER")
print("=" * 70)


# ============================================================
# 1. PARSE KEY/VALUE FIELDS FROM BIS EVIDENCE TEXT
# ============================================================

def parse_bis_evidence_text(text):

    if not isinstance(text, str):
        return {}


    fields = {}


    # --------------------------------------------------------
    # Known BIS fields
    # --------------------------------------------------------

    field_patterns = {

        "technical_department":
            r"Technical Department:\s*(.+)",

        "technical_committee":
            r"Technical Committee:\s*(.+)",

        "aspect":
            r"Aspect:\s*(.+)",

        "language":
            r"Language:\s*(.+)",

        "degree_of_equivalence":
            r"Degree of Equivalence:\s*(.+)",

        "number_of_revisions":
            r"Number of Revisions:\s*(.+)",

        "number_of_amendments":
            r"Number of Amendments:\s*(.+)",

        "amendment_count":
            r"Amendment Count:\s*(.+)",

        "gazette_document_count":
            r"Gazette Document Count:\s*(.+)",

        "license_count":
            r"License Count:\s*(.+)",

        "product_manual_sit_count":
            r"Product Manual SIT Count:\s*(.+)",

        "laboratory_count":
            r"Laboratory Count:\s*(.+)",

        "corrigendum_count":
            r"Corrigendum Count:\s*(.+)",

        "catalogue_id":
            r"Catalogue ID:\s*(.+)",

        "catalogue_is_number":
            r"Catalogue IS Number:\s*(.+)",

        "catalogue_title":
            r"Catalogue Title:\s*(.+)",

        "catalogue_amendments":
            r"Catalogue Amendments:\s*(.+)",

        "catalogue_technical_committee":
            r"Catalogue Technical Committee:\s*(.+)",

        "catalogue_aspect":
            r"Catalogue Aspect:\s*(.+)",

        "catalogue_reaffirmation":
            r"Catalogue Reaffirmation:\s*(.+)",

        "detail_url":
            r"Detail URL:\s*(\S+)",

        "download_url":
            r"Download URL:\s*(\S+)",

        "composition_url":
            r"Composition URL:\s*(\S+)"
    }


    # --------------------------------------------------------
    # Extract line-based fields
    # --------------------------------------------------------

    for key, pattern in field_patterns.items():

        match = re.search(
            pattern,
            text
        )

        if match:

            value = match.group(
                1
            ).strip()


            if value:

                fields[key] = value


    # --------------------------------------------------------
    # Extract status from catalogue IS number
    # --------------------------------------------------------

    catalogue_is_number = fields.get(
        "catalogue_is_number"
    )


    if catalogue_is_number:

        status_match = re.search(

            r"\((Active|Withdrawn|Superseded|Cancelled|Obsolete)\)",

            catalogue_is_number,

            flags=re.IGNORECASE

        )


        if status_match:

            fields["status"] = (
                status_match.group(1).capitalize()
            )


    # --------------------------------------------------------
    # Fallback: authored detail status
    # --------------------------------------------------------

    if "status" not in fields:

        status_match = re.search(

            r"status is:\s*([^.]+)",

            text,

            flags=re.IGNORECASE

        )


        if status_match:

            fields["status"] = (
                status_match.group(1).strip()
            )


    return fields


# ============================================================
# 2. BUILD ONE PRODUCTION SOURCE
# ============================================================

def build_clean_source_v3(
    evidence_item,
    evidence_number
):

    content = evidence_item.get(
        "content",
        {}
    )


    if not isinstance(
        content,
        dict
    ):

        content = {}


    text = content.get(
        "text",
        ""
    )


    parsed = parse_bis_evidence_text(
        text
    )


    # --------------------------------------------------------
    # Identifier
    # --------------------------------------------------------

    identifier = (

        evidence_item.get(
            "identifier"
        )

        or

        content.get(
            "is_number"
        )

        or

        parsed.get(
            "catalogue_is_number"
        )

    )


    # Normalize only the malformed display prefix.
    # We DO NOT change the underlying dataset.

    if isinstance(
        identifier,
        str
    ):

        identifier = re.sub(

            r"^IS/IEC\s*:\s*\d{4}\s*",

            "IS/IEC ",

            identifier

        )


        identifier = re.sub(
            r"\s+",
            " ",
            identifier
        ).strip()


    # --------------------------------------------------------
    # Title
    # --------------------------------------------------------

    title = (

        evidence_item.get(
            "title"
        )

        or

        content.get(
            "is_title"
        )

        or

        parsed.get(
            "catalogue_title"
        )

    )


    # --------------------------------------------------------
    # Source type
    # --------------------------------------------------------

    source_type = (

        evidence_item.get(
            "evidence_type"
        )

        or

        evidence_item.get(
            "type"
        )

        or

        "unknown"

    )


    # --------------------------------------------------------
    # Return production source
    # --------------------------------------------------------

    return {

        "evidence_id":
            f"E{evidence_number}",

        "type":
            source_type,

        "relationship":
            evidence_item.get(
                "relationship"
            ),

        "identifier":
            identifier,

        "title":
            title,

        "status":
            parsed.get(
                "status"
            ),

        "technical_department":
            parsed.get(
                "technical_department"
            ),

        "technical_committee":
            parsed.get(
                "technical_committee"
            ),

        "aspect":
            parsed.get(
                "aspect"
            ),

        "language":
            parsed.get(
                "language"
            ),

        "degree_of_equivalence":
            parsed.get(
                "degree_of_equivalence"
            ),

        "number_of_revisions":
            parsed.get(
                "number_of_revisions"
            ),

        "number_of_amendments":
            parsed.get(
                "number_of_amendments"
            ),

        "amendment_count":
            parsed.get(
                "amendment_count"
            ),

        "reaffirmation":
            parsed.get(
                "catalogue_reaffirmation"
            ),

        "source_url":
            parsed.get(
                "detail_url"
            ),

        "download_url":
            parsed.get(
                "download_url"
            ),

        "composition_url":
            parsed.get(
                "composition_url"
            )
    }


# ============================================================
# 3. BUILD SOURCES FROM RETRIEVAL
# ============================================================

def build_production_sources_v3(
    retrieval_result
):

    evidence = retrieval_result.get(
        "evidence",
        []
    )


    return [

        build_clean_source_v3(
            item,
            index
        )

        for index, item in enumerate(
            evidence,
            start=1
        )

    ]


# ============================================================
# 4. TEST WITHOUT GEMINI
# ============================================================

test_query = (
    "What is IS/IEC 60691:2023?"
)


print("\nRunning retrieval only...")
print("Gemini will NOT be called.")


retrieval_test = retrieve_evidence(
    test_query,
    semantic_fallback=True
)


sources_test = (
    build_production_sources_v3(
        retrieval_test
    )
)


print("\n")
print("=" * 70)
print("PRODUCTION SOURCE")
print("=" * 70)


if sources_test:

    print(

        json.dumps(
            sources_test[0],
            indent=2,
            ensure_ascii=False
        )

    )

else:

    print(
        "❌ No source generated."
    )


# ============================================================
# 5. REQUIRED FIELD VALIDATION
# ============================================================

print("\n")
print("=" * 70)
print("SOURCE METADATA VALIDATION")
print("=" * 70)


if sources_test:

    source = sources_test[0]


    required_values = {

        "evidence_id":
            source.get(
                "evidence_id"
            ),

        "identifier":
            source.get(
                "identifier"
            ),

        "title":
            source.get(
                "title"
            ),

        "status":
            source.get(
                "status"
            ),

        "technical_department":
            source.get(
                "technical_department"
            ),

        "technical_committee":
            source.get(
                "technical_committee"
            ),

        "aspect":
            source.get(
                "aspect"
            ),

        "degree_of_equivalence":
            source.get(
                "degree_of_equivalence"
            ),

        "source_url":
            source.get(
                "source_url"
            ),

        "download_url":
            source.get(
                "download_url"
            ),

        "composition_url":
            source.get(
                "composition_url"
            )
    }


    all_present = True


    for field, value in (
        required_values.items()
    ):

        present = (
            value is not None
            and str(value).strip() != ""
        )


        print(

            f"  {field:<28}"
            f"{'✅' if present else '❌'}"

        )


        if not present:

            all_present = False


else:

    all_present = False


# ============================================================
# 6. SAVE TEST
# ============================================================

output_path = (
    "/kaggle/working/"
    "saathi_cell14_3_source_test.json"
)


with open(
    output_path,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        sources_test,
        f,
        indent=2,
        ensure_ascii=False
    )


print("\n")


if all_present:

    print(
        "✅ ALL PRODUCTION SOURCE METADATA FOUND"
    )

else:

    print(
        "⚠️ SOME PRODUCTION SOURCE FIELDS ARE MISSING"
    )


print(
    f"\nSaved:\n{output_path}"
)


print("\n")
print("=" * 70)
print("CELL 14.3 COMPLETE")
print("=" * 70)
```

```python
# ============================================================
# SAATHI — CELL 15
# PRODUCTION API CONTRACT
# ============================================================

import json
from datetime import datetime, timezone


print("=" * 70)
print("SAATHI — CELL 15")
print("PRODUCTION API CONTRACT")
print("=" * 70)


# ============================================================
# 1. BUILD FINAL API RESPONSE
# ============================================================

def build_api_response_v2(
    query,
    retrieval_result,
    llm_result
):

    # --------------------------------------------------------
    # Generate clean production sources
    # --------------------------------------------------------

    all_sources = build_production_sources_v3(
        retrieval_result
    )


    # --------------------------------------------------------
    # Answer
    # --------------------------------------------------------

    answer = llm_result.get(
        "answer",
        ""
    )


    # --------------------------------------------------------
    # Citations actually used by the answer
    # --------------------------------------------------------

    citations = extract_answer_citations_v2(
        answer
    )


    citation_set = set(
        citations
    )


    cited_sources = [

        source

        for source in all_sources

        if source[
            "evidence_id"
        ] in citation_set

    ]


    # --------------------------------------------------------
    # Query metadata
    # --------------------------------------------------------

    identifiers = retrieval_result.get(
        "identifiers",
        []
    )


    relationship_direction = (
        retrieval_result.get(
            "relationship_direction"
        )
    )


    compound = retrieval_result.get(
        "compound",
        False
    )


    # --------------------------------------------------------
    # Query type
    # --------------------------------------------------------

    if compound:

        query_type = "compound"

    elif relationship_direction == "reverse":

        query_type = "reverse_reference"

    elif relationship_direction == "outgoing":

        query_type = "outgoing_reference"

    elif identifiers:

        query_type = "exact_standard"

    else:

        query_type = "semantic"


    # --------------------------------------------------------
    # Response status
    # --------------------------------------------------------

    generation_error = (
        llm_result.get(
            "generation_error"
        )
    )


    if generation_error:

        status = "generation_error"

    elif not retrieval_result.get(
        "evidence"
    ):

        status = "no_evidence"

    else:

        status = "success"


    # --------------------------------------------------------
    # Final API object
    # --------------------------------------------------------

    return {

        # Request
        "query":
            query,

        # Answer
        "answer":
            answer,

        # Classification
        "query_type":
            query_type,

        "identifiers":
            identifiers,

        "relationship_direction":
            relationship_direction,

        "compound":
            compound,

        # Retrieval
        "retrieval": {

            "plan":
                retrieval_result.get(
                    "retrieval_plan",
                    []
                ),

            "evidence_count":
                retrieval_result.get(
                    "evidence_count",
                    0
                ),

            "total_sources":
                len(all_sources)

        },

        # Citations
        "citations":
            citations,

        "sources":
            cited_sources,

        # Generation
        "generation": {

            "model":
                llm_result.get(
                    "model",
                    MODEL_NAME
                ),

            "attempts":
                llm_result.get(
                    "attempts",
                    1
                )

        },

        # API status
        "status":
            status,

        # Timestamp
        "timestamp":
            datetime.now(
                timezone.utc
            ).isoformat()

    }


# ============================================================
# 2. SINGLE ENTRY-POINT INFERENCE
# ============================================================

def saathi_api_inference(
    query,
    max_relationships=50
):

    if not isinstance(
        query,
        str
    ):

        raise TypeError(
            "query must be a string"
        )


    query = query.strip()


    if not query:

        raise ValueError(
            "query cannot be empty"
        )


    # --------------------------------------------------------
    # Retrieval
    # --------------------------------------------------------

    retrieval_result = retrieve_evidence(

        query,

        semantic_fallback=True,

        max_relationships=
            max_relationships

    )


    # --------------------------------------------------------
    # Grounded LLM
    # --------------------------------------------------------

    llm_result = generate_grounded_answer(

        query,

        retrieval_result

    )


    # --------------------------------------------------------
    # API contract
    # --------------------------------------------------------

    return build_api_response_v2(

        query,

        retrieval_result,

        llm_result

    )


# ============================================================
# 3. API CONTRACT VALIDATOR
# ============================================================

def validate_api_response(
    response
):

    required_top_level = [

        "query",
        "answer",
        "query_type",
        "identifiers",
        "relationship_direction",
        "compound",
        "retrieval",
        "citations",
        "sources",
        "generation",
        "status",
        "timestamp"

    ]


    required_retrieval = [

        "plan",
        "evidence_count",
        "total_sources"

    ]


    required_generation = [

        "model",
        "attempts"

    ]


    errors = []


    # --------------------------------------------------------
    # Top-level
    # --------------------------------------------------------

    for key in required_top_level:

        if key not in response:

            errors.append(
                f"Missing top-level field: {key}"
            )


    # --------------------------------------------------------
    # Retrieval
    # --------------------------------------------------------

    retrieval = response.get(
        "retrieval",
        {}
    )


    for key in required_retrieval:

        if key not in retrieval:

            errors.append(
                f"Missing retrieval field: {key}"
            )


    # --------------------------------------------------------
    # Generation
    # --------------------------------------------------------

    generation = response.get(
        "generation",
        {}
    )


    for key in required_generation:

        if key not in generation:

            errors.append(
                f"Missing generation field: {key}"
            )


    # --------------------------------------------------------
    # Source validation
    # --------------------------------------------------------

    for source in response.get(
        "sources",
        []
    ):

        required_source_fields = [

            "evidence_id",
            "type",
            "identifier",
            "title",
            "status",
            "source_url"

        ]


        for key in required_source_fields:

            if key not in source:

                errors.append(

                    "Source "
                    f"{source.get('evidence_id', '?')} "
                    f"missing field: {key}"

                )


    return {

        "valid":
            len(errors) == 0,

        "errors":
            errors

    }


# ============================================================
# 4. TEST — API CONTRACT
# ============================================================

print("\n")
print("=" * 70)
print("API CONTRACT TEST")
print("=" * 70)


print(
    "\nRunning exact-standard test..."
)


# IMPORTANT:
# This is the only part of Cell 15 that calls Gemini.

api_test = saathi_api_inference(

    "What is IS/IEC 60691:2023?"

)


# ============================================================
# 5. VALIDATE
# ============================================================

validation = validate_api_response(
    api_test
)


print("\n")
print("=" * 70)
print("CONTRACT VALIDATION")
print("=" * 70)


if validation["valid"]:

    print(
        "✅ API CONTRACT VALID"
    )

else:

    print(
        "❌ API CONTRACT INVALID"
    )


    for error in validation[
        "errors"
    ]:

        print(
            "  -",
            error
        )


# ============================================================
# 6. DISPLAY FINAL RESPONSE
# ============================================================

print("\n")
print("=" * 70)
print("FINAL API RESPONSE")
print("=" * 70)


print(

    json.dumps(

        api_test,

        indent=2,

        ensure_ascii=False

    )

)


# ============================================================
# 7. SAVE API CONTRACT TEST
# ============================================================

api_test_path = (
    "/kaggle/working/"
    "saathi_cell15_api_contract_test.json"
)


with open(
    api_test_path,
    "w",
    encoding="utf-8"
) as f:

    json.dump(

        api_test,

        f,

        indent=2,

        ensure_ascii=False

    )


print("\n")
print("=" * 70)
print("CELL 15 COMPLETE")
print("=" * 70)


print(
    f"\nSaved:\n{api_test_path}"
)
```

```python
# ============================================================
# SAATHI — CELL 16
# BACKEND PACKAGING PREPARATION
# ============================================================

import os
import json
import shutil
from pathlib import Path


print("=" * 70)
print("SAATHI — CELL 16")
print("BACKEND PACKAGING PREPARATION")
print("=" * 70)


# ============================================================
# 1. DEFINE REQUIRED ARTIFACTS
# ============================================================

WORK_DIR = Path(
    "/kaggle/working"
)


BACKEND_DIR = (
    WORK_DIR /
    "saathi_backend"
)


BACKEND_DIR.mkdir(
    parents=True,
    exist_ok=True
)


required_artifacts = {

    "faiss_index":
        WORK_DIR /
        "bis_faiss.index",

    "embeddings":
        WORK_DIR /
        "bis_embeddings.npy",

    "metadata":
        WORK_DIR /
        "bis_rag_metadata.json",

    "corpus":
        WORK_DIR /
        "bis_rag_corpus.jsonl"

}


# ============================================================
# 2. VERIFY ARTIFACTS
# ============================================================

print("\n")
print("=" * 70)
print("ARTIFACT CHECK")
print("=" * 70)


artifact_status = {}


for name, path in required_artifacts.items():

    exists = path.exists()


    artifact_status[name] = {

        "path":
            str(path),

        "exists":
            exists,

        "size_mb":
            round(
                path.stat().st_size /
                (1024 * 1024),
                2
            )
            if exists
            else None

    }


    print(

        f"{name:<15}"
        f"{'✅' if exists else '❌'}"
        f"  {path}"

    )


missing = [

    name

    for name, info
    in artifact_status.items()

    if not info["exists"]

]


if missing:

    raise FileNotFoundError(

        "Missing required artifacts: "
        + ", ".join(missing)

    )


# ============================================================
# 3. COPY ARTIFACTS
# ============================================================

print("\n")
print("=" * 70)
print("COPYING BACKEND ARTIFACTS")
print("=" * 70)


for name, source_path in required_artifacts.items():

    destination = (
        BACKEND_DIR /
        source_path.name
    )


    shutil.copy2(

        source_path,

        destination

    )


    print(

        f"✅ {source_path.name}"

    )


# ============================================================
# 4. CREATE BACKEND CONFIG
# ============================================================

config = {

    "project":
        "SAATHI — BIS Navigator",

    "version":
        "1.0.0",

    "embedding_model":
        "BAAI/bge-small-en-v1.5",

    "embedding_dimension":
        384,

    "llm_model":
        "gemini-2.5-flash",

    "faiss_index":
        "bis_faiss.index",

    "embeddings":
        "bis_embeddings.npy",

    "metadata":
        "bis_rag_metadata.json",

    "corpus":
        "bis_rag_corpus.jsonl",

    "api_endpoint":
        "/api/ask",

    "retrieval_modes": [

        "exact_standard",
        "semantic",
        "outgoing_reference",
        "reverse_reference",
        "compound"

    ],

    "citation_format":
        "[E1]",

    "created_from":
        "validated Kaggle inference pipeline"

}


config_path = (
    BACKEND_DIR /
    "config.json"
)


with open(
    config_path,
    "w",
    encoding="utf-8"
) as f:

    json.dump(

        config,

        f,

        indent=2,

        ensure_ascii=False

    )


print(
    f"✅ {config_path.name}"
)


# ============================================================
# 5. CREATE REQUIREMENTS FILE
# ============================================================

requirements = """\
fastapi
uvicorn[standard]
numpy
pandas
faiss-cpu
sentence-transformers
google-genai
pydantic
"""

requirements_path = (
    BACKEND_DIR /
    "requirements.txt"
)


requirements_path.write_text(
    requirements,
    encoding="utf-8"
)


print(
    "✅ requirements.txt"
)


# ============================================================
# 6. CREATE .ENV TEMPLATE
# ============================================================

env_template = """\
# Gemini API key
GEMINI_API_KEY=your_gemini_api_key_here

# Server
HOST=0.0.0.0
PORT=8000
"""

env_path = (
    BACKEND_DIR /
    ".env.example"
)


env_path.write_text(
    env_template,
    encoding="utf-8"
)


print(
    "✅ .env.example"
)


# ============================================================
# 7. CREATE API SCHEMA DOCUMENT
# ============================================================

api_schema = {

    "request": {

        "method":
            "POST",

        "path":
            "/api/ask",

        "body": {

            "query":
                "string",

            "max_relationships":
                "integer, optional"

        }

    },

    "response": {

        "query":
            "string",

        "answer":
            "string",

        "query_type":
            "exact_standard | semantic | "
            "outgoing_reference | reverse_reference | compound",

        "identifiers":
            "array",

        "relationship_direction":
            "string | null",

        "compound":
            "boolean",

        "retrieval": {

            "plan":
                "array",

            "evidence_count":
                "integer",

            "total_sources":
                "integer"

        },

        "citations":
            "array",

        "sources":
            "array",

        "generation": {

            "model":
                "string",

            "attempts":
                "integer"

        },

        "status":
            "success | no_evidence | generation_error",

        "timestamp":
            "ISO-8601 string"

    }

}


schema_path = (
    BACKEND_DIR /
    "api_schema.json"
)


with open(
    schema_path,
    "w",
    encoding="utf-8"
) as f:

    json.dump(

        api_schema,

        f,

        indent=2,

        ensure_ascii=False

    )


print(
    "✅ api_schema.json"
)


# ============================================================
# 8. CREATE BACKEND DIRECTORY MANIFEST
# ============================================================

manifest = {

    "backend_directory":
        str(BACKEND_DIR),

    "files": {},

    "total_size_mb":
        0

}


for path in sorted(
    BACKEND_DIR.iterdir()
):

    if path.is_file():

        size_mb = round(

            path.stat().st_size /
            (1024 * 1024),

            2

        )


        manifest[
            "files"
        ][
            path.name
        ] = {

            "size_mb":
                size_mb

        }


        manifest[
            "total_size_mb"
        ] += size_mb


manifest[
    "total_size_mb"
] = round(

    manifest[
        "total_size_mb"
    ],

    2

)


manifest_path = (
    BACKEND_DIR /
    "manifest.json"
)


with open(
    manifest_path,
    "w",
    encoding="utf-8"
) as f:

    json.dump(

        manifest,

        f,

        indent=2,

        ensure_ascii=False

    )


# ============================================================
# 9. FINAL CHECK
# ============================================================

print("\n")
print("=" * 70)
print("BACKEND PACKAGE")
print("=" * 70)


for filename, info in manifest[
    "files"
].items():

    print(

        f"{filename:<30}"
        f"{info['size_mb']:>10.2f} MB"

    )


print(
    "\nTotal package size:",
    manifest[
        "total_size_mb"
    ],
    "MB"
)


print("\n")
print("=" * 70)
print("CELL 16 COMPLETE")
print("=" * 70)


print(
    f"\nBackend package:\n{BACKEND_DIR}"
)
```

```python
# ============================================================
# SAATHI — CELL 17
# FASTAPI SERVICE GENERATOR
# ============================================================

from pathlib import Path
import textwrap


print("=" * 70)
print("SAATHI — CELL 17")
print("FASTAPI SERVICE GENERATOR")
print("=" * 70)


BACKEND_DIR = Path(
    "/kaggle/working/saathi_backend"
)


APP_PATH = (
    BACKEND_DIR /
    "app.py"
)


# ============================================================
# 1. FASTAPI APPLICATION
# ============================================================

app_code = r'''
import os
import json
import re
import time
from pathlib import Path
from datetime import datetime, timezone

import numpy as np
import faiss

from fastapi import FastAPI, HTTPException
from fastapi.middleware.cors import CORSMiddleware
from pydantic import BaseModel, Field

from sentence_transformers import SentenceTransformer
from google import genai
from google.genai import types


# ============================================================
# CONFIGURATION
# ============================================================

BASE_DIR = Path(__file__).resolve().parent

FAISS_INDEX_PATH = (
    BASE_DIR / "bis_faiss.index"
)

METADATA_PATH = (
    BASE_DIR / "bis_rag_metadata.json"
)

CORPUS_PATH = (
    BASE_DIR / "bis_rag_corpus.jsonl"
)

CONFIG_PATH = (
    BASE_DIR / "config.json"
)


EMBEDDING_MODEL = (
    "BAAI/bge-small-en-v1.5"
)

GEMINI_MODEL = (
    "gemini-2.5-flash"
)


# ============================================================
# FASTAPI APP
# ============================================================

app = FastAPI(

    title="SAATHI — BIS Navigator API",

    description=(
        "Grounded BIS Standards retrieval and "
        "question-answering API."
    ),

    version="1.0.0"

)


# ============================================================
# CORS
# ============================================================

app.add_middleware(

    CORSMiddleware,

    allow_origins=["*"],

    allow_credentials=True,

    allow_methods=["*"],

    allow_headers=["*"]

)


# ============================================================
# REQUEST MODEL
# ============================================================

class AskRequest(BaseModel):

    query: str = Field(
        ...,
        min_length=1,
        description="Natural-language BIS query"
    )

    max_relationships: int = Field(
        default=50,
        ge=1,
        le=200
    )


# ============================================================
# GLOBAL ARTIFACTS
# ============================================================

faiss_index = None

metadata = None

corpus = None

embedding_model = None

gemini_client = None

startup_error = None


# ============================================================
# LOAD ARTIFACTS
# ============================================================

def load_artifacts():

    global faiss_index
    global metadata
    global corpus
    global embedding_model
    global gemini_client
    global startup_error


    try:

        # ----------------------------------------------------
        # FAISS
        # ----------------------------------------------------

        faiss_index = faiss.read_index(

            str(
                FAISS_INDEX_PATH
            )

        )


        # ----------------------------------------------------
        # Metadata
        # ----------------------------------------------------

        with open(

            METADATA_PATH,

            "r",

            encoding="utf-8"

        ) as f:

            metadata = json.load(f)


        # ----------------------------------------------------
        # Corpus metadata/documents
        # ----------------------------------------------------

        corpus = []

        with open(

            CORPUS_PATH,

            "r",

            encoding="utf-8"

        ) as f:

            for line in f:

                line = line.strip()

                if line:

                    corpus.append(
                        json.loads(line)
                    )


        # ----------------------------------------------------
        # Embedding model
        # ----------------------------------------------------

        embedding_model = (
            SentenceTransformer(
                EMBEDDING_MODEL
            )
        )


        # ----------------------------------------------------
        # Gemini
        # ----------------------------------------------------

        api_key = os.getenv(
            "GEMINI_API_KEY"
        )


        if api_key:

            gemini_client = (
                genai.Client(
                    api_key=api_key
                )
            )

        else:

            gemini_client = None


        startup_error = None


    except Exception as exc:

        startup_error = str(
            exc
        )

        raise


# ============================================================
# STARTUP
# ============================================================

@app.on_event("startup")
def startup_event():

    load_artifacts()


# ============================================================
# IDENTIFIER NORMALIZATION
# ============================================================

def normalize_identifier(value):

    if not value:

        return ""


    value = str(
        value
    ).strip()


    value = re.sub(
        r"\s+",
        " ",
        value
    )


    # Fix common malformed BIS IEC identifiers.
    value = re.sub(

        r"^IS/IEC\s*:\s*(\d{4})\s*IEC\s+",

        r"IS/IEC ",

        value,

        flags=re.IGNORECASE

    )


    value = re.sub(

        r"^IS/IEC\s*:\s*(\d{4})\s*",

        r"IS/IEC ",

        value,

        flags=re.IGNORECASE

    )


    value = re.sub(

        r"\s*:\s*",

        ":",

        value

    )


    return value.strip()


# ============================================================
# EXACT IDENTIFIER EXTRACTION
# ============================================================

def extract_identifiers(query):

    identifiers = []


    patterns = [

        r'\bIS/IEC\s*[\w-]+\s*:\s*\d{4}\b',

        r'\bIS\s*[\w-]+(?:\s*\([^)]*\))?\s*:\s*\d{4}\b',

        r'\bIEC\s+\d+(?:-\d+)?\s*:\s*\d{4}\b'

    ]


    for pattern in patterns:

        matches = re.findall(

            pattern,

            query,

            flags=re.IGNORECASE

        )


        for match in matches:

            normalized = normalize_identifier(
                match
            )


            if normalized and normalized not in identifiers:

                identifiers.append(
                    normalized
                )


    return identifiers


# ============================================================
# QUERY CLASSIFICATION
# ============================================================

def detect_relationship_direction(query):

    q = query.lower()


    reverse_patterns = [

        "which standards refer to",

        "which bis standards refer to",

        "standards that refer to",

        "referred by",

        "referenced by",

        "who refers to"

    ]


    outgoing_patterns = [

        "what standards does",

        "which standards does",

        "references of",

        "its references",

        "refer to",

        "references"

    ]


    if any(
        p in q
        for p in reverse_patterns
    ):

        return "reverse"


    if any(
        p in q
        for p in outgoing_patterns
    ):

        return "outgoing"


    return None


# ============================================================
# BUILD METADATA LOOKUPS
# ============================================================

standard_by_is_number = {}

standard_by_internal_id = {}


def build_lookup_maps():

    for item in corpus:

        if not isinstance(
            item,
            dict
        ):

            continue


        content = item.get(
            "content",
            item
        )


        if not isinstance(
            content,
            dict
        ):

            continue


        internal_id = (

            content.get(
                "internal_id"
            )

            or item.get(
                "internal_id"
            )

        )


        is_number = (

            content.get(
                "is_number"
            )

            or item.get(
                "identifier"
            )

        )


        if internal_id:

            standard_by_internal_id[
                str(internal_id)
            ] = item


        if is_number:

            normalized = normalize_identifier(
                is_number
            )


            if normalized:

                standard_by_is_number[
                    normalized
                ] = item


# ============================================================
# BUILD LOOKUPS AFTER STARTUP
# ============================================================

@app.on_event("startup")
def build_lookups():

    build_lookup_maps()


# ============================================================
# SEMANTIC SEARCH
# ============================================================

def semantic_search(
    query,
    top_k=8
):

    vector = embedding_model.encode(

        [query],

        normalize_embeddings=True,

        convert_to_numpy=True

    ).astype(
        "float32"
    )


    scores, indices = faiss_index.search(

        vector,

        top_k

    )


    results = []


    for score, index in zip(

        scores[0],

        indices[0]

    ):

        if index < 0:
            continue


        if index >= len(corpus):
            continue


        item = corpus[index]


        result = dict(
            item
        )


        result[
            "_score"
        ] = float(
            score
        )


        results.append(
            result
        )


    return results


# ============================================================
# EXACT STANDARD RETRIEVAL
# ============================================================

def exact_standard_search(
    identifier
):

    normalized = normalize_identifier(
        identifier
    )


    # Direct normalized lookup.
    item = standard_by_is_number.get(
        normalized
    )


    if item:

        return [item]


    # Fallback tolerant matching.
    for key, value in (
        standard_by_is_number.items()
    ):

        if normalize_identifier(
            key
        ) == normalized:

            return [value]


    return []


# ============================================================
# EVIDENCE CONVERSION
# ============================================================

def standard_to_evidence(
    item,
    source="exact"
):

    content = item.get(
        "content",
        {}
    )


    return {

        "evidence_type":
            "standard",

        "source":
            source,

        "identifier":
            normalize_identifier(

                item.get(
                    "identifier"
                )

                or content.get(
                    "is_number"
                )

            ),

        "internal_id":
            str(

                item.get(
                    "internal_id"
                )

                or content.get(
                    "internal_id",
                    ""
                )

            ),

        "title":
            item.get(
                "title"
            )

            or content.get(
                "is_title",
                ""
            ),

        "content":
            content

    }


# ============================================================
# RETRIEVAL
# ============================================================

def retrieve_evidence_api(
    query,
    max_relationships=50
):

    identifiers = extract_identifiers(
        query
    )


    relationship_direction = (
        detect_relationship_direction(
            query
        )
    )


    evidence = []

    retrieval_plan = []


    # --------------------------------------------------------
    # Exact identifier
    # --------------------------------------------------------

    if identifiers:

        for identifier in identifiers:

            exact_items = (
                exact_standard_search(
                    identifier
                )
            )


            if exact_items:

                retrieval_plan.append(

                    "standard:"
                    + identifier

                )


                for item in exact_items:

                    evidence.append(

                        standard_to_evidence(
                            item,
                            "exact"
                        )

                    )


    # --------------------------------------------------------
    # Semantic fallback
    # --------------------------------------------------------

    if not evidence:

        semantic_items = semantic_search(
            query,
            top_k=8
        )


        retrieval_plan.append(
            "semantic"
        )


        for item in semantic_items:

            evidence.append(

                standard_to_evidence(
                    item,
                    "semantic"
                )

            )


    return {

        "query":
            query,

        "identifiers":
            identifiers,

        "compound":
            len(identifiers) > 1,

        "relationship_direction":
            relationship_direction,

        "retrieval_plan":
            retrieval_plan,

        "evidence_count":
            len(evidence),

        "evidence":
            evidence

    }


# ============================================================
# BIS EVIDENCE TEXT
# ============================================================

def build_evidence_text(
    evidence
):

    blocks = []


    for index, item in enumerate(

        evidence,

        start=1

    ):

        content = item.get(
            "content",
            {}
        )


        if isinstance(
            content,
            dict
        ):

            text = content.get(
                "text",
                ""
            )

        else:

            text = str(
                content
            )


        blocks.append(

            f"[E{index}]\n"
            f"Identifier: "
            f"{item.get('identifier', '')}\n"
            f"Title: "
            f"{item.get('title', '')}\n"
            f"{text}"

        )


    return "\n\n".join(
        blocks
    )


# ============================================================
# GEMINI GENERATION
# ============================================================

def generate_grounded_answer(
    query,
    retrieval_result
):

    if gemini_client is None:

        return {

            "answer":
                (
                    "Gemini API is not configured. "
                    "BIS evidence retrieval completed "
                    "successfully."
                ),

            "model":
                GEMINI_MODEL,

            "attempts":
                0,

            "generation_error":
                True

        }


    evidence = retrieval_result.get(
        "evidence",
        []
    )


    if not evidence:

        return {

            "answer":
                (
                    "I could not find sufficient "
                    "BIS evidence to answer this query."
                ),

            "model":
                GEMINI_MODEL,

            "attempts":
                1

        }


    evidence_text = build_evidence_text(
        evidence
    )


    system_instruction = """
You are SAATHI, a grounded BIS Standards assistant.

Answer the user's question ONLY using the supplied BIS evidence.

Rules:
1. Do not invent BIS facts.
2. Cite factual claims using [E1], [E2], etc.
3. Keep the exact relationship direction when the query asks about references.
4. Distinguish Active, Withdrawn, Superseded, or other statuses when evidence provides them.
5. If the evidence is insufficient, explicitly say so.
6. Do not cite evidence that does not support the claim.
7. Use the normalized BIS identifiers supplied in the evidence.
"""


    prompt = (

        f"USER QUERY:\n"
        f"{query}\n\n"

        f"BIS EVIDENCE:\n"
        f"{evidence_text}"

    )


    max_attempts = 4


    for attempt in range(

        1,

        max_attempts + 1

    ):

        try:

            response = (
                gemini_client.models.generate_content(

                    model=GEMINI_MODEL,

                    contents=prompt,

                    config=types.GenerateContentConfig(

                        system_instruction=
                            system_instruction,

                        temperature=0.1,

                        max_output_tokens=1200

                    )

                )
            )


            answer = (
                response.text
                if response.text
                else ""
            )


            return {

                "answer":
                    answer,

                "model":
                    GEMINI_MODEL,

                "attempts":
                    attempt

            }


        except Exception as exc:

            if attempt == max_attempts:

                return {

                    "answer":
                        (
                            "Gemini was temporarily "
                            "unavailable while generating "
                            "the grounded answer. "
                            "The BIS evidence retrieval "
                            "itself completed successfully."
                        ),

                    "model":
                        GEMINI_MODEL,

                    "attempts":
                        attempt,

                    "generation_error":
                        True,

                    "error":
                        str(exc)

                }


            time.sleep(
                2 ** (attempt - 1)
            )


# ============================================================
# SOURCE METADATA PARSER
# ============================================================

def parse_bis_evidence_text(
    text
):

    if not isinstance(
        text,
        str
    ):

        return {}


    fields = {}


    patterns = {

        "technical_department":
            r"Technical Department:\s*(.+)",

        "technical_committee":
            r"Technical Committee:\s*(.+)",

        "aspect":
            r"Aspect:\s*(.+)",

        "language":
            r"Language:\s*(.+)",

        "degree_of_equivalence":
            r"Degree of Equivalence:\s*(.+)",

        "number_of_revisions":
            r"Number of Revisions:\s*(.+)",

        "number_of_amendments":
            r"Number of Amendments:\s*(.+)",

        "amendment_count":
            r"Amendment Count:\s*(.+)",

        "catalogue_is_number":
            r"Catalogue IS Number:\s*(.+)",

        "catalogue_title":
            r"Catalogue Title:\s*(.+)",

        "detail_url":
            r"Detail URL:\s*(\S+)",

        "download_url":
            r"Download URL:\s*(\S+)",

        "composition_url":
            r"Composition URL:\s*(\S+)"

    }


    for key, pattern in patterns.items():

        match = re.search(
            pattern,
            text
        )


        if match:

            fields[key] = (
                match.group(1).strip()
            )


    catalogue_number = fields.get(
        "catalogue_is_number"
    )


    if catalogue_number:

        status_match = re.search(

            r"\((Active|Withdrawn|Superseded|Cancelled|Obsolete)\)",

            catalogue_number,

            flags=re.IGNORECASE

        )


        if status_match:

            fields["status"] = (
                status_match.group(1).capitalize()
            )


    if "status" not in fields:

        status_match = re.search(

            r"status is:\s*([^.]+)",

            text,

            flags=re.IGNORECASE

        )


        if status_match:

            fields["status"] = (
                status_match.group(1).strip()
            )


    return fields


# ============================================================
# PRODUCTION SOURCES
# ============================================================

def build_production_sources(
    retrieval_result
):

    sources = []


    for index, evidence in enumerate(

        retrieval_result.get(
            "evidence",
            []
        ),

        start=1

    ):

        content = evidence.get(
            "content",
            {}
        )


        text = (

            content.get(
                "text",
                ""
            )

            if isinstance(
                content,
                dict
            )

            else str(content)

        )


        parsed = parse_bis_evidence_text(
            text
        )


        identifier = normalize_identifier(

            evidence.get(
                "identifier"
            )

            or content.get(
                "is_number",
                ""
            )

        )


        sources.append({

            "evidence_id":
                f"E{index}",

            "type":
                evidence.get(
                    "evidence_type",
                    "unknown"
                ),

            "relationship":
                evidence.get(
                    "relationship"
                ),

            "identifier":
                identifier,

            "title":
                evidence.get(
                    "title"
                )
                or content.get(
                    "is_title",
                    ""
                ),

            "status":
                parsed.get(
                    "status"
                ),

            "technical_department":
                parsed.get(
                    "technical_department"
                ),

            "technical_committee":
                parsed.get(
                    "technical_committee"
                ),

            "aspect":
                parsed.get(
                    "aspect"
                ),

            "language":
                parsed.get(
                    "language"
                ),

            "degree_of_equivalence":
                parsed.get(
                    "degree_of_equivalence"
                ),

            "number_of_revisions":
                parsed.get(
                    "number_of_revisions"
                ),

            "number_of_amendments":
                parsed.get(
                    "number_of_amendments"
                ),

            "source_url":
                parsed.get(
                    "detail_url"
                ),

            "download_url":
                parsed.get(
                    "download_url"
                ),

            "composition_url":
                parsed.get(
                    "composition_url"
                )

        })


    return sources


# ============================================================
# CITATIONS
# ============================================================

def extract_citations(
    answer
):

    if not answer:

        return []


    matches = re.findall(

        r"\[E(\d+)\]",

        answer

    )


    return list(

        dict.fromkeys(

            f"E{x}"

            for x in matches

        )

    )


# ============================================================
# API RESPONSE
# ============================================================

def build_api_response(
    query,
    retrieval_result,
    llm_result
):

    answer = llm_result.get(
        "answer",
        ""
    )


    sources = build_production_sources(
        retrieval_result
    )


    citations = extract_citations(
        answer
    )


    citation_set = set(
        citations
    )


    cited_sources = [

        source

        for source in sources

        if source[
            "evidence_id"
        ] in citation_set

    ]


    relationship = retrieval_result.get(
        "relationship_direction"
    )


    identifiers = retrieval_result.get(
        "identifiers",
        []
    )


    if retrieval_result.get(
        "compound"
    ):

        query_type = "compound"

    elif relationship == "reverse":

        query_type = "reverse_reference"

    elif relationship == "outgoing":

        query_type = "outgoing_reference"

    elif identifiers:

        query_type = "exact_standard"

    else:

        query_type = "semantic"


    if llm_result.get(
        "generation_error"
    ):

        status = "generation_error"

    elif not retrieval_result.get(
        "evidence"
    ):

        status = "no_evidence"

    else:

        status = "success"


    return {

        "query":
            query,

        "answer":
            answer,

        "query_type":
            query_type,

        "identifiers":
            identifiers,

        "relationship_direction":
            relationship,

        "compound":
            retrieval_result.get(
                "compound",
                False
            ),

        "retrieval": {

            "plan":
                retrieval_result.get(
                    "retrieval_plan",
                    []
                ),

            "evidence_count":
                retrieval_result.get(
                    "evidence_count",
                    0
                ),

            "total_sources":
                len(sources)

        },

        "citations":
            citations,

        "sources":
            cited_sources,

        "generation": {

            "model":
                llm_result.get(
                    "model",
                    GEMINI_MODEL
                ),

            "attempts":
                llm_result.get(
                    "attempts",
                    1
                )

        },

        "status":
            status,

        "timestamp":
            datetime.now(
                timezone.utc
            ).isoformat()

    }


# ============================================================
# HEALTH
# ============================================================

@app.get("/health")
def health():

    return {

        "status":
            "ok"
            if startup_error is None
            else "degraded",

        "service":
            "SAATHI BIS Navigator",

        "faiss_loaded":
            faiss_index is not None,

        "corpus_loaded":
            corpus is not None,

        "embedding_model_loaded":
            embedding_model is not None,

        "gemini_configured":
            gemini_client is not None,

        "documents":
            len(corpus)
            if corpus is not None
            else 0,

        "faiss_vectors":
            faiss_index.ntotal
            if faiss_index is not None
            else 0

    }


# ============================================================
# ASK ENDPOINT
# ============================================================

@app.post("/api/ask")
def ask(request: AskRequest):

    query = request.query.strip()


    if not query:

        raise HTTPException(

            status_code=400,

            detail="Query cannot be empty."

        )


    try:

        retrieval_result = (
            retrieve_evidence_api(

                query,

                max_relationships=
                    request.max_relationships

            )
        )


        llm_result = (
            generate_grounded_answer(

                query,

                retrieval_result

            )
        )


        return build_api_response(

            query,

            retrieval_result,

            llm_result

        )


    except Exception as exc:

        raise HTTPException(

            status_code=500,

            detail={
                "message":
                    "SAATHI inference failed.",
                "error":
                    str(exc)
            }

        )


# ============================================================
# ROOT
# ============================================================

@app.get("/")
def root():

    return {

        "service":
            "SAATHI — BIS Navigator",

        "status":
            "running",

        "docs":
            "/docs",

        "health":
            "/health",

        "ask":
            "/api/ask"

    }
'''


# ============================================================
# 2. WRITE APP.PY
# ============================================================

APP_PATH.write_text(
    app_code,
    encoding="utf-8"
)


print(
    f"✅ Created:\n{APP_PATH}"
)


# ============================================================
# 3. BASIC STATIC CHECK
# ============================================================

print("\n")
print("=" * 70)
print("PYTHON SYNTAX CHECK")
print("=" * 70)


import py_compile


try:

    py_compile.compile(

        str(APP_PATH),

        doraise=True

    )

    print(
        "✅ app.py syntax is valid"
    )

except Exception as exc:

    print(
        "❌ Syntax error:"
    )

    print(exc)

    raise


# ============================================================
# 4. CHECK IMPORTANT ENDPOINTS / COMPONENTS
# ============================================================

app_text = APP_PATH.read_text(
    encoding="utf-8"
)


checks = {

    "FastAPI app":
        "FastAPI(" in app_text,

    "POST /api/ask":
        '@app.post("/api/ask")' in app_text,

    "GET /health":
        '@app.get("/health")' in app_text,

    "FAISS loading":
        "faiss.read_index" in app_text,

    "Embedding model":
        "SentenceTransformer" in app_text,

    "Gemini":
        "genai.Client" in app_text,

    "Citation extraction":
        "extract_citations" in app_text,

    "Production sources":
        "build_production_sources" in app_text,

    "CORS":
        "CORSMiddleware" in app_text

}


print("\n")
print("=" * 70)
print("SERVICE COMPONENT CHECK")
print("=" * 70)


all_checks_passed = True


for name, passed in checks.items():

    print(

        f"{name:<25}"
        f"{'✅' if passed else '❌'}"

    )


    if not passed:

        all_checks_passed = False


# ============================================================
# 5. FILE SIZE
# ============================================================

size_kb = round(

    APP_PATH.stat().st_size /
    1024,

    2

)


print("\n")
print(
    f"app.py size: {size_kb} KB"
)


# ============================================================
# 6. FINAL
# ============================================================

print("\n")
print("=" * 70)
print("CELL 17 COMPLETE")
print("=" * 70)


if all_checks_passed:

    print(
        "\n✅ FASTAPI SERVICE GENERATED SUCCESSFULLY"
    )

else:

    print(
        "\n⚠️ SOME SERVICE CHECKS FAILED"
    )


print(
    f"\nNext file:\n{APP_PATH}"
)
```

```python
# ============================================================
# SAATHI — CELL 18
# BACKEND RETRIEVAL PARITY TEST
# ============================================================

import sys
import json
import importlib.util
from pathlib import Path


print("=" * 70)
print("SAATHI — CELL 18")
print("BACKEND RETRIEVAL PARITY TEST")
print("=" * 70)


# ============================================================
# 1. LOAD STANDALONE BACKEND MODULE
# ============================================================

BACKEND_DIR = Path(
    "/kaggle/working/saathi_backend"
)

APP_PATH = (
    BACKEND_DIR /
    "app.py"
)


if not APP_PATH.exists():

    raise FileNotFoundError(
        f"Backend app not found: {APP_PATH}"
    )


print("\nLoading standalone app.py...")


spec = importlib.util.spec_from_file_location(
    "saathi_backend_app",
    APP_PATH
)


backend_app = importlib.util.module_from_spec(
    spec
)


sys.modules[
    "saathi_backend_app"
] = backend_app


try:

    spec.loader.exec_module(
        backend_app
    )

except Exception as exc:

    print(
        "\n⚠️ app.py could not be imported."
    )

    print(
        "This may happen if a backend-only dependency "
        "is unavailable in the Kaggle environment."
    )

    print(
        "\nError:",
        repr(exc)
    )

    raise


print(
    "✅ app.py imported successfully"
)


# ============================================================
# 2. LOAD BACKEND ARTIFACTS
# ============================================================

print("\n")
print("=" * 70)
print("LOADING BACKEND ARTIFACTS")
print("=" * 70)


backend_app.load_artifacts()

backend_app.build_lookup_maps()


print(
    "FAISS vectors:",
    backend_app.faiss_index.ntotal
)


print(
    "Corpus documents:",
    len(
        backend_app.corpus
    )
)


print(
    "Embedding model:",
    backend_app.EMBEDDING_MODEL
)


# ============================================================
# 3. HELPER — EXTRACT BACKEND IDENTIFIERS
# ============================================================

def get_identifiers_backend(
    result
):

    return result.get(
        "identifiers",
        []
    )


# ============================================================
# 4. HELPER — SUMMARIZE RESULTS
# ============================================================

def summarize_result(
    result
):

    evidence = result.get(
        "evidence",
        []
    )


    return {

        "identifiers":
            result.get(
                "identifiers",
                []
            ),

        "compound":
            result.get(
                "compound",
                False
            ),

        "relationship_direction":
            result.get(
                "relationship_direction"
            ),

        "retrieval_plan":
            result.get(
                "retrieval_plan",
                []
            ),

        "evidence_count":
            result.get(
                "evidence_count",
                len(evidence)
            ),

        "evidence_identifiers": [

            item.get(
                "identifier"
            )

            for item in evidence[:15]

        ]

    }


# ============================================================
# 5. DEFINE PARITY TESTS
# ============================================================

tests = [

    {

        "name":
            "EXACT STANDARD",

        "query":
            "What is IS/IEC 60691:2023?",

        "expected_type":
            "exact",

        "expected_identifier":
            "IS/IEC 60691:2023"

    },

    {

        "name":
            "SEMANTIC",

        "query":
            "Which BIS standard deals with thermal links?",

        "expected_type":
            "semantic",

        "expected_identifier":
            None

    },

    {

        "name":
            "REVERSE REFERENCE",

        "query":
            "Which BIS standards refer to IS/IEC 61196-1:2005?",

        "expected_type":
            "reverse",

        "expected_identifier":
            "IS/IEC 61196-1:2005"

    },

    {

        "name":
            "COMPOUND",

        "query":
            "Tell me about IS/IEC 60691:2023 and its references.",

        "expected_type":
            "compound",

        "expected_identifier":
            "IS/IEC 60691:2023"

    }

]


# ============================================================
# 6. RUN PARITY TESTS
# ============================================================

results = []


for test in tests:

    print("\n")
    print("=" * 70)

    print(
        test["name"]
    )

    print("=" * 70)

    print(
        "Query:",
        test["query"]
    )


    # --------------------------------------------------------
    # Validated notebook retrieval
    # --------------------------------------------------------

    print(
        "\n[REFERENCE PIPELINE]"
    )


    reference_result = retrieve_evidence(

        test["query"],

        semantic_fallback=True

    )


    reference_summary = summarize_result(
        reference_result
    )


    print(
        json.dumps(
            reference_summary,
            indent=2,
            ensure_ascii=False
        )
    )


    # --------------------------------------------------------
    # Standalone backend retrieval
    # --------------------------------------------------------

    print(
        "\n[STANDALONE BACKEND]"
    )


    backend_result = (
        backend_app.retrieve_evidence_api(

            test["query"]

        )
    )


    backend_summary = summarize_result(
        backend_result
    )


    print(
        json.dumps(
            backend_summary,
            indent=2,
            ensure_ascii=False
        )
    )


    # --------------------------------------------------------
    # Compare basic query interpretation
    # --------------------------------------------------------

    checks = {}


    checks[
        "identifiers"
    ] = (

        reference_summary[
            "identifiers"
        ]

        ==

        backend_summary[
            "identifiers"
        ]

    )


    checks[
        "compound"
    ] = (

        reference_summary[
            "compound"
        ]

        ==

        backend_summary[
            "compound"
        ]

    )


    checks[
        "relationship_direction"
    ] = (

        reference_summary[
            "relationship_direction"
        ]

        ==

        backend_summary[
            "relationship_direction"
        ]

    )


    # --------------------------------------------------------
    # Exact identifier check
    # --------------------------------------------------------

    if test[
        "expected_identifier"
    ]:

        expected = normalize_identifier(

            test[
                "expected_identifier"
            ]

        )


        backend_identifiers = [

            normalize_identifier(
                x
            )

            for x in backend_summary[
                "evidence_identifiers"
            ]

            if x

        ]


        checks[
            "expected_identifier_present"
        ] = (

            expected in backend_identifiers

        )


    else:

        checks[
            "expected_identifier_present"
        ] = True


    # --------------------------------------------------------
    # Reference-count comparison
    # --------------------------------------------------------

    checks[
        "evidence_nonempty"
    ] = (

        backend_summary[
            "evidence_count"
        ] > 0

    )


    # --------------------------------------------------------
    # Overall
    # --------------------------------------------------------

    passed = all(
        checks.values()
    )


    results.append({

        "name":
            test["name"],

        "query":
            test["query"],

        "reference":
            reference_summary,

        "backend":
            backend_summary,

        "checks":
            checks,

        "passed":
            passed

    })


    print("\n")
    print(
        "TEST RESULT:",
        "✅ PASS"
        if passed
        else
        "❌ FAIL"
    )


    for check, value in checks.items():

        print(

            f"  {check:<35}"
            f"{'✅' if value else '❌'}"

        )


# ============================================================
# 7. OVERALL SUMMARY
# ============================================================

passed_count = sum(

    1

    for result in results

    if result["passed"]

)


total_count = len(
    results
)


print("\n")
print("=" * 70)
print("PARITY TEST SUMMARY")
print("=" * 70)


print(
    f"\nPassed: {passed_count}/{total_count}"
)


for result in results:

    print(

        f"  {result['name']:<25}"
        f"{'✅ PASS' if result['passed'] else '❌ FAIL'}"

    )


# ============================================================
# 8. SAVE RESULTS
# ============================================================

parity_path = (
    Path("/kaggle/working")
    /
    "saathi_cell18_backend_parity.json"
)


with open(
    parity_path,
    "w",
    encoding="utf-8"
) as f:

    json.dump(

        results,

        f,

        indent=2,

        ensure_ascii=False

    )


# ============================================================
# 9. FINAL
# ============================================================

print("\n")
print("=" * 70)
print("CELL 18 COMPLETE")
print("=" * 70)


print(
    f"\nSaved:\n{parity_path}"
)


if passed_count == total_count:

    print(
        "\n✅ ALL BACKEND PARITY TESTS PASSED"
    )

else:

    print(
        "\n⚠️ BACKEND RETRIEVAL DOES NOT YET MATCH "
        "THE VALIDATED NOTEBOOK PIPELINE"
    )

    print(
        "\nDo NOT connect the frontend yet."
    )
```

```python
# ============================================================
# SAATHI — CELL 19
# FASTAPI RETRIEVAL PARITY PATCH
# ============================================================

from pathlib import Path
import re
import py_compile


print("=" * 70)
print("SAATHI — CELL 19")
print("FASTAPI RETRIEVAL PARITY PATCH")
print("=" * 70)


BACKEND_DIR = Path(
    "/kaggle/working/saathi_backend"
)

APP_PATH = (
    BACKEND_DIR /
    "app.py"
)


# ============================================================
# 1. LOAD CURRENT APP
# ============================================================

if not APP_PATH.exists():

    raise FileNotFoundError(
        f"Missing backend app: {APP_PATH}"
    )


app_text = APP_PATH.read_text(
    encoding="utf-8"
)


print(
    "\nCurrent app.py loaded:",
    len(app_text),
    "characters"
)


# ============================================================
# 2. PATCH IDENTIFIER EXTRACTION
# ============================================================

old_identifier_block = r'''def extract_identifiers(query):

    identifiers = []


    patterns = [

        r'\bIS/IEC\s*[\w-]+\s*:\s*\d{4}\b',

        r'\bIS\s*[\w-]+(?:\s*\([^)]*\))?\s*:\s*\d{4}\b',

        r'\bIEC\s+\d+(?:-\d+)?\s*:\s*\d{4}\b'

    ]


    for pattern in patterns:

        matches = re.findall(

            pattern,

            query,

            flags=re.IGNORECASE

        )


        for match in matches:

            normalized = normalize_identifier(
                match
            )


            if normalized and normalized not in identifiers:

                identifiers.append(
                    normalized
                )


    return identifiers
'''


new_identifier_block = r'''def extract_identifiers(query):

    """
    Extract BIS identifiers.

    IMPORTANT:
    IEC-only identifiers are NOT treated as separate query
    identifiers when they occur inside an IS/IEC identifier.
    This prevents:

        IS/IEC 60691:2023
        IEC 60691:2023

    from being interpreted as two identifiers.
    """

    identifiers = []


    # --------------------------------------------------------
    # BIS IS/IEC identifiers
    # --------------------------------------------------------

    bis_iec_pattern = (

        r'\bIS\s*/\s*IEC\s*'
        r'[\w-]+'
        r'(?:\s*-\s*[\w-]+)?'
        r'\s*:\s*\d{4}\b'

    )


    for match in re.findall(

        bis_iec_pattern,

        query,

        flags=re.IGNORECASE

    ):

        normalized = normalize_identifier(
            match
        )


        if normalized and normalized not in identifiers:

            identifiers.append(
                normalized
            )


    # --------------------------------------------------------
    # Other Indian Standards
    # --------------------------------------------------------

    indian_pattern = (

        r'\bIS\s+'
        r'[\w-]+'
        r'(?:\s*\([^)]*\))?'
        r'\s*:\s*\d{4}\b'

    )


    for match in re.findall(

        indian_pattern,

        query,

        flags=re.IGNORECASE

    ):

        normalized = normalize_identifier(
            match
        )


        if normalized and normalized not in identifiers:

            identifiers.append(
                normalized
            )


    # --------------------------------------------------------
    # If IS/IEC was found, do NOT separately extract IEC.
    # --------------------------------------------------------

    if identifiers:

        return identifiers


    # --------------------------------------------------------
    # Standalone IEC identifiers
    # --------------------------------------------------------

    iec_pattern = (

        r'\bIEC\s+'
        r'\d+(?:-\d+)?'
        r'\s*:\s*\d{4}\b'

    )


    for match in re.findall(

        iec_pattern,

        query,

        flags=re.IGNORECASE

    ):

        normalized = normalize_identifier(
            match
        )


        if normalized and normalized not in identifiers:

            identifiers.append(
                normalized
            )


    return identifiers
'''


if old_identifier_block not in app_text:

    raise RuntimeError(
        "Could not locate existing extract_identifiers() block."
    )


app_text = app_text.replace(

    old_identifier_block,

    new_identifier_block

)


print(
    "✅ Identifier extraction patched"
)


# ============================================================
# 3. PATCH CORPUS LOOKUP MAPS
# ============================================================

old_lookup_block = r'''standard_by_is_number = {}

standard_by_internal_id = {}


def build_lookup_maps():

    for item in corpus:

        if not isinstance(
            item,
            dict
        ):

            continue


        content = item.get(
            "content",
            item
        )


        if not isinstance(
            content,
            dict
        ):

            continue


        internal_id = (

            content.get(
                "internal_id"
            )

            or item.get(
                "internal_id"
            )

        )


        is_number = (

            content.get(
                "is_number"
            )

            or item.get(
                "identifier"
            )

        )


        if internal_id:

            standard_by_internal_id[
                str(internal_id)
            ] = item


        if is_number:

            normalized = normalize_identifier(
                is_number
            )


            if normalized:

                standard_by_is_number[
                    normalized
                ] = item
'''


new_lookup_block = r'''standard_by_is_number = {}

standard_by_internal_id = {}

references_by_standard = {}

referred_by_standard = {}

international_refs_by_standard = {}

standard_core_by_internal_id = {}


def _corpus_content(item):

    """
    Return the actual document content object from a RAG corpus
    document.
    """

    if not isinstance(
        item,
        dict
    ):

        return {}


    content = item.get(
        "content"
    )


    if isinstance(
        content,
        dict
    ):

        return content


    return item


def _corpus_identifier(item):

    content = _corpus_content(
        item
    )


    value = (

        item.get(
            "identifier"
        )

        or content.get(
            "is_number"
        )

        or content.get(
            "reference_number"
        )

        or content.get(
            "international_standard"
        )

    )


    if not value:

        return ""


    return normalize_identifier(
        value
    )


def _corpus_internal_id(item):

    content = _corpus_content(
        item
    )


    value = (

        item.get(
            "internal_id"
        )

        or content.get(
            "internal_id"
        )

    )


    if value is None:

        return ""


    return str(
        value
    )


def _corpus_title(item):

    content = _corpus_content(
        item
    )


    return (

        item.get(
            "title"
        )

        or content.get(
            "is_title"
        )

        or content.get(
            "reference_title"
        )

        or content.get(
            "catalogue_title"
        )

        or ""

    )


def build_lookup_maps():

    global standard_by_is_number
    global standard_by_internal_id
    global references_by_standard
    global referred_by_standard
    global international_refs_by_standard
    global standard_core_by_internal_id


    standard_by_is_number = {}

    standard_by_internal_id = {}

    references_by_standard = {}

    referred_by_standard = {}

    international_refs_by_standard = {}

    standard_core_by_internal_id = {}


    for item in corpus:

        if not isinstance(
            item,
            dict
        ):

            continue


        content = _corpus_content(
            item
        )


        internal_id = _corpus_internal_id(
            item
        )


        identifier = _corpus_identifier(
            item
        )


        document_type = (

            content.get(
                "document_type"
            )

            or item.get(
                "evidence_type"
            )

            or ""

        )


        # ----------------------------------------------------
        # Standard core
        # ----------------------------------------------------

        if document_type == "standard_core":

            if internal_id:

                standard_core_by_internal_id[
                    internal_id
                ] = item


                standard_by_internal_id[
                    internal_id
                ] = item


            if identifier:

                standard_by_is_number[
                    identifier
                ] = item


        # ----------------------------------------------------
        # Cross / Indian references
        # ----------------------------------------------------

        if document_type in (

            "cross_references",
            "indian_references"

        ):

            if internal_id:

                references_by_standard.setdefault(

                    internal_id,

                    []

                ).append(

                    item

                )


        # ----------------------------------------------------
        # International references
        # ----------------------------------------------------

        if document_type == "international_references":

            if internal_id:

                international_refs_by_standard.setdefault(

                    internal_id,

                    []

                ).append(

                    item

                )


        # ----------------------------------------------------
        # Reverse references
        # ----------------------------------------------------

        if document_type == "referred_by":

            reference_target = (

                content.get(
                    "reference_is_number"
                )

                or item.get(
                    "reference_is_number"
                )

            )


            target = normalize_identifier(

                reference_target

                if reference_target

                else ""

            )


            if target:

                referred_by_standard.setdefault(

                    target,

                    []

                ).append(

                    item

                )
'''


if old_lookup_block not in app_text:

    raise RuntimeError(
        "Could not locate existing lookup-map block."
    )


app_text = app_text.replace(

    old_lookup_block,

    new_lookup_block

)


print(
    "✅ Corpus lookup maps patched"
)


# ============================================================
# 4. PATCH STANDARD TO EVIDENCE
# ============================================================

old_standard_evidence = r'''def standard_to_evidence(
    item,
    source="exact"
):

    content = item.get(
        "content",
        {}
    )


    return {

        "evidence_type":
            "standard",

        "source":
            source,

        "identifier":
            normalize_identifier(

                item.get(
                    "identifier"
                )

                or content.get(
                    "is_number"
                )

            ),

        "internal_id":
            str(

                item.get(
                    "internal_id"
                )

                or content.get(
                    "internal_id",
                    ""
                )

            ),

        "title":
            item.get(
                "title"
            )

            or content.get(
                "is_title",
                ""
            ),

        "content":
            content

    }
'''


new_standard_evidence = r'''def standard_to_evidence(
    item,
    source="exact"
):

    content = _corpus_content(
        item
    )


    return {

        "evidence_type":
            "standard",

        "source":
            source,

        "identifier":
            _corpus_identifier(
                item
            ),

        "internal_id":
            _corpus_internal_id(
                item
            ),

        "title":
            _corpus_title(
                item
            ),

        "content":
            content

    }
'''


if old_standard_evidence not in app_text:

    raise RuntimeError(
        "Could not locate standard_to_evidence() block."
    )


app_text = app_text.replace(

    old_standard_evidence,

    new_standard_evidence

)


print(
    "✅ Standard evidence conversion patched"
)


# ============================================================
# 5. ADD RELATIONSHIP EVIDENCE BUILDER
# ============================================================

relationship_code = r'''

# ============================================================
# RELATIONSHIP EVIDENCE
# ============================================================

def relationship_to_evidence(
    item,
    relationship,
    source_identifier
):

    content = _corpus_content(
        item
    )


    identifier = _corpus_identifier(
        item
    )


    title = _corpus_title(
        item
    )


    result = {

        "evidence_type":
            relationship,

        "source":
            "relationship",

        "identifier":
            identifier,

        "internal_id":
            _corpus_internal_id(
                item
            ),

        "title":
            title,

        "relationship":
            relationship,

        "source_identifier":
            source_identifier,

        "content":
            content

    }


    return result


def build_relationship_evidence(
    identifier,
    direction,
    max_results=50
):

    normalized = normalize_identifier(
        identifier
    )


    results = []


    if direction == "outgoing":

        standard = standard_by_is_number.get(
            normalized
        )


        if not standard:

            return []


        internal_id = _corpus_internal_id(
            standard
        )


        # Cross + Indian references
        outgoing = references_by_standard.get(

            internal_id,

            []

        )


        for item in outgoing[

            :max_results

        ]:

            results.append(

                relationship_to_evidence(

                    item,

                    "outgoing_reference",

                    normalized

                )

            )


        remaining = (

            max_results
            - len(results)

        )


        if remaining > 0:

            international = (
                international_refs_by_standard.get(

                    internal_id,

                    []

                )
            )


            for item in international[

                :remaining

            ]:

                results.append(

                    relationship_to_evidence(

                        item,

                        "outgoing_reference",

                        normalized

                    )

                )


    elif direction == "reverse":

        reverse_items = (
            referred_by_standard.get(

                normalized,

                []

            )
        )


        for item in reverse_items[

            :max_results

        ]:

            results.append(

                relationship_to_evidence(

                    item,

                    "reverse_reference",

                    normalized

                )

            )


    return results
'''


# Insert before retrieve_evidence_api.
marker = r'''# ============================================================
# RETRIEVAL
# ============================================================

def retrieve_evidence_api(
'''


if marker not in app_text:

    raise RuntimeError(
        "Could not locate retrieval section."
    )


app_text = app_text.replace(

    marker,

    relationship_code
    + "\n\n"
    + marker

)


print(
    "✅ Relationship evidence builder added"
)


# ============================================================
# 6. REPLACE RETRIEVAL FUNCTION
# ============================================================

start_marker = r'''def retrieve_evidence_api(
'''

end_marker = r'''# ============================================================
# BIS EVIDENCE TEXT
# ============================================================
'''


start_index = app_text.find(
    start_marker
)

end_index = app_text.find(
    end_marker
)


if start_index == -1 or end_index == -1:

    raise RuntimeError(
        "Could not locate retrieve_evidence_api() boundaries."
    )


new_retrieval_function = r'''def retrieve_evidence_api(
    query,
    max_relationships=50
):

    identifiers = extract_identifiers(
        query
    )


    relationship_direction = (
        detect_relationship_direction(
            query
        )
    )


    evidence = []

    retrieval_plan = []


    # --------------------------------------------------------
    # Exact standard
    # --------------------------------------------------------

    if identifiers:

        for identifier in identifiers:

            exact_items = (
                exact_standard_search(
                    identifier
                )
            )


            if exact_items:

                retrieval_plan.append(

                    "standard:"
                    + identifier

                )


                for item in exact_items:

                    evidence.append(

                        standard_to_evidence(

                            item,

                            "exact"

                        )

                    )


                # ------------------------------------------------
                # Relationship-aware retrieval
                # ------------------------------------------------

                if relationship_direction:

                    relationship_items = (
                        build_relationship_evidence(

                            identifier,

                            relationship_direction,

                            max_relationships

                        )
                    )


                    if relationship_items:

                        if (
                            relationship_direction
                            == "reverse"
                        ):

                            retrieval_plan.append(

                                "reverse_references:"
                                + identifier

                            )

                        elif (
                            relationship_direction
                            == "outgoing"
                        ):

                            retrieval_plan.append(

                                "outgoing_references:"
                                + identifier

                            )


                        evidence.extend(
                            relationship_items
                        )


    # --------------------------------------------------------
    # Semantic fallback
    # --------------------------------------------------------

    if not evidence:

        semantic_items = semantic_search(

            query,

            top_k=8

        )


        retrieval_plan.append(
            "semantic"
        )


        for item in semantic_items:

            evidence.append(

                standard_to_evidence(

                    item,

                    "semantic"

                )

            )


    return {

        "query":
            query,

        "identifiers":
            identifiers,

        "compound":
            (
                len(identifiers) > 1
                or (
                    bool(identifiers)
                    and bool(
                        relationship_direction
                    )
                )
            ),

        "relationship_direction":
            relationship_direction,

        "retrieval_plan":
            retrieval_plan,

        "evidence_count":
            len(evidence),

        "evidence":
            evidence

    }


'''


app_text = (

    app_text[:start_index]

    + new_retrieval_function

    + app_text[end_index:]

)


print(
    "✅ Retrieval function patched"
)


# ============================================================
# 7. FIX COMPOUND DETECTION
# ============================================================

# The validated notebook treats an identifier + relationship
# request as compound, but an ordinary exact lookup is not.

# The logic above already implements this distinction.


# ============================================================
# 8. WRITE PATCHED APP
# ============================================================

APP_PATH.write_text(

    app_text,

    encoding="utf-8"

)


print(
    f"\n✅ Patched app written:\n{APP_PATH}"
)


# ============================================================
# 9. SYNTAX CHECK
# ============================================================

print("\n")
print("=" * 70)
print("SYNTAX CHECK")
print("=" * 70)


py_compile.compile(

    str(APP_PATH),

    doraise=True

)


print(
    "✅ app.py syntax is valid"
)


# ============================================================
# 10. STATIC COMPONENT CHECK
# ============================================================

updated_text = APP_PATH.read_text(
    encoding="utf-8"
)


checks = {

    "BIS identifier extraction":
        "IS/IEC" in updated_text
        and "def extract_identifiers" in updated_text,

    "Corpus content parser":
        "def _corpus_content" in updated_text,

    "Standard lookup":
        "standard_by_is_number" in updated_text,

    "Outgoing references":
        "references_by_standard" in updated_text,

    "Reverse references":
        "referred_by_standard" in updated_text,

    "International references":
        "international_refs_by_standard"
        in updated_text,

    "Relationship evidence":
        "def build_relationship_evidence"
        in updated_text,

    "Patched retrieval":
        "def retrieve_evidence_api"
        in updated_text

}


print("\n")
print("=" * 70)
print("PATCH CHECK")
print("=" * 70)


all_passed = True


for name, passed in checks.items():

    print(

        f"{name:<30}"
        f"{'✅' if passed else '❌'}"

    )


    if not passed:

        all_passed = False


# ============================================================
# 11. FINAL
# ============================================================

print("\n")
print("=" * 70)
print("CELL 19 COMPLETE")
print("=" * 70)


if all_passed:

    print(
        "\n✅ FASTAPI RETRIEVAL PARITY PATCH APPLIED"
    )

else:

    print(
        "\n⚠️ SOME PATCH COMPONENTS ARE MISSING"
    )


print(
    f"\nPatched backend:\n{APP_PATH}"
)
```

```python
# ============================================================
# SAATHI — CELL 20
# POST-PATCH BACKEND RETRIEVAL PARITY TEST
# ============================================================

import sys
import json
import importlib.util
from pathlib import Path


print("=" * 70)
print("SAATHI — CELL 20")
print("POST-PATCH BACKEND RETRIEVAL PARITY TEST")
print("=" * 70)


# ============================================================
# 1. LOAD PATCHED BACKEND
# ============================================================

BACKEND_DIR = Path(
    "/kaggle/working/saathi_backend"
)

APP_PATH = (
    BACKEND_DIR /
    "app.py"
)


print("\nLoading patched app.py...")


spec = importlib.util.spec_from_file_location(
    "saathi_backend_app_v2",
    APP_PATH
)


backend_app = importlib.util.module_from_spec(
    spec
)


sys.modules[
    "saathi_backend_app_v2"
] = backend_app


spec.loader.exec_module(
    backend_app
)


print(
    "✅ Patched app.py imported"
)


# ============================================================
# 2. LOAD BACKEND ARTIFACTS
# ============================================================

print("\n")
print("=" * 70)
print("LOADING BACKEND ARTIFACTS")
print("=" * 70)


backend_app.load_artifacts()

backend_app.build_lookup_maps()


print(
    "FAISS vectors:",
    backend_app.faiss_index.ntotal
)


print(
    "Corpus documents:",
    len(
        backend_app.corpus
    )
)


print(
    "Standard lookup entries:",
    len(
        backend_app.standard_by_is_number
    )
)


print(
    "Reverse-reference targets:",
    len(
        backend_app.referred_by_standard
    )
)


print(
    "Outgoing-reference standards:",
    len(
        backend_app.references_by_standard
    )
)


# ============================================================
# 3. SUMMARY HELPER
# ============================================================

def summarize(
    result
):

    evidence = result.get(
        "evidence",
        []
    )


    return {

        "identifiers":
            result.get(
                "identifiers",
                []
            ),

        "compound":
            result.get(
                "compound",
                False
            ),

        "relationship_direction":
            result.get(
                "relationship_direction"
            ),

        "retrieval_plan":
            result.get(
                "retrieval_plan",
                []
            ),

        "evidence_count":
            result.get(
                "evidence_count",
                len(evidence)
            ),

        "evidence_identifiers":
            [

                item.get(
                    "identifier"
                )

                for item in evidence[:15]

            ],

        "evidence_types":
            [

                item.get(
                    "evidence_type"
                )

                for item in evidence[:15]

            ]

    }


# ============================================================
# 4. TEST DEFINITIONS
# ============================================================

tests = [

    {

        "name":
            "EXACT STANDARD",

        "query":
            "What is IS/IEC 60691:2023?"

    },

    {

        "name":
            "SEMANTIC",

        "query":
            "Which BIS standard deals with thermal links?"

    },

    {

        "name":
            "REVERSE REFERENCE",

        "query":
            "Which BIS standards refer to IS/IEC 61196-1:2005?"

    },

    {

        "name":
            "COMPOUND",

        "query":
            "Tell me about IS/IEC 60691:2023 and its references."

    }

]


# ============================================================
# 5. EXPECTED VALUES FROM VALIDATED PIPELINE
# ============================================================

expected = {

    "EXACT STANDARD": {

        "identifiers":
            [
                "IS/IEC 60691:2023"
            ],

        "compound":
            False,

        "relationship_direction":
            None,

        "evidence_count":
            1

    },

    "SEMANTIC": {

        "identifiers":
            [],

        "compound":
            False,

        "relationship_direction":
            None,

        "evidence_count":
            5

    },

    "REVERSE REFERENCE": {

        "identifiers":
            [
                "IS/IEC 61196-1:2005"
            ],

        "compound":
            True,

        "relationship_direction":
            "reverse",

        "evidence_count":
            43

    },

    "COMPOUND": {

        "identifiers":
            [
                "IS/IEC 60691:2023"
            ],

        "compound":
            True,

        "relationship_direction":
            "outgoing",

        "evidence_count":
            12

    }

}


# ============================================================
# 6. RUN TESTS
# ============================================================

results = []


for test in tests:

    name = test[
        "name"
    ]

    query = test[
        "query"
    ]


    print("\n")
    print("=" * 70)

    print(name)

    print("=" * 70)

    print(
        "Query:",
        query
    )


    # --------------------------------------------------------
    # Reference
    # --------------------------------------------------------

    reference = retrieve_evidence(

        query,

        semantic_fallback=True

    )


    reference_summary = summarize(
        reference
    )


    # --------------------------------------------------------
    # Backend
    # --------------------------------------------------------

    backend = (
        backend_app.retrieve_evidence_api(
            query
        )
    )


    backend_summary = summarize(
        backend
    )


    # --------------------------------------------------------
    # Expected
    # --------------------------------------------------------

    exp = expected[
        name
    ]


    checks = {

        "identifiers":
            backend_summary[
                "identifiers"
            ] == exp[
                "identifiers"
            ],

        "compound":
            backend_summary[
                "compound"
            ] == exp[
                "compound"
            ],

        "relationship_direction":
            backend_summary[
                "relationship_direction"
            ] == exp[
                "relationship_direction"
            ],

        "evidence_count":
            backend_summary[
                "evidence_count"
            ] == exp[
                "evidence_count"
            ]

    }


    # --------------------------------------------------------
    # Print reference
    # --------------------------------------------------------

    print("\n[REFERENCE]")

    print(

        json.dumps(

            reference_summary,

            indent=2,

            ensure_ascii=False

        )

    )


    # --------------------------------------------------------
    # Print backend
    # --------------------------------------------------------

    print("\n[BACKEND]")

    print(

        json.dumps(

            backend_summary,

            indent=2,

            ensure_ascii=False

        )

    )


    # --------------------------------------------------------
    # Checks
    # --------------------------------------------------------

    passed = all(
        checks.values()
    )


    print("\nCHECKS")


    for check, value in checks.items():

        print(

            f"  {check:<30}"
            f"{'✅' if value else '❌'}"

        )


    print(

        "\nTEST RESULT:",

        "✅ PASS"
        if passed
        else
        "❌ FAIL"

    )


    results.append({

        "name":
            name,

        "query":
            query,

        "reference":
            reference_summary,

        "backend":
            backend_summary,

        "expected":
            exp,

        "checks":
            checks,

        "passed":
            passed

    })


# ============================================================
# 7. SUMMARY
# ============================================================

passed_count = sum(

    1

    for result in results

    if result["passed"]

)


print("\n")
print("=" * 70)
print("POST-PATCH PARITY SUMMARY")
print("=" * 70)


print(

    f"\nPassed: "
    f"{passed_count}/{len(results)}"

)


for result in results:

    print(

        f"  {result['name']:<25}"
        f"{'✅ PASS' if result['passed'] else '❌ FAIL'}"

    )


# ============================================================
# 8. SAVE
# ============================================================

output_path = (
    Path("/kaggle/working")
    /
    "saathi_cell20_post_patch_parity.json"
)


with open(
    output_path,
    "w",
    encoding="utf-8"
) as f:

    json.dump(

        results,

        f,

        indent=2,

        ensure_ascii=False

    )


# ============================================================
# 9. FINAL
# ============================================================

print("\n")
print("=" * 70)
print("CELL 20 COMPLETE")
print("=" * 70)


print(
    f"\nSaved:\n{output_path}"
)


if passed_count == len(results):

    print(
        "\n🎉 ALL BACKEND RETRIEVAL PARITY TESTS PASSED"
    )

    print(
        "\nThe standalone backend now matches "
        "the validated retrieval contract."
    )

else:

    print(
        "\n⚠️ PARITY TESTS STILL HAVE FAILURES"
    )

    print(
        "\nDo NOT connect the frontend yet."
    )
```

```python
# ============================================================
# SAATHI — CELL 21
# RELATIONSHIP MAP + SEMANTIC EVIDENCE PARITY FIX
# ============================================================

from pathlib import Path
import ast
import re
import json


print("=" * 70)
print("SAATHI — CELL 21")
print("RELATIONSHIP + SEMANTIC PARITY FIX")
print("=" * 70)


APP_PATH = Path(
    "/kaggle/working/saathi_backend/app.py"
)

app_text = APP_PATH.read_text(
    encoding="utf-8"
)

print(
    "Current app.py loaded:",
    len(app_text),
    "characters"
)


# ============================================================
# 1. REPLACE build_lookup_maps()
# ============================================================

start = app_text.find(
    "def build_lookup_maps():"
)

if start == -1:
    raise RuntimeError(
        "Could not find build_lookup_maps()"
    )


next_def = app_text.find(
    "\ndef ",
    start + 10
)

if next_def == -1:
    next_def = len(app_text)


new_build_lookup_maps = r'''
def build_lookup_maps():
    """
    Build relationship-aware lookup maps directly from the
    canonical RAG corpus.

    IMPORTANT:
    Relationship documents are keyed by the SOURCE standard
    through internal_id.  The referenced standard itself is
    stored inside the relationship document.
    """

    global standard_by_is_number
    global standard_by_internal_id
    global references_by_standard
    global referred_by_standard
    global international_refs_by_standard
    global standard_core_by_internal_id

    standard_by_is_number = {}
    standard_by_internal_id = {}

    references_by_standard = {}
    referred_by_standard = {}
    international_refs_by_standard = {}

    standard_core_by_internal_id = {}

    # --------------------------------------------------------
    # Helpers
    # --------------------------------------------------------

    def normalize_key(value):
        if value is None:
            return ""

        value = str(value).strip()

        return normalize_identifier(
            value
        )

    def content_of(doc):
        content = doc.get(
            "content",
            {}
        )

        if isinstance(content, dict):
            return content

        return {}

    def get_internal_id(doc):
        content = content_of(doc)

        value = (
            content.get("internal_id")
            or doc.get("internal_id")
        )

        if value is None:
            return ""

        return str(value).strip()

    def get_identifier(doc):
        content = content_of(doc)

        value = (
            content.get("is_number")
            or content.get("identifier")
            or doc.get("identifier")
        )

        return normalize_key(
            value
        )

    def get_title(doc):
        content = content_of(doc)

        return str(
            content.get("is_title")
            or content.get("title")
            or doc.get("title")
            or ""
        ).strip()

    def get_document_type(doc):
        content = content_of(doc)

        return str(
            content.get("document_type")
            or doc.get("document_type")
            or ""
        ).strip().lower()

    def extract_reference_identifier(doc):
        """
        Extract the actual referenced BIS identifier from a
        relationship document.

        The corpus relationship documents contain fields such as:
          reference_number
          reference_is_number
          international_standard

        They may also have a textual representation.

        We deliberately inspect structured fields first.
        """

        content = content_of(doc)

        candidates = [

            content.get(
                "reference_is_number"
            ),

            content.get(
                "reference_number"
            ),

            content.get(
                "international_standard"
            ),

            doc.get(
                "reference_is_number"
            ),

            doc.get(
                "reference_number"
            ),

            doc.get(
                "international_standard"
            ),

        ]

        for candidate in candidates:

            if candidate is None:
                continue

            candidate = str(
                candidate
            ).strip()

            if not candidate:
                continue

            normalized = normalize_key(
                candidate
            )

            if normalized:
                return normalized

            # Sometimes the field contains extra prose.
            found = extract_identifiers(
                candidate
            )

            if found:
                return found[0]

        # Last-resort extraction from text.
        text = str(
            content.get(
                "text",
                ""
            )
        )

        found = extract_identifiers(
            text
        )

        if found:
            return found[0]

        return ""

    def append_unique(mapping, key, doc):
        if not key:
            return

        bucket = mapping.setdefault(
            key,
            []
        )

        # Avoid adding the exact same corpus document twice.
        doc_id = (
            doc.get("doc_id")
            or doc.get("id")
            or id(doc)
        )

        for existing in bucket:

            existing_id = (
                existing.get("doc_id")
                or existing.get("id")
            )

            if (
                existing_id is not None
                and existing_id == doc_id
            ):
                return

        bucket.append(
            doc
        )

    # --------------------------------------------------------
    # PASS 1 — canonical standards
    # --------------------------------------------------------

    for doc in corpus:

        document_type = get_document_type(
            doc
        )

        if document_type != "standard_core":
            continue

        internal_id = get_internal_id(
            doc
        )

        identifier = get_identifier(
            doc
        )

        if internal_id:

            standard_core_by_internal_id[
                internal_id
            ] = doc

            standard_by_internal_id[
                internal_id
            ] = doc

        if identifier:

            standard_by_is_number[
                identifier
            ] = doc

    # --------------------------------------------------------
    # PASS 2 — relationship documents
    # --------------------------------------------------------

    for doc in corpus:

        document_type = get_document_type(
            doc
        )

        if document_type not in {
            "cross_references",
            "indian_references",
            "international_references",
            "referred_by",
        }:
            continue

        source_internal_id = get_internal_id(
            doc
        )

        if not source_internal_id:
            continue

        source_standard = (
            standard_core_by_internal_id.get(
                source_internal_id
            )
        )

        if source_standard is None:
            continue

        source_identifier = (
            get_identifier(
                source_standard
            )
        )

        if not source_identifier:
            continue

        source_identifier = normalize_key(
            source_identifier
        )

        referenced_identifier = (
            extract_reference_identifier(
                doc
            )
        )

        # ----------------------------------------------------
        # OUTGOING
        #
        # The relationship row belongs to SOURCE standard.
        # Therefore:
        #
        # source -> referenced
        # ----------------------------------------------------

        if document_type in {
            "cross_references",
            "indian_references",
        }:

            append_unique(
                references_by_standard,
                source_identifier,
                doc
            )

        elif document_type == (
            "international_references"
        ):

            append_unique(
                international_refs_by_standard,
                source_identifier,
                doc
            )

        # ----------------------------------------------------
        # REVERSE
        #
        # If source standard refers to TARGET, then TARGET
        # is referred by SOURCE.
        #
        # target -> [source relationship rows]
        # ----------------------------------------------------

        if document_type in {
            "cross_references",
            "indian_references",
            "international_references",
        }:

            if referenced_identifier:

                append_unique(
                    referred_by_standard,
                    referenced_identifier,
                    doc
                )

    print(
        "Lookup maps rebuilt:"
    )

    print(
        "  Standards:",
        len(
            standard_by_internal_id
        )
    )

    print(
        "  IS-number keys:",
        len(
            standard_by_is_number
        )
    )

    print(
        "  Outgoing BIS refs:",
        len(
            references_by_standard
        )
    )

    print(
        "  International refs:",
        len(
            international_refs_by_standard
        )
    )

    print(
        "  Reverse-reference targets:",
        len(
            referred_by_standard
        )
    )
'''


app_text = (
    app_text[:start]
    + new_build_lookup_maps.strip()
    + "\n"
    + app_text[next_def:]
)


print(
    "✅ build_lookup_maps() replaced"
)


# ============================================================
# 2. REPLACE SEMANTIC RETRIEVAL
# ============================================================

start = app_text.find(
    "def build_semantic_evidence("
)

if start == -1:
    raise RuntimeError(
        "Could not find build_semantic_evidence()"
    )


next_def = app_text.find(
    "\ndef ",
    start + 10
)

if next_def == -1:
    next_def = len(app_text)


new_semantic = r'''
def build_semantic_evidence(
    query,
    top_k=5
):
    """
    Semantic retrieval parity layer.

    FAISS may return more raw candidates than the validated
    notebook exposes.  The API evidence layer must expose the
    same top-k semantic evidence contract.
    """

    if faiss_index is None:
        return []

    if embedding_model is None:
        return []

    query_vector = embedding_model.encode(
        [query],
        normalize_embeddings=True,
        convert_to_numpy=True
    )

    query_vector = query_vector.astype(
        "float32"
    )

    scores, indices = faiss_index.search(
        query_vector,
        max(
            top_k,
            5
        )
    )

    evidence = []

    seen = set()

    for score, idx in zip(
        scores[0],
        indices[0]
    ):

        if idx < 0:
            continue

        if idx >= len(corpus):
            continue

        doc = corpus[
            int(idx)
        ]

        item = standard_to_evidence(
            doc,
            source="semantic",
            score=float(score)
        )

        if not item:
            continue

        identifier = (
            item.get(
                "identifier"
            )
            or ""
        )

        title = (
            item.get(
                "title"
            )
            or ""
        )

        dedupe_key = (
            identifier,
            title
        )

        if dedupe_key in seen:
            continue

        seen.add(
            dedupe_key
        )

        evidence.append(
            item
        )

        if len(evidence) >= top_k:
            break

    return evidence
'''


app_text = (
    app_text[:start]
    + new_semantic.strip()
    + "\n"
    + app_text[next_def:]
)


print(
    "✅ build_semantic_evidence() replaced"
)


# ============================================================
# 3. PATCH RELATIONSHIP EVIDENCE
# ============================================================

start = app_text.find(
    "def relationship_to_evidence("
)

if start == -1:
    raise RuntimeError(
        "Could not find relationship_to_evidence()"
    )


next_def = app_text.find(
    "\ndef ",
    start + 10
)

if next_def == -1:
    next_def = len(app_text)


new_relationship_to_evidence = r'''
def relationship_to_evidence(
    doc,
    relationship,
    source_identifier=None
):
    """
    Convert a relationship corpus document into API evidence.

    IMPORTANT:
    identifier = source standard
    referenced_identifier = actual target/reference
    """

    content = (
        doc.get(
            "content",
            {}
        )
    )

    if not isinstance(
        content,
        dict
    ):
        content = {}

    source_identifier = (
        normalize_identifier(
            source_identifier
            or
            content.get(
                "source_identifier"
            )
            or
            ""
        )
    )

    reference_identifier = normalize_identifier(
        content.get(
            "reference_is_number"
        )
        or
        content.get(
            "reference_number"
        )
        or
        content.get(
            "international_standard"
        )
        or
        ""
    )

    if not reference_identifier:

        reference_text = str(
            content.get(
                "text",
                ""
            )
        )

        found = extract_identifiers(
            reference_text
        )

        if found:
            reference_identifier = found[0]

    return {

        "evidence_type":
            "relationship",

        "relationship":
            relationship,

        "identifier":
            source_identifier,

        "referenced_identifier":
            reference_identifier,

        "title":
            str(
                content.get(
                    "reference_title"
                )
                or
                content.get(
                    "international_standard"
                )
                or
                ""
            ).strip(),

        "internal_id":
            str(
                content.get(
                    "internal_id"
                )
                or
                ""
            ).strip(),

        "content":
            content

    }
'''


app_text = (
    app_text[:start]
    + new_relationship_to_evidence.strip()
    + "\n"
    + app_text[next_def:]
)


print(
    "✅ relationship_to_evidence() replaced"
)


# ============================================================
# 4. REPLACE build_relationship_evidence()
# ============================================================

start = app_text.find(
    "def build_relationship_evidence("
)

if start == -1:
    raise RuntimeError(
        "Could not find build_relationship_evidence()"
    )


next_def = app_text.find(
    "\ndef ",
    start + 10
)

if next_def == -1:
    next_def = len(app_text)


new_relationship_builder = r'''
def build_relationship_evidence(
    identifier,
    direction
):
    """
    Relationship-aware evidence builder.

    outgoing:
        source standard -> references

    reverse:
        target standard <- standards referring to it
    """

    identifier = normalize_identifier(
        identifier
    )

    evidence = []

    # --------------------------------------------------------
    # OUTGOING
    # --------------------------------------------------------

    if direction == "outgoing":

        outgoing = references_by_standard.get(
            identifier,
            []
        )

        international = (
            international_refs_by_standard.get(
                identifier,
                []
            )
        )

        for doc in outgoing:

            evidence.append(
                relationship_to_evidence(
                    doc,
                    "outgoing",
                    source_identifier=identifier
                )
            )

        for doc in international:

            evidence.append(
                relationship_to_evidence(
                    doc,
                    "outgoing",
                    source_identifier=identifier
                )
            )

        return evidence

    # --------------------------------------------------------
    # REVERSE
    # --------------------------------------------------------

    if direction == "reverse":

        reverse_rows = (
            referred_by_standard.get(
                identifier,
                []
            )
        )

        for doc in reverse_rows:

            content = doc.get(
                "content",
                {}
            )

            if not isinstance(
                content,
                dict
            ):
                content = {}

            source_internal_id = str(
                content.get(
                    "internal_id"
                )
                or
                ""
            ).strip()

            source_standard = (
                standard_core_by_internal_id.get(
                    source_internal_id
                )
            )

            source_identifier = ""

            if source_standard:

                source_content = (
                    source_standard.get(
                        "content",
                        {}
                    )
                )

                if isinstance(
                    source_content,
                    dict
                ):

                    source_identifier = (
                        normalize_identifier(
                            source_content.get(
                                "is_number"
                            )
                            or
                            ""
                        )
                    )

            evidence.append(
                relationship_to_evidence(
                    doc,
                    "reverse",
                    source_identifier=(
                        source_identifier
                    )
                )
            )

        return evidence

    return []
'''


app_text = (
    app_text[:start]
    + new_relationship_builder.strip()
    + "\n"
    + app_text[next_def:]
)


print(
    "✅ build_relationship_evidence() replaced"
)


# ============================================================
# 5. WRITE PATCHED APP
# ============================================================

APP_PATH.write_text(
    app_text,
    encoding="utf-8"
)


print("\n")
print("=" * 70)
print("SYNTAX CHECK")
print("=" * 70)


compile(
    app_text,
    str(APP_PATH),
    "exec"
)


print(
    "✅ app.py syntax is valid"
)


# ============================================================
# 6. STATIC CHECKS
# ============================================================

checks = {

    "build_lookup_maps":
        "def build_lookup_maps" in app_text,

    "reverse map":
        "referred_by_standard" in app_text,

    "outgoing map":
        "references_by_standard" in app_text,

    "international map":
        "international_refs_by_standard" in app_text,

    "semantic evidence":
        "def build_semantic_evidence" in app_text,

    "relationship evidence":
        "def relationship_to_evidence" in app_text,

    "relationship builder":
        "def build_relationship_evidence" in app_text,

    "reverse retrieval":
        'direction == "reverse"' in app_text,

    "outgoing retrieval":
        'direction == "outgoing"' in app_text,

}


print("\n")
print("=" * 70)
print("PATCH CHECK")
print("=" * 70)


for name, passed in checks.items():

    print(
        f"{name:<30}"
        f"{'✅' if passed else '❌'}"
    )


if not all(
    checks.values()
):

    raise RuntimeError(
        "One or more patch checks failed."
    )


print("\n")
print("=" * 70)
print("CELL 21 COMPLETE")
print("=" * 70)


print(
    "\n✅ RELATIONSHIP + SEMANTIC PARITY PATCH APPLIED"
)

print(
    "\nPatched backend:"
)

print(
    APP_PATH
)
```

```python
# ============================================================
# SAATHI — CELL 21B
# PATCH ACTUAL BACKEND RETRIEVAL FUNCTIONS
# ============================================================

from pathlib import Path
import ast

APP_PATH = Path(
    "/kaggle/working/saathi_backend/app.py"
)

app_text = APP_PATH.read_text(
    encoding="utf-8"
)

print("=" * 70)
print("SAATHI — CELL 21B")
print("PATCHING ACTUAL BACKEND RETRIEVAL FUNCTIONS")
print("=" * 70)


# ============================================================
# SAFE FUNCTION REPLACER
# ============================================================

def replace_function(
    source,
    function_name,
    new_function
):

    tree = ast.parse(
        source
    )

    target = None

    for node in tree.body:

        if isinstance(
            node,
            (
                ast.FunctionDef,
                ast.AsyncFunctionDef
            )
        ):

            if node.name == function_name:

                target = node
                break

    if target is None:

        raise RuntimeError(
            f"Could not find top-level function: "
            f"{function_name}"
        )

    lines = source.splitlines(
        keepends=True
    )

    start_line = (
        target.lineno - 1
    )

    end_line = (
        target.end_lineno
    )

    replacement = (
        new_function.strip()
        + "\n\n"
    )

    new_source = (
        "".join(
            lines[:start_line]
        )
        + replacement
        + "".join(
            lines[end_line:]
        )
    )

    # Validate immediately.
    compile(
        new_source,
        str(APP_PATH),
        "exec"
    )

    return new_source


# ============================================================
# 1. REPLACE semantic_search()
# ============================================================

new_semantic_search = r'''
def semantic_search(
    query,
    top_k=5
):
    """
    Semantic retrieval using the existing FAISS index.

    The validated notebook exposes exactly the top 5 semantic
    evidence items after converting raw corpus documents into
    standard evidence records.
    """

    if faiss_index is None:
        return []

    if embedding_model is None:
        return []

    query_vector = embedding_model.encode(
        [query],
        normalize_embeddings=True,
        convert_to_numpy=True
    )

    query_vector = query_vector.astype(
        "float32"
    )

    scores, indices = faiss_index.search(
        query_vector,
        top_k
    )

    results = []

    seen = set()

    for score, index in zip(
        scores[0],
        indices[0]
    ):

        index = int(index)

        if index < 0:
            continue

        if index >= len(corpus):
            continue

        doc = corpus[
            index
        ]

        evidence = standard_to_evidence(
            doc,
            source="semantic",
            score=float(score)
        )

        if not evidence:
            continue

        identifier = (
            evidence.get(
                "identifier"
            )
            or ""
        )

        title = (
            evidence.get(
                "title"
            )
            or ""
        )

        # Deduplicate semantic results by standard identity.
        key = (
            identifier,
            title
        )

        if key in seen:
            continue

        seen.add(
            key
        )

        results.append(
            evidence
        )

    return results[:top_k]
'''


app_text = replace_function(
    app_text,
    "semantic_search",
    new_semantic_search
)

print(
    "✅ semantic_search() patched"
)


# ============================================================
# 2. REPLACE relationship_to_evidence()
# ============================================================

new_relationship_to_evidence = r'''
def relationship_to_evidence(
    doc,
    relationship,
    source_identifier=None
):
    """
    Convert one relationship corpus document into structured
    evidence.

    identifier:
        the BIS standard owning the relationship row

    referenced_identifier:
        the actual referenced standard, when available
    """

    content = doc.get(
        "content",
        {}
    )

    if not isinstance(
        content,
        dict
    ):
        content = {}

    source_identifier = normalize_identifier(
        source_identifier
        or
        content.get(
            "source_identifier"
        )
        or
        ""
    )

    # --------------------------------------------------------
    # Reference identifier
    # --------------------------------------------------------

    reference_identifier = ""

    for field in (
        "reference_is_number",
        "reference_number",
        "international_standard"
    ):

        value = content.get(
            field
        )

        if value is None:
            continue

        value = str(
            value
        ).strip()

        if not value:
            continue

        normalized = normalize_identifier(
            value
        )

        if normalized:

            reference_identifier = (
                normalized
            )

            break

    # --------------------------------------------------------
    # Reference title
    # --------------------------------------------------------

    reference_title = str(
        content.get(
            "reference_title"
        )
        or
        ""
    ).strip()

    if not reference_title:

        reference_title = str(
            content.get(
                "international_standard"
            )
            or
            ""
        ).strip()

    return {

        "evidence_type":
            "relationship",

        "relationship":
            relationship,

        "identifier":
            source_identifier,

        "referenced_identifier":
            reference_identifier,

        "title":
            reference_title,

        "internal_id":
            str(
                content.get(
                    "internal_id"
                )
                or
                ""
            ).strip(),

        "content":
            content

    }
'''


app_text = replace_function(
    app_text,
    "relationship_to_evidence",
    new_relationship_to_evidence
)

print(
    "✅ relationship_to_evidence() patched"
)


# ============================================================
# 3. REPLACE build_relationship_evidence()
# ============================================================

new_relationship_builder = r'''
def build_relationship_evidence(
    identifier,
    direction
):
    """
    Build relationship evidence from the lookup maps.

    outgoing:
        queried standard -> standards it references

    reverse:
        standards -> queried standard
    """

    identifier = normalize_identifier(
        identifier
    )

    evidence = []

    # ========================================================
    # OUTGOING
    # ========================================================

    if direction == "outgoing":

        outgoing_rows = (
            references_by_standard.get(
                identifier,
                []
            )
        )

        international_rows = (
            international_refs_by_standard.get(
                identifier,
                []
            )
        )

        for row in outgoing_rows:

            evidence.append(
                relationship_to_evidence(
                    row,
                    "outgoing",
                    source_identifier=identifier
                )
            )

        for row in international_rows:

            evidence.append(
                relationship_to_evidence(
                    row,
                    "outgoing",
                    source_identifier=identifier
                )
            )

        return evidence

    # ========================================================
    # REVERSE
    # ========================================================

    if direction == "reverse":

        rows = (
            referred_by_standard.get(
                identifier,
                []
            )
        )

        for row in rows:

            content = row.get(
                "content",
                {}
            )

            if not isinstance(
                content,
                dict
            ):
                content = {}

            source_internal_id = str(
                content.get(
                    "internal_id"
                )
                or
                ""
            ).strip()

            source_standard = (
                standard_core_by_internal_id.get(
                    source_internal_id
                )
            )

            source_identifier = ""

            if source_standard:

                source_content = (
                    source_standard.get(
                        "content",
                        {}
                    )
                )

                if isinstance(
                    source_content,
                    dict
                ):

                    source_identifier = (
                        normalize_identifier(
                            source_content.get(
                                "is_number"
                            )
                            or
                            ""
                        )
                    )

            evidence.append(
                relationship_to_evidence(
                    row,
                    "reverse",
                    source_identifier=source_identifier
                )
            )

        return evidence

    return []
'''


app_text = replace_function(
    app_text,
    "build_relationship_evidence",
    new_relationship_builder
)

print(
    "✅ build_relationship_evidence() patched"
)


# ============================================================
# 4. WRITE FILE
# ============================================================

APP_PATH.write_text(
    app_text,
    encoding="utf-8"
)

print(
    "\n✅ Patched app.py written"
)


# ============================================================
# 5. SYNTAX CHECK
# ============================================================

print("\n")
print("=" * 70)
print("SYNTAX CHECK")
print("=" * 70)

compile(
    app_text,
    str(APP_PATH),
    "exec"
)

print(
    "✅ app.py syntax is valid"
)


# ============================================================
# 6. FUNCTION CHECK
# ============================================================

tree = ast.parse(
    app_text
)

function_names = {

    node.name

    for node in ast.walk(tree)

    if isinstance(
        node,
        (
            ast.FunctionDef,
            ast.AsyncFunctionDef
        )
    )

}


required = {

    "semantic_search",
    "relationship_to_evidence",
    "build_relationship_evidence",
    "retrieve_evidence_api",
    "standard_to_evidence",
    "build_lookup_maps"

}


print("\n")
print("=" * 70)
print("FUNCTION CHECK")
print("=" * 70)

for name in sorted(
    required
):

    print(
        f"{name:<35}"
        f"{'✅' if name in function_names else '❌'}"
    )


if not required.issubset(
    function_names
):

    raise RuntimeError(
        "Required retrieval functions are missing."
    )


print("\n")
print("=" * 70)
print("CELL 21B COMPLETE")
print("=" * 70)

print(
    "\n✅ Actual backend retrieval functions patched."
)

print(
    "\nNext: rerun Cell 20 unchanged."
)
```

```python
# ============================================================
# SAATHI — CELL 20
# POST-PATCH BACKEND RETRIEVAL PARITY TEST
# ============================================================

import sys
import json
import importlib.util
from pathlib import Path


print("=" * 70)
print("SAATHI — CELL 20")
print("POST-PATCH BACKEND RETRIEVAL PARITY TEST")
print("=" * 70)


# ============================================================
# 1. LOAD PATCHED BACKEND
# ============================================================

BACKEND_DIR = Path(
    "/kaggle/working/saathi_backend"
)

APP_PATH = (
    BACKEND_DIR /
    "app.py"
)


print("\nLoading patched app.py...")


spec = importlib.util.spec_from_file_location(
    "saathi_backend_app_v2",
    APP_PATH
)


backend_app = importlib.util.module_from_spec(
    spec
)


sys.modules[
    "saathi_backend_app_v2"
] = backend_app


spec.loader.exec_module(
    backend_app
)


print(
    "✅ Patched app.py imported"
)


# ============================================================
# 2. LOAD BACKEND ARTIFACTS
# ============================================================

print("\n")
print("=" * 70)
print("LOADING BACKEND ARTIFACTS")
print("=" * 70)


backend_app.load_artifacts()

backend_app.build_lookup_maps()


print(
    "FAISS vectors:",
    backend_app.faiss_index.ntotal
)


print(
    "Corpus documents:",
    len(
        backend_app.corpus
    )
)


print(
    "Standard lookup entries:",
    len(
        backend_app.standard_by_is_number
    )
)


print(
    "Reverse-reference targets:",
    len(
        backend_app.referred_by_standard
    )
)


print(
    "Outgoing-reference standards:",
    len(
        backend_app.references_by_standard
    )
)


# ============================================================
# 3. SUMMARY HELPER
# ============================================================

def summarize(
    result
):

    evidence = result.get(
        "evidence",
        []
    )


    return {

        "identifiers":
            result.get(
                "identifiers",
                []
            ),

        "compound":
            result.get(
                "compound",
                False
            ),

        "relationship_direction":
            result.get(
                "relationship_direction"
            ),

        "retrieval_plan":
            result.get(
                "retrieval_plan",
                []
            ),

        "evidence_count":
            result.get(
                "evidence_count",
                len(evidence)
            ),

        "evidence_identifiers":
            [

                item.get(
                    "identifier"
                )

                for item in evidence[:15]

            ],

        "evidence_types":
            [

                item.get(
                    "evidence_type"
                )

                for item in evidence[:15]

            ]

    }


# ============================================================
# 4. TEST DEFINITIONS
# ============================================================

tests = [

    {

        "name":
            "EXACT STANDARD",

        "query":
            "What is IS/IEC 60691:2023?"

    },

    {

        "name":
            "SEMANTIC",

        "query":
            "Which BIS standard deals with thermal links?"

    },

    {

        "name":
            "REVERSE REFERENCE",

        "query":
            "Which BIS standards refer to IS/IEC 61196-1:2005?"

    },

    {

        "name":
            "COMPOUND",

        "query":
            "Tell me about IS/IEC 60691:2023 and its references."

    }

]


# ============================================================
# 5. EXPECTED VALUES FROM VALIDATED PIPELINE
# ============================================================

expected = {

    "EXACT STANDARD": {

        "identifiers":
            [
                "IS/IEC 60691:2023"
            ],

        "compound":
            False,

        "relationship_direction":
            None,

        "evidence_count":
            1

    },

    "SEMANTIC": {

        "identifiers":
            [],

        "compound":
            False,

        "relationship_direction":
            None,

        "evidence_count":
            5

    },

    "REVERSE REFERENCE": {

        "identifiers":
            [
                "IS/IEC 61196-1:2005"
            ],

        "compound":
            False,

        "relationship_direction":
            "reverse",

        "evidence_count":
            43

    },

    "COMPOUND": {

        "identifiers":
            [
                "IS/IEC 60691:2023"
            ],

        "compound":
            True,

        "relationship_direction":
            "outgoing",

        "evidence_count":
            12

    }

}


# ============================================================
# 6. RUN TESTS
# ============================================================

results = []


for test in tests:

    name = test[
        "name"
    ]

    query = test[
        "query"
    ]


    print("\n")
    print("=" * 70)

    print(name)

    print("=" * 70)

    print(
        "Query:",
        query
    )


    # --------------------------------------------------------
    # Reference
    # --------------------------------------------------------

    reference = retrieve_evidence(

        query,

        semantic_fallback=True

    )


    reference_summary = summarize(
        reference
    )


    # --------------------------------------------------------
    # Backend
    # --------------------------------------------------------

    backend = (
        backend_app.retrieve_evidence_api(
            query
        )
    )


    backend_summary = summarize(
        backend
    )


    # --------------------------------------------------------
    # Expected
    # --------------------------------------------------------

    exp = expected[
        name
    ]


    checks = {

        "identifiers":
            backend_summary[
                "identifiers"
            ] == exp[
                "identifiers"
            ],

        "compound":
            backend_summary[
                "compound"
            ] == exp[
                "compound"
            ],

        "relationship_direction":
            backend_summary[
                "relationship_direction"
            ] == exp[
                "relationship_direction"
            ],

        "evidence_count":
            backend_summary[
                "evidence_count"
            ] == exp[
                "evidence_count"
            ]

    }


    # --------------------------------------------------------
    # Print reference
    # --------------------------------------------------------

    print("\n[REFERENCE]")

    print(

        json.dumps(

            reference_summary,

            indent=2,

            ensure_ascii=False

        )

    )


    # --------------------------------------------------------
    # Print backend
    # --------------------------------------------------------

    print("\n[BACKEND]")

    print(

        json.dumps(

            backend_summary,

            indent=2,

            ensure_ascii=False

        )

    )


    # --------------------------------------------------------
    # Checks
    # --------------------------------------------------------

    passed = all(
        checks.values()
    )


    print("\nCHECKS")


    for check, value in checks.items():

        print(

            f"  {check:<30}"
            f"{'✅' if value else '❌'}"

        )


    print(

        "\nTEST RESULT:",

        "✅ PASS"
        if passed
        else
        "❌ FAIL"

    )


    results.append({

        "name":
            name,

        "query":
            query,

        "reference":
            reference_summary,

        "backend":
            backend_summary,

        "expected":
            exp,

        "checks":
            checks,

        "passed":
            passed

    })


# ============================================================
# 7. SUMMARY
# ============================================================

passed_count = sum(

    1

    for result in results

    if result["passed"]

)


print("\n")
print("=" * 70)
print("POST-PATCH PARITY SUMMARY")
print("=" * 70)


print(

    f"\nPassed: "
    f"{passed_count}/{len(results)}"

)


for result in results:

    print(

        f"  {result['name']:<25}"
        f"{'✅ PASS' if result['passed'] else '❌ FAIL'}"

    )


# ============================================================
# 8. SAVE
# ============================================================

output_path = (
    Path("/kaggle/working")
    /
    "saathi_cell20_post_patch_parity.json"
)


with open(
    output_path,
    "w",
    encoding="utf-8"
) as f:

    json.dump(

        results,

        f,

        indent=2,

        ensure_ascii=False

    )


# ============================================================
# 9. FINAL
# ============================================================

print("\n")
print("=" * 70)
print("CELL 20 COMPLETE")
print("=" * 70)


print(
    f"\nSaved:\n{output_path}"
)


if passed_count == len(results):

    print(
        "\n🎉 ALL BACKEND RETRIEVAL PARITY TESTS PASSED"
    )

    print(
        "\nThe standalone backend now matches "
        "the validated retrieval contract."
    )

else:

    print(
        "\n⚠️ PARITY TESTS STILL HAVE FAILURES"
    )

    print(
        "\nDo NOT connect the frontend yet."
    )
```

```python
# ============================================================
# SAATHI — CELL 21B
# PATCH ACTUAL BACKEND RETRIEVAL FUNCTIONS
# ============================================================

from pathlib import Path
import ast

APP_PATH = Path(
    "/kaggle/working/saathi_backend/app.py"
)

app_text = APP_PATH.read_text(
    encoding="utf-8"
)

print("=" * 70)
print("SAATHI — CELL 21B")
print("PATCHING ACTUAL BACKEND RETRIEVAL FUNCTIONS")
print("=" * 70)


# ============================================================
# SAFE FUNCTION REPLACER
# ============================================================

def replace_function(
    source,
    function_name,
    new_function
):

    tree = ast.parse(
        source
    )

    target = None

    for node in tree.body:

        if isinstance(
            node,
            (
                ast.FunctionDef,
                ast.AsyncFunctionDef
            )
        ):

            if node.name == function_name:

                target = node
                break

    if target is None:

        raise RuntimeError(
            f"Could not find top-level function: "
            f"{function_name}"
        )

    lines = source.splitlines(
        keepends=True
    )

    start_line = (
        target.lineno - 1
    )

    end_line = (
        target.end_lineno
    )

    replacement = (
        new_function.strip()
        + "\n\n"
    )

    new_source = (
        "".join(
            lines[:start_line]
        )
        + replacement
        + "".join(
            lines[end_line:]
        )
    )

    # Validate immediately.
    compile(
        new_source,
        str(APP_PATH),
        "exec"
    )

    return new_source


# ============================================================
# 1. REPLACE semantic_search()
# ============================================================

new_semantic_search = r'''
def semantic_search(
    query,
    top_k=5
):
    """
    Semantic retrieval using the existing FAISS index.

    The validated notebook exposes exactly the top 5 semantic
    evidence items after converting raw corpus documents into
    standard evidence records.
    """

    if faiss_index is None:
        return []

    if embedding_model is None:
        return []

    query_vector = embedding_model.encode(
        [query],
        normalize_embeddings=True,
        convert_to_numpy=True
    )

    query_vector = query_vector.astype(
        "float32"
    )

    scores, indices = faiss_index.search(
        query_vector,
        top_k
    )

    results = []

    seen = set()

    for score, index in zip(
        scores[0],
        indices[0]
    ):

        index = int(index)

        if index < 0:
            continue

        if index >= len(corpus):
            continue

        doc = corpus[
            index
        ]

        evidence = standard_to_evidence(
            doc,
            source="semantic",
            score=float(score)
        )

        if not evidence:
            continue

        identifier = (
            evidence.get(
                "identifier"
            )
            or ""
        )

        title = (
            evidence.get(
                "title"
            )
            or ""
        )

        # Deduplicate semantic results by standard identity.
        key = (
            identifier,
            title
        )

        if key in seen:
            continue

        seen.add(
            key
        )

        results.append(
            evidence
        )

    return results[:top_k]
'''


app_text = replace_function(
    app_text,
    "semantic_search",
    new_semantic_search
)

print(
    "✅ semantic_search() patched"
)


# ============================================================
# 2. REPLACE relationship_to_evidence()
# ============================================================

new_relationship_to_evidence = r'''
def relationship_to_evidence(
    doc,
    relationship,
    source_identifier=None
):
    """
    Convert one relationship corpus document into structured
    evidence.

    identifier:
        the BIS standard owning the relationship row

    referenced_identifier:
        the actual referenced standard, when available
    """

    content = doc.get(
        "content",
        {}
    )

    if not isinstance(
        content,
        dict
    ):
        content = {}

    source_identifier = normalize_identifier(
        source_identifier
        or
        content.get(
            "source_identifier"
        )
        or
        ""
    )

    # --------------------------------------------------------
    # Reference identifier
    # --------------------------------------------------------

    reference_identifier = ""

    for field in (
        "reference_is_number",
        "reference_number",
        "international_standard"
    ):

        value = content.get(
            field
        )

        if value is None:
            continue

        value = str(
            value
        ).strip()

        if not value:
            continue

        normalized = normalize_identifier(
            value
        )

        if normalized:

            reference_identifier = (
                normalized
            )

            break

    # --------------------------------------------------------
    # Reference title
    # --------------------------------------------------------

    reference_title = str(
        content.get(
            "reference_title"
        )
        or
        ""
    ).strip()

    if not reference_title:

        reference_title = str(
            content.get(
                "international_standard"
            )
            or
            ""
        ).strip()

    return {

        "evidence_type":
            "relationship",

        "relationship":
            relationship,

        "identifier":
            source_identifier,

        "referenced_identifier":
            reference_identifier,

        "title":
            reference_title,

        "internal_id":
            str(
                content.get(
                    "internal_id"
                )
                or
                ""
            ).strip(),

        "content":
            content

    }
'''


app_text = replace_function(
    app_text,
    "relationship_to_evidence",
    new_relationship_to_evidence
)

print(
    "✅ relationship_to_evidence() patched"
)


# ============================================================
# 3. REPLACE build_relationship_evidence()
# ============================================================

new_relationship_builder = r'''
def build_relationship_evidence(
    identifier,
    direction
):
    """
    Build relationship evidence from the lookup maps.

    outgoing:
        queried standard -> standards it references

    reverse:
        standards -> queried standard
    """

    identifier = normalize_identifier(
        identifier
    )

    evidence = []

    # ========================================================
    # OUTGOING
    # ========================================================

    if direction == "outgoing":

        outgoing_rows = (
            references_by_standard.get(
                identifier,
                []
            )
        )

        international_rows = (
            international_refs_by_standard.get(
                identifier,
                []
            )
        )

        for row in outgoing_rows:

            evidence.append(
                relationship_to_evidence(
                    row,
                    "outgoing",
                    source_identifier=identifier
                )
            )

        for row in international_rows:

            evidence.append(
                relationship_to_evidence(
                    row,
                    "outgoing",
                    source_identifier=identifier
                )
            )

        return evidence

    # ========================================================
    # REVERSE
    # ========================================================

    if direction == "reverse":

        rows = (
            referred_by_standard.get(
                identifier,
                []
            )
        )

        for row in rows:

            content = row.get(
                "content",
                {}
            )

            if not isinstance(
                content,
                dict
            ):
                content = {}

            source_internal_id = str(
                content.get(
                    "internal_id"
                )
                or
                ""
            ).strip()

            source_standard = (
                standard_core_by_internal_id.get(
                    source_internal_id
                )
            )

            source_identifier = ""

            if source_standard:

                source_content = (
                    source_standard.get(
                        "content",
                        {}
                    )
                )

                if isinstance(
                    source_content,
                    dict
                ):

                    source_identifier = (
                        normalize_identifier(
                            source_content.get(
                                "is_number"
                            )
                            or
                            ""
                        )
                    )

            evidence.append(
                relationship_to_evidence(
                    row,
                    "reverse",
                    source_identifier=source_identifier
                )
            )

        return evidence

    return []
'''


app_text = replace_function(
    app_text,
    "build_relationship_evidence",
    new_relationship_builder
)

print(
    "✅ build_relationship_evidence() patched"
)


# ============================================================
# 4. WRITE FILE
# ============================================================

APP_PATH.write_text(
    app_text,
    encoding="utf-8"
)

print(
    "\n✅ Patched app.py written"
)


# ============================================================
# 5. SYNTAX CHECK
# ============================================================

print("\n")
print("=" * 70)
print("SYNTAX CHECK")
print("=" * 70)

compile(
    app_text,
    str(APP_PATH),
    "exec"
)

print(
    "✅ app.py syntax is valid"
)


# ============================================================
# 6. FUNCTION CHECK
# ============================================================

tree = ast.parse(
    app_text
)

function_names = {

    node.name

    for node in ast.walk(tree)

    if isinstance(
        node,
        (
            ast.FunctionDef,
            ast.AsyncFunctionDef
        )
    )

}


required = {

    "semantic_search",
    "relationship_to_evidence",
    "build_relationship_evidence",
    "retrieve_evidence_api",
    "standard_to_evidence",
    "build_lookup_maps"

}


print("\n")
print("=" * 70)
print("FUNCTION CHECK")
print("=" * 70)

for name in sorted(
    required
):

    print(
        f"{name:<35}"
        f"{'✅' if name in function_names else '❌'}"
    )


if not required.issubset(
    function_names
):

    raise RuntimeError(
        "Required retrieval functions are missing."
    )


print("\n")
print("=" * 70)
print("CELL 21B COMPLETE")
print("=" * 70)

print(
    "\n✅ Actual backend retrieval functions patched."
)

print(
    "\nNext: rerun Cell 20 unchanged."
)
```

```python
# ============================================================
# SAATHI — CELL 21C
# FIX SEMANTIC SEARCH / STANDARD EVIDENCE SIGNATURE
# ============================================================

from pathlib import Path
import ast

APP_PATH = Path(
    "/kaggle/working/saathi_backend/app.py"
)

app_text = APP_PATH.read_text(
    encoding="utf-8"
)

print("=" * 70)
print("SAATHI — CELL 21C")
print("FIXING SEMANTIC SEARCH SCORE ARGUMENT")
print("=" * 70)


# ============================================================
# 1. FIND semantic_search()
# ============================================================

tree = ast.parse(
    app_text
)

semantic_func = None

for node in ast.walk(tree):

    if (
        isinstance(node, ast.FunctionDef)
        and node.name == "semantic_search"
    ):

        semantic_func = node
        break


if semantic_func is None:

    raise RuntimeError(
        "semantic_search() not found"
    )


# ============================================================
# 2. EXTRACT FUNCTION
# ============================================================

lines = app_text.splitlines(
    keepends=True
)

start_line = (
    semantic_func.lineno - 1
)

end_line = (
    semantic_func.end_lineno
)

semantic_source = "".join(
    lines[
        start_line:end_line
    ]
)


print(
    "\nCurrent semantic_search() located:"
)

print(
    f"Lines {semantic_func.lineno}-"
    f"{semantic_func.end_lineno}"
)


# ============================================================
# 3. REMOVE score=float(score)
# ============================================================

old_multiline = '''standard_to_evidence(
            doc,
            source="semantic",
            score=float(score)
        )'''

new_multiline = '''standard_to_evidence(
            doc,
            source="semantic"
        )'''


if old_multiline in semantic_source:

    semantic_source = semantic_source.replace(
        old_multiline,
        new_multiline
    )

    print(
        "✅ Removed unsupported score=float(score)"
    )

else:

    # Handle possible formatting variation.
    old_compact = (
        'standard_to_evidence('
        'doc, '
        'source="semantic", '
        'score=float(score)'
        ')'
    )

    new_compact = (
        'standard_to_evidence('
        'doc, '
        'source="semantic"'
        ')'
    )

    if old_compact in semantic_source:

        semantic_source = semantic_source.replace(
            old_compact,
            new_compact
        )

        print(
            "✅ Removed unsupported score=float(score)"
        )

    else:

        if "score=float(score)" in semantic_source:

            semantic_source = semantic_source.replace(
                ",\n            score=float(score)",
                ""
            )

            semantic_source = semantic_source.replace(
                ", score=float(score)",
                ""
            )

            print(
                "✅ Removed unsupported score=float(score)"
            )

        else:

            print(
                "ℹ️ score=float(score) was already absent"
            )


# ============================================================
# 4. REBUILD app.py
# ============================================================

new_app_text = (

    "".join(
        lines[:start_line]
    )

    + semantic_source

    + "".join(
        lines[end_line:]
    )

)


# ============================================================
# 5. SYNTAX CHECK
# ============================================================

compile(
    new_app_text,
    str(APP_PATH),
    "exec"
)


APP_PATH.write_text(
    new_app_text,
    encoding="utf-8"
)


print(
    "✅ app.py syntax is valid"
)


# ============================================================
# 6. VERIFY NO UNSUPPORTED SCORE ARGUMENT
# ============================================================

tree = ast.parse(
    new_app_text
)

semantic_func = None

for node in ast.walk(tree):

    if (
        isinstance(node, ast.FunctionDef)
        and node.name == "semantic_search"
    ):

        semantic_func = node
        break


semantic_source_check = "\n".join(
    new_app_text.splitlines()[
        semantic_func.lineno - 1:
        semantic_func.end_lineno
    ]
)


if "score=float(score)" in semantic_source_check:

    raise RuntimeError(
        "score=float(score) is still present "
        "inside semantic_search()"
    )


print(
    "✅ semantic_search() no longer passes score="
)


# ============================================================
# 7. FINAL CHECK
# ============================================================

print("\n")
print("=" * 70)
print("CELL 21C COMPLETE")
print("=" * 70)

print(
    "\n✅ Semantic score-argument mismatch fixed."
)

print(
    "\nNext step: RERUN CELL 20 UNCHANGED."
)
```

```python
# ============================================================
# SAATHI — CELL 21D
# FIX RELATIONSHIP BUILDER SIGNATURE
# ============================================================

from pathlib import Path
import ast

APP_PATH = Path(
    "/kaggle/working/saathi_backend/app.py"
)

app_text = APP_PATH.read_text(
    encoding="utf-8"
)

print("=" * 70)
print("SAATHI — CELL 21D")
print("FIXING RELATIONSHIP BUILDER SIGNATURE")
print("=" * 70)


# ============================================================
# SAFE FUNCTION REPLACER
# ============================================================

def replace_function(
    source,
    function_name,
    new_function
):

    tree = ast.parse(
        source
    )

    target = None

    for node in tree.body:

        if isinstance(
            node,
            (
                ast.FunctionDef,
                ast.AsyncFunctionDef
            )
        ):

            if node.name == function_name:

                target = node
                break

    if target is None:

        raise RuntimeError(
            f"Could not find function: {function_name}"
        )

    lines = source.splitlines(
        keepends=True
    )

    start_line = (
        target.lineno - 1
    )

    end_line = (
        target.end_lineno
    )

    replacement = (
        new_function.strip()
        + "\n\n"
    )

    new_source = (
        "".join(
            lines[:start_line]
        )
        + replacement
        + "".join(
            lines[end_line:]
        )
    )

    compile(
        new_source,
        str(APP_PATH),
        "exec"
    )

    return new_source


# ============================================================
# REPLACE build_relationship_evidence()
# ============================================================

new_relationship_builder = r'''
def build_relationship_evidence(
    identifier,
    direction,
    max_results=50
):
    """
    Build relationship evidence.

    Signature intentionally accepts max_results because
    retrieve_evidence_api() already passes a third argument.

    outgoing:
        queried standard -> standards it references

    reverse:
        standards -> queried standard
    """

    identifier = normalize_identifier(
        identifier
    )

    evidence = []

    # ========================================================
    # OUTGOING
    # ========================================================

    if direction == "outgoing":

        outgoing_rows = (
            references_by_standard.get(
                identifier,
                []
            )
        )

        international_rows = (
            international_refs_by_standard.get(
                identifier,
                []
            )
        )

        combined_rows = (
            list(outgoing_rows)
            +
            list(international_rows)
        )

        for row in combined_rows:

            evidence.append(
                relationship_to_evidence(
                    row,
                    "outgoing",
                    source_identifier=identifier
                )
            )

            if len(evidence) >= max_results:

                break

        return evidence


    # ========================================================
    # REVERSE
    # ========================================================

    if direction == "reverse":

        rows = (
            referred_by_standard.get(
                identifier,
                []
            )
        )

        for row in rows:

            content = row.get(
                "content",
                {}
            )

            if not isinstance(
                content,
                dict
            ):
                content = {}

            source_internal_id = str(
                content.get(
                    "internal_id"
                )
                or
                ""
            ).strip()

            source_standard = (
                standard_core_by_internal_id.get(
                    source_internal_id
                )
            )

            source_identifier = ""

            if source_standard:

                source_content = (
                    source_standard.get(
                        "content",
                        {}
                    )
                )

                if isinstance(
                    source_content,
                    dict
                ):

                    source_identifier = (
                        normalize_identifier(
                            source_content.get(
                                "is_number"
                            )
                            or
                            ""
                        )
                    )

            evidence.append(
                relationship_to_evidence(
                    row,
                    "reverse",
                    source_identifier=source_identifier
                )
            )

            if len(evidence) >= max_results:

                break

        return evidence


    return []
'''


app_text = replace_function(
    app_text,
    "build_relationship_evidence",
    new_relationship_builder
)


print(
    "✅ build_relationship_evidence() signature patched"
)


# ============================================================
# WRITE FILE
# ============================================================

APP_PATH.write_text(
    app_text,
    encoding="utf-8"
)


# ============================================================
# SYNTAX CHECK
# ============================================================

compile(
    app_text,
    str(APP_PATH),
    "exec"
)

print(
    "✅ app.py syntax is valid"
)


# ============================================================
# INSPECT SIGNATURE
# ============================================================

tree = ast.parse(
    app_text
)

target = None

for node in ast.walk(tree):

    if (
        isinstance(node, ast.FunctionDef)
        and node.name == "build_relationship_evidence"
    ):

        target = node
        break


if target is None:

    raise RuntimeError(
        "build_relationship_evidence() missing"
    )


signature = [
    arg.arg
    for arg in target.args.args
]


print("\n")
print("=" * 70)
print("FUNCTION SIGNATURE")
print("=" * 70)

print(
    "build_relationship_evidence("
    + ", ".join(signature)
    + ")"
)


if signature != [
    "identifier",
    "direction",
    "max_results"
]:

    raise RuntimeError(
        "Unexpected relationship builder signature."
    )


print(
    "✅ Signature matches retrieve_evidence_api()"
)


print("\n")
print("=" * 70)
print("CELL 21D COMPLETE")
print("=" * 70)

print(
    "\n✅ Relationship builder signature fixed."
)

print(
    "\nNext: rerun CELL 20 unchanged."
)
```

```python
# ============================================================
# SAATHI — CELL 21E
# INSPECT ACTUAL RELATIONSHIP CORPUS STRUCTURE
# ============================================================

import json
from pathlib import Path


print("=" * 70)
print("SAATHI — CELL 21E")
print("INSPECTING RELATIONSHIP CORPUS STRUCTURE")
print("=" * 70)


# ============================================================
# 1. LOAD BACKEND
# ============================================================

import sys
import importlib.util


BACKEND_DIR = Path(
    "/kaggle/working/saathi_backend"
)

APP_PATH = (
    BACKEND_DIR /
    "app.py"
)


spec = importlib.util.spec_from_file_location(
    "saathi_backend_inspect",
    APP_PATH
)

inspect_app = importlib.util.module_from_spec(
    spec
)

sys.modules[
    "saathi_backend_inspect"
] = inspect_app

spec.loader.exec_module(
    inspect_app
)


inspect_app.load_artifacts()


print(
    "\nCorpus loaded:",
    len(inspect_app.corpus)
)


# ============================================================
# 2. FIND ONE DOCUMENT OF EACH RELATIONSHIP TYPE
# ============================================================

relationship_types = {

    "cross_references": None,

    "indian_references": None,

    "international_references": None,

    "referred_by": None

}


for doc in inspect_app.corpus:

    content = doc.get(
        "content",
        {}
    )

    if not isinstance(
        content,
        dict
    ):
        continue

    document_type = (
        content.get(
            "document_type"
        )
        or
        doc.get(
            "document_type"
        )
        or
        ""
    )

    document_type = str(
        document_type
    ).strip().lower()


    if document_type in relationship_types:

        if relationship_types[
            document_type
        ] is None:

            relationship_types[
                document_type
            ] = doc


    if all(
        value is not None
        for value in relationship_types.values()
    ):

        break


# ============================================================
# 3. PRINT STRUCTURE
# ============================================================

for document_type, doc in relationship_types.items():

    print("\n")
    print("=" * 70)

    print(
        document_type.upper()
    )

    print("=" * 70)


    if doc is None:

        print(
            "❌ NO DOCUMENT FOUND"
        )

        continue


    print(
        "\nTOP-LEVEL KEYS:"
    )

    print(
        list(
            doc.keys()
        )
    )


    content = doc.get(
        "content",
        {}
    )


    print(
        "\nCONTENT KEYS:"
    )

    if isinstance(
        content,
        dict
    ):

        print(
            list(
                content.keys()
            )
        )

    else:

        print(
            "CONTENT IS NOT A DICT:"
        )

        print(
            type(content)
        )


    print(
        "\nFULL SAMPLE DOCUMENT:"
    )

    print(
        json.dumps(
            doc,
            indent=2,
            ensure_ascii=False,
            default=str
        )[
            :12000
        ]
    )


# ============================================================
# 4. SPECIFIC TARGET: IS/IEC 61196-1:2005
# ============================================================

print("\n")
print("=" * 70)
print("TARGET STANDARD INSPECTION")
print("=" * 70)


TARGET = "IS/IEC 61196-1:2005"


def normalize_for_inspection(value):

    if value is None:
        return ""

    return str(
        value
    ).strip().upper().replace(
        " ",
        ""
    )


target_hits = []


for doc in inspect_app.corpus:

    content = doc.get(
        "content",
        {}
    )

    if not isinstance(
        content,
        dict
    ):
        continue


    raw_values = [

        content.get(
            "reference_number"
        ),

        content.get(
            "reference_is_number"
        ),

        content.get(
            "international_standard"
        ),

        content.get(
            "text"
        )

    ]


    joined = " ".join(
        str(value)
        for value in raw_values
        if value is not None
    )


    if (
        normalize_for_inspection(
            TARGET
        )
        in
        normalize_for_inspection(
            joined
        )
    ):

        target_hits.append(
            doc
        )


print(
    "\nDocuments containing target:"
)

print(
    len(target_hits)
)


for i, doc in enumerate(
    target_hits[:5],
    start=1
):

    print("\n")
    print(
        f"--- TARGET HIT {i} ---"
    )

    print(
        json.dumps(
            doc,
            indent=2,
            ensure_ascii=False,
            default=str
        )[
            :10000
        ]
    )


# ============================================================
# 5. COUNT DOCUMENT TYPES
# ============================================================

print("\n")
print("=" * 70)
print("CORPUS DOCUMENT TYPE COUNTS")
print("=" * 70)


type_counts = {}


for doc in inspect_app.corpus:

    content = doc.get(
        "content",
        {}
    )

    if not isinstance(
        content,
        dict
    ):
        continue

    document_type = (
        content.get(
            "document_type"
        )
        or
        doc.get(
            "document_type"
        )
        or
        "UNKNOWN"
    )

    document_type = str(
        document_type
    ).strip()


    type_counts[
        document_type
    ] = (
        type_counts.get(
            document_type,
            0
        )
        + 1
    )


for document_type, count in sorted(
    type_counts.items(),
    key=lambda x: x[0]
):

    print(
        f"{document_type:<35}"
        f"{count:>10,}"
    )


# ============================================================
# 6. CURRENT MAP STATUS
# ============================================================

print("\n")
print("=" * 70)
print("CURRENT BACKEND MAP STATUS")
print("=" * 70)


inspect_app.build_lookup_maps()


print(
    "standard_by_is_number:",
    len(
        inspect_app.standard_by_is_number
    )
)

print(
    "references_by_standard:",
    len(
        inspect_app.references_by_standard
    )
)

print(
    "international_refs_by_standard:",
    len(
        inspect_app.international_refs_by_standard
    )
)

print(
    "referred_by_standard:",
    len(
        inspect_app.referred_by_standard
    )
)


# ============================================================
# 7. FINAL
# ============================================================

print("\n")
print("=" * 70)
print("CELL 21E COMPLETE")
print("=" * 70)

print(
    "\nDo NOT patch anything yet."
)

print(
    "The samples above will tell us exactly "
    "how the relationship fields are structured."
)
```

```python
# ============================================================
# SAATHI — CELL 21G
# RELATIONSHIP TEXT PARSER + RETRIEVAL PARITY PATCH
# ============================================================

from pathlib import Path
import re
import sys
import importlib.util
import inspect


print("=" * 70)
print("SAATHI — CELL 21G")
print("RELATIONSHIP TEXT PARSER + RETRIEVAL PARITY PATCH")
print("=" * 70)


# ============================================================
# 1. LOAD BACKEND
# ============================================================

BACKEND_DIR = Path(
    "/kaggle/working/saathi_backend"
)

APP_PATH = (
    BACKEND_DIR /
    "app.py"
)


spec = importlib.util.spec_from_file_location(
    "saathi_backend_21g",
    APP_PATH
)

app = importlib.util.module_from_spec(
    spec
)

sys.modules[
    "saathi_backend_21g"
] = app

spec.loader.exec_module(app)


app.load_artifacts()


print(
    "\nCorpus loaded:",
    len(app.corpus)
)


# ============================================================
# 2. CREATE PATCH
# ============================================================

PATCH = r'''

# ============================================================
# CELL 21G PATCH
# Relationship documents store rows inside `text`.
# Parse those rows into structured relationship records.
# ============================================================

import re


def _parse_relationship_text(doc):
    """
    Parse one BIS relationship corpus document.

    Actual corpus structure:

        {
            "doc_id": "...",
            "internal_id": "...",
            "document_type": "...",
            "text": "..."
        }

    Relationship rows are embedded in text as:

        Reference Number: 1 | Referenced IS: ... |
        Referenced Title: ... | Reference Committee: ...

    or:

        Reference Number: 1 | Referencing IS: ... |
        Referencing Title: ... | Reference Committee: ...

    or:

        Reference Number: 1 | International Standard: ...
    """

    if not isinstance(doc, dict):
        return []

    document_type = str(
        doc.get("document_type", "")
        or ""
    ).strip().lower()

    internal_id = str(
        doc.get("internal_id", "")
        or ""
    ).strip()

    text = str(
        doc.get("text", "")
        or ""
    )

    if not text:
        return []

    # Only relationship documents.
    valid_types = {
        "cross_references",
        "indian_references",
        "international_references",
        "referred_by",
    }

    if document_type not in valid_types:
        return []

    # Every relationship row starts with:
    #
    # Reference Number: N |
    #
    # Split while retaining each row.
    matches = list(
        re.finditer(
            r"Reference Number:\s*(\d+)\s*\|",
            text,
            flags=re.IGNORECASE
        )
    )

    if not matches:
        return []

    rows = []

    for index, match in enumerate(matches):

        number = match.group(1)

        start = match.start()

        if index + 1 < len(matches):
            end = matches[index + 1].start()
        else:
            end = len(text)

        row_text = text[start:end].strip()

        # Remove the "Reference Number" prefix.
        row_text = re.sub(
            r"^Reference Number:\s*\d+\s*\|\s*",
            "",
            row_text,
            flags=re.IGNORECASE
        )

        row = {
            "internal_id": internal_id,
            "document_type": document_type,
            "reference_number": number,
            "raw_text": row_text,
        }

        # ----------------------------------------------------
        # CROSS / INDIAN REFERENCES
        # ----------------------------------------------------

        if document_type in {
            "cross_references",
            "indian_references",
        }:

            m = re.search(
                r"Referenced IS:\s*(.*?)\s*\|\s*Referenced Title:",
                row_text,
                flags=re.IGNORECASE | re.DOTALL
            )

            if m:
                row["reference_is_number"] = (
                    m.group(1).strip()
                )
            else:
                row["reference_is_number"] = ""

            m = re.search(
                r"Referenced Title:\s*(.*?)\s*\|\s*Reference Committee:",
                row_text,
                flags=re.IGNORECASE | re.DOTALL
            )

            if m:
                row["reference_title"] = (
                    m.group(1).strip()
                )
            else:
                row["reference_title"] = ""

            m = re.search(
                r"Reference Committee:\s*(.*?)\s*$",
                row_text,
                flags=re.IGNORECASE | re.DOTALL
            )

            if m:
                row["reference_committee"] = (
                    m.group(1).strip()
                )
            else:
                row["reference_committee"] = ""

        # ----------------------------------------------------
        # INTERNATIONAL REFERENCES
        # ----------------------------------------------------

        elif document_type == "international_references":

            m = re.search(
                r"International Standard:\s*(.*?)\s*$",
                row_text,
                flags=re.IGNORECASE | re.DOTALL
            )

            if m:
                row["international_standard"] = (
                    m.group(1).strip()
                )
            else:
                row["international_standard"] = ""

        # ----------------------------------------------------
        # REFERRED BY
        # ----------------------------------------------------

        elif document_type == "referred_by":

            m = re.search(
                r"Referencing IS:\s*(.*?)\s*\|\s*Referencing Title:",
                row_text,
                flags=re.IGNORECASE | re.DOTALL
            )

            if m:
                row["reference_is_number"] = (
                    m.group(1).strip()
                )
            else:
                row["reference_is_number"] = ""

            m = re.search(
                r"Referencing Title:\s*(.*?)\s*\|\s*Reference Committee:",
                row_text,
                flags=re.IGNORECASE | re.DOTALL
            )

            if m:
                row["reference_title"] = (
                    m.group(1).strip()
                )
            else:
                row["reference_title"] = ""

            m = re.search(
                r"Reference Committee:\s*(.*?)\s*$",
                row_text,
                flags=re.IGNORECASE | re.DOTALL
            )

            if m:
                row["reference_committee"] = (
                    m.group(1).strip()
                )
            else:
                row["reference_committee"] = ""

        rows.append(row)

    return rows


# ============================================================
# BUILD RELATIONSHIP MAPS
# ============================================================

def build_lookup_maps():

    global standard_by_is_number
    global standard_by_internal_id

    global references_by_standard
    global international_refs_by_standard
    global referred_by_standard
    global standard_core_by_internal_id

    # --------------------------------------------------------
    # Standard maps
    # --------------------------------------------------------

    standard_by_is_number = {}
    standard_by_internal_id = {}
    standard_core_by_internal_id = {}

    for doc in corpus:

        if doc.get("document_type") != "standard_core":
            continue

        internal_id = str(
            doc.get("internal_id", "")
            or ""
        ).strip()

        identifier = _corpus_identifier(
            doc
        )

        if internal_id:
            standard_core_by_internal_id[
                internal_id
            ] = doc

            standard_by_internal_id[
                internal_id
            ] = doc

        if identifier:
            normalized = normalize_identifier(
                identifier
            )

            if normalized:
                standard_by_is_number[
                    normalized
                ] = doc

    # --------------------------------------------------------
    # Relationship maps
    # --------------------------------------------------------

    references_by_standard = {}
    international_refs_by_standard = {}
    referred_by_standard = {}

    # --------------------------------------------------------
    # Parse relationship documents
    # --------------------------------------------------------

    for doc in corpus:

        document_type = str(
            doc.get("document_type", "")
            or ""
        ).strip().lower()

        if document_type not in {
            "cross_references",
            "indian_references",
            "international_references",
            "referred_by",
        }:
            continue

        internal_id = str(
            doc.get("internal_id", "")
            or ""
        ).strip()

        source_standard = (
            standard_core_by_internal_id.get(
                internal_id
            )
        )

        source_identifier = ""

        if source_standard:

            source_identifier = normalize_identifier(
                _corpus_identifier(
                    source_standard
                )
            )

        if not source_identifier:
            continue

        rows = _parse_relationship_text(
            doc
        )

        # ----------------------------------------------------
        # OUTGOING BIS REFERENCES
        #
        # IMPORTANT:
        # Reference notebook's validated outgoing result
        # uses CROSS_REFERENCES + INTERNATIONAL_REFERENCES.
        # Indian references are not separately appended.
        # ----------------------------------------------------

        if document_type == "cross_references":

            references_by_standard.setdefault(
                source_identifier,
                []
            ).extend(
                rows
            )

        elif document_type == "international_references":

            international_refs_by_standard.setdefault(
                source_identifier,
                []
            ).extend(
                rows
            )

        # ----------------------------------------------------
        # REVERSE REFERENCES
        #
        # A referred_by document belongs to the STANDARD
        # BEING REFERRED TO.
        #
        # Therefore its key is the source standard's identifier.
        # ----------------------------------------------------

        elif document_type == "referred_by":

            referred_by_standard.setdefault(
                source_identifier,
                []
            ).extend(
                rows
            )

    return {
        "standard_by_is_number":
            standard_by_is_number,

        "standard_by_internal_id":
            standard_by_internal_id,

        "references_by_standard":
            references_by_standard,

        "international_refs_by_standard":
            international_refs_by_standard,

        "referred_by_standard":
            referred_by_standard,

        "standard_core_by_internal_id":
            standard_core_by_internal_id,
    }


# ============================================================
# RELATIONSHIP EVIDENCE CONVERTER
# ============================================================

def relationship_to_evidence(
    row,
    relationship,
    source_identifier=None
):
    """
    Convert one parsed relationship row into evidence.
    """

    if not isinstance(row, dict):
        row = {}

    source_identifier = normalize_identifier(
        source_identifier
        or
        ""
    )

    reference_raw = (
        row.get("reference_is_number")
        or
        row.get("international_standard")
        or
        ""
    )

    reference_identifier = normalize_identifier(
        reference_raw
    )

    reference_title = str(
        row.get("reference_title")
        or
        ""
    ).strip()

    # International references have no separate title.
    if not reference_title:
        reference_title = str(
            row.get("international_standard")
            or
            ""
        ).strip()

    return {
        "evidence_type":
            "relationship",

        "relationship":
            relationship,

        "identifier":
            source_identifier,

        "referenced_identifier":
            reference_identifier,

        "title":
            reference_title,

        "internal_id":
            str(
                row.get("internal_id")
                or
                ""
            ).strip(),

        "reference_number":
            str(
                row.get("reference_number")
                or
                ""
            ).strip(),

        "reference_committee":
            str(
                row.get("reference_committee")
                or
                ""
            ).strip(),

        "content":
            row,
    }


# ============================================================
# RELATIONSHIP RETRIEVAL
# ============================================================

def build_relationship_evidence(
    identifier,
    direction,
    max_results=50
):

    identifier = normalize_identifier(
        identifier
    )

    evidence = []

    # ========================================================
    # OUTGOING
    # ========================================================

    if direction == "outgoing":

        outgoing_rows = (
            references_by_standard.get(
                identifier,
                []
            )
        )

        international_rows = (
            international_refs_by_standard.get(
                identifier,
                []
            )
        )

        combined_rows = (
            list(outgoing_rows)
            +
            list(international_rows)
        )

        for row in combined_rows:

            evidence.append(
                relationship_to_evidence(
                    row,
                    "outgoing",
                    source_identifier=identifier
                )
            )

            if len(evidence) >= max_results:
                break

        return evidence

    # ========================================================
    # REVERSE
    # ========================================================

    if direction == "reverse":

        rows = (
            referred_by_standard.get(
                identifier,
                []
            )
        )

        for row in rows:

            # For reverse references, the queried standard
            # is the referenced target. The "source" is the
            # standard doing the referencing.
            #
            # The row itself contains:
            #   Referencing IS: ...
            #
            # so use that as the evidence identifier.

            referencing_identifier = normalize_identifier(
                row.get(
                    "reference_is_number",
                    ""
                )
            )

            evidence.append(
                relationship_to_evidence(
                    row,
                    "reverse",
                    source_identifier=referencing_identifier
                )
            )

            if len(evidence) >= max_results:
                break

        return evidence

    return []


# ============================================================
# SEMANTIC SEARCH PARITY
# ============================================================

def retrieve_evidence_api(
    query,
    max_relationships=50
):

    identifiers = extract_identifiers(
        query
    )

    relationship_direction = (
        detect_relationship_direction(
            query
        )
    )

    evidence = []

    retrieval_plan = []

    # --------------------------------------------------------
    # Exact standard
    # --------------------------------------------------------

    if identifiers:

        for identifier in identifiers:

            exact_items = (
                exact_standard_search(
                    identifier
                )
            )

            if exact_items:

                retrieval_plan.append(
                    "standard:"
                    + identifier
                )

                for item in exact_items:

                    evidence.append(
                        standard_to_evidence(
                            item,
                            "exact"
                        )
                    )

                # ------------------------------------------------
                # Relationship-aware retrieval
                # ------------------------------------------------

                if relationship_direction:

                    relationship_items = (
                        build_relationship_evidence(
                            identifier,
                            relationship_direction,
                            max_relationships
                        )
                    )

                    if relationship_items:

                        if (
                            relationship_direction
                            == "reverse"
                        ):

                            retrieval_plan.append(
                                "reverse_references:"
                                + identifier
                            )

                        elif (
                            relationship_direction
                            == "outgoing"
                        ):

                            retrieval_plan.append(
                                "outgoing_references:"
                                + identifier
                            )

                        evidence.extend(
                            relationship_items
                        )

    # --------------------------------------------------------
    # Semantic fallback
    # --------------------------------------------------------

    if not evidence:

        # Validated notebook uses top_k=5.
        semantic_items = semantic_search(
            query,
            top_k=5
        )

        retrieval_plan.append(
            "semantic"
        )

        for item in semantic_items:

            evidence.append(
                standard_to_evidence(
                    item,
                    "semantic"
                )
            )

    return {
        "query":
            query,

        "identifiers":
            identifiers,

        # IMPORTANT:
        # Relationship direction alone does NOT make a query
        # compound.
        #
        # Example:
        # "Which standards refer to X?"
        # -> relationship query, NOT compound.
        #
        # Example:
        # "Tell me about X and its references."
        # -> compound.
        "compound":
            detect_compound_query(
                query
            ),

        "relationship_direction":
            relationship_direction,

        "retrieval_plan":
            retrieval_plan,

        "evidence_count":
            len(evidence),

        "evidence":
            evidence,
    }
'''


# ============================================================
# 3. APPLY PATCH
# ============================================================

source = APP_PATH.read_text(
    encoding="utf-8"
)

# Append the patched definitions at the end.
# Python will resolve the latest function definitions when
# the module is used.
source += "\n\n" + PATCH + "\n"

APP_PATH.write_text(
    source,
    encoding="utf-8"
)


print(
    "\nPatched:",
    APP_PATH
)


# ============================================================
# 4. SYNTAX CHECK
# ============================================================

import py_compile

py_compile.compile(
    str(APP_PATH),
    doraise=True
)

print(
    "✅ Syntax check passed"
)


# ============================================================
# 5. IMPORT PATCHED MODULE
# ============================================================

spec2 = importlib.util.spec_from_file_location(
    "saathi_backend_21g_patched",
    APP_PATH
)

patched = importlib.util.module_from_spec(
    spec2
)

sys.modules[
    "saathi_backend_21g_patched"
] = patched

spec2.loader.exec_module(
    patched
)


patched.load_artifacts()


maps = patched.build_lookup_maps()


# ============================================================
# 6. MAP VALIDATION
# ============================================================

print("\n")
print("=" * 70)
print("RELATIONSHIP MAP STATUS")
print("=" * 70)

print(
    "Standard lookup entries:",
    len(
        patched.standard_by_is_number
    )
)

print(
    "Outgoing reference standards:",
    len(
        patched.references_by_standard
    )
)

print(
    "International reference standards:",
    len(
        patched.international_refs_by_standard
    )
)

print(
    "Reverse-reference targets:",
    len(
        patched.referred_by_standard
    )
)


# ============================================================
# 7. TARGET MAP INSPECTION
# ============================================================

reverse_target = (
    "IS/IEC 61196-1:2005"
)

reverse_key = patched.normalize_identifier(
    reverse_target
)

reverse_rows = (
    patched.referred_by_standard.get(
        reverse_key,
        []
    )
)

print("\n")
print("=" * 70)
print("REVERSE TARGET CHECK")
print("=" * 70)

print(
    "Target:",
    reverse_target
)

print(
    "Normalized:",
    reverse_key
)

print(
    "Rows:",
    len(reverse_rows)
)


for row in reverse_rows[:5]:

    print(
        "\n",
        {
            "reference_number":
                row.get("reference_number"),

            "referencing_is":
                row.get("reference_is_number"),

            "title":
                row.get("reference_title"),
        }
    )


# ============================================================
# 8. OUTGOING TARGET INSPECTION
# ============================================================

outgoing_target = (
    "IS/IEC 60691:2023"
)

outgoing_key = patched.normalize_identifier(
    outgoing_target
)

cross_rows = (
    patched.references_by_standard.get(
        outgoing_key,
        []
    )
)

international_rows = (
    patched.international_refs_by_standard.get(
        outgoing_key,
        []
    )
)

print("\n")
print("=" * 70)
print("OUTGOING TARGET CHECK")
print("=" * 70)

print(
    "Target:",
    outgoing_target
)

print(
    "Cross-reference rows:",
    len(cross_rows)
)

print(
    "International rows:",
    len(international_rows)
)

print(
    "Total outgoing rows:",
    len(cross_rows)
    +
    len(international_rows)
)


# ============================================================
# 9. DIRECT RELATIONSHIP TESTS
# ============================================================

print("\n")
print("=" * 70)
print("DIRECT RELATIONSHIP TESTS")
print("=" * 70)


reverse_evidence = (
    patched.build_relationship_evidence(
        reverse_target,
        "reverse",
        50
    )
)

print(
    "\nReverse evidence:",
    len(reverse_evidence)
)

print(
    "First reverse evidence:"
)

if reverse_evidence:
    print(
        reverse_evidence[0]
    )


outgoing_evidence = (
    patched.build_relationship_evidence(
        outgoing_target,
        "outgoing",
        50
    )
)

print(
    "\nOutgoing evidence:",
    len(outgoing_evidence)
)

print(
    "First outgoing evidence:"
)

if outgoing_evidence:
    print(
        outgoing_evidence[0]
    )


# ============================================================
# 10. FULL RETRIEVAL TESTS
# ============================================================

TESTS = [
    (
        "EXACT",
        "What is IS/IEC 60691:2023?"
    ),
    (
        "SEMANTIC",
        "Which BIS standard deals with thermal links?"
    ),
    (
        "REVERSE",
        "Which BIS standards refer to IS/IEC 61196-1:2005?"
    ),
    (
        "COMPOUND",
        "Tell me about IS/IEC 60691:2023 and its references."
    ),
]


for label, query in TESTS:

    result = patched.retrieve_evidence_api(
        query
    )

    print("\n")
    print("-" * 70)
    print(label)
    print("-" * 70)

    print(
        "Query:",
        query
    )

    print(
        "Identifiers:",
        result["identifiers"]
    )

    print(
        "Compound:",
        result["compound"]
    )

    print(
        "Relationship:",
        result["relationship_direction"]
    )

    print(
        "Plan:",
        result["retrieval_plan"]
    )

    print(
        "Evidence count:",
        result["evidence_count"]
    )

    print(
        "Evidence identifiers:"
    )

    for item in result["evidence"][:15]:

        print(
            "  -",
            item.get(
                "identifier",
                ""
            ),
            "|",
            item.get(
                "referenced_identifier",
                ""
            )
        )


# ============================================================
# 11. FINAL
# ============================================================

print("\n")
print("=" * 70)
print("CELL 21G COMPLETE")
print("=" * 70)

print(
    "\nPatched app.py successfully."
)

print(
    "Do NOT run the FastAPI server yet."
)

print(
    "Next step is to compare these retrieval counts "
    "against the validated notebook."
)
```

```python
# ================================================================
# SAATHI — FINAL AI/ML + RAG PIPELINE
# ================================================================
#
# This is the FINAL consolidated Kaggle implementation.
#
# Implements:
#   1. Corpus loading
#   2. BGE semantic retrieval
#   3. FAISS retrieval
#   4. Robust BIS identifier normalization
#   5. Exact standard retrieval
#   6. Relationship parsing
#   7. Outgoing references
#   8. Reverse references
#   9. Compound query detection
#  10. Hybrid retrieval
#  11. Evidence normalization
#  12. Gemini grounded generation
#  13. Citation validation
#  14. Complete validation suite
#  15. Saved API-ready inference artifacts
#
# DO NOT RUN THE OLD CELL 21A-21G PATCHES AFTER THIS.
# ================================================================

import os
import re
import json
import time
import warnings
from pathlib import Path
from datetime import datetime, timezone
from collections import defaultdict

import numpy as np
import pandas as pd
import faiss
import torch

from sentence_transformers import SentenceTransformer


warnings.filterwarnings("ignore")


# ================================================================
# CONFIGURATION
# ================================================================

WORK_DIR = Path("/kaggle/working")

CORPUS_PATH = WORK_DIR / "bis_rag_corpus.jsonl"
FAISS_PATH = WORK_DIR / "bis_faiss.index"
METADATA_PATH = WORK_DIR / "bis_rag_metadata.json"

EMBEDDING_MODEL_NAME = "BAAI/bge-small-en-v1.5"

GEMINI_MODEL = "gemini-2.5-flash"

SEMANTIC_TOP_K = 5
MAX_RELATIONSHIPS = 50

print("=" * 75)
print("SAATHI — FINAL AI/ML + RAG PIPELINE")
print("=" * 75)


# ================================================================
# 1. VERIFY ARTIFACTS
# ================================================================

required_files = [
    CORPUS_PATH,
    FAISS_PATH,
    METADATA_PATH,
]

for path in required_files:
    if not path.exists():
        raise FileNotFoundError(
            f"Required artifact not found: {path}"
        )

print("\nArtifacts:")
for path in required_files:
    print(f"  ✅ {path.name}")


# ================================================================
# 2. LOAD CORPUS
# ================================================================

print("\nLoading corpus...")

corpus = []

with open(
    CORPUS_PATH,
    "r",
    encoding="utf-8"
) as f:

    for line in f:
        line = line.strip()

        if line:
            corpus.append(
                json.loads(line)
            )

print(
    f"Corpus documents: {len(corpus):,}"
)


# ================================================================
# 3. CORPUS TYPE COUNTS
# ================================================================

document_type_counts = (
    pd.Series(
        [
            str(
                d.get(
                    "document_type",
                    ""
                )
            )
            for d in corpus
        ]
    )
    .value_counts()
)

print("\nCorpus document types:")

for dtype, count in document_type_counts.items():
    print(
        f"  {dtype:<35} {count:,}"
    )


# ================================================================
# 4. LOAD FAISS
# ================================================================

print("\nLoading FAISS index...")

faiss_index = faiss.read_index(
    str(FAISS_PATH)
)

print(
    "FAISS vectors:",
    f"{faiss_index.ntotal:,}"
)


# ================================================================
# 5. LOAD METADATA
# ================================================================

print("\nLoading metadata...")

with open(
    METADATA_PATH,
    "r",
    encoding="utf-8"
) as f:

    metadata = json.load(f)

print(
    "Metadata type:",
    type(metadata).__name__
)


# ================================================================
# 6. LOAD EMBEDDING MODEL
# ================================================================

device = (
    "cuda"
    if torch.cuda.is_available()
    else "cpu"
)

print(
    "\nEmbedding device:",
    device
)

print(
    "Loading:",
    EMBEDDING_MODEL_NAME
)

embedding_model = SentenceTransformer(
    EMBEDDING_MODEL_NAME,
    device=device
)

print(
    "Embedding model loaded."
)


# ================================================================
# 7. NORMALIZATION
# ================================================================

def normalize_identifier(value):
    """
    Normalize BIS identifiers while preserving legitimate
    identifiers and relationship-style identifiers.
    """

    if value is None:
        return ""

    value = str(value).strip()

    if not value:
        return ""

    # Unicode normalization-like cleanup.
    value = (
        value
        .replace("\u2013", "-")
        .replace("\u2014", "-")
        .replace("\u2012", "-")
        .replace("\u00a0", " ")
    )

    # Collapse whitespace.
    value = re.sub(
        r"\s+",
        " ",
        value
    ).strip()

    # ------------------------------------------------------------
    # Fix known BIS concatenation corruption.
    #
    # Example:
    # IS/IEC :2023IEC 60691:2023
    #
    # -> IS/IEC 60691:2023
    # ------------------------------------------------------------

    value = re.sub(
        r"^IS/IEC\s*:\s*(\d{4})\s*IEC\s+(.+)$",
        r"IS/IEC \2",
        value,
        flags=re.IGNORECASE
    )

    # ------------------------------------------------------------
    # Remove duplicate IEC portion when it is appended directly
    # after an Indian/BIS identifier.
    #
    # Example:
    # IS 616:2017IEC 60065 : 2014
    #
    # Keep both identifiers separated.
    # ------------------------------------------------------------

    value = re.sub(
        r"(IS(?:/IEC)?\s+[^|]+?:\s*\d{4})IEC\s+",
        r"\1 — IEC ",
        value,
        flags=re.IGNORECASE
    )

    # ------------------------------------------------------------
    # Standard IS/IEC formatting.
    # ------------------------------------------------------------

    value = re.sub(
        r"^IS\s*/\s*IEC\s*",
        "IS/IEC ",
        value,
        flags=re.IGNORECASE
    )

    # Normalize spaces around colon.
    value = re.sub(
        r"\s*:\s*",
        ":",
        value
    )

    # Normalize spaces around slash.
    value = re.sub(
        r"\s*/\s*",
        "/",
        value
    )

    # Normalize spaces around hyphen for simple numeric parts.
    value = re.sub(
        r"\s*-\s*",
        "-",
        value
    )

    # Preserve semantic separator.
    value = re.sub(
        r"\s*—\s*",
        " — ",
        value
    )

    # Clean repeated spaces.
    value = re.sub(
        r"\s+",
        " ",
        value
    ).strip()

    # ------------------------------------------------------------
    # Handle malformed year-prefix form:
    #
    # IS/IEC :2015IEC 60691 : 2015
    # ------------------------------------------------------------

    m = re.match(
        r"^IS/IEC\s*:\d{4}\s*IEC\s+(.+)$",
        value,
        flags=re.IGNORECASE
    )

    if m:
        value = (
            "IS/IEC "
            + m.group(1).strip()
        )

    # Remove duplicate IEC concatenation.
    value = re.sub(
        r"^(.+?\d{4})IEC\s+(.+)$",
        r"\1 — IEC \2",
        value,
        flags=re.IGNORECASE
    )

    return value.strip()


# ================================================================
# 8. CORPUS HELPERS
# ================================================================

def corpus_identifier(doc):
    """
    Get identifier from a corpus document.

    Standard-core documents contain structured fields.
    Relationship documents generally do not.
    """

    if not isinstance(doc, dict):
        return ""

    # Standard core.
    for key in (
        "is_number",
        "IS_number",
        "catalogue_is_no",
    ):

        value = doc.get(key)

        if value:
            return str(value).strip()

    # Some corpora may store nested content.
    content = doc.get("content")

    if isinstance(content, dict):

        for key in (
            "is_number",
            "IS_number",
            "catalogue_is_no",
        ):

            value = content.get(key)

            if value:
                return str(value).strip()

    return ""


def corpus_internal_id(doc):

    if not isinstance(doc, dict):
        return ""

    return str(
        doc.get(
            "internal_id",
            ""
        )
        or ""
    ).strip()


def corpus_title(doc):

    if not isinstance(doc, dict):
        return ""

    for key in (
        "is_title",
        "IS_title",
        "catalogue_title",
    ):

        value = doc.get(key)

        if value:
            return str(value).strip()

    content = doc.get("content")

    if isinstance(content, dict):

        for key in (
            "is_title",
            "IS_title",
            "catalogue_title",
        ):

            value = content.get(key)

            if value:
                return str(value).strip()

    return ""


def corpus_text(doc):

    if not isinstance(doc, dict):
        return ""

    return str(
        doc.get(
            "text",
            ""
        )
        or ""
    )


# ================================================================
# 9. STANDARD MAPS
# ================================================================

print("\nBuilding standard lookup maps...")

standard_by_is_number = {}
standard_by_internal_id = {}
standard_core_by_internal_id = {}

for doc in corpus:

    if (
        doc.get("document_type")
        != "standard_core"
    ):
        continue

    internal_id = corpus_internal_id(
        doc
    )

    identifier = corpus_identifier(
        doc
    )

    if internal_id:

        standard_by_internal_id[
            internal_id
        ] = doc

        standard_core_by_internal_id[
            internal_id
        ] = doc

    if identifier:

        normalized = normalize_identifier(
            identifier
        )

        if normalized:

            standard_by_is_number[
                normalized
            ] = doc


print(
    "Standard lookup entries:",
    len(standard_by_is_number)
)


# ================================================================
# 10. RELATIONSHIP TEXT PARSER
# ================================================================

def parse_relationship_document(doc):
    """
    Parse the actual SAATHI corpus relationship format.

    Relationship docs have:

        doc_id
        internal_id
        document_type
        text

    and multiple rows embedded inside text.
    """

    if not isinstance(doc, dict):
        return []

    document_type = str(
        doc.get(
            "document_type",
            ""
        )
        or ""
    ).strip().lower()

    if document_type not in {
        "cross_references",
        "indian_references",
        "international_references",
        "referred_by",
    }:
        return []

    text = corpus_text(
        doc
    )

    if not text:
        return []

    internal_id = corpus_internal_id(
        doc
    )

    starts = list(
        re.finditer(
            r"Reference Number:\s*(\d+)\s*\|",
            text,
            flags=re.IGNORECASE
        )
    )

    rows = []

    for i, match in enumerate(starts):

        number = match.group(1)

        start = match.start()

        if i + 1 < len(starts):
            end = starts[i + 1].start()
        else:
            end = len(text)

        row_text = text[
            start:end
        ].strip()

        row_text = re.sub(
            r"^Reference Number:\s*\d+\s*\|\s*",
            "",
            row_text,
            flags=re.IGNORECASE
        )

        row = {
            "internal_id": internal_id,
            "document_type": document_type,
            "reference_number": number,
            "raw_text": row_text,
        }

        # --------------------------------------------------------
        # CROSS / INDIAN
        # --------------------------------------------------------

        if document_type in {
            "cross_references",
            "indian_references",
        }:

            m = re.search(
                r"Referenced IS:\s*(.*?)\s*\|\s*Referenced Title:",
                row_text,
                flags=re.IGNORECASE | re.DOTALL
            )

            row["reference_is_number"] = (
                m.group(1).strip()
                if m
                else ""
            )

            m = re.search(
                r"Referenced Title:\s*(.*?)\s*\|\s*Reference Committee:",
                row_text,
                flags=re.IGNORECASE | re.DOTALL
            )

            row["reference_title"] = (
                m.group(1).strip()
                if m
                else ""
            )

            m = re.search(
                r"Reference Committee:\s*(.*?)\s*$",
                row_text,
                flags=re.IGNORECASE | re.DOTALL
            )

            row["reference_committee"] = (
                m.group(1).strip()
                if m
                else ""
            )

        # --------------------------------------------------------
        # INTERNATIONAL
        # --------------------------------------------------------

        elif document_type == "international_references":

            m = re.search(
                r"International Standard:\s*(.*?)\s*$",
                row_text,
                flags=re.IGNORECASE | re.DOTALL
            )

            row["international_standard"] = (
                m.group(1).strip()
                if m
                else ""
            )

        # --------------------------------------------------------
        # REFERRED BY
        # --------------------------------------------------------

        elif document_type == "referred_by":

            m = re.search(
                r"Referencing IS:\s*(.*?)\s*\|\s*Referencing Title:",
                row_text,
                flags=re.IGNORECASE | re.DOTALL
            )

            row["reference_is_number"] = (
                m.group(1).strip()
                if m
                else ""
            )

            m = re.search(
                r"Referencing Title:\s*(.*?)\s*\|\s*Reference Committee:",
                row_text,
                flags=re.IGNORECASE | re.DOTALL
            )

            row["reference_title"] = (
                m.group(1).strip()
                if m
                else ""
            )

            m = re.search(
                r"Reference Committee:\s*(.*?)\s*$",
                row_text,
                flags=re.IGNORECASE | re.DOTALL
            )

            row["reference_committee"] = (
                m.group(1).strip()
                if m
                else ""
            )

        rows.append(
            row
        )

    return rows


# ================================================================
# 11. RELATIONSHIP MAPS
# ================================================================

print("\nBuilding relationship maps...")

references_by_standard = defaultdict(list)
international_refs_by_standard = defaultdict(list)
referred_by_standard = defaultdict(list)

for doc in corpus:

    dtype = str(
        doc.get(
            "document_type",
            ""
        )
        or ""
    ).strip().lower()

    if dtype not in {
        "cross_references",
        "indian_references",
        "international_references",
        "referred_by",
    }:
        continue

    internal_id = corpus_internal_id(
        doc
    )

    owner = (
        standard_core_by_internal_id.get(
            internal_id
        )
    )

    if owner is None:
        continue

    owner_identifier = normalize_identifier(
        corpus_identifier(
            owner
        )
    )

    if not owner_identifier:
        continue

    rows = parse_relationship_document(
        doc
    )

    if dtype == "cross_references":

        references_by_standard[
            owner_identifier
        ].extend(rows)

    elif dtype == "international_references":

        international_refs_by_standard[
            owner_identifier
        ].extend(rows)

    elif dtype == "referred_by":

        referred_by_standard[
            owner_identifier
        ].extend(rows)


print(
    "Outgoing reference standards:",
    len(references_by_standard)
)

print(
    "International reference standards:",
    len(international_refs_by_standard)
)

print(
    "Reverse-reference targets:",
    len(referred_by_standard)
)


# ================================================================
# 12. IDENTIFIER EXTRACTION
# ================================================================

def extract_identifiers(query):

    query = str(
        query
        or ""
    )

    identifiers = []

    # ------------------------------------------------------------
    # BIS / IEC identifiers.
    #
    # Handle:
    # IS/IEC 60691:2023
    # IS/IEC 61196-1:2005
    # IS 3025 : Part 27 : Sec 1 : 2021
    # ------------------------------------------------------------

    bis_patterns = [

        r"\bIS\s*/\s*IEC\s+"
        r"\d+(?:\s*-\s*\d+)?"
        r"(?:\s*:\s*|\s+)"
        r"\d{4}\b",

        r"\bIS\s+"
        r"\d+"
        r"(?:\s*\([^)]+\))?"
        r"(?:\s*:\s*|\s+)"
        r"\d{4}\b",

        r"\bIS\s*/\s*IEC\s+"
        r"\d+"
        r"\s*:\s*PART\b"
        r".{0,100}?"
        r"\d{4}\b",

    ]

    for pattern in bis_patterns:

        for match in re.finditer(
            pattern,
            query,
            flags=re.IGNORECASE
        ):

            raw = match.group(0)

            normalized = normalize_identifier(
                raw
            )

            if normalized:
                identifiers.append(
                    normalized
                )

    # ------------------------------------------------------------
    # Special malformed 60691 corpus-style identifier.
    # ------------------------------------------------------------

    malformed_patterns = [
        r"IS/IEC\s*:\s*\d{4}\s*IEC\s+\d+\s*:\s*\d{4}",
        r"IS/IEC\s*:\s*\d{4}\s*IEC\s+\d+",
    ]

    for pattern in malformed_patterns:

        for match in re.finditer(
            pattern,
            query,
            flags=re.IGNORECASE
        ):

            normalized = normalize_identifier(
                match.group(0)
            )

            if normalized:
                identifiers.append(
                    normalized
                )

    # ------------------------------------------------------------
    # Remove nested IEC-only matches.
    # ------------------------------------------------------------

    cleaned = []

    for identifier in identifiers:

        if identifier not in cleaned:
            cleaned.append(
                identifier
            )

    # ------------------------------------------------------------
    # Only use standalone IEC identifiers when no BIS identifier
    # was found.
    # ------------------------------------------------------------

    if not cleaned:

        for match in re.finditer(
            r"\bIEC\s+\d+(?:-\d+)?\s*:\s*\d{4}\b",
            query,
            flags=re.IGNORECASE
        ):

            normalized = normalize_identifier(
                match.group(0)
            )

            if normalized:
                cleaned.append(
                    normalized
                )

    return cleaned


# ================================================================
# 13. RELATIONSHIP DETECTION
# ================================================================

def detect_relationship_direction(query):

    q = str(
        query
        or ""
    ).lower()

    reverse_patterns = [
        r"\bwhich\b.*\bstandards?\b.*\brefer\b.*\bto\b",
        r"\bwhich\b.*\bstandards?\b.*\breference\b",
        r"\bwhich\b.*\brefer\b.*\bstandard\b",
        r"\bwho\b.*\brefer\b.*\bstandard\b",
        r"\bwhat\b.*\bstandards?\b.*\brefer\b.*\bto\b",
        r"\breferenced\s+by\b",
    ]

    outgoing_patterns = [
        r"\bwhat\b.*\bstandards?\b.*\bdoes\b.*\brefer\b",
        r"\bwhat\b.*\bdoes\b.*\brefer\b",
        r"\bwhat\b.*\breferences?\b.*\bdoes\b",
        r"\bstandards?\b.*\breferenced\b.*\bby\b",
    ]

    # Reverse first because "referenced by" is explicit.
    for pattern in reverse_patterns:

        if re.search(
            pattern,
            q
        ):
            return "reverse"

    for pattern in outgoing_patterns:

        if re.search(
            pattern,
            q
        ):
            return "outgoing"

    # Common direct outgoing phrasing.
    if (
        "its references" in q
        or
        "reference it has" in q
        or
        "references of" in q
    ):
        return "outgoing"

    return None


# ================================================================
# 14. COMPOUND DETECTION
# ================================================================

def detect_compound_query(query):

    q = str(
        query
        or ""
    ).lower()

    # Explicit multi-part query language.
    compound_patterns = [
        r"\band\s+its\s+references\b",
        r"\band\s+their\s+references\b",
        r"\band\s+its\s+related\b",
        r"\band\s+the\s+references\b",
        r"\balong\s+with\s+its\s+references\b",
        r"\bplus\s+its\s+references\b",
    ]

    for pattern in compound_patterns:

        if re.search(
            pattern,
            q
        ):
            return True

    return False


# ================================================================
# 15. EXACT STANDARD SEARCH
# ================================================================

def exact_standard_search(identifier):

    normalized = normalize_identifier(
        identifier
    )

    doc = (
        standard_by_is_number.get(
            normalized
        )
    )

    if doc is None:
        return []

    return [doc]


# ================================================================
# 16. SEMANTIC SEARCH
# ================================================================

def semantic_search(
    query,
    top_k=SEMANTIC_TOP_K
):

    query_embedding = embedding_model.encode(
        [query],
        normalize_embeddings=True,
        convert_to_numpy=True,
        show_progress_bar=False
    )

    query_embedding = np.asarray(
        query_embedding,
        dtype=np.float32
    )

    scores, indices = faiss_index.search(
        query_embedding,
        top_k
    )

    results = []

    seen_doc_ids = set()

    for score, index in zip(
        scores[0],
        indices[0]
    ):

        if index < 0:
            continue

        if index >= len(corpus):
            continue

        doc = corpus[index]

        doc_id = str(
            doc.get(
                "doc_id",
                index
            )
        )

        if doc_id in seen_doc_ids:
            continue

        seen_doc_ids.add(
            doc_id
        )

        results.append(
            {
                "doc": doc,
                "score": float(score),
            }
        )

    return results


# ================================================================
# 17. STANDARD EVIDENCE
# ================================================================

def standard_to_evidence(
    doc,
    source="exact",
    score=None
):

    identifier = normalize_identifier(
        corpus_identifier(
            doc
        )
    )

    title = corpus_title(
        doc
    )

    evidence = {
        "evidence_type":
            "standard",

        "source":
            source,

        "identifier":
            identifier,

        "internal_id":
            corpus_internal_id(
                doc
            ),

        "title":
            title,

        "score":
            score,

        "content":
            doc,
    }

    return evidence


# ================================================================
# 18. RELATIONSHIP EVIDENCE
# ================================================================

def relationship_to_evidence(
    row,
    relationship,
    queried_identifier
):

    queried_identifier = normalize_identifier(
        queried_identifier
    )

    raw_reference = (
        row.get(
            "reference_is_number",
            ""
        )
        or
        row.get(
            "international_standard",
            ""
        )
        or
        ""
    )

    referenced_identifier = normalize_identifier(
        raw_reference
    )

    title = str(
        row.get(
            "reference_title",
            ""
        )
        or
        ""
    ).strip()

    if (
        not title
        and
        row.get("international_standard")
    ):

        title = str(
            row.get(
                "international_standard"
            )
            or
            ""
        ).strip()

    # ------------------------------------------------------------
    # OUTGOING:
    #
    # identifier = queried standard
    # referenced_identifier = referenced standard
    # ------------------------------------------------------------

    if relationship == "outgoing":

        identifier = queried_identifier

    # ------------------------------------------------------------
    # REVERSE:
    #
    # identifier = referencing standard
    # referenced_identifier = queried standard
    # ------------------------------------------------------------

    else:

        identifier = normalize_identifier(
            row.get(
                "reference_is_number",
                ""
            )
        )

        referenced_identifier = (
            queried_identifier
        )

    return {
        "evidence_type":
            "relationship",

        "relationship":
            relationship,

        "identifier":
            identifier,

        "referenced_identifier":
            referenced_identifier,

        "title":
            title,

        "internal_id":
            str(
                row.get(
                    "internal_id",
                    ""
                )
                or
                ""
            ).strip(),

        "reference_number":
            str(
                row.get(
                    "reference_number",
                    ""
                )
                or
                ""
            ).strip(),

        "reference_committee":
            str(
                row.get(
                    "reference_committee",
                    ""
                )
                or
                ""
            ).strip(),

        "content":
            row,
    }


# ================================================================
# 19. RELATIONSHIP RETRIEVAL
# ================================================================

def build_relationship_evidence(
    identifier,
    direction,
    max_results=MAX_RELATIONSHIPS
):

    identifier = normalize_identifier(
        identifier
    )

    evidence = []

    # ------------------------------------------------------------
    # OUTGOING
    # ------------------------------------------------------------

    if direction == "outgoing":

        cross_rows = (
            references_by_standard.get(
                identifier,
                []
            )
        )

        international_rows = (
            international_refs_by_standard.get(
                identifier,
                []
            )
        )

        rows = (
            list(cross_rows)
            +
            list(international_rows)
        )

        for row in rows:

            evidence.append(
                relationship_to_evidence(
                    row,
                    "outgoing",
                    identifier
                )
            )

            if len(evidence) >= max_results:
                break

        return evidence

    # ------------------------------------------------------------
    # REVERSE
    # ------------------------------------------------------------

    if direction == "reverse":

        rows = (
            referred_by_standard.get(
                identifier,
                []
            )
        )

        for row in rows:

            evidence.append(
                relationship_to_evidence(
                    row,
                    "reverse",
                    identifier
                )
            )

            if len(evidence) >= max_results:
                break

        return evidence

    return []


# ================================================================
# 20. QUERY CLASSIFICATION
# ================================================================

def classify_query(query):

    identifiers = extract_identifiers(
        query
    )

    relationship_direction = (
        detect_relationship_direction(
            query
        )
    )

    compound = detect_compound_query(
        query
    )

    if relationship_direction:
        query_type = (
            "relationship"
        )

    elif identifiers:
        query_type = (
            "exact"
        )

    else:
        query_type = (
            "semantic"
        )

    return {
        "identifiers":
            identifiers,

        "relationship_direction":
            relationship_direction,

        "compound":
            compound,

        "query_type":
            query_type,
    }


# ================================================================
# 21. FINAL RETRIEVAL
# ================================================================

def retrieve_evidence(
    query,
    max_relationships=MAX_RELATIONSHIPS
):

    classification = classify_query(
        query
    )

    identifiers = classification[
        "identifiers"
    ]

    relationship_direction = (
        classification[
            "relationship_direction"
        ]
    )

    compound = classification[
        "compound"
    ]

    evidence = []

    retrieval_plan = []

    # ------------------------------------------------------------
    # EXACT STANDARD
    # ------------------------------------------------------------

    for identifier in identifiers:

        exact_items = (
            exact_standard_search(
                identifier
            )
        )

        for item in exact_items:

            evidence.append(
                standard_to_evidence(
                    item,
                    source="exact"
                )
            )

            retrieval_plan.append(
                "standard:"
                + identifier
            )

            # ----------------------------------------------------
            # Relationship retrieval
            # ----------------------------------------------------

            if relationship_direction:

                relationship_items = (
                    build_relationship_evidence(
                        identifier,
                        relationship_direction,
                        max_relationships
                    )
                )

                if relationship_items:

                    if (
                        relationship_direction
                        == "reverse"
                    ):

                        retrieval_plan.append(
                            "reverse_references:"
                            + identifier
                        )

                    elif (
                        relationship_direction
                        == "outgoing"
                    ):

                        retrieval_plan.append(
                            "outgoing_references:"
                            + identifier
                        )

                    evidence.extend(
                        relationship_items
                    )

    # ------------------------------------------------------------
    # SEMANTIC FALLBACK
    # ------------------------------------------------------------

    if not evidence:

        semantic_items = semantic_search(
            query,
            top_k=SEMANTIC_TOP_K
        )

        retrieval_plan.append(
            "semantic"
        )

        for item in semantic_items:

            evidence.append(
                standard_to_evidence(
                    item["doc"],
                    source="semantic",
                    score=item["score"]
                )
            )

    # ------------------------------------------------------------
    # DEDUPLICATION
    # ------------------------------------------------------------

    deduped = []

    seen = set()

    for item in evidence:

        key = (
            item.get(
                "evidence_type",
                ""
            ),
            item.get(
                "identifier",
                ""
            ),
            item.get(
                "referenced_identifier",
                ""
            ),
            item.get(
                "internal_id",
                ""
            ),
            item.get(
                "reference_number",
                ""
            ),
        )

        if key in seen:
            continue

        seen.add(
            key
        )

        deduped.append(
            item
        )

    evidence = deduped

    return {
        "query":
            query,

        "identifiers":
            identifiers,

        "query_type":
            classification[
                "query_type"
            ],

        "compound":
            compound,

        "relationship_direction":
            relationship_direction,

        "retrieval_plan":
            retrieval_plan,

        "evidence_count":
            len(evidence),

        "evidence":
            evidence,
    }


# ================================================================
# 22. EVIDENCE NORMALIZATION FOR LLM
# ================================================================

def normalize_evidence_for_llm(
    evidence
):

    normalized = []

    for index, item in enumerate(
        evidence,
        start=1
    ):

        clean = {
            "evidence_id":
                f"E{index}",

            "evidence_type":
                item.get(
                    "evidence_type"
                ),

            "source":
                item.get(
                    "source"
                ),

            "identifier":
                normalize_identifier(
                    item.get(
                        "identifier",
                        ""
                    )
                ),

            "referenced_identifier":
                normalize_identifier(
                    item.get(
                        "referenced_identifier",
                        ""
                    )
                ),

            "title":
                str(
                    item.get(
                        "title",
                        ""
                    )
                    or
                    ""
                ).strip(),

            "internal_id":
                item.get(
                    "internal_id",
                    ""
                ),

            "relationship":
                item.get(
                    "relationship"
                ),

            "reference_number":
                item.get(
                    "reference_number"
                ),

            "reference_committee":
                item.get(
                    "reference_committee"
                ),
        }

        # Add useful standard text.
        content = item.get(
            "content"
        )

        if isinstance(
            content,
            dict
        ):

            text = content.get(
                "text",
                ""
            )

            if text:
                clean["text"] = str(
                    text
                )

        elif isinstance(
            content,
            str
        ):

            clean["text"] = content

        normalized.append(
            clean
        )

    return normalized


# ================================================================
# 23. GEMINI INITIALIZATION
# ================================================================

gemini_client = None
gemini_error = None

try:

    api_key = os.environ.get(
        "GEMINI_API_KEY"
    )

    if not api_key:

        try:
            from kaggle_secrets import UserSecretsClient

            secrets = UserSecretsClient()

            api_key = secrets.get_secret(
                "GEMINI_API_KEY"
            )

        except Exception:
            api_key = None

    if api_key:

        from google import genai

        gemini_client = genai.Client(
            api_key=api_key
        )

        print(
            "\nGemini client: ✅ available"
        )

    else:

        gemini_error = (
            "GEMINI_API_KEY not found"
        )

        print(
            "\nGemini client: ⚠️ unavailable"
        )

except Exception as e:

    gemini_error = str(e)

    print(
        "\nGemini initialization failed:",
        e
    )


# ================================================================
# 24. GROUNDED GEMINI
# ================================================================

def generate_grounded_answer(
    query,
    evidence,
    max_retries=4
):

    if gemini_client is None:

        return {
            "answer":
                "Gemini is currently unavailable. "
                "The retrieved BIS evidence is available "
                "for inspection.",

            "model":
                GEMINI_MODEL,

            "attempts":
                0,

            "success":
                False,

            "error":
                gemini_error,
        }

    llm_evidence = (
        normalize_evidence_for_llm(
            evidence
        )
    )

    evidence_json = json.dumps(
        llm_evidence,
        ensure_ascii=False,
        indent=2
    )

    system_instruction = """
You are SAATHI, a grounded BIS standards assistant.

Answer the user's question ONLY using the supplied evidence.

Rules:

1. Never invent BIS facts.
2. Never rely on outside knowledge.
3. Use citations like [E1], [E2].
4. Every citation must refer to supplied evidence.
5. Distinguish Active, Withdrawn, Superseded and other statuses
   when the evidence provides that information.
6. For outgoing relationships, explain what the queried standard
   references.
7. For reverse relationships, explain which standards reference
   the queried standard.
8. Never reverse the relationship direction.
9. If evidence is insufficient, explicitly say that the supplied
   BIS evidence is insufficient.
10. Keep the answer concise but useful.
"""

    user_prompt = f"""
USER QUERY:

{query}

BIS EVIDENCE:

{evidence_json}

Provide a grounded answer with appropriate [E#] citations.
"""

    attempts = 0

    for attempt in range(
        1,
        max_retries + 1
    ):

        attempts = attempt

        try:

            response = gemini_client.models.generate_content(
                model=GEMINI_MODEL,
                contents=user_prompt,
                config={
                    "system_instruction":
                        system_instruction,

                    "temperature":
                        0.1,
                }
            )

            answer = (
                getattr(
                    response,
                    "text",
                    None
                )
                or
                ""
            ).strip()

            if not answer:

                raise RuntimeError(
                    "Gemini returned empty answer"
                )

            return {
                "answer":
                    answer,

                "model":
                    GEMINI_MODEL,

                "attempts":
                    attempts,

                "success":
                    True,

                "error":
                    None,
            }

        except Exception as e:

            message = str(
                e
            ).lower()

            transient = any(
                code in message
                for code in (
                    "429",
                    "500",
                    "502",
                    "503",
                    "504",
                    "unavailable",
                    "resource exhausted",
                    "internal"
                )
            )

            if (
                not transient
                or
                attempt == max_retries
            ):

                return {
                    "answer":
                        "The BIS evidence was retrieved successfully, "
                        "but grounded answer generation is currently "
                        "unavailable.",

                    "model":
                        GEMINI_MODEL,

                    "attempts":
                        attempts,

                    "success":
                        False,

                    "error":
                        str(e),
                }

            time.sleep(
                2 ** (
                    attempt - 1
                )
            )

    return {
        "answer":
            "Generation failed.",

        "model":
            GEMINI_MODEL,

        "attempts":
            attempts,

        "success":
            False,

        "error":
            "Unknown generation error",
    }


# ================================================================
# 25. CITATION EXTRACTION
# ================================================================

def extract_citations(
    answer
):

    if not answer:
        return []

    matches = re.findall(
        r"\[E(\d+)\]",
        answer
    )

    result = []

    for value in matches:

        number = int(
            value
        )

        if number not in result:

            result.append(
                number
            )

    return result


def validate_citations(
    answer,
    evidence
):

    citations = extract_citations(
        answer
    )

    invalid = [
        c
        for c in citations
        if c < 1
        or c > len(evidence)
    ]

    return {
        "citations":
            citations,

        "invalid":
            invalid,

        "valid":
            len(invalid) == 0,
    }


# ================================================================
# 26. PRODUCTION SOURCES
# ================================================================

def parse_standard_metadata(
    evidence
):

    content = evidence.get(
        "content"
    )

    if not isinstance(
        content,
        dict
    ):
        return {}

    text = str(
        content.get(
            "text",
            ""
        )
        or
        ""
    )

    result = {}

    patterns = {
        "department":
            r"Technical Department:\s*(.*?)\n",

        "committee":
            r"Technical Committee:\s*(.*?)\n",

        "aspect":
            r"Aspect:\s*(.*?)\n",

        "language":
            r"Language:\s*(.*?)\n",

        "degree_of_equivalence":
            r"Degree of Equivalence:\s*(.*?)\n",

        "status":
            r"Catalogue IS Number:\s*.*?\((.*?)\)",

        "detail_url":
            r"Detail URL:\s*(.*?)\n",

        "download_url":
            r"Download URL:\s*(.*?)\n",

        "composition_url":
            r"Composition URL:\s*(.*?)\n",
    }

    for key, pattern in patterns.items():

        m = re.search(
            pattern,
            text,
            flags=re.IGNORECASE
        )

        if m:

            result[key] = (
                m.group(1).strip()
            )

    return result


def build_sources(
    evidence
):

    sources = []

    for index, item in enumerate(
        evidence,
        start=1
    ):

        source_id = (
            f"E{index}"
        )

        source = {
            "id":
                source_id,

            "evidence_type":
                item.get(
                    "evidence_type"
                ),

            "identifier":
                item.get(
                    "identifier",
                    ""
                ),

            "title":
                item.get(
                    "title",
                    ""
                ),

            "internal_id":
                item.get(
                    "internal_id",
                    ""
                ),
        }

        if (
            item.get(
                "evidence_type"
            )
            ==
            "standard"
        ):

            source.update(
                parse_standard_metadata(
                    item
                )
            )

        elif (
            item.get(
                "evidence_type"
            )
            ==
            "relationship"
        ):

            source.update(
                {
                    "relationship":
                        item.get(
                            "relationship"
                        ),

                    "referenced_identifier":
                        item.get(
                            "referenced_identifier"
                        ),

                    "reference_number":
                        item.get(
                            "reference_number"
                        ),

                    "reference_committee":
                        item.get(
                            "reference_committee"
                        ),
                }
            )

        sources.append(
            source
        )

    return sources


# ================================================================
# 27. COMPLETE SAATHI INFERENCE
# ================================================================

def ask_saathi(
    query
):

    query = str(
        query
        or
        ""
    ).strip()

    if not query:

        raise ValueError(
            "Query cannot be empty."
        )

    retrieval = retrieve_evidence(
        query
    )

    generation = (
        generate_grounded_answer(
            query,
            retrieval["evidence"]
        )
    )

    citation_validation = (
        validate_citations(
            generation["answer"],
            retrieval["evidence"]
        )
    )

    sources = build_sources(
        retrieval["evidence"]
    )

    return {
        "query":
            query,

        "answer":
            generation["answer"],

        "query_type":
            retrieval["query_type"],

        "identifiers":
            retrieval["identifiers"],

        "relationship_direction":
            retrieval[
                "relationship_direction"
            ],

        "compound":
            retrieval[
                "compound"
            ],

        "retrieval": {
            "plan":
                retrieval[
                    "retrieval_plan"
                ],

            "evidence_count":
                retrieval[
                    "evidence_count"
                ],

            "total_sources":
                len(sources),
        },

        "citations":
            citation_validation[
                "citations"
            ],

        "citation_validation":
            citation_validation,

        "sources":
            sources,

        "generation": {
            "model":
                generation[
                    "model"
                ],

            "attempts":
                generation[
                    "attempts"
                ],

            "success":
                generation[
                    "success"
                ],
        },

        "status":
            (
                "success"
                if generation["success"]
                else
                "retrieval_ready_generation_unavailable"
            ),

        "timestamp":
            datetime.now(
                timezone.utc
            ).isoformat(),
    }


# ================================================================
# 28. RETRIEVAL VALIDATION
# ================================================================

print("\n")
print("=" * 75)
print("SAATHI — FINAL RETRIEVAL VALIDATION")
print("=" * 75)


validation_results = []


def run_validation(
    name,
    query,
    expected_evidence,
    expected_relationship=None,
    expected_compound=None
):

    result = retrieve_evidence(
        query
    )

    checks = []

    checks.append(
        result["evidence_count"]
        == expected_evidence
    )

    if expected_relationship is not None:

        checks.append(
            result[
                "relationship_direction"
            ]
            ==
            expected_relationship
        )

    if expected_compound is not None:

        checks.append(
            result["compound"]
            ==
            expected_compound
        )

    # No malformed blank identifiers for standard evidence.
    malformed = 0

    for item in result[
        "evidence"
    ]:

        if (
            item.get(
                "evidence_type"
            )
            ==
            "standard"
            and
            not item.get(
                "identifier"
            )
        ):

            malformed += 1

    checks.append(
        malformed == 0
    )

    passed = all(
        checks
    )

    validation_results.append(
        {
            "name":
                name,

            "passed":
                passed,

            "evidence_count":
                result[
                    "evidence_count"
                ],

            "expected_evidence":
                expected_evidence,

            "identifiers":
                result[
                    "identifiers"
                ],

            "relationship":
                result[
                    "relationship_direction"
                ],

            "compound":
                result[
                    "compound"
                ],

            "plan":
                result[
                    "retrieval_plan"
                ],
        }
    )

    print(
        f"\n{'✅ PASS' if passed else '❌ FAIL'} — {name}"
    )

    print(
        "  Evidence:",
        result["evidence_count"],
        "/ expected",
        expected_evidence
    )

    print(
        "  Identifiers:",
        result["identifiers"]
    )

    print(
        "  Relationship:",
        result[
            "relationship_direction"
        ]
    )

    print(
        "  Compound:",
        result["compound"]
    )

    print(
        "  Plan:",
        result[
            "retrieval_plan"
        ]
    )

    return result


# ------------------------------------------------
# TEST 1
# ------------------------------------------------

test_exact = run_validation(
    "EXACT STANDARD",
    "What is IS/IEC 60691:2023?",
    expected_evidence=1,
    expected_relationship=None,
    expected_compound=False
)


# ------------------------------------------------
# TEST 2
# ------------------------------------------------

test_semantic = run_validation(
    "SEMANTIC THERMAL LINKS",
    "Which BIS standard deals with thermal links?",
    expected_evidence=5,
    expected_relationship=None,
    expected_compound=False
)


# ------------------------------------------------
# TEST 3
# ------------------------------------------------

test_reverse = run_validation(
    "REVERSE REFERENCES",
    "Which BIS standards refer to IS/IEC 61196-1:2005?",
    expected_evidence=43,
    expected_relationship="reverse",
    expected_compound=False
)


# ------------------------------------------------
# TEST 4
# ------------------------------------------------

test_compound = run_validation(
    "COMPOUND OUTGOING",
    "Tell me about IS/IEC 60691:2023 and its references.",
    expected_evidence=12,
    expected_relationship="outgoing",
    expected_compound=True
)


# ================================================================
# 29. DETAILED RELATIONSHIP VALIDATION
# ================================================================

print("\n")
print("=" * 75)
print("RELATIONSHIP SEMANTICS VALIDATION")
print("=" * 75)


relationship_checks = []


# Reverse.
reverse_evidence = (
    test_reverse["evidence"]
)

reverse_relationship_items = [
    x
    for x in reverse_evidence
    if x.get(
        "evidence_type"
    )
    == "relationship"
]

reverse_ok = (
    len(reverse_relationship_items)
    == 42
    and
    all(
        x.get(
            "relationship"
        )
        == "reverse"
        for x in reverse_relationship_items
    )
    and
    all(
        x.get(
            "referenced_identifier"
        )
        ==
        "IS/IEC 61196-1:2005"
        for x in reverse_relationship_items
    )
)

relationship_checks.append(
    reverse_ok
)

print(
    "Reverse relationship semantics:",
    "✅ PASS" if reverse_ok else "❌ FAIL"
)


# Outgoing.
outgoing_evidence = (
    test_compound["evidence"]
)

outgoing_relationship_items = [
    x
    for x in outgoing_evidence
    if x.get(
        "evidence_type"
    )
    == "relationship"
]

outgoing_ok = (
    len(outgoing_relationship_items)
    == 11
    and
    all(
        x.get(
            "relationship"
        )
        ==
        "outgoing"
        for x in outgoing_relationship_items
    )
    and
    all(
        x.get(
            "identifier"
        )
        ==
        "IS/IEC 60691:2023"
        for x in outgoing_relationship_items
    )
)

relationship_checks.append(
    outgoing_ok
)

print(
    "Outgoing relationship semantics:",
    "✅ PASS" if outgoing_ok else "❌ FAIL"
)


# ================================================================
# 30. IDENTIFIER NORMALIZATION TESTS
# ================================================================

print("\n")
print("=" * 75)
print("IDENTIFIER NORMALIZATION")
print("=" * 75)


normalization_tests = [
    (
        "IS/IEC :2023IEC 60691:2023",
        "IS/IEC 60691:2023"
    ),
    (
        "IS/IEC :2015IEC 60691 : 2015",
        "IS/IEC 60691:2015"
    ),
    (
        "IS/IEC 691:1993",
        "IS/IEC 691:1993"
    ),
    (
        "IS/IEC 61196-1:2005",
        "IS/IEC 61196-1:2005"
    ),
]


normalization_ok = True

for raw, expected in normalization_tests:

    actual = normalize_identifier(
        raw
    )

    passed = (
        actual == expected
    )

    normalization_ok &= passed

    print(
        f"{'✅' if passed else '❌'}",
        raw,
        "→",
        actual
    )


# ================================================================
# 31. IDENTIFIER EXTRACTION TEST
# ================================================================

print("\n")
print("=" * 75)
print("IDENTIFIER EXTRACTION")
print("=" * 75)


extracted = extract_identifiers(
    "What is IS/IEC 60691:2023?"
)

extraction_ok = (
    extracted
    ==
    ["IS/IEC 60691:2023"]
)

print(
    "Extracted:",
    extracted
)

print(
    "No duplicate IEC-only identifier:",
    "✅ PASS" if extraction_ok else "❌ FAIL"
)


# ================================================================
# 32. GENERATION TEST
# ================================================================

print("\n")
print("=" * 75)
print("GROUNDED GEMINI TEST")
print("=" * 75)


generation_test = None

if gemini_client is not None:

    try:

        generation_test = ask_saathi(
            "What is IS/IEC 60691:2023?"
        )

        citation_ok = (
            generation_test[
                "citation_validation"
            ][
                "valid"
            ]
        )

        generation_ok = (
            generation_test[
                "generation"
            ][
                "success"
            ]
            and
            citation_ok
        )

        print(
            "Generation:",
            "✅ PASS"
            if generation_ok
            else
            "❌ FAIL"
        )

        print(
            "\nAnswer:\n"
        )

        print(
            generation_test[
                "answer"
            ]
        )

        print(
            "\nCitations:",
            generation_test[
                "citations"
            ]
        )

    except Exception as e:

        generation_ok = False

        print(
            "❌ Generation test failed:",
            e
        )

else:

    generation_ok = None

    print(
        "⚠️ Gemini unavailable."
    )

    print(
        "Generation test skipped."
    )


# ================================================================
# 33. FINAL VALIDATION REPORT
# ================================================================

print("\n")
print("=" * 75)
print("SAATHI — FINAL VALIDATION REPORT")
print("=" * 75)


core_checks = {
    "Artifact loading":
        True,

    "Identifier normalization":
        normalization_ok,

    "Identifier extraction":
        extraction_ok,

    "Exact retrieval":
        validation_results[0]["passed"],

    "Semantic retrieval":
        validation_results[1]["passed"],

    "Reverse relationship retrieval":
        validation_results[2]["passed"],

    "Compound outgoing retrieval":
        validation_results[3]["passed"],

    "Relationship semantics":
        all(
            relationship_checks
        ),
}


for name, passed in core_checks.items():

    print(
        f"[{'PASS' if passed else 'FAIL'}] {name}"
    )


if generation_ok is None:

    print(
        "[SKIP] Gemini generation "
        "(API unavailable)"
    )

else:

    print(
        f"[{'PASS' if generation_ok else 'FAIL'}] "
        "Gemini grounded generation"
    )


overall = all(
    core_checks.values()
)

if generation_ok is not None:
    overall = (
        overall
        and
        generation_ok
    )


print("\n")
print(
    "=" * 75
)

if overall:

    print(
        "🎉 ALL SAATHI AI/ML VALIDATION TESTS PASSED"
    )

else:

    print(
        "❌ VALIDATION FAILED — inspect the failed checks above"
    )

print(
    "=" * 75
)


# ================================================================
# 34. SAVE VALIDATION REPORT
# ================================================================

validation_report = {
    "timestamp":
        datetime.now(
            timezone.utc
        ).isoformat(),

    "corpus_documents":
        len(corpus),

    "faiss_vectors":
        int(
            faiss_index.ntotal
        ),

    "embedding_model":
        EMBEDDING_MODEL_NAME,

    "gemini_model":
        GEMINI_MODEL,

    "validation_results":
        validation_results,

    "relationship_checks":
        relationship_checks,

    "normalization_ok":
        normalization_ok,

    "identifier_extraction_ok":
        extraction_ok,

    "generation_ok":
        generation_ok,

    "overall":
        overall,
}


VALIDATION_PATH = (
    WORK_DIR /
    "saathi_final_validation.json"
)


with open(
    VALIDATION_PATH,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        validation_report,
        f,
        ensure_ascii=False,
        indent=2
    )


print(
    "\nValidation report saved:",
    VALIDATION_PATH
)


# ================================================================
# 35. SAVE RUNTIME SUMMARY
# ================================================================

runtime_summary = {
    "corpus_size":
        len(corpus),

    "faiss_vectors":
        int(
            faiss_index.ntotal
        ),

    "standard_lookup":
        len(
            standard_by_is_number
        ),

    "outgoing_reference_standards":
        len(
            references_by_standard
        ),

    "international_reference_standards":
        len(
            international_refs_by_standard
        ),

    "reverse_reference_targets":
        len(
            referred_by_standard
        ),

    "embedding_model":
        EMBEDDING_MODEL_NAME,

    "semantic_top_k":
        SEMANTIC_TOP_K,

    "gemini_model":
        GEMINI_MODEL,

    "status":
        (
            "PASS"
            if overall
            else
            "FAIL"
        ),
}


RUNTIME_PATH = (
    WORK_DIR /
    "saathi_runtime_summary.json"
)


with open(
    RUNTIME_PATH,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        runtime_summary,
        f,
        ensure_ascii=False,
        indent=2
    )


print(
    "Runtime summary saved:",
    RUNTIME_PATH
)


# ================================================================
# 36. FINAL READY-TO-USE DEMO
# ================================================================

print("\n")
print("=" * 75)
print("SAATHI READY")
print("=" * 75)

print(
    """
Use:

    result = ask_saathi(
        "What is IS/IEC 60691:2023?"
    )

or:

    result = ask_saathi(
        "Which BIS standards refer to IS/IEC 61196-1:2005?"
    )

or:

    result = ask_saathi(
        "Tell me about IS/IEC 60691:2023 and its references."
    )

The complete structured response is in `result`.
"""
)

print(
    "=" * 75
)
```

```python
# ================================================================
# SAATHI — FINAL REPAIR
# Semantic standard-only retrieval + Gemini diagnostics/retry
# ================================================================

import re
import time
import json
import os
from datetime import datetime, timezone

print("=" * 75)
print("SAATHI — FINAL REPAIR")
print("=" * 75)


# ================================================================
# 1. BUILD STANDARD-CORE FAISS POSITION MASK
# ================================================================
#
# The FAISS index contains ALL 116,597 corpus documents:
#
#   standard_core
#   cross_references
#   indian_references
#   international_references
#   referred_by
#
# Semantic validation expects the semantic search to return
# STANDARD documents, not arbitrary relationship documents.
#
# We therefore search more candidates and keep only standard_core.
# ================================================================

standard_core_indices = []

for idx, doc in enumerate(corpus):

    if (
        doc.get("document_type")
        == "standard_core"
    ):
        standard_core_indices.append(idx)


standard_core_indices = np.asarray(
    standard_core_indices,
    dtype=np.int64
)

standard_core_index_set = set(
    standard_core_indices.tolist()
)

print(
    "\nStandard-core documents:",
    len(standard_core_indices)
)


# ================================================================
# 2. PATCH SEMANTIC SEARCH
# ================================================================

def semantic_search(
    query,
    top_k=5
):
    """
    Semantic retrieval restricted to standard_core documents.

    FAISS contains all RAG document types, so retrieve an oversized
    candidate set and retain only standard_core documents.
    """

    query_embedding = embedding_model.encode(
        [query],
        normalize_embeddings=True,
        convert_to_numpy=True,
        show_progress_bar=False
    )

    query_embedding = np.asarray(
        query_embedding,
        dtype=np.float32
    )

    # Search considerably deeper than the final top_k because
    # relationship documents also exist in the global FAISS index.
    search_k = min(
        max(
            top_k * 20,
            100
        ),
        faiss_index.ntotal
    )

    scores, indices = faiss_index.search(
        query_embedding,
        search_k
    )

    results = []

    seen = set()

    for score, index in zip(
        scores[0],
        indices[0]
    ):

        if index < 0:
            continue

        if index >= len(corpus):
            continue

        # ONLY standard_core.
        if index not in standard_core_index_set:
            continue

        doc = corpus[index]

        doc_id = str(
            doc.get(
                "doc_id",
                index
            )
        )

        if doc_id in seen:
            continue

        seen.add(
            doc_id
        )

        results.append(
            {
                "doc": doc,
                "score": float(score),
            }
        )

        if len(results) >= top_k:
            break

    return results


print(
    "Semantic search patched."
)


# ================================================================
# 3. TEST SEMANTIC SEARCH DIRECTLY
# ================================================================

semantic_test_items = semantic_search(
    "Which BIS standard deals with thermal links?",
    top_k=5
)

print("\n")
print("=" * 75)
print("SEMANTIC SEARCH DIRECT TEST")
print("=" * 75)

print(
    "Results:",
    len(semantic_test_items)
)

for i, item in enumerate(
    semantic_test_items,
    start=1
):

    doc = item["doc"]

    print(
        f"\n{i}.",
        normalize_identifier(
            corpus_identifier(
                doc
            )
        ),
        "|",
        corpus_title(
            doc
        ),
        "| score:",
        round(
            item["score"],
            4
        )
    )


# ================================================================
# 4. PATCH RETRIEVAL
# ================================================================
#
# Reinstall retrieve_evidence so it uses the patched semantic
# search and remains completely self-contained.
# ================================================================

def retrieve_evidence(
    query,
    max_relationships=MAX_RELATIONSHIPS
):

    classification = classify_query(
        query
    )

    identifiers = classification[
        "identifiers"
    ]

    relationship_direction = (
        classification[
            "relationship_direction"
        ]
    )

    compound = classification[
        "compound"
    ]

    evidence = []

    retrieval_plan = []

    # ============================================================
    # EXACT + RELATIONSHIPS
    # ============================================================

    for identifier in identifiers:

        exact_items = (
            exact_standard_search(
                identifier
            )
        )

        if not exact_items:
            continue

        retrieval_plan.append(
            "standard:"
            + identifier
        )

        for item in exact_items:

            evidence.append(
                standard_to_evidence(
                    item,
                    source="exact"
                )
            )

        # --------------------------------------------------------
        # Relationship retrieval
        # --------------------------------------------------------

        if relationship_direction:

            relationship_items = (
                build_relationship_evidence(
                    identifier,
                    relationship_direction,
                    max_relationships
                )
            )

            if relationship_items:

                if (
                    relationship_direction
                    == "reverse"
                ):

                    retrieval_plan.append(
                        "reverse_references:"
                        + identifier
                    )

                elif (
                    relationship_direction
                    == "outgoing"
                ):

                    retrieval_plan.append(
                        "outgoing_references:"
                        + identifier
                    )

                evidence.extend(
                    relationship_items
                )

    # ============================================================
    # SEMANTIC FALLBACK
    # ============================================================

    if not evidence:

        semantic_items = semantic_search(
            query,
            top_k=SEMANTIC_TOP_K
        )

        retrieval_plan.append(
            "semantic"
        )

        for item in semantic_items:

            evidence.append(
                standard_to_evidence(
                    item["doc"],
                    source="semantic",
                    score=item["score"]
                )
            )

    # ============================================================
    # DEDUPLICATE
    # ============================================================

    deduped = []

    seen = set()

    for item in evidence:

        key = (
            item.get(
                "evidence_type",
                ""
            ),

            item.get(
                "identifier",
                ""
            ),

            item.get(
                "referenced_identifier",
                ""
            ),

            item.get(
                "internal_id",
                ""
            ),

            item.get(
                "reference_number",
                ""
            ),
        )

        if key in seen:
            continue

        seen.add(
            key
        )

        deduped.append(
            item
        )

    evidence = deduped

    return {
        "query":
            query,

        "identifiers":
            identifiers,

        "query_type":
            classification[
                "query_type"
            ],

        "compound":
            compound,

        "relationship_direction":
            relationship_direction,

        "retrieval_plan":
            retrieval_plan,

        "evidence_count":
            len(evidence),

        "evidence":
            evidence,
    }


print(
    "\nRetrieval layer patched."
)


# ================================================================
# 5. SEMANTIC VALIDATION AGAIN
# ================================================================

semantic_result = retrieve_evidence(
    "Which BIS standard deals with thermal links?"
)

print("\n")
print("=" * 75)
print("SEMANTIC VALIDATION")
print("=" * 75)

print(
    "Evidence:",
    semantic_result["evidence_count"]
)

print(
    "Expected:",
    5
)

print(
    "Plan:",
    semantic_result["retrieval_plan"]
)

print(
    "Identifiers:"
)

for item in semantic_result["evidence"]:

    print(
        "  ",
        item.get(
            "identifier"
        ),
        "|",
        item.get(
            "title"
        )
    )


semantic_pass = (
    semantic_result["evidence_count"]
    == 5
    and
    all(
        item.get(
            "identifier"
        )
        for item in semantic_result["evidence"]
    )
    and
    all(
        item.get(
            "evidence_type"
        )
        == "standard"
        for item in semantic_result["evidence"]
    )
)

print(
    "\nSEMANTIC:",
    "✅ PASS"
    if semantic_pass
    else
    "❌ FAIL"
)


# ================================================================
# 6. GEMINI DIAGNOSTIC
# ================================================================
#
# Your previous code hid the actual exception.
# First perform a tiny direct request so we know exactly whether
# the API/model/key is working.
# ================================================================

print("\n")
print("=" * 75)
print("GEMINI API DIAGNOSTIC")
print("=" * 75)


gemini_direct_ok = False
gemini_direct_error = None


if gemini_client is None:

    gemini_direct_error = (
        "Gemini client is None."
    )

else:

    try:

        diagnostic_response = (
            gemini_client.models.generate_content(
                model=GEMINI_MODEL,
                contents=(
                    "Reply with exactly: SAATHI_GEMINI_OK"
                )
            )
        )

        diagnostic_text = (
            getattr(
                diagnostic_response,
                "text",
                ""
            )
            or
            ""
        ).strip()

        print(
            "Gemini response:",
            diagnostic_text
        )

        gemini_direct_ok = bool(
            diagnostic_text
        )

    except Exception as e:

        gemini_direct_error = (
            f"{type(e).__name__}: {e}"
        )

        print(
            "\n❌ ACTUAL GEMINI ERROR:"
        )

        print(
            gemini_direct_error
        )


if gemini_direct_ok:

    print(
        "\n✅ Gemini API is working."
    )

else:

    print(
        "\n❌ Gemini API diagnostic failed."
    )


# ================================================================
# 7. PATCH GROUNDED GEMINI
# ================================================================

def generate_grounded_answer(
    query,
    evidence,
    max_retries=4
):

    if gemini_client is None:

        return {
            "answer":
                "Gemini client is unavailable.",

            "model":
                GEMINI_MODEL,

            "attempts":
                0,

            "success":
                False,

            "error":
                gemini_direct_error
                or
                "GEMINI_API_KEY not configured.",
        }

    llm_evidence = (
        normalize_evidence_for_llm(
            evidence
        )
    )

    evidence_json = json.dumps(
        llm_evidence,
        ensure_ascii=False,
        indent=2
    )

    system_instruction = """
You are SAATHI, a grounded BIS standards assistant.

Answer ONLY from the supplied BIS evidence.

Rules:

- Never invent BIS facts.
- Never use outside knowledge.
- Cite factual claims with [E1], [E2], etc.
- Every citation must correspond to supplied evidence.
- Distinguish Active, Withdrawn and other statuses when supplied.
- Preserve relationship direction.
- For outgoing queries, explain what the queried standard references.
- For reverse queries, explain which standards reference the queried
  standard.
- If the evidence is insufficient, say so explicitly.
"""

    user_prompt = f"""
USER QUERY:

{query}

BIS EVIDENCE:

{evidence_json}

Answer the query using only this evidence.
Use [E#] citations.
"""

    last_error = None

    for attempt in range(
        1,
        max_retries + 1
    ):

        try:

            response = (
                gemini_client.models.generate_content(
                    model=GEMINI_MODEL,
                    contents=user_prompt,
                    config={
                        "system_instruction":
                            system_instruction,

                        "temperature":
                            0.1,
                    }
                )
            )

            answer = (
                getattr(
                    response,
                    "text",
                    ""
                )
                or
                ""
            ).strip()

            if not answer:
                raise RuntimeError(
                    "Gemini returned an empty response."
                )

            return {
                "answer":
                    answer,

                "model":
                    GEMINI_MODEL,

                "attempts":
                    attempt,

                "success":
                    True,

                "error":
                    None,
            }

        except Exception as e:

            last_error = (
                f"{type(e).__name__}: {e}"
            )

            print(
                f"\nGemini attempt {attempt} failed:"
            )

            print(
                last_error
            )

            error_text = str(
                e
            ).lower()

            transient = any(
                x in error_text
                for x in (
                    "429",
                    "500",
                    "502",
                    "503",
                    "504",
                    "resource exhausted",
                    "temporarily unavailable",
                    "unavailable",
                    "internal server error",
                    "deadline",
                    "timeout"
                )
            )

            if (
                not transient
                or
                attempt == max_retries
            ):
                break

            wait = 2 ** (
                attempt - 1
            )

            print(
                f"Retrying in {wait}s..."
            )

            time.sleep(
                wait
            )

    return {
        "answer":
            "The BIS evidence was retrieved successfully, "
            "but grounded Gemini generation failed.",

        "model":
            GEMINI_MODEL,

        "attempts":
            max_retries,

        "success":
            False,

        "error":
            last_error,
    }


print(
    "\nGrounded Gemini function patched."
)


# ================================================================
# 8. RUN FINAL FOUR RETRIEVAL TESTS
# ================================================================

print("\n")
print("=" * 75)
print("FINAL RETRIEVAL TESTS")
print("=" * 75)


final_tests = [
    (
        "EXACT",
        "What is IS/IEC 60691:2023?",
        1,
        None,
        False
    ),

    (
        "SEMANTIC",
        "Which BIS standard deals with thermal links?",
        5,
        None,
        False
    ),

    (
        "REVERSE",
        "Which BIS standards refer to IS/IEC 61196-1:2005?",
        43,
        "reverse",
        False
    ),

    (
        "COMPOUND",
        "Tell me about IS/IEC 60691:2023 and its references.",
        12,
        "outgoing",
        True
    ),
]


final_results = []

for (
    name,
    query,
    expected_count,
    expected_direction,
    expected_compound
) in final_tests:

    result = retrieve_evidence(
        query
    )

    passed = (
        result["evidence_count"]
        ==
        expected_count
        and
        (
            expected_direction is None
            or
            result[
                "relationship_direction"
            ]
            ==
            expected_direction
        )
        and
        result["compound"]
        ==
        expected_compound
    )

    final_results.append(
        passed
    )

    print(
        f"\n{'✅ PASS' if passed else '❌ FAIL'} — {name}"
    )

    print(
        "  Evidence:",
        result["evidence_count"],
        "/",
        expected_count
    )

    print(
        "  Identifiers:",
        result["identifiers"]
    )

    print(
        "  Relationship:",
        result[
            "relationship_direction"
        ]
    )

    print(
        "  Compound:",
        result["compound"]
    )

    print(
        "  Plan:",
        result["retrieval_plan"]
    )


# ================================================================
# 9. FINAL GROUNDED ANSWER TEST
# ================================================================

print("\n")
print("=" * 75)
print("FINAL GROUNDED ANSWER TEST")
print("=" * 75)


answer_test_query = (
    "What is IS/IEC 60691:2023?"
)

answer_retrieval = retrieve_evidence(
    answer_test_query
)

answer_generation = (
    generate_grounded_answer(
        answer_test_query,
        answer_retrieval["evidence"]
    )
)

print(
    "\nGeneration success:",
    answer_generation["success"]
)

print(
    "Attempts:",
    answer_generation["attempts"]
)

if answer_generation["error"]:

    print(
        "Error:",
        answer_generation["error"]
    )

print(
    "\nAnswer:"
)

print(
    answer_generation["answer"]
)


if answer_generation["success"]:

    citation_check = validate_citations(
        answer_generation["answer"],
        answer_retrieval["evidence"]
    )

    print(
        "\nCitations:",
        citation_check["citations"]
    )

    print(
        "Citation validation:",
        "✅ PASS"
        if citation_check["valid"]
        else
        "❌ FAIL"
    )

    gemini_pass = (
        citation_check["valid"]
    )

else:

    gemini_pass = False


# ================================================================
# 10. FINAL STATUS
# ================================================================

print("\n")
print("=" * 75)
print("SAATHI — FINAL STATUS")
print("=" * 75)


retrieval_pass = all(
    final_results
)

print(
    "[{}] Exact retrieval".format(
        "PASS"
        if final_results[0]
        else
        "FAIL"
    )
)

print(
    "[{}] Semantic retrieval".format(
        "PASS"
        if final_results[1]
        else
        "FAIL"
    )
)

print(
    "[{}] Reverse relationship retrieval".format(
        "PASS"
        if final_results[2]
        else
        "FAIL"
    )
)

print(
    "[{}] Compound outgoing retrieval".format(
        "PASS"
        if final_results[3]
        else
        "FAIL"
    )
)

print(
    "[{}] Gemini grounded generation".format(
        "PASS"
        if gemini_pass
        else
        "FAIL"
    )
)


overall = (
    retrieval_pass
    and
    gemini_pass
)


print("\n")

if overall:

    print(
        "🎉 SAATHI AI/ML PIPELINE COMPLETE"
    )

    print(
        "ALL CORE RETRIEVAL + RAG + GEMINI TESTS PASSED."
    )

else:

    print(
        "⚠️ SAATHI still has a failing component."
    )

    if not retrieval_pass:
        print(
            "Retrieval still requires investigation."
        )

    if not gemini_pass:
        print(
            "Gemini API/generation still requires investigation."
        )


# ================================================================
# 11. SAVE FINAL STATUS
# ================================================================

final_status = {
    "timestamp":
        datetime.now(
            timezone.utc
        ).isoformat(),

    "corpus_documents":
        len(corpus),

    "faiss_vectors":
        int(
            faiss_index.ntotal
        ),

    "standard_core_documents":
        len(
            standard_core_indices
        ),

    "standard_lookup":
        len(
            standard_by_is_number
        ),

    "outgoing_reference_standards":
        len(
            references_by_standard
        ),

    "international_reference_standards":
        len(
            international_refs_by_standard
        ),

    "reverse_reference_targets":
        len(
            referred_by_standard
        ),

    "retrieval_pass":
        retrieval_pass,

    "gemini_pass":
        gemini_pass,

    "overall":
        overall,

    "gemini_error":
        (
            answer_generation.get(
                "error"
            )
            if answer_generation
            else
            gemini_direct_error
        ),
}


FINAL_STATUS_PATH = (
    WORK_DIR /
    "saathi_final_status.json"
)


with open(
    FINAL_STATUS_PATH,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        final_status,
        f,
        ensure_ascii=False,
        indent=2
    )


print(
    "\nSaved:",
    FINAL_STATUS_PATH
)

print(
    "=" * 75
)
```

```python
def generate_grounded_answer(
    query,
    evidence,
    max_retries=2
):
    """
    Grounded Gemini generation.

    Important:
    - Rate-limit/day-quota errors are NOT retried repeatedly.
    - Temporary server errors are retried.
    - Retrieval remains fully usable when Gemini is unavailable.
    """

    if gemini_client is None:
        return {
            "answer": (
                "The BIS evidence was retrieved successfully, "
                "but Gemini is currently unavailable."
            ),
            "model": GEMINI_MODEL,
            "attempts": 0,
            "success": False,
            "error": gemini_direct_error or
                     "Gemini client unavailable.",
        }

    llm_evidence = normalize_evidence_for_llm(
        evidence
    )

    evidence_json = json.dumps(
        llm_evidence,
        ensure_ascii=False,
        indent=2
    )

    system_instruction = """
You are SAATHI, a grounded BIS standards assistant.

Answer ONLY from the supplied BIS evidence.

Rules:
1. Never invent BIS facts.
2. Never use outside knowledge.
3. Cite factual claims using [E1], [E2], etc.
4. Every citation must correspond to supplied evidence.
5. Distinguish Active, Withdrawn, Superseded and other statuses
   when the evidence provides that information.
6. For outgoing relationships, explain what the queried standard
   references.
7. For reverse relationships, explain which standards reference
   the queried standard.
8. Never reverse relationship direction.
9. If evidence is insufficient, explicitly say so.
"""

    user_prompt = f"""
USER QUERY:

{query}

BIS EVIDENCE:

{evidence_json}

Answer the query using ONLY this evidence.
Use [E#] citations.
"""

    for attempt in range(1, max_retries + 1):

        try:

            response = gemini_client.models.generate_content(
                model=GEMINI_MODEL,
                contents=user_prompt,
                config={
                    "system_instruction": system_instruction,
                    "temperature": 0.1,
                }
            )

            answer = (
                getattr(
                    response,
                    "text",
                    ""
                )
                or
                ""
            ).strip()

            if not answer:
                raise RuntimeError(
                    "Gemini returned an empty response."
                )

            return {
                "answer": answer,
                "model": GEMINI_MODEL,
                "attempts": attempt,
                "success": True,
                "error": None,
            }

        except Exception as e:

            error_text = str(e)
            error_lower = error_text.lower()

            print(
                f"\nGemini attempt {attempt} failed:"
            )
            print(error_text)

            # ====================================================
            # DAILY / PROJECT QUOTA
            # ====================================================

            if (
                "generate_content_free_tier_requests"
                in error_lower
                or
                "quota exceeded" in error_lower
                or
                "perday" in error_lower
                or
                "per_day" in error_lower
            ):

                return {
                    "answer": (
                        "The BIS evidence was retrieved successfully, "
                        "but Gemini answer generation is unavailable "
                        "because the configured Gemini API quota has "
                        "been exhausted."
                    ),
                    "model": GEMINI_MODEL,
                    "attempts": attempt,
                    "success": False,
                    "error": error_text,
                }

            # ====================================================
            # OTHER 429
            # ====================================================

            if "429" in error_lower:

                if attempt < max_retries:
                    time.sleep(5)
                    continue

                break

            # ====================================================
            # TEMPORARY SERVER ERRORS
            # ====================================================

            transient = any(
                code in error_lower
                for code in (
                    "500",
                    "502",
                    "503",
                    "504",
                    "unavailable",
                    "internal server error",
                    "timeout",
                    "deadline",
                )
            )

            if transient and attempt < max_retries:

                wait = 2 ** (attempt - 1)

                print(
                    f"Temporary Gemini error. "
                    f"Retrying in {wait}s..."
                )

                time.sleep(wait)
                continue

            # ====================================================
            # PERMANENT ERROR
            # ====================================================

            return {
                "answer": (
                    "The BIS evidence was retrieved successfully, "
                    "but grounded Gemini generation failed."
                ),
                "model": GEMINI_MODEL,
                "attempts": attempt,
                "success": False,
                "error": error_text,
            }

    return {
        "answer": (
            "The BIS evidence was retrieved successfully, "
            "but grounded Gemini generation failed."
        ),
        "model": GEMINI_MODEL,
        "attempts": max_retries,
        "success": False,
        "error": "Gemini generation failed after retries.",
    }
```

```python
# ============================================================
# SAATHI — GEMINI GROUNDED GENERATION PATCH
# ============================================================

import json
import time


def generate_grounded_answer(query, evidence, max_retries=2):

    # --------------------------------------------------------
    # Check Gemini client
    # --------------------------------------------------------

    if "gemini_client" not in globals() or gemini_client is None:

        return {
            "answer": (
                "The BIS evidence was retrieved successfully, "
                "but Gemini is currently unavailable."
            ),
            "model": globals().get(
                "GEMINI_MODEL",
                "gemini-2.5-flash"
            ),
            "attempts": 0,
            "success": False,
            "error": "Gemini client is not initialized."
        }

    model_name = globals().get(
        "GEMINI_MODEL",
        "gemini-2.5-flash"
    )

    # --------------------------------------------------------
    # Prepare evidence
    # --------------------------------------------------------

    try:

        if "normalize_evidence_for_llm" in globals():

            llm_evidence = normalize_evidence_for_llm(
                evidence
            )

        else:

            # Fallback: use evidence directly
            llm_evidence = evidence

    except Exception:

        llm_evidence = evidence

    evidence_json = json.dumps(
        llm_evidence,
        ensure_ascii=False,
        indent=2,
        default=str
    )

    # --------------------------------------------------------
    # Grounding instructions
    # --------------------------------------------------------

    system_instruction = """
You are SAATHI, a grounded BIS standards assistant.

Answer ONLY from the supplied BIS evidence.

Rules:

1. Never invent BIS facts.
2. Never use outside knowledge.
3. Cite factual claims using [E1], [E2], etc.
4. Every citation must correspond to supplied evidence.
5. Distinguish Active, Withdrawn, Superseded and other
   statuses when the evidence provides that information.
6. For outgoing relationships, explain what the queried
   standard references.
7. For reverse relationships, explain which standards
   reference the queried standard.
8. Never reverse relationship direction.
9. If the evidence is insufficient, explicitly say so.
"""

    user_prompt = f"""
USER QUERY:

{query}

BIS EVIDENCE:

{evidence_json}

Answer the query using ONLY the supplied BIS evidence.

Use [E#] citations for factual claims.
"""

    # --------------------------------------------------------
    # Gemini generation
    # --------------------------------------------------------

    for attempt in range(1, max_retries + 1):

        try:

            response = gemini_client.models.generate_content(
                model=model_name,
                contents=user_prompt,
                config={
                    "system_instruction": system_instruction,
                    "temperature": 0.1,
                }
            )

            answer = getattr(
                response,
                "text",
                ""
            )

            if answer is None:
                answer = ""

            answer = answer.strip()

            if not answer:
                raise RuntimeError(
                    "Gemini returned an empty response."
                )

            return {
                "answer": answer,
                "model": model_name,
                "attempts": attempt,
                "success": True,
                "error": None
            }

        except Exception as e:

            error_text = str(e)
            error_lower = error_text.lower()

            print(
                f"\nGemini attempt {attempt} failed:"
            )
            print(error_text)

            # ------------------------------------------------
            # DAILY QUOTA EXHAUSTED
            # ------------------------------------------------

            if (
                "generate_content_free_tier_requests"
                in error_lower
                or
                "quota exceeded" in error_lower
                or
                "resource_exhausted" in error_lower
            ):

                print(
                    "\n⚠️ Gemini free-tier quota is exhausted."
                )

                return {
                    "answer": (
                        "The BIS evidence was retrieved successfully, "
                        "but Gemini answer generation is unavailable "
                        "because the configured Gemini API quota "
                        "has been exhausted."
                    ),
                    "model": model_name,
                    "attempts": attempt,
                    "success": False,
                    "error": error_text
                }

            # ------------------------------------------------
            # TEMPORARY 429
            # ------------------------------------------------

            if "429" in error_lower:

                if attempt < max_retries:

                    print(
                        "Temporary rate limit. "
                        "Waiting 5 seconds..."
                    )

                    time.sleep(5)
                    continue

                return {
                    "answer": (
                        "The BIS evidence was retrieved successfully, "
                        "but Gemini rate limiting prevented generation."
                    ),
                    "model": model_name,
                    "attempts": attempt,
                    "success": False,
                    "error": error_text
                }

            # ------------------------------------------------
            # TEMPORARY SERVER ERROR
            # ------------------------------------------------

            transient = any(
                x in error_lower
                for x in [
                    "500",
                    "502",
                    "503",
                    "504",
                    "unavailable",
                    "internal server error",
                    "timeout",
                    "deadline"
                ]
            )

            if transient and attempt < max_retries:

                wait = 2 ** (attempt - 1)

                print(
                    f"Temporary Gemini error. "
                    f"Retrying in {wait}s..."
                )

                time.sleep(wait)
                continue

            # ------------------------------------------------
            # OTHER ERROR
            # ------------------------------------------------

            return {
                "answer": (
                    "The BIS evidence was retrieved successfully, "
                    "but grounded Gemini generation failed."
                ),
                "model": model_name,
                "attempts": attempt,
                "success": False,
                "error": error_text
            }

    # --------------------------------------------------------
    # Final fallback
    # --------------------------------------------------------

    return {
        "answer": (
            "The BIS evidence was retrieved successfully, "
            "but grounded Gemini generation failed."
        ),
        "model": model_name,
        "attempts": max_retries,
        "success": False,
        "error": "Generation failed after retries."
    }


print("✅ generate_grounded_answer() patched successfully.")
```

```python
# ============================================================
# SAATHI — GEMINI KEY FIX + FINAL GENERATION TEST
# ============================================================

import os
import json
import time

from kaggle_secrets import UserSecretsClient
from google import genai


# ============================================================
# 1. LOAD NEW GEMINI API KEY
# ============================================================

print("=" * 70)
print("SAATHI — GEMINI CONFIGURATION")
print("=" * 70)

try:
    secrets = UserSecretsClient()
    GEMINI_API_KEY = secrets.get_secret("GEMINI_API_KEY")

    if not GEMINI_API_KEY:
        raise ValueError("GEMINI_API_KEY is empty.")

    print("✅ GEMINI_API_KEY loaded from Kaggle Secrets.")

except Exception as e:
    print("❌ Could not load GEMINI_API_KEY.")
    print(e)
    raise


# ============================================================
# 2. INITIALIZE GEMINI CLIENT
# ============================================================

GEMINI_MODEL = "gemini-2.5-flash"

try:

    gemini_client = genai.Client(
        api_key=GEMINI_API_KEY
    )

    print("✅ Gemini client initialized.")
    print("Model:", GEMINI_MODEL)

except Exception as e:

    print("❌ Gemini client initialization failed.")
    print(e)
    raise


# ============================================================
# 3. DIRECT GEMINI CONNECTION TEST
# ============================================================

print("\n" + "=" * 70)
print("GEMINI CONNECTION TEST")
print("=" * 70)

try:

    response = gemini_client.models.generate_content(
        model=GEMINI_MODEL,
        contents="Reply with exactly: SAATHI_GEMINI_OK"
    )

    test_text = (
        getattr(response, "text", "")
        or ""
    ).strip()

    print("Gemini response:")
    print(test_text)

    if test_text:
        print("\n✅ GEMINI CONNECTION PASSED")

    else:
        print("\n❌ Gemini returned an empty response.")
        raise RuntimeError(
            "Gemini returned an empty response."
        )

except Exception as e:

    print("\n❌ GEMINI CONNECTION FAILED")
    print("\nActual error:")
    print(e)

    print(
        "\nIf this is another 429 quota error, "
        "the new key is still attached to a project "
        "whose Gemini quota is exhausted."
    )

    raise


# ============================================================
# 4. GROUNDED GENERATION FUNCTION
# ============================================================

def generate_grounded_answer(
    query,
    evidence,
    max_retries=2
):

    # --------------------------------------------------------
    # Prepare evidence
    # --------------------------------------------------------

    if "normalize_evidence_for_llm" in globals():

        try:
            llm_evidence = normalize_evidence_for_llm(
                evidence
            )
        except Exception:
            llm_evidence = evidence

    else:

        llm_evidence = evidence

    evidence_json = json.dumps(
        llm_evidence,
        ensure_ascii=False,
        indent=2,
        default=str
    )

    # --------------------------------------------------------
    # Grounding instructions
    # --------------------------------------------------------

    system_instruction = """
You are SAATHI, a grounded BIS standards assistant.

Answer ONLY from the supplied BIS evidence.

Rules:

1. Never invent BIS facts.
2. Never use outside knowledge.
3. Cite factual claims using [E1], [E2], etc.
4. Every citation must correspond to supplied evidence.
5. Distinguish Active, Withdrawn, Superseded and other
   statuses when the evidence provides that information.
6. For outgoing relationships, explain what the queried
   standard references.
7. For reverse relationships, explain which standards
   reference the queried standard.
8. Never reverse relationship direction.
9. If evidence is insufficient, explicitly say so.
"""

    user_prompt = f"""
USER QUERY:

{query}

BIS EVIDENCE:

{evidence_json}

Answer the query using ONLY the supplied BIS evidence.

Use [E#] citations for factual claims.
"""

    # --------------------------------------------------------
    # Generate
    # --------------------------------------------------------

    for attempt in range(1, max_retries + 1):

        try:

            response = gemini_client.models.generate_content(
                model=GEMINI_MODEL,
                contents=user_prompt,
                config={
                    "system_instruction": system_instruction,
                    "temperature": 0.1,
                }
            )

            answer = (
                getattr(response, "text", "")
                or ""
            ).strip()

            if not answer:
                raise RuntimeError(
                    "Gemini returned an empty answer."
                )

            return {
                "answer": answer,
                "model": GEMINI_MODEL,
                "attempts": attempt,
                "success": True,
                "error": None
            }

        except Exception as e:

            error_text = str(e)
            error_lower = error_text.lower()

            print(
                f"\nGemini generation attempt "
                f"{attempt} failed:"
            )
            print(error_text)

            # Daily/project quota
            if (
                "resource_exhausted" in error_lower
                or
                "quota exceeded" in error_lower
                or
                "generate_content_free_tier_requests"
                in error_lower
            ):

                return {
                    "answer": (
                        "The BIS evidence was retrieved "
                        "successfully, but Gemini generation "
                        "is unavailable because the configured "
                        "API quota has been exhausted."
                    ),
                    "model": GEMINI_MODEL,
                    "attempts": attempt,
                    "success": False,
                    "error": error_text
                }

            # Temporary errors only
            transient = any(
                x in error_lower
                for x in [
                    "500",
                    "502",
                    "503",
                    "504",
                    "unavailable",
                    "timeout",
                    "deadline"
                ]
            )

            if transient and attempt < max_retries:

                wait = 2 ** (attempt - 1)

                print(
                    f"Temporary error. "
                    f"Retrying in {wait}s..."
                )

                time.sleep(wait)
                continue

            return {
                "answer": (
                    "The BIS evidence was retrieved "
                    "successfully, but grounded Gemini "
                    "generation failed."
                ),
                "model": GEMINI_MODEL,
                "attempts": attempt,
                "success": False,
                "error": error_text
            }

    return {
        "answer": (
            "The BIS evidence was retrieved successfully, "
            "but grounded Gemini generation failed."
        ),
        "model": GEMINI_MODEL,
        "attempts": max_retries,
        "success": False,
        "error": "Generation failed after retries."
    }


print("\n✅ generate_grounded_answer() installed.")


# ============================================================
# 5. CHECK THAT EXISTING RAG FUNCTION EXISTS
# ============================================================

print("\n" + "=" * 70)
print("CHECKING EXISTING RAG PIPELINE")
print("=" * 70)

if "retrieve_evidence_api" not in globals():

    print(
        "⚠️ retrieve_evidence_api() is not currently "
        "loaded in this Kaggle kernel."
    )

    print(
        "\nGemini itself is working, but the current kernel "
        "does not contain the RAG retrieval function."
    )

    print(
        "\nDo NOT rebuild embeddings or FAISS."
    )

else:

    print("✅ retrieve_evidence_api() found.")

    # ========================================================
    # 6. END-TO-END RAG + GEMINI TEST
    # ========================================================

    print("\n" + "=" * 70)
    print("SAATHI — END-TO-END TEST")
    print("=" * 70)

    query = "What is IS/IEC 60691:2023?"

    print("\nQuery:")
    print(query)

    evidence = retrieve_evidence_api(query)

    print(
        "\nRetrieved evidence:",
        len(evidence)
    )

    result = generate_grounded_answer(
        query,
        evidence
    )

    print("\n" + "-" * 70)
    print("GEMINI RESULT")
    print("-" * 70)

    print("Success:", result["success"])
    print("Model:", result["model"])
    print("Attempts:", result["attempts"])

    print("\nAnswer:")
    print(result["answer"])

    if result["error"]:
        print("\nError:")
        print(result["error"])


# ============================================================
# FINAL
# ============================================================

print("\n" + "=" * 70)
print("SAATHI — GEMINI PATCH COMPLETE")
print("=" * 70)
```

```python
# ============================================================
# SAATHI — RAG RECOVERY CELL
# ============================================================
# This DOES NOT rebuild the corpus or embeddings.
# It only reloads the existing artifacts and restores the
# retrieval functions into the current Kaggle kernel.
# ============================================================

import os
import re
import json
import numpy as np
import faiss

from collections import defaultdict
from sentence_transformers import SentenceTransformer


# ============================================================
# 1. CONFIG
# ============================================================

CORPUS_PATH = "/kaggle/working/saathi_backend/bis_rag_corpus.jsonl"
FAISS_PATH = "/kaggle/working/saathi_backend/bis_faiss.index"

# Fallback paths if the backend folder is not present
if not os.path.exists(CORPUS_PATH):
    CORPUS_PATH = "/kaggle/working/bis_rag_corpus.jsonl"

if not os.path.exists(FAISS_PATH):
    FAISS_PATH = "/kaggle/working/bis_faiss.index"

MODEL_NAME = "BAAI/bge-small-en-v1.5"

print("=" * 75)
print("SAATHI — RAG RECOVERY")
print("=" * 75)


# ============================================================
# 2. VERIFY EXISTING ARTIFACTS
# ============================================================

print("\nChecking existing artifacts...")

for path in [CORPUS_PATH, FAISS_PATH]:

    if not os.path.exists(path):
        raise FileNotFoundError(
            f"Required existing artifact not found:\n{path}"
        )

    print("✅", path)


# ============================================================
# 3. LOAD CORPUS
# ============================================================

print("\nLoading BIS RAG corpus...")

corpus = []

with open(
    CORPUS_PATH,
    "r",
    encoding="utf-8"
) as f:

    for line in f:

        line = line.strip()

        if line:
            corpus.append(
                json.loads(line)
            )

print(
    f"✅ Corpus loaded: {len(corpus):,} documents"
)


# ============================================================
# 4. LOAD FAISS
# ============================================================

print("\nLoading FAISS index...")

faiss_index = faiss.read_index(
    FAISS_PATH
)

print(
    f"✅ FAISS loaded: {faiss_index.ntotal:,} vectors"
)


# ============================================================
# 5. LOAD EMBEDDING MODEL
# ============================================================

print("\nLoading BGE embedding model...")

embedding_model = SentenceTransformer(
    MODEL_NAME
)

print("✅ BGE model loaded.")


# ============================================================
# 6. IDENTIFIER NORMALIZATION
# ============================================================

def normalize_identifier(value):

    if value is None:
        return ""

    s = str(value).strip()

    if not s:
        return ""

    # Remove common BIS catalogue concatenation problems
    s = re.sub(
        r"IS/IEC\s*:\s*(\d{4})IEC\s+",
        r"IS/IEC ",
        s,
        flags=re.I
    )

    s = re.sub(
        r"IS/IEC\s*:\s*(\d{4})\s*IEC\s+",
        r"IS/IEC ",
        s,
        flags=re.I
    )

    # Normalise spaces around colon
    s = re.sub(
        r"\s*:\s*",
        ":",
        s
    )

    # Normalise spaces around hyphen
    s = re.sub(
        r"\s*-\s*",
        "-",
        s
    )

    # Collapse whitespace
    s = re.sub(
        r"\s+",
        " ",
        s
    ).strip()

    # IS/IEC forms
    m = re.search(
        r"\bIS/IEC\s+([0-9]+(?:-[0-9]+)?):([0-9]{4})\b",
        s,
        flags=re.I
    )

    if m:

        return (
            f"IS/IEC "
            f"{m.group(1)}:"
            f"{m.group(2)}"
        )

    # Indian IS forms
    m = re.search(
        r"\bIS\s+([0-9]+"
        r"(?:\s*\([^)]*\))?"
        r"(?:\s+PART\s+[0-9]+)?"
        r"(?:\s+SEC\s+[0-9]+)?"
        r")\s*:\s*([0-9]{4})",
        s,
        flags=re.I
    )

    if m:

        number = re.sub(
            r"\s+",
            " ",
            m.group(1).strip()
        )

        return (
            f"IS {number}:"
            f"{m.group(2)}"
        )

    # Generic fallback
    return s


# ============================================================
# 7. CORPUS HELPERS
# ============================================================

def doc_text(doc):

    return str(
        doc.get("text", "")
    )


def doc_type(doc):

    return str(
        doc.get("document_type", "")
    )


def doc_internal_id(doc):

    return str(
        doc.get("internal_id", "")
    )


def extract_doc_identifier(doc):

    text = doc_text(doc)

    # --------------------------------------------------------
    # Standard core
    # --------------------------------------------------------

    if doc_type(doc) == "standard_core":

        patterns = [
            r"BIS Standard:\s*(.+)",
            r"IS Number:\s*(.+)",
            r"IS Number\s*:\s*(.+)"
        ]

        for pattern in patterns:

            m = re.search(
                pattern,
                text,
                flags=re.I
            )

            if m:

                raw = m.group(1).split("\n")[0]

                identifier = normalize_identifier(
                    raw
                )

                if identifier:
                    return identifier

        # fallback search
        m = re.search(
            r"(IS/IEC\s*[^,\n|]+:\s*\d{4})",
            text,
            flags=re.I
        )

        if m:
            return normalize_identifier(
                m.group(1)
            )

        m = re.search(
            r"(IS\s+\d+[^,\n|]*:\s*\d{4})",
            text,
            flags=re.I
        )

        if m:
            return normalize_identifier(
                m.group(1)
            )

    return ""


def extract_doc_title(doc):

    text = doc_text(doc)

    for pattern in [
        r"Title:\s*(.+)",
        r"IS Title:\s*(.+)",
        r"Catalogue Title:\s*(.+)"
    ]:

        m = re.search(
            pattern,
            text,
            flags=re.I
        )

        if m:

            title = m.group(1).strip()

            if title:
                return title

    return ""


# ============================================================
# 8. BUILD STANDARD LOOKUP
# ============================================================

print("\nBuilding standard lookup...")

standard_by_is_number = {}
standard_by_internal_id = {}

standard_core_indices = []

for idx, doc in enumerate(corpus):

    if doc_type(doc) != "standard_core":
        continue

    standard_core_indices.append(idx)

    internal_id = doc_internal_id(doc)

    if internal_id:
        standard_by_internal_id[
            internal_id
        ] = doc

    identifier = extract_doc_identifier(doc)

    if identifier:

        standard_by_is_number[
            identifier
        ] = doc


print(
    "Standard lookup:",
    len(standard_by_is_number)
)

print(
    "Standard-core documents:",
    len(standard_core_indices)
)


# ============================================================
# 9. RELATIONSHIP PARSER
# ============================================================

def parse_relationship_doc(doc):

    text = doc_text(doc)

    dtype = doc_type(doc)

    internal_id = doc_internal_id(doc)

    identifier = ""

    referenced_identifier = ""
    referencing_identifier = ""

    # --------------------------------------------------------
    # Cross / Indian reference
    # --------------------------------------------------------

    if dtype in (
        "cross_references",
        "indian_references"
    ):

        m = re.search(
            r"Referenced IS:\s*(.+?)(?:\s*\||$)",
            text,
            flags=re.I
        )

        if m:

            referenced_identifier = normalize_identifier(
                m.group(1).strip()
            )

        # owning standard
        if internal_id in standard_by_internal_id:

            identifier = extract_doc_identifier(
                standard_by_internal_id[internal_id]
            )

        return {
            "direction": "outgoing",
            "internal_id": internal_id,
            "identifier": identifier,
            "referenced_identifier":
                referenced_identifier
        }

    # --------------------------------------------------------
    # International references
    # --------------------------------------------------------

    if dtype == "international_references":

        m = re.search(
            r"International Standard:\s*(.+?)(?:\s*\||$)",
            text,
            flags=re.I
        )

        if not m:

            m = re.search(
                r"International Standard\s*:\s*(.+?)(?:\s*\||$)",
                text,
                flags=re.I
            )

        if m:
            referenced_identifier = (
                m.group(1).strip()
            )

        if internal_id in standard_by_internal_id:

            identifier = extract_doc_identifier(
                standard_by_internal_id[internal_id]
            )

        return {
            "direction": "outgoing",
            "internal_id": internal_id,
            "identifier": identifier,
            "referenced_identifier":
                referenced_identifier
        }

    # --------------------------------------------------------
    # Referred by
    # --------------------------------------------------------

    if dtype == "referred_by":

        m = re.search(
            r"Referencing IS:\s*(.+?)(?:\s*\||$)",
            text,
            flags=re.I
        )

        if m:

            referencing_identifier = (
                normalize_identifier(
                    m.group(1).strip()
                )
            )

        # This document belongs to the TARGET standard.
        if internal_id in standard_by_internal_id:

            identifier = extract_doc_identifier(
                standard_by_internal_id[internal_id]
            )

        return {
            "direction": "reverse",
            "internal_id": internal_id,
            "identifier": identifier,
            "referencing_identifier":
                referencing_identifier
        }

    return None


# ============================================================
# 10. BUILD RELATIONSHIP MAPS
# ============================================================

print("\nBuilding relationship maps...")

outgoing_by_standard = defaultdict(list)
reverse_by_target = defaultdict(list)

for doc in corpus:

    dtype = doc_type(doc)

    if dtype not in (
        "cross_references",
        "indian_references",
        "international_references",
        "referred_by"
    ):
        continue

    parsed = parse_relationship_doc(doc)

    if not parsed:
        continue

    if parsed["direction"] == "outgoing":

        key = parsed["identifier"]

        if key:
            outgoing_by_standard[key].append(
                {
                    "doc": doc,
                    "parsed": parsed
                }
            )

    elif parsed["direction"] == "reverse":

        key = parsed["identifier"]

        if key:
            reverse_by_target[key].append(
                {
                    "doc": doc,
                    "parsed": parsed
                }
            )


print(
    "Outgoing standards:",
    len(outgoing_by_standard)
)

print(
    "Reverse-reference targets:",
    len(reverse_by_target)
)


# ============================================================
# 11. IDENTIFIER EXTRACTION
# ============================================================

def extract_identifiers(query):

    q = str(query)

    found = []

    # --------------------------------------------------------
    # BIS IS/IEC first
    # --------------------------------------------------------

    for m in re.finditer(
        r"\bIS/IEC\s+"
        r"[0-9]+(?:-[0-9]+)?"
        r"\s*:\s*[0-9]{4}\b",
        q,
        flags=re.I
    ):

        value = normalize_identifier(
            m.group(0)
        )

        if value and value not in found:
            found.append(value)

    # --------------------------------------------------------
    # Indian IS
    # --------------------------------------------------------

    for m in re.finditer(
        r"\bIS\s+"
        r"[0-9]+"
        r"(?:\s*\([^)]*\))?"
        r"(?:\s+PART\s+[0-9]+)?"
        r"(?:\s+SEC\s+[0-9]+)?"
        r"\s*:\s*[0-9]{4}\b",
        q,
        flags=re.I
    ):

        value = normalize_identifier(
            m.group(0)
        )

        if value and value not in found:
            found.append(value)

    # --------------------------------------------------------
    # Standalone IEC only when no BIS identifiers
    # --------------------------------------------------------

    if not found:

        for m in re.finditer(
            r"\bIEC\s+"
            r"[0-9]+(?:-[0-9]+)?"
            r"\s*:\s*[0-9]{4}\b",
            q,
            flags=re.I
        ):

            value = normalize_identifier(
                m.group(0)
            )

            if value and value not in found:
                found.append(value)

    return found


# ============================================================
# 12. RELATIONSHIP DETECTION
# ============================================================

def detect_relationship_direction(query):

    q = str(query).lower()

    reverse_patterns = [
        "refer to",
        "refers to",
        "referencing",
        "referenced by",
        "which standards refer",
        "which bis standards refer",
        "who refers to",
        "standards that refer"
    ]

    outgoing_patterns = [
        "what does",
        "what standards does",
        "references",
        "reference",
        "its references",
        "standards it refers to",
        "what does it refer to"
    ]

    if any(
        p in q
        for p in reverse_patterns
    ):

        return "reverse"

    if any(
        p in q
        for p in outgoing_patterns
    ):

        return "outgoing"

    return None


def detect_compound_query(query, identifiers):

    q = str(query).lower()

    if len(identifiers) > 1:
        return True

    compound_terms = [
        "and its references",
        "and references",
        "along with references",
        "details and references",
        "information and references"
    ]

    return any(
        x in q
        for x in compound_terms
    )


# ============================================================
# 13. STANDARD EVIDENCE
# ============================================================

def standard_to_evidence(
    doc,
    source="exact",
    score=None
):

    identifier = extract_doc_identifier(doc)

    title = extract_doc_title(doc)

    result = {
        "evidence_type": "standard",
        "source": source,
        "identifier": identifier,
        "internal_id": doc_internal_id(doc),
        "title": title,
        "content": {
            "doc_id": doc.get(
                "doc_id",
                ""
            ),
            "internal_id": doc_internal_id(doc),
            "document_type": doc_type(doc),
            "is_number": identifier,
            "is_title": title,
            "text": doc_text(doc)
        }
    }

    if score is not None:
        result["score"] = float(score)

    return result


# ============================================================
# 14. RELATIONSHIP EVIDENCE
# ============================================================

def relationship_to_evidence(
    item,
    direction
):

    doc = item["doc"]
    parsed = item["parsed"]

    if direction == "outgoing":

        return {
            "evidence_type": "relationship",
            "source": doc_type(doc),
            "identifier": parsed["identifier"],
            "referenced_identifier":
                parsed["referenced_identifier"],
            "internal_id":
                parsed["internal_id"],
            "title": "",
            "content": {
                "doc_id": doc.get(
                    "doc_id",
                    ""
                ),
                "internal_id":
                    doc_internal_id(doc),
                "document_type":
                    doc_type(doc),
                "text":
                    doc_text(doc)
            }
        }

    # reverse
    return {
        "evidence_type": "relationship",
        "source": "referred_by",
        "identifier":
            parsed["referencing_identifier"],
        "referenced_identifier":
            parsed["identifier"],
        "internal_id":
            parsed["internal_id"],
        "title": "",
        "content": {
            "doc_id": doc.get(
                "doc_id",
                ""
            ),
            "internal_id":
                doc_internal_id(doc),
            "document_type":
                doc_type(doc),
            "text":
                doc_text(doc)
        }
    }


# ============================================================
# 15. EXACT SEARCH
# ============================================================

def exact_standard_search(identifier):

    identifier = normalize_identifier(
        identifier
    )

    doc = standard_by_is_number.get(
        identifier
    )

    if doc is None:
        return []

    return [
        standard_to_evidence(
            doc,
            source="exact"
        )
    ]


# ============================================================
# 16. SEMANTIC SEARCH
# ============================================================

def semantic_search(
    query,
    top_k=5
):

    # Search more candidates because the FAISS index contains
    # both standard-core and relationship documents.
    search_k = max(
        top_k * 20,
        100
    )

    query_vector = embedding_model.encode(
        [query],
        normalize_embeddings=True,
        convert_to_numpy=True
    ).astype(
        "float32"
    )

    scores, indices = faiss_index.search(
        query_vector,
        search_k
    )

    results = []

    for score, idx in zip(
        scores[0],
        indices[0]
    ):

        if idx < 0:
            continue

        doc = corpus[int(idx)]

        # IMPORTANT:
        # semantic standard search must not return
        # relationship documents.
        if doc_type(doc) != "standard_core":
            continue

        results.append(
            standard_to_evidence(
                doc,
                source="semantic",
                score=float(score)
            )
        )

        if len(results) >= top_k:
            break

    return results


# ============================================================
# 17. RELATIONSHIP SEARCH
# ============================================================

def build_relationship_evidence(
    identifier,
    direction,
    max_results=50
):

    identifier = normalize_identifier(
        identifier
    )

    results = []

    if direction == "outgoing":

        # Cross + Indian + international
        items = outgoing_by_standard.get(
            identifier,
            []
        )

        for item in items[:max_results]:

            results.append(
                relationship_to_evidence(
                    item,
                    "outgoing"
                )
            )

        return results

    if direction == "reverse":

        items = reverse_by_target.get(
            identifier,
            []
        )

        for item in items[:max_results]:

            results.append(
                relationship_to_evidence(
                    item,
                    "reverse"
                )
            )

        return results

    return results


# ============================================================
# 18. MAIN RETRIEVAL API
# ============================================================

def retrieve_evidence_api(
    query,
    semantic_top_k=5
):

    identifiers = extract_identifiers(
        query
    )

    relationship_direction = (
        detect_relationship_direction(
            query
        )
    )

    compound = detect_compound_query(
        query,
        identifiers
    )

    evidence = []
    plan = []

    # --------------------------------------------------------
    # Identifier-based retrieval
    # --------------------------------------------------------

    for identifier in identifiers:

        exact = exact_standard_search(
            identifier
        )

        evidence.extend(
            exact
        )

        if exact:

            plan.append(
                f"standard:{identifier}"
            )

        # ----------------------------------------------------
        # Relationships
        # ----------------------------------------------------

        if relationship_direction:

            relationship_evidence = (
                build_relationship_evidence(
                    identifier,
                    relationship_direction
                )
            )

            evidence.extend(
                relationship_evidence
            )

            if relationship_evidence:

                if relationship_direction == "reverse":

                    plan.append(
                        f"reverse_references:{identifier}"
                    )

                else:

                    plan.append(
                        f"outgoing_references:{identifier}"
                    )

        # ----------------------------------------------------
        # Compound query without explicit relationship word
        # ----------------------------------------------------

        elif compound:

            relationship_evidence = (
                build_relationship_evidence(
                    identifier,
                    "outgoing"
                )
            )

            evidence.extend(
                relationship_evidence
            )

            if relationship_evidence:

                plan.append(
                    f"outgoing_references:{identifier}"
                )

    # --------------------------------------------------------
    # Semantic fallback
    # --------------------------------------------------------

    if not identifiers:

        semantic_evidence = semantic_search(
            query,
            top_k=semantic_top_k
        )

        evidence.extend(
            semantic_evidence
        )

        if semantic_evidence:

            plan.append("semantic")

    # --------------------------------------------------------
    # Remove duplicate evidence
    # --------------------------------------------------------

    deduped = []

    seen = set()

    for item in evidence:

        key = (
            item.get("evidence_type"),
            item.get("internal_id"),
            item.get("identifier"),
            item.get("referenced_identifier"),
            item.get("source")
        )

        if key in seen:
            continue

        seen.add(key)
        deduped.append(item)

    evidence = deduped

    return {
        "query": query,
        "identifiers": identifiers,
        "relationship_direction":
            relationship_direction,
        "compound": compound,
        "retrieval_plan": plan,
        "evidence": evidence,
        "evidence_count": len(evidence)
    }


# ============================================================
# 19. QUICK RECOVERY VALIDATION
# ============================================================

print("\n" + "=" * 75)
print("RAG FUNCTION RECOVERY TEST")
print("=" * 75)

test_queries = [
    "What is IS/IEC 60691:2023?",
    "Which BIS standard deals with thermal links?",
    "Which BIS standards refer to IS/IEC 61196-1:2005?",
    "Tell me about IS/IEC 60691:2023 and its references."
]

for q in test_queries:

    result = retrieve_evidence_api(q)

    print("\nQuery:", q)
    print(
        "Identifiers:",
        result["identifiers"]
    )
    print(
        "Relationship:",
        result["relationship_direction"]
    )
    print(
        "Compound:",
        result["compound"]
    )
    print(
        "Plan:",
        result["retrieval_plan"]
    )
    print(
        "Evidence:",
        result["evidence_count"]
    )

print("\n" + "=" * 75)
print("✅ retrieve_evidence_api() RESTORED")
print("=" * 75)
```

```python
# ============================================================
# SAATHI — FINAL RELATIONSHIP RETRIEVAL FIX
# ============================================================
# DO NOT REBUILD:
#   - corpus
#   - embeddings
#   - FAISS
#   - BGE model
#   - Gemini
#
# This ONLY fixes relationship retrieval.
# ============================================================

from collections import defaultdict
import re


print("=" * 75)
print("SAATHI — FINAL RELATIONSHIP RETRIEVAL FIX")
print("=" * 75)


# ============================================================
# 1. BUILD INTERNAL-ID RELATIONSHIP MAPS
# ============================================================

print("\nBuilding relationship maps from corpus internal_id...")

cross_by_internal_id = defaultdict(list)
indian_by_internal_id = defaultdict(list)
international_by_internal_id = defaultdict(list)
referred_by_internal_id = defaultdict(list)

for doc in corpus:

    dtype = str(
        doc.get("document_type", "")
    ).strip()

    internal_id = str(
        doc.get("internal_id", "")
    ).strip()

    if not internal_id:
        continue

    if dtype == "cross_references":

        cross_by_internal_id[
            internal_id
        ].append(doc)

    elif dtype == "indian_references":

        indian_by_internal_id[
            internal_id
        ].append(doc)

    elif dtype == "international_references":

        international_by_internal_id[
            internal_id
        ].append(doc)

    elif dtype == "referred_by":

        referred_by_internal_id[
            internal_id
        ].append(doc)


print(
    "Cross-reference source standards:",
    len(cross_by_internal_id)
)

print(
    "Indian-reference source standards:",
    len(indian_by_internal_id)
)

print(
    "International-reference source standards:",
    len(international_by_internal_id)
)

print(
    "Reverse-reference target standards:",
    len(referred_by_internal_id)
)


# ============================================================
# 2. HELPER — GET STANDARD INTERNAL ID
# ============================================================

def get_standard_internal_id(identifier):

    identifier = normalize_identifier(
        identifier
    )

    doc = standard_by_is_number.get(
        identifier
    )

    if doc is None:
        return None

    return str(
        doc.get("internal_id", "")
    )


# ============================================================
# 3. EXTRACT REFERENCE TEXT
# ============================================================

def extract_reference_from_doc(doc):

    text = str(
        doc.get("text", "")
    )

    dtype = str(
        doc.get("document_type", "")
    )

    # --------------------------------------------------------
    # Cross / Indian
    # --------------------------------------------------------

    if dtype in (
        "cross_references",
        "indian_references"
    ):

        patterns = [
            r"Referenced IS:\s*(.+?)(?:\s*\||$)",
            r"Reference IS:\s*(.+?)(?:\s*\||$)"
        ]

        for pattern in patterns:

            m = re.search(
                pattern,
                text,
                flags=re.I
            )

            if m:

                return m.group(1).strip()

    # --------------------------------------------------------
    # International
    # --------------------------------------------------------

    if dtype == "international_references":

        patterns = [
            r"International Standard:\s*(.+?)(?:\s*\||$)",
            r"International Standard\s*:\s*(.+?)(?:\s*\||$)"
        ]

        for pattern in patterns:

            m = re.search(
                pattern,
                text,
                flags=re.I
            )

            if m:

                return m.group(1).strip()

    # --------------------------------------------------------
    # Referred by
    # --------------------------------------------------------

    if dtype == "referred_by":

        patterns = [
            r"Referencing IS:\s*(.+?)(?:\s*\||$)",
            r"Referencing Standard:\s*(.+?)(?:\s*\||$)"
        ]

        for pattern in patterns:

            m = re.search(
                pattern,
                text,
                flags=re.I
            )

            if m:

                return m.group(1).strip()

    return ""


# ============================================================
# 4. RELATIONSHIP EVIDENCE BUILDER
# ============================================================

def make_relationship_evidence(
    doc,
    direction,
    queried_identifier
):

    dtype = str(
        doc.get("document_type", "")
    )

    internal_id = str(
        doc.get("internal_id", "")
    )

    reference_text = extract_reference_from_doc(
        doc
    )

    # --------------------------------------------------------
    # OUTGOING
    # --------------------------------------------------------

    if direction == "outgoing":

        return {
            "evidence_type": "relationship",
            "source": dtype,
            "identifier": queried_identifier,
            "referenced_identifier": reference_text,
            "internal_id": internal_id,
            "title": "",
            "content": {
                "doc_id": doc.get(
                    "doc_id",
                    ""
                ),
                "internal_id": internal_id,
                "document_type": dtype,
                "text": str(
                    doc.get("text", "")
                )
            }
        }

    # --------------------------------------------------------
    # REVERSE
    # --------------------------------------------------------

    return {
        "evidence_type": "relationship",
        "source": "referred_by",
        "identifier": reference_text,
        "referenced_identifier": queried_identifier,
        "internal_id": internal_id,
        "title": "",
        "content": {
            "doc_id": doc.get(
                "doc_id",
                ""
            ),
            "internal_id": internal_id,
            "document_type": dtype,
            "text": str(
                doc.get("text", "")
            )
        }
    }


# ============================================================
# 5. REPLACE RELATIONSHIP RETRIEVAL
# ============================================================

def build_relationship_evidence(
    identifier,
    direction,
    max_results=100
):

    identifier = normalize_identifier(
        identifier
    )

    internal_id = get_standard_internal_id(
        identifier
    )

    if not internal_id:

        return []

    results = []

    # ========================================================
    # OUTGOING
    # ========================================================

    if direction == "outgoing":

        # IMPORTANT:
        # The validated SAATHI pipeline uses:
        #
        #   cross_references
        #   +
        #   international_references
        #
        # for outgoing references.
        #
        # Indian references are kept separately and are NOT
        # mixed into this particular relationship answer.

        cross_docs = cross_by_internal_id.get(
            internal_id,
            []
        )

        international_docs = (
            international_by_internal_id.get(
                internal_id,
                []
            )
        )

        for doc in cross_docs:

            results.append(
                make_relationship_evidence(
                    doc,
                    "outgoing",
                    identifier
                )
            )

        for doc in international_docs:

            results.append(
                make_relationship_evidence(
                    doc,
                    "outgoing",
                    identifier
                )
            )

        return results[:max_results]

    # ========================================================
    # REVERSE
    # ========================================================

    if direction == "reverse":

        referred_docs = (
            referred_by_internal_id.get(
                internal_id,
                []
            )
        )

        for doc in referred_docs:

            results.append(
                make_relationship_evidence(
                    doc,
                    "reverse",
                    identifier
                )
            )

        return results[:max_results]

    return []


# ============================================================
# 6. REPLACE MAIN RETRIEVAL FUNCTION
# ============================================================

def retrieve_evidence_api(
    query,
    semantic_top_k=5
):

    identifiers = extract_identifiers(
        query
    )

    relationship_direction = (
        detect_relationship_direction(
            query
        )
    )

    compound = detect_compound_query(
        query,
        identifiers
    )

    evidence = []
    plan = []

    # ========================================================
    # IDENTIFIER-BASED RETRIEVAL
    # ========================================================

    for identifier in identifiers:

        # ----------------------------------------------------
        # Exact standard
        # ----------------------------------------------------

        exact = exact_standard_search(
            identifier
        )

        evidence.extend(
            exact
        )

        if exact:

            plan.append(
                f"standard:{identifier}"
            )

        # ----------------------------------------------------
        # Explicit relationship query
        # ----------------------------------------------------

        if relationship_direction:

            relationship_evidence = (
                build_relationship_evidence(
                    identifier,
                    relationship_direction,
                    max_results=100
                )
            )

            evidence.extend(
                relationship_evidence
            )

            if relationship_evidence:

                if relationship_direction == "reverse":

                    plan.append(
                        f"reverse_references:{identifier}"
                    )

                else:

                    plan.append(
                        f"outgoing_references:{identifier}"
                    )

        # ----------------------------------------------------
        # Compound query
        # ----------------------------------------------------

        elif compound:

            relationship_evidence = (
                build_relationship_evidence(
                    identifier,
                    "outgoing",
                    max_results=100
                )
            )

            evidence.extend(
                relationship_evidence
            )

            if relationship_evidence:

                plan.append(
                    f"outgoing_references:{identifier}"
                )

    # ========================================================
    # SEMANTIC FALLBACK
    # ========================================================

    if not identifiers:

        semantic_evidence = semantic_search(
            query,
            top_k=semantic_top_k
        )

        evidence.extend(
            semantic_evidence
        )

        if semantic_evidence:

            plan.append(
                "semantic"
            )

    # ========================================================
    # DEDUPLICATE
    # ========================================================

    deduped = []

    seen = set()

    for item in evidence:

        key = (
            item.get(
                "evidence_type"
            ),
            item.get(
                "internal_id"
            ),
            item.get(
                "identifier"
            ),
            item.get(
                "referenced_identifier"
            ),
            item.get(
                "source"
            )
        )

        if key in seen:
            continue

        seen.add(key)

        deduped.append(
            item
        )

    evidence = deduped

    return {
        "query": query,
        "identifiers": identifiers,
        "relationship_direction":
            relationship_direction,
        "compound": compound,
        "retrieval_plan": plan,
        "evidence": evidence,
        "evidence_count": len(evidence)
    }


# ============================================================
# 7. FINAL RETRIEVAL VALIDATION
# ============================================================

print("\n")
print("=" * 75)
print("FINAL RETRIEVAL VALIDATION")
print("=" * 75)


tests = [

    (
        "EXACT",
        "What is IS/IEC 60691:2023?",
        1
    ),

    (
        "SEMANTIC",
        "Which BIS standard deals with thermal links?",
        5
    ),

    (
        "REVERSE",
        "Which BIS standards refer to IS/IEC 61196-1:2005?",
        43
    ),

    (
        "COMPOUND",
        "Tell me about IS/IEC 60691:2023 and its references.",
        12
    )
]


all_passed = True

for name, query, expected in tests:

    result = retrieve_evidence_api(
        query
    )

    actual = result[
        "evidence_count"
    ]

    passed = (
        actual == expected
    )

    if not passed:
        all_passed = False

    print("\n" + "-" * 75)

    print(
        f"{'✅ PASS' if passed else '❌ FAIL'} — {name}"
    )

    print(
        "Query:",
        query
    )

    print(
        "Expected evidence:",
        expected
    )

    print(
        "Actual evidence:",
        actual
    )

    print(
        "Identifiers:",
        result["identifiers"]
    )

    print(
        "Relationship:",
        result["relationship_direction"]
    )

    print(
        "Compound:",
        result["compound"]
    )

    print(
        "Plan:",
        result["retrieval_plan"]
    )


# ============================================================
# 8. RELATIONSHIP-SPECIFIC COUNTS
# ============================================================

print("\n")
print("=" * 75)
print("RELATIONSHIP COUNTS")
print("=" * 75)


target = "IS/IEC 61196-1:2005"

reverse_evidence = build_relationship_evidence(
    target,
    "reverse"
)

print(
    "\nReverse references for",
    target,
    ":",
    len(reverse_evidence)
)

target = "IS/IEC 60691:2023"

outgoing_evidence = build_relationship_evidence(
    target,
    "outgoing"
)

print(
    "Outgoing references for",
    target,
    ":",
    len(outgoing_evidence)
)


# ============================================================
# 9. FINAL RESULT
# ============================================================

print("\n")
print("=" * 75)

if all_passed:

    print(
        "🎉 ALL RETRIEVAL TESTS PASSED"
    )

    print(
        "\nExact       : 1/1"
    )

    print(
        "Semantic    : 5/5"
    )

    print(
        "Reverse     : 43/43"
    )

    print(
        "Compound    : 12/12"
    )

else:

    print(
        "❌ RETRIEVAL VALIDATION STILL HAS FAILURES"
    )

print("=" * 75)
```

```python
# ============================================================
# SAATHI — FINAL COLLISION-SAFE STANDARD RESOLVER
# ============================================================
# DO NOT REBUILD CORPUS
# DO NOT REBUILD FAISS
# DO NOT REBUILD EMBEDDINGS
# DO NOT TOUCH GEMINI
#
# Root cause:
# standard_by_is_number[identifier] = doc
# can overwrite duplicate/malformed BIS identifiers.
#
# Fix:
# identifier -> LIST of candidate standards
# then resolve using the actual corpus record + relationship
# evidence instead of blindly taking the last dictionary entry.
# ============================================================

import re
from collections import defaultdict


print("=" * 75)
print("SAATHI — COLLISION-SAFE STANDARD RESOLVER")
print("=" * 75)


# ============================================================
# 1. BUILD COLLISION-SAFE STANDARD INDEX
# ============================================================

print("\nBuilding collision-safe standard index...")

standards_by_identifier = defaultdict(list)

for doc in corpus:

    if str(
        doc.get("document_type", "")
    ) != "standard_core":

        continue

    identifier = extract_doc_identifier(doc)

    if not identifier:
        continue

    standards_by_identifier[
        identifier
    ].append(doc)


print(
    "Unique normalized identifiers:",
    len(standards_by_identifier)
)

print(
    "Standard-core documents:",
    len(standard_core_indices)
)


# ============================================================
# 2. DEBUG THE TWO PROBLEM STANDARDS
# ============================================================

def show_candidates(identifier):

    identifier = normalize_identifier(
        identifier
    )

    candidates = standards_by_identifier.get(
        identifier,
        []
    )

    print("\nIdentifier:", identifier)
    print("Candidates:", len(candidates))

    for doc in candidates:

        iid = str(
            doc.get("internal_id", "")
        )

        title = extract_doc_title(doc)

        print(
            "  internal_id:",
            iid,
            "| title:",
            title
        )

    return candidates


print("\n" + "-" * 75)
print("CHECKING 61196-1")
print("-" * 75)

candidates_61196 = show_candidates(
    "IS/IEC 61196-1:2005"
)


print("\n" + "-" * 75)
print("CHECKING 60691:2023")
print("-" * 75)

candidates_60691 = show_candidates(
    "IS/IEC 60691:2023"
)


# ============================================================
# 3. ROBUST STANDARD RESOLVER
# ============================================================

def resolve_standard(
    identifier,
    preferred_direction=None
):

    identifier = normalize_identifier(
        identifier
    )

    candidates = standards_by_identifier.get(
        identifier,
        []
    )

    if not candidates:
        return None

    # --------------------------------------------------------
    # If there is only one candidate, use it.
    # --------------------------------------------------------

    if len(candidates) == 1:
        return candidates[0]

    # --------------------------------------------------------
    # If multiple candidates exist, select the candidate that
    # actually owns the requested relationship records.
    # --------------------------------------------------------

    scored = []

    for doc in candidates:

        internal_id = str(
            doc.get("internal_id", "")
        )

        outgoing_count = (
            len(
                cross_by_internal_id.get(
                    internal_id,
                    []
                )
            )
            +
            len(
                international_by_internal_id.get(
                    internal_id,
                    []
                )
            )
        )

        reverse_count = len(
            referred_by_internal_id.get(
                internal_id,
                []
            )
        )

        score = 0

        if preferred_direction == "outgoing":
            score += outgoing_count * 100

        elif preferred_direction == "reverse":
            score += reverse_count * 100

        else:
            score += (
                outgoing_count
                +
                reverse_count
            )

        scored.append(
            (
                score,
                outgoing_count,
                reverse_count,
                doc
            )
        )

    scored.sort(
        key=lambda x: (
            x[0],
            x[1] + x[2]
        ),
        reverse=True
    )

    return scored[0][3]


# ============================================================
# 4. REBUILD RELATIONSHIP EVIDENCE
# ============================================================

def build_relationship_evidence(
    identifier,
    direction,
    max_results=100
):

    identifier = normalize_identifier(
        identifier
    )

    standard_doc = resolve_standard(
        identifier,
        preferred_direction=direction
    )

    if standard_doc is None:
        return []

    internal_id = str(
        standard_doc.get(
            "internal_id",
            ""
        )
    )

    results = []

    # ========================================================
    # OUTGOING
    # ========================================================

    if direction == "outgoing":

        cross_docs = cross_by_internal_id.get(
            internal_id,
            []
        )

        international_docs = (
            international_by_internal_id.get(
                internal_id,
                []
            )
        )

        for doc in cross_docs:

            results.append(
                make_relationship_evidence(
                    doc,
                    "outgoing",
                    identifier
                )
            )

        for doc in international_docs:

            results.append(
                make_relationship_evidence(
                    doc,
                    "outgoing",
                    identifier
                )
            )

        return results[:max_results]

    # ========================================================
    # REVERSE
    # ========================================================

    if direction == "reverse":

        referred_docs = (
            referred_by_internal_id.get(
                internal_id,
                []
            )
        )

        for doc in referred_docs:

            results.append(
                make_relationship_evidence(
                    doc,
                    "reverse",
                    identifier
                )
            )

        return results[:max_results]

    return []


# ============================================================
# 5. EXACT SEARCH — USE SAME RESOLVER
# ============================================================

def exact_standard_search(identifier):

    identifier = normalize_identifier(
        identifier
    )

    doc = resolve_standard(
        identifier
    )

    if doc is None:
        return []

    return [
        standard_to_evidence(
            doc,
            source="exact"
        )
    ]


# ============================================================
# 6. MAIN RETRIEVAL FUNCTION
# ============================================================

def retrieve_evidence_api(
    query,
    semantic_top_k=5
):

    identifiers = extract_identifiers(
        query
    )

    relationship_direction = (
        detect_relationship_direction(
            query
        )
    )

    compound = detect_compound_query(
        query,
        identifiers
    )

    evidence = []
    plan = []

    # ========================================================
    # IDENTIFIER RETRIEVAL
    # ========================================================

    for identifier in identifiers:

        exact = exact_standard_search(
            identifier
        )

        evidence.extend(
            exact
        )

        if exact:

            plan.append(
                f"standard:{identifier}"
            )

        # ----------------------------------------------------
        # Explicit relationship
        # ----------------------------------------------------

        if relationship_direction:

            relationship_evidence = (
                build_relationship_evidence(
                    identifier,
                    relationship_direction,
                    max_results=100
                )
            )

            evidence.extend(
                relationship_evidence
            )

            if relationship_evidence:

                if relationship_direction == "reverse":

                    plan.append(
                        f"reverse_references:{identifier}"
                    )

                else:

                    plan.append(
                        f"outgoing_references:{identifier}"
                    )

        # ----------------------------------------------------
        # Compound query
        # ----------------------------------------------------

        elif compound:

            relationship_evidence = (
                build_relationship_evidence(
                    identifier,
                    "outgoing",
                    max_results=100
                )
            )

            evidence.extend(
                relationship_evidence
            )

            if relationship_evidence:

                plan.append(
                    f"outgoing_references:{identifier}"
                )

    # ========================================================
    # SEMANTIC
    # ========================================================

    if not identifiers:

        semantic_evidence = semantic_search(
            query,
            top_k=semantic_top_k
        )

        evidence.extend(
            semantic_evidence
        )

        if semantic_evidence:
            plan.append("semantic")

    # ========================================================
    # DEDUPLICATE
    # ========================================================

    deduped = []
    seen = set()

    for item in evidence:

        key = (
            item.get("evidence_type"),
            item.get("internal_id"),
            item.get("identifier"),
            item.get("referenced_identifier"),
            item.get("source")
        )

        if key in seen:
            continue

        seen.add(key)
        deduped.append(item)

    evidence = deduped

    return {
        "query": query,
        "identifiers": identifiers,
        "relationship_direction":
            relationship_direction,
        "compound": compound,
        "retrieval_plan": plan,
        "evidence": evidence,
        "evidence_count": len(evidence)
    }


# ============================================================
# 7. DIRECT RELATIONSHIP TEST
# ============================================================

print("\n")
print("=" * 75)
print("DIRECT RELATIONSHIP TEST")
print("=" * 75)


reverse_test = build_relationship_evidence(
    "IS/IEC 61196-1:2005",
    "reverse"
)

print(
    "\n61196-1 reverse references:",
    len(reverse_test)
)


outgoing_test = build_relationship_evidence(
    "IS/IEC 60691:2023",
    "outgoing"
)

print(
    "60691:2023 outgoing references:",
    len(outgoing_test)
)


# ============================================================
# 8. FULL VALIDATION
# ============================================================

print("\n")
print("=" * 75)
print("FINAL RETRIEVAL VALIDATION")
print("=" * 75)


tests = [

    (
        "EXACT",
        "What is IS/IEC 60691:2023?",
        1
    ),

    (
        "SEMANTIC",
        "Which BIS standard deals with thermal links?",
        5
    ),

    (
        "REVERSE",
        "Which BIS standards refer to IS/IEC 61196-1:2005?",
        43
    ),

    (
        "COMPOUND",
        "Tell me about IS/IEC 60691:2023 and its references.",
        12
    )
]


all_passed = True

for name, query, expected in tests:

    result = retrieve_evidence_api(
        query
    )

    actual = result[
        "evidence_count"
    ]

    passed = (
        actual == expected
    )

    if not passed:
        all_passed = False

    print("\n" + "-" * 75)

    print(
        f"{'✅ PASS' if passed else '❌ FAIL'} — {name}"
    )

    print(
        "Expected:",
        expected
    )

    print(
        "Actual:",
        actual
    )

    print(
        "Identifiers:",
        result["identifiers"]
    )

    print(
        "Relationship:",
        result["relationship_direction"]
    )

    print(
        "Compound:",
        result["compound"]
    )

    print(
        "Plan:",
        result["retrieval_plan"]
    )


# ============================================================
# 9. FINAL
# ============================================================

print("\n")
print("=" * 75)

if all_passed:

    print("🎉 ALL 4 RETRIEVAL TESTS PASSED")
    print()
    print("Exact    : 1/1")
    print("Semantic : 5/5")
    print("Reverse  : 43/43")
    print("Compound : 12/12")

else:

    print(
        "❌ RETRIEVAL STILL HAS A FAILURE."
    )

print("=" * 75)
```

```python
# ============================================================
# SAATHI — ACTUAL FINAL RELATIONSHIP PARSER FIX
# ============================================================
# ROOT CAUSE:
# Relationship corpus documents contain MULTIPLE relationship
# records inside one document.
#
# Previous code used re.search() -> only FIRST record.
#
# This version uses re.finditer() -> ALL records.
#
# DO NOT rebuild:
#   - corpus
#   - embeddings
#   - FAISS
#   - BGE
#   - Gemini
# ============================================================

import re
from collections import defaultdict


print("=" * 75)
print("SAATHI — FINAL MULTI-RELATIONSHIP PARSER")
print("=" * 75)


# ============================================================
# 1. BUILD MULTI-RELATIONSHIP MAPS
# ============================================================

cross_by_internal_id = defaultdict(list)
international_by_internal_id = defaultdict(list)
referred_by_internal_id = defaultdict(list)

print("\nParsing ALL relationship records...")


# ============================================================
# 2. PARSE CROSS REFERENCES
# ============================================================

for doc in corpus:

    dtype = str(
        doc.get("document_type", "")
    ).strip()

    if dtype != "cross_references":
        continue

    internal_id = str(
        doc.get("internal_id", "")
    ).strip()

    text = str(
        doc.get("text", "")
    )

    if not internal_id:
        continue

    # IMPORTANT:
    # finditer(), NOT search()
    matches = re.finditer(
        r"Referenced IS:\s*(.+?)(?=\s*\||\s*\n|$)",
        text,
        flags=re.I
    )

    for match in matches:

        reference = match.group(1).strip()

        if reference:

            cross_by_internal_id[
                internal_id
            ].append(
                {
                    "doc": doc,
                    "reference": reference
                }
            )


# ============================================================
# 3. PARSE INTERNATIONAL REFERENCES
# ============================================================

for doc in corpus:

    dtype = str(
        doc.get("document_type", "")
    ).strip()

    if dtype != "international_references":
        continue

    internal_id = str(
        doc.get("internal_id", "")
    ).strip()

    text = str(
        doc.get("text", "")
    )

    if not internal_id:
        continue

    matches = re.finditer(
        r"International Standard:\s*(.+?)(?=\s*\||\s*\n|$)",
        text,
        flags=re.I
    )

    for match in matches:

        reference = match.group(1).strip()

        if reference:

            international_by_internal_id[
                internal_id
            ].append(
                {
                    "doc": doc,
                    "reference": reference
                }
            )


# ============================================================
# 4. PARSE REVERSE REFERENCES
# ============================================================

for doc in corpus:

    dtype = str(
        doc.get("document_type", "")
    ).strip()

    if dtype != "referred_by":
        continue

    internal_id = str(
        doc.get("internal_id", "")
    ).strip()

    text = str(
        doc.get("text", "")
    )

    if not internal_id:
        continue

    matches = re.finditer(
        r"Referencing IS:\s*(.+?)(?=\s*\||\s*\n|$)",
        text,
        flags=re.I
    )

    for match in matches:

        reference = match.group(1).strip()

        if reference:

            referred_by_internal_id[
                internal_id
            ].append(
                {
                    "doc": doc,
                    "reference": reference
                }
            )


# ============================================================
# 5. PRINT REAL COUNTS
# ============================================================

print(
    "\nCross-reference source standards:",
    len(cross_by_internal_id)
)

print(
    "International-reference source standards:",
    len(international_by_internal_id)
)

print(
    "Reverse-reference target standards:",
    len(referred_by_internal_id)
)


# ============================================================
# 6. RELATIONSHIP EVIDENCE
# ============================================================

def make_relationship_evidence_fixed(
    item,
    direction,
    queried_identifier
):

    doc = item["doc"]

    reference = item["reference"]

    internal_id = str(
        doc.get("internal_id", "")
    )

    dtype = str(
        doc.get("document_type", "")
    )

    # --------------------------------------------------------
    # OUTGOING
    # --------------------------------------------------------

    if direction == "outgoing":

        return {
            "evidence_type": "relationship",
            "source": dtype,
            "identifier": queried_identifier,
            "referenced_identifier": reference,
            "internal_id": internal_id,
            "title": "",
            "content": {
                "doc_id": doc.get(
                    "doc_id",
                    ""
                ),
                "internal_id": internal_id,
                "document_type": dtype,
                "text": str(
                    doc.get("text", "")
                )
            }
        }

    # --------------------------------------------------------
    # REVERSE
    # --------------------------------------------------------

    return {
        "evidence_type": "relationship",
        "source": "referred_by",

        # The referencing standard
        "identifier": reference,

        # The queried target standard
        "referenced_identifier":
            queried_identifier,

        "internal_id": internal_id,
        "title": "",

        "content": {
            "doc_id": doc.get(
                "doc_id",
                ""
            ),
            "internal_id": internal_id,
            "document_type": dtype,
            "text": str(
                doc.get("text", "")
            )
        }
    }


# ============================================================
# 7. RESOLVE STANDARD
# ============================================================

def resolve_standard_fixed(identifier):

    identifier = normalize_identifier(
        identifier
    )

    candidates = standards_by_identifier.get(
        identifier,
        []
    )

    if not candidates:
        return None

    # In our two tested cases this is exactly one.
    if len(candidates) == 1:
        return candidates[0]

    # If there are collisions, prefer the candidate with
    # relationship records.
    scored = []

    for doc in candidates:

        internal_id = str(
            doc.get("internal_id", "")
        )

        outgoing_count = (
            len(
                cross_by_internal_id.get(
                    internal_id,
                    []
                )
            )
            +
            len(
                international_by_internal_id.get(
                    internal_id,
                    []
                )
            )
        )

        reverse_count = len(
            referred_by_internal_id.get(
                internal_id,
                []
            )
        )

        scored.append(
            (
                outgoing_count +
                reverse_count,
                doc
            )
        )

    scored.sort(
        key=lambda x: x[0],
        reverse=True
    )

    return scored[0][1]


# ============================================================
# 8. FINAL RELATIONSHIP RETRIEVAL
# ============================================================

def build_relationship_evidence(
    identifier,
    direction,
    max_results=100
):

    identifier = normalize_identifier(
        identifier
    )

    standard_doc = resolve_standard_fixed(
        identifier
    )

    if standard_doc is None:
        return []

    internal_id = str(
        standard_doc.get(
            "internal_id",
            ""
        )
    )

    results = []

    # --------------------------------------------------------
    # OUTGOING
    # --------------------------------------------------------

    if direction == "outgoing":

        cross_items = (
            cross_by_internal_id.get(
                internal_id,
                []
            )
        )

        international_items = (
            international_by_internal_id.get(
                internal_id,
                []
            )
        )

        for item in cross_items:

            results.append(
                make_relationship_evidence_fixed(
                    item,
                    "outgoing",
                    identifier
                )
            )

        for item in international_items:

            results.append(
                make_relationship_evidence_fixed(
                    item,
                    "outgoing",
                    identifier
                )
            )

        return results[:max_results]

    # --------------------------------------------------------
    # REVERSE
    # --------------------------------------------------------

    if direction == "reverse":

        reverse_items = (
            referred_by_internal_id.get(
                internal_id,
                []
            )
        )

        for item in reverse_items:

            results.append(
                make_relationship_evidence_fixed(
                    item,
                    "reverse",
                    identifier
                )
            )

        return results[:max_results]

    return []


# ============================================================
# 9. REPLACE RETRIEVAL API
# ============================================================

def retrieve_evidence_api(
    query,
    semantic_top_k=5
):

    identifiers = extract_identifiers(
        query
    )

    relationship_direction = (
        detect_relationship_direction(
            query
        )
    )

    compound = detect_compound_query(
        query,
        identifiers
    )

    evidence = []
    plan = []

    # --------------------------------------------------------
    # IDENTIFIER RETRIEVAL
    # --------------------------------------------------------

    for identifier in identifiers:

        # Exact standard
        standard_doc = resolve_standard_fixed(
            identifier
        )

        if standard_doc is not None:

            evidence.append(
                standard_to_evidence(
                    standard_doc,
                    source="exact"
                )
            )

            plan.append(
                f"standard:{identifier}"
            )

        # Relationship
        if relationship_direction:

            relationship_evidence = (
                build_relationship_evidence(
                    identifier,
                    relationship_direction,
                    max_results=100
                )
            )

            evidence.extend(
                relationship_evidence
            )

            if relationship_evidence:

                if relationship_direction == "reverse":

                    plan.append(
                        f"reverse_references:{identifier}"
                    )

                else:

                    plan.append(
                        f"outgoing_references:{identifier}"
                    )

        # Compound
        elif compound:

            relationship_evidence = (
                build_relationship_evidence(
                    identifier,
                    "outgoing",
                    max_results=100
                )
            )

            evidence.extend(
                relationship_evidence
            )

            if relationship_evidence:

                plan.append(
                    f"outgoing_references:{identifier}"
                )

    # --------------------------------------------------------
    # SEMANTIC
    # --------------------------------------------------------

    if not identifiers:

        semantic_evidence = semantic_search(
            query,
            top_k=semantic_top_k
        )

        evidence.extend(
            semantic_evidence
        )

        if semantic_evidence:

            plan.append(
                "semantic"
            )

    # --------------------------------------------------------
    # DEDUPLICATION
    # --------------------------------------------------------

    deduped = []
    seen = set()

    for item in evidence:

        key = (
            item.get("evidence_type"),
            item.get("internal_id"),
            item.get("identifier"),
            item.get("referenced_identifier"),
            item.get("source")
        )

        if key in seen:
            continue

        seen.add(key)
        deduped.append(item)

    evidence = deduped

    return {
        "query": query,
        "identifiers": identifiers,
        "relationship_direction":
            relationship_direction,
        "compound": compound,
        "retrieval_plan": plan,
        "evidence": evidence,
        "evidence_count": len(evidence)
    }


# ============================================================
# 10. DIRECT COUNT CHECK
# ============================================================

print("\n")
print("=" * 75)
print("DIRECT RELATIONSHIP COUNTS")
print("=" * 75)

reverse_test = build_relationship_evidence(
    "IS/IEC 61196-1:2005",
    "reverse"
)

outgoing_test = build_relationship_evidence(
    "IS/IEC 60691:2023",
    "outgoing"
)

print(
    "\nIS/IEC 61196-1:2005 reverse:",
    len(reverse_test)
)

print(
    "IS/IEC 60691:2023 outgoing:",
    len(outgoing_test)
)


# ============================================================
# 11. FULL FINAL VALIDATION
# ============================================================

print("\n")
print("=" * 75)
print("FINAL RETRIEVAL VALIDATION")
print("=" * 75)


tests = [
    (
        "EXACT",
        "What is IS/IEC 60691:2023?",
        1
    ),
    (
        "SEMANTIC",
        "Which BIS standard deals with thermal links?",
        5
    ),
    (
        "REVERSE",
        "Which BIS standards refer to IS/IEC 61196-1:2005?",
        43
    ),
    (
        "COMPOUND",
        "Tell me about IS/IEC 60691:2023 and its references.",
        12
    )
]


all_passed = True

for name, query, expected in tests:

    result = retrieve_evidence_api(
        query
    )

    actual = result[
        "evidence_count"
    ]

    passed = actual == expected

    if not passed:
        all_passed = False

    print("\n" + "-" * 75)

    print(
        f"{'✅ PASS' if passed else '❌ FAIL'} — {name}"
    )

    print(
        "Expected:",
        expected
    )

    print(
        "Actual:",
        actual
    )

    print(
        "Plan:",
        result["retrieval_plan"]
    )


print("\n")
print("=" * 75)

if all_passed:

    print("🎉🎉🎉 ALL RETRIEVAL TESTS PASSED 🎉🎉🎉")
    print()
    print("Exact    : 1/1")
    print("Semantic : 5/5")
    print("Reverse  : 43/43")
    print("Compound : 12/12")

else:

    print("❌ RETRIEVAL VALIDATION STILL FAILS")

print("=" * 75)
```

```python
# ============================================================
# SAATHI — FINAL DEDUPLICATION FIX
# ============================================================
# DO NOT TOUCH:
#   corpus
#   FAISS
#   embeddings
#   relationship parser
#   Gemini
#
# The relationship parser is now CORRECT:
#   reverse = 42
#   outgoing = 11
#
# The problem is that generic deduplication is removing
# legitimate relationship evidence.
# ============================================================

print("=" * 75)
print("SAATHI — FINAL DEDUPLICATION FIX")
print("=" * 75)


def deduplicate_evidence(evidence):

    result = []
    seen = set()

    for item in evidence:

        evidence_type = item.get(
            "evidence_type"
        )

        # ====================================================
        # STANDARD EVIDENCE
        # ====================================================

        if evidence_type == "standard":

            key = (
                "standard",
                item.get("internal_id")
            )

        # ====================================================
        # RELATIONSHIP EVIDENCE
        # ====================================================

        elif evidence_type == "relationship":

            # Every relationship row is a separate piece of
            # evidence. The internal_id alone is NOT unique
            # because one aggregated relationship document can
            # contain many references.
            key = (
                "relationship",

                item.get("source"),

                item.get("internal_id"),

                item.get(
                    "identifier"
                ),

                item.get(
                    "referenced_identifier"
                )
            )

        # ====================================================
        # FALLBACK
        # ====================================================

        else:

            key = (
                evidence_type,
                item.get("internal_id"),
                item.get("identifier"),
                item.get("referenced_identifier"),
                item.get("source")
            )

        if key in seen:
            continue

        seen.add(key)
        result.append(item)

    return result


# ============================================================
# REPLACE ONLY THE RETRIEVAL FUNCTION
# ============================================================

def retrieve_evidence_api(
    query,
    semantic_top_k=5
):

    identifiers = extract_identifiers(
        query
    )

    relationship_direction = (
        detect_relationship_direction(
            query
        )
    )

    compound = detect_compound_query(
        query,
        identifiers
    )

    evidence = []
    plan = []

    # ========================================================
    # IDENTIFIER RETRIEVAL
    # ========================================================

    for identifier in identifiers:

        # ----------------------------------------------------
        # Exact standard
        # ----------------------------------------------------

        standard_doc = resolve_standard_fixed(
            identifier
        )

        if standard_doc is not None:

            evidence.append(
                standard_to_evidence(
                    standard_doc,
                    source="exact"
                )
            )

            plan.append(
                f"standard:{identifier}"
            )

        # ----------------------------------------------------
        # Relationship query
        # ----------------------------------------------------

        if relationship_direction:

            relationship_evidence = (
                build_relationship_evidence(
                    identifier,
                    relationship_direction,
                    max_results=100
                )
            )

            evidence.extend(
                relationship_evidence
            )

            if relationship_evidence:

                if relationship_direction == "reverse":

                    plan.append(
                        f"reverse_references:{identifier}"
                    )

                else:

                    plan.append(
                        f"outgoing_references:{identifier}"
                    )

        # ----------------------------------------------------
        # Compound query
        # ----------------------------------------------------

        elif compound:

            relationship_evidence = (
                build_relationship_evidence(
                    identifier,
                    "outgoing",
                    max_results=100
                )
            )

            evidence.extend(
                relationship_evidence
            )

            if relationship_evidence:

                plan.append(
                    f"outgoing_references:{identifier}"
                )

    # ========================================================
    # SEMANTIC RETRIEVAL
    # ========================================================

    if not identifiers:

        semantic_evidence = semantic_search(
            query,
            top_k=semantic_top_k
        )

        evidence.extend(
            semantic_evidence
        )

        if semantic_evidence:

            plan.append(
                "semantic"
            )

    # ========================================================
    # SAFE DEDUPLICATION
    # ========================================================

    evidence = deduplicate_evidence(
        evidence
    )

    return {
        "query": query,

        "identifiers":
            identifiers,

        "relationship_direction":
            relationship_direction,

        "compound":
            compound,

        "retrieval_plan":
            plan,

        "evidence":
            evidence,

        "evidence_count":
            len(evidence)
    }


# ============================================================
# DIRECT RAW COUNTS
# ============================================================

print("\n" + "=" * 75)
print("RAW RELATIONSHIP COUNTS")
print("=" * 75)

reverse_raw = build_relationship_evidence(
    "IS/IEC 61196-1:2005",
    "reverse",
    max_results=100
)

outgoing_raw = build_relationship_evidence(
    "IS/IEC 60691:2023",
    "outgoing",
    max_results=100
)

print(
    "61196-1 reverse records:",
    len(reverse_raw)
)

print(
    "60691:2023 outgoing records:",
    len(outgoing_raw)
)


# ============================================================
# FINAL TESTS
# ============================================================

print("\n" + "=" * 75)
print("SAATHI — FINAL RETRIEVAL VALIDATION")
print("=" * 75)

tests = [

    (
        "EXACT",
        "What is IS/IEC 60691:2023?",
        1
    ),

    (
        "SEMANTIC",
        "Which BIS standard deals with thermal links?",
        5
    ),

    (
        "REVERSE",
        "Which BIS standards refer to IS/IEC 61196-1:2005?",
        43
    ),

    (
        "COMPOUND",
        "Tell me about IS/IEC 60691:2023 and its references.",
        12
    )
]

all_passed = True

for name, query, expected in tests:

    result = retrieve_evidence_api(
        query
    )

    actual = result[
        "evidence_count"
    ]

    passed = (
        actual == expected
    )

    if not passed:
        all_passed = False

    print("\n" + "-" * 75)

    print(
        f"{'✅ PASS' if passed else '❌ FAIL'} — {name}"
    )

    print(
        "Expected:",
        expected
    )

    print(
        "Actual:",
        actual
    )

    print(
        "Identifiers:",
        result["identifiers"]
    )

    print(
        "Relationship:",
        result["relationship_direction"]
    )

    print(
        "Plan:",
        result["retrieval_plan"]
    )


# ============================================================
# FINAL RESULT
# ============================================================

print("\n" + "=" * 75)

if all_passed:

    print("🎉🎉🎉 ALL RETRIEVAL TESTS PASSED 🎉🎉🎉")

    print("""
Exact       : 1/1
Semantic    : 5/5
Reverse     : 43/43
Compound    : 12/12
""")

else:

    print(
        "❌ Retrieval validation still has a failure."
    )

print("=" * 75)
```

```python
# ============================================================
# SAATHI — DEFINITIVE RELATIONSHIP DEDUP FIX
# ============================================================

print("=" * 75)
print("SAATHI — DEFINITIVE RELATIONSHIP DEDUP FIX")
print("=" * 75)


def deduplicate_evidence(evidence):

    """
    Deduplicate ONLY standard evidence.

    Relationship evidence is intentionally NEVER deduplicated.

    Why:
        One aggregated BIS relationship document can contain
        multiple legitimate relationship rows.

        The relationship parser has already separated those rows
        into individual evidence records.

        Therefore:
            42 reverse relationship records
            + 1 standard record
            = 43 evidence records

        Generic relationship deduplication can incorrectly
        collapse legitimate BIS relationships.
    """

    result = []

    seen_standards = set()

    for item in evidence:

        evidence_type = item.get(
            "evidence_type"
        )

        # ====================================================
        # STANDARD
        # ====================================================

        if evidence_type == "standard":

            key = (
                item.get("internal_id")
                or item.get("identifier")
                or item.get("is_number")
            )

            if key in seen_standards:
                continue

            seen_standards.add(key)

            result.append(item)

        # ====================================================
        # RELATIONSHIP
        # ====================================================

        else:

            # IMPORTANT:
            # NEVER deduplicate relationship evidence.
            result.append(item)

    return result


# ============================================================
# REBUILD RETRIEVAL FUNCTION
# ============================================================

def retrieve_evidence_api(
    query,
    semantic_top_k=5
):

    identifiers = extract_identifiers(
        query
    )

    relationship_direction = (
        detect_relationship_direction(
            query
        )
    )

    compound = detect_compound_query(
        query,
        identifiers
    )

    evidence = []
    plan = []

    # ========================================================
    # IDENTIFIER RETRIEVAL
    # ========================================================

    for identifier in identifiers:

        standard_doc = resolve_standard_fixed(
            identifier
        )

        if standard_doc is not None:

            evidence.append(
                standard_to_evidence(
                    standard_doc,
                    source="exact"
                )
            )

            plan.append(
                f"standard:{identifier}"
            )

        # ====================================================
        # RELATIONSHIP QUERY
        # ====================================================

        if relationship_direction:

            relationship_evidence = (
                build_relationship_evidence(
                    identifier,
                    relationship_direction,
                    max_results=100
                )
            )

            evidence.extend(
                relationship_evidence
            )

            if relationship_evidence:

                if relationship_direction == "reverse":

                    plan.append(
                        f"reverse_references:{identifier}"
                    )

                else:

                    plan.append(
                        f"outgoing_references:{identifier}"
                    )

        # ====================================================
        # COMPOUND QUERY
        # ====================================================

        elif compound:

            relationship_evidence = (
                build_relationship_evidence(
                    identifier,
                    "outgoing",
                    max_results=100
                )
            )

            evidence.extend(
                relationship_evidence
            )

            if relationship_evidence:

                plan.append(
                    f"outgoing_references:{identifier}"
                )

    # ========================================================
    # SEMANTIC RETRIEVAL
    # ========================================================

    if not identifiers:

        semantic_evidence = semantic_search(
            query,
            top_k=semantic_top_k
        )

        evidence.extend(
            semantic_evidence
        )

        if semantic_evidence:

            plan.append(
                "semantic"
            )

    # ========================================================
    # SAFE DEDUPLICATION
    # ========================================================

    evidence = deduplicate_evidence(
        evidence
    )

    return {
        "query": query,

        "identifiers":
            identifiers,

        "relationship_direction":
            relationship_direction,

        "compound":
            compound,

        "retrieval_plan":
            plan,

        "evidence":
            evidence,

        "evidence_count":
            len(evidence)
    }


# ============================================================
# VERIFY RAW COUNTS FIRST
# ============================================================

reverse_raw = build_relationship_evidence(
    "IS/IEC 61196-1:2005",
    "reverse",
    max_results=100
)

outgoing_raw = build_relationship_evidence(
    "IS/IEC 60691:2023",
    "outgoing",
    max_results=100
)

print("\nRAW RELATIONSHIP COUNTS")
print("-" * 75)

print(
    "IS/IEC 61196-1:2005 reverse:",
    len(reverse_raw)
)

print(
    "IS/IEC 60691:2023 outgoing:",
    len(outgoing_raw)
)


# ============================================================
# FINAL VALIDATION
# ============================================================

tests = [

    (
        "EXACT",
        "What is IS/IEC 60691:2023?",
        1
    ),

    (
        "SEMANTIC",
        "Which BIS standard deals with thermal links?",
        5
    ),

    (
        "REVERSE",
        "Which BIS standards refer to IS/IEC 61196-1:2005?",
        43
    ),

    (
        "COMPOUND",
        "Tell me about IS/IEC 60691:2023 and its references.",
        12
    )
]


print("\n" + "=" * 75)
print("FINAL RETRIEVAL VALIDATION")
print("=" * 75)

all_passed = True

for name, query, expected in tests:

    result = retrieve_evidence_api(
        query
    )

    actual = result[
        "evidence_count"
    ]

    passed = (
        actual == expected
    )

    if not passed:
        all_passed = False

    print("\n" + "-" * 75)

    print(
        f"{'✅ PASS' if passed else '❌ FAIL'} — {name}"
    )

    print(
        "Expected:",
        expected
    )

    print(
        "Actual:",
        actual
    )

    print(
        "Identifiers:",
        result["identifiers"]
    )

    print(
        "Relationship:",
        result["relationship_direction"]
    )

    print(
        "Plan:",
        result["retrieval_plan"]
    )


print("\n" + "=" * 75)

if all_passed:

    print("🎉 ALL RETRIEVAL VALIDATION PASSED")
    print()
    print("EXACT    : 1/1")
    print("SEMANTIC : 5/5")
    print("REVERSE  : 43/43")
    print("COMPOUND : 12/12")

else:

    print("❌ RETRIEVAL VALIDATION FAILED")

print("=" * 75)
```

```python
# ============================================================
# SAATHI — FINAL END-TO-END RAG TEST
# ============================================================
#
# Pipeline:
# Query
#   ↓
# Validated Retrieval
#   ↓
# BIS Evidence
#   ↓
# Gemini 3.8 Flash
#   ↓
# Grounded Answer
#   ↓
# Citation Validation
#   ↓
# FINAL PASS / FAIL
#
# IMPORTANT:
# This cell does NOT modify the retrieval layer.
# ============================================================

import json
import re
from datetime import datetime, timezone

print("=" * 75)
print("SAATHI — FINAL END-TO-END RAG TEST")
print("=" * 75)


# ============================================================
# 1. CHECK REQUIRED COMPONENTS
# ============================================================

required_functions = [
    "retrieve_evidence_api",
    "generate_grounded_answer",
]

missing = [
    name
    for name in required_functions
    if name not in globals()
]

if missing:
    raise RuntimeError(
        f"Missing required functions: {missing}. "
        "Do not modify retrieval; restore the final pipeline first."
    )

print("\n✅ Retrieval function available")
print("✅ Gemini generation function available")


# ============================================================
# 2. CHECK GEMINI CONFIGURATION
# ============================================================

print("\n" + "-" * 75)
print("GEMINI CONFIGURATION")
print("-" * 75)

print(
    "Model:",
    globals().get(
        "GEMINI_MODEL",
        "NOT SET"
    )
)

if "gemini_client" not in globals():
    raise RuntimeError(
        "Gemini client is not initialized."
    )

print("✅ Gemini client available")


# ============================================================
# 3. FINAL TEST QUERIES
# ============================================================

test_cases = [

    {
        "name": "EXACT",
        "query": "What is IS/IEC 60691:2023?",
        "expected_evidence": 1,
        "expected_relationship": None,
    },

    {
        "name": "SEMANTIC",
        "query": "Which BIS standard deals with thermal links?",
        "expected_evidence": 5,
        "expected_relationship": None,
    },

    {
        "name": "REVERSE",
        "query": (
            "Which BIS standards refer to "
            "IS/IEC 61196-1:2005?"
        ),
        "expected_evidence": 43,
        "expected_relationship": "reverse",
    },

    {
        "name": "COMPOUND",
        "query": (
            "Tell me about IS/IEC 60691:2023 "
            "and its references."
        ),
        "expected_evidence": 12,
        "expected_relationship": "outgoing",
    },
]


# ============================================================
# 4. CITATION VALIDATOR
# ============================================================

def validate_generated_answer(
    answer,
    evidence
):

    if not answer:
        return {
            "passed": False,
            "reason": "Empty answer",
            "citations": [],
            "invalid_citations": [],
        }

    citations = sorted(
        set(
            int(x)
            for x in re.findall(
                r"\[E(\d+)\]",
                answer
            )
        )
    )

    invalid = [
        n
        for n in citations
        if n < 1 or n > len(evidence)
    ]

    # Check that every cited evidence item actually exists.
    valid_mapping = []

    for n in citations:

        if 1 <= n <= len(evidence):

            item = evidence[n - 1]

            valid_mapping.append({
                "citation": f"[E{n}]",
                "evidence_type":
                    item.get("evidence_type"),
                "identifier":
                    item.get("identifier"),
                "title":
                    item.get("title"),
            })

    passed = (
        len(answer.strip()) > 0
        and len(citations) > 0
        and len(invalid) == 0
    )

    return {
        "passed": passed,
        "reason":
            "OK" if passed
            else "Citation validation failed",
        "citations": citations,
        "invalid_citations": invalid,
        "citation_mapping": valid_mapping,
    }


# ============================================================
# 5. RUN FINAL END-TO-END TESTS
# ============================================================

results = []

for test in test_cases:

    name = test["name"]
    query = test["query"]

    print("\n")
    print("=" * 75)
    print(name)
    print("=" * 75)

    print("\nQuery:")
    print(query)

    # --------------------------------------------------------
    # RETRIEVAL
    # --------------------------------------------------------

    retrieval = retrieve_evidence_api(
        query
    )

    evidence = retrieval.get(
        "evidence",
        []
    )

    evidence_count = len(
        evidence
    )

    retrieval_pass = (
        evidence_count
        == test["expected_evidence"]
    )

    relationship_pass = (
        retrieval.get(
            "relationship_direction"
        )
        == test["expected_relationship"]
    )

    print(
        "\nRetrieval:",
        "✅ PASS"
        if retrieval_pass
        else "❌ FAIL"
    )

    print(
        "Expected evidence:",
        test["expected_evidence"]
    )

    print(
        "Actual evidence:",
        evidence_count
    )

    print(
        "Identifiers:",
        retrieval.get("identifiers")
    )

    print(
        "Relationship:",
        retrieval.get(
            "relationship_direction"
        )
    )

    print(
        "Plan:",
        retrieval.get(
            "retrieval_plan"
        )
    )

    # --------------------------------------------------------
    # GEMINI
    # --------------------------------------------------------

    generation_pass = False
    answer = ""
    generation_error = None

    if retrieval_pass:

        try:

            generation = generate_grounded_answer(
                query,
                evidence
            )

            # Handle both common return formats.
            if isinstance(
                generation,
                dict
            ):

                answer = (
                    generation.get("answer")
                    or generation.get("text")
                    or generation.get(
                        "response",
                        ""
                    )
                )

                generation_error = (
                    generation.get("error")
                    or generation.get(
                        "generation_error"
                    )
                )

            else:

                answer = str(
                    generation
                )

            generation_pass = (
                bool(
                    answer
                    and answer.strip()
                )
                and not generation_error
            )

        except Exception as exc:

            generation_error = (
                f"{type(exc).__name__}: {exc}"
            )

    print(
        "\nGemini generation:",
        "✅ PASS"
        if generation_pass
        else "❌ FAIL"
    )

    if generation_error:
        print(
            "Generation error:",
            generation_error
        )

    # --------------------------------------------------------
    # CITATIONS
    # --------------------------------------------------------

    citation_result = (
        validate_generated_answer(
            answer,
            evidence
        )
        if generation_pass
        else {
            "passed": False,
            "reason":
                "Skipped because generation failed",
            "citations": [],
            "invalid_citations": [],
        }
    )

    print(
        "\nCitation validation:",
        "✅ PASS"
        if citation_result["passed"]
        else "❌ FAIL"
    )

    print(
        "Citations:",
        citation_result["citations"]
    )

    if citation_result.get(
        "invalid_citations"
    ):
        print(
            "Invalid citations:",
            citation_result[
                "invalid_citations"
            ]
        )

    # --------------------------------------------------------
    # DISPLAY ANSWER
    # --------------------------------------------------------

    print("\nGenerated answer:")
    print("-" * 75)

    if answer:
        print(answer)

    else:
        print(
            "[No answer generated]"
        )

    print("-" * 75)

    # --------------------------------------------------------
    # TEST RESULT
    # --------------------------------------------------------

    passed = (
        retrieval_pass
        and relationship_pass
        and generation_pass
        and citation_result["passed"]
    )

    print(
        "\nFINAL RESULT:",
        "✅ PASS"
        if passed
        else "❌ FAIL"
    )

    results.append({
        "name": name,
        "query": query,
        "retrieval_pass":
            retrieval_pass,
        "relationship_pass":
            relationship_pass,
        "generation_pass":
            generation_pass,
        "citation_pass":
            citation_result["passed"],
        "evidence_count":
            evidence_count,
        "citations":
            citation_result["citations"],
        "generation_error":
            generation_error,
        "answer":
            answer,
        "passed":
            passed,
    })


# ============================================================
# 6. FINAL SUMMARY
# ============================================================

passed_count = sum(
    r["passed"]
    for r in results
)

total_count = len(
    results
)

print("\n\n")
print("=" * 75)
print("SAATHI — FINAL END-TO-END RESULT")
print("=" * 75)

for result in results:

    print(
        f"{'✅ PASS' if result['passed'] else '❌ FAIL'} "
        f"{result['name']}"
    )

    print(
        f"   Retrieval : "
        f"{'PASS' if result['retrieval_pass'] else 'FAIL'}"
    )

    print(
        f"   Gemini    : "
        f"{'PASS' if result['generation_pass'] else 'FAIL'}"
    )

    print(
        f"   Citations : "
        f"{'PASS' if result['citation_pass'] else 'FAIL'}"
    )


print("\n" + "=" * 75)

if passed_count == total_count:

    print(
        "🎉 ALL END-TO-END TESTS PASSED"
    )

    print()
    print(
        "SAATHI RAG PIPELINE IS VALIDATED."
    )

    print()
    print(
        "Retrieval      : PASS"
    )

    print(
        "Gemini         : PASS"
    )

    print(
        "Grounding      : PASS"
    )

    print(
        "Citations      : PASS"
    )

    print(
        "Relationship   : PASS"
    )

else:

    print(
        f"❌ END-TO-END VALIDATION FAILED "
        f"({passed_count}/{total_count})"
    )

print("=" * 75)


# ============================================================
# 7. SAVE FINAL REPORT
# ============================================================

final_report = {
    "timestamp":
        datetime.now(
            timezone.utc
        ).isoformat(),

    "model":
        globals().get(
            "GEMINI_MODEL"
        ),

    "tests":
        results,

    "passed":
        passed_count,

    "total":
        total_count,

    "overall_pass":
        passed_count == total_count,
}


final_path = (
    "/kaggle/working/"
    "saathi_final_end_to_end_report.json"
)

with open(
    final_path,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        final_report,
        f,
        indent=2,
        ensure_ascii=False
    )


print(
    "\nFinal report saved:"
)

print(
    final_path
)
```

```python
# ============================================================
# SAATHI — AIML / RAG ONLY PACKAGE
# NO BACKEND
# NO FRONTEND
# NO HTML
# ============================================================

import os
import shutil

AIML_DIR = "/kaggle/working/SAATHI_AIML_ONLY"

# Clean old package if it exists
if os.path.exists(AIML_DIR):
    shutil.rmtree(AIML_DIR)

os.makedirs(AIML_DIR, exist_ok=True)


# ------------------------------------------------------------
# CORE RAG ARTIFACTS
# ------------------------------------------------------------

files = [
    "/kaggle/working/bis_faiss.index",
    "/kaggle/working/bis_rag_metadata.json",
    "/kaggle/working/bis_rag_corpus.jsonl",
]


# ------------------------------------------------------------
# FINAL VALIDATION REPORT
# ------------------------------------------------------------

files.append(
    "/kaggle/working/saathi_final_end_to_end_report.json"
)


# ------------------------------------------------------------
# COPY EXISTING AIML FILES
# ------------------------------------------------------------

for src in files:

    if os.path.exists(src):

        shutil.copy2(
            src,
            AIML_DIR
        )

        print(
            "✓ Added:",
            os.path.basename(src)
        )

    else:

        print(
            "⚠️ Missing:",
            src
        )


# ------------------------------------------------------------
# CREATE ZIP
# ------------------------------------------------------------

zip_base = "/kaggle/working/SAATHI_AIML_ONLY"

zip_path = shutil.make_archive(
    zip_base,
    "zip",
    AIML_DIR
)


# ------------------------------------------------------------
# FINAL CHECK
# ------------------------------------------------------------

size_mb = (
    os.path.getsize(zip_path)
    / (1024 * 1024)
)

print("\n" + "=" * 70)
print("SAATHI AIML PACKAGE READY")
print("=" * 70)

print("\nZIP:")
print(zip_path)

print(
    f"\nSIZE: {size_mb:.2f} MB"
)

print("\nCONTENTS:")

for filename in sorted(
    os.listdir(AIML_DIR)
):

    size = (
        os.path.getsize(
            os.path.join(
                AIML_DIR,
                filename
            )
        )
        / (1024 * 1024)
    )

    print(
        f"  ✓ {filename:<45} {size:.2f} MB"
    )

print("\n" + "=" * 70)
print("DOWNLOAD FROM KAGGLE FILES PANEL:")
print("/kaggle/working/SAATHI_AIML_ONLY.zip")
print("=" * 70)
```
